# Divided differences

Divided differences describe how tabulated values change on distinct nodes with arbitrary spacing. We construct their table, derive a formula for arbitrary data, and relate the entries to interpolation and derivatives. On a uniform grid they also give the [finite differences](../01.2/lecture01_2_eng.ipynb) studied in the next lecture.

We write $f\in C^r(J)$ if the derivatives through order $r$ are continuous on $J$. At an endpoint we use one-sided derivatives.

## The divided-difference table

**Definition 1 (divided differences).** Let $x_0<\cdots<x_n$ be distinct nodes with prescribed values $y_j$, where $n\geq0$. Write $f[x_i]=y_i$. The first-order divided difference is

$$
f[x_i,x_{i+1}]=\frac{y_{i+1}-y_i}{x_{i+1}-x_i}.\tag{1}
$$

For $k\geq2$, define the divided differences recursively by

$$
f[x_i,\ldots,x_{i+k}]=\frac{f[x_{i+1},\ldots,x_{i+k}]-f[x_i,\ldots,x_{i+k-1}]}{x_{i+k}-x_i}.\tag{2}
$$

Here $1\leq k\leq n$ and $0\leq i\leq n-k$. When the data come from a function $f$, we have $y_j=f(x_j)$. The recursion uses only the nodes and the prescribed values.

Each entry is placed between the two from which it is computed. Table 1 shows the layout for four nodes.

***Table 1.*** *Staggered divided differences for four distinct nodes*

$$
\begin{array}{c|c|c|c|c}
x & y & k=1 & k=2 & k=3 \\
\hline
x_0 & y_0 & & & \\
& & f[x_0,x_1] & & \\
x_1 & y_1 & & f[x_0,x_1,x_2] & \\
& & f[x_1,x_2] & & f[x_0,x_1,x_2,x_3] \\
x_2 & y_2 & & f[x_1,x_2,x_3] & \\
& & f[x_2,x_3] & & \\
x_3 & y_3 & & &
\end{array}
$$

For $n+1$ nodes with data rows numbered $0,2,\ldots,2n$, the entry $f[x_i,\ldots,x_{i+k}]$ is at row $2i+k$. The last column contains just $f[x_0,\ldots,x_n]$, at row $n$. The full table contains $n(n+1)/2$ differences of positive order, each requiring a subtraction and a division once the node spacings are known.

**Example 1 (unequally spaced nodes).** For $f(t)=t^2$, take the nodes $0,1,3,4$ and values $0,1,9,16$. The first-order differences are $1,4,7$. The next two columns give

$$
f[0,1,3]=\frac{4-1}{3-0}=1,\quad f[1,3,4]=\frac{7-4}{4-1}=1,\quad f[0,1,3,4]=\frac{1-1}{4-0}=0.\tag{3}
$$

The second-order differences equal the leading coefficient of $t^2$ and the third-order difference vanishes. The next proposition explains this behavior for polynomial data.

## Explicit formula and interpolation

**Proposition 1 (explicit formula and symmetry).** For distinct nodes $x_i,\ldots,x_{i+k}$,

$$
f[x_i,\ldots,x_{i+k}]=\sum_{j=i}^{i+k}
\frac{y_j}{\displaystyle\prod_{\substack{i\leq l\leq i+k\\l\ne j}}(x_j-x_l)}.\tag{4}
$$

where an empty product equals $1$. The result is invariant under permutations of the node–value pairs.

**Proof.** For $k=0$ this is $f[x_i]=y_i$. At the induction step, apply (2) to the formulas for the two differences of order $k-1$ (for $k=1$ use (1)). With the common product in the denominator in (4), the coefficient of an interior value $y_j$ has numerator $(x_j-x_i)-(x_j-x_{i+k})=x_{i+k}-x_i$. At each endpoint just one of the two shorter differences contributes and yields the same factor. It cancels the denominator of the recurrence. Since the resulting expression is symmetric in the node–value pairs, the recursion gives the same divided difference in any order. $\square$

**Corollary 1 (coefficient of the interpolant).** The divided difference $f[x_i,\ldots,x_{i+k}]$ is the coefficient of $t^k$ in the interpolation polynomial

$$
Q(t)=\sum_{j=i}^{i+k}y_j
\prod_{\substack{i\leq l\leq i+k\\l\ne j}}\frac{t-x_l}{x_j-x_l}.\tag{5}
$$

**Proof.** At $t=x_j$ only the $j$th term contributes, and $Q(x_j)=y_j$. Each product of numerator factors is monic of degree $k$. Hence its contribution to the coefficient of $t^k$ is the corresponding term in (4). $\square$

**Remark 1 (polynomials and the Newton form).** For data from a polynomial of degree at most $k$, the polynomial (5) reproduces it exactly. The divided difference of order $k$ then equals its coefficient of $t^k$. In the Newton form this is the coefficient multiplying the monic product $(t-x_i)\cdots(t-x_{i+k-1})$.

## Mean value and coincident nodes

**Theorem 1 (mean value of a divided difference).** Let $k\geq1$, let $x_i<\cdots<x_{i+k}$ be distinct nodes, and let $f\in C^k([x_i,x_{i+k}])$. The divided differences here use the exact values $y_j=f(x_j)$. There is a point $\xi$ such that

$$
f[x_i,\ldots,x_{i+k}]=\frac{f^{(k)}(\xi)}{k!},\quad \xi\in(x_i,x_{i+k}).\tag{6}
$$

**Proof.** Let $Q$ be the interpolation polynomial (5). The function $f-Q$ has $k+1$ distinct zeros at the nodes. Applying Rolle's theorem $k$ times gives $(f-Q)^{(k)}(\xi)=0$ at an interior point. By Corollary 1, $Q^{(k)}(\xi)=k!f[x_i,\ldots,x_{i+k}]$, which gives (6). $\square$

**Remark 2 (coincident nodes).** Suppose $f\in C^k(J)$ and all $k+1$ distinct nodes in $J$ approach the same point $x\in J$. The point $\xi$ in (6) tends to $x$, so the limit exists and defines

$$
f[\underbrace{x,\ldots,x}_{k+1}]=\frac{f^{(k)}(x)}{k!}.\tag{7}
$$

For $k=0$ this is just $f[x]=f(x)$.
