# Разделенные разности

Разделенные разности описывают изменение табличных значений в различных узлах с произвольным расстоянием между ними. Построим таблицу разделенных разностей, выведем явную формулу для произвольных данных и установим связь с интерполяцией и производными. На равномерной сетке из них также получаются [конечные разности](../01.2/lecture04_rus.ipynb), которые изучаются в следующей лекции.

Запись $f\in C^r(J)$ означает, что производные до порядка $r$ включительно непрерывны на $J$. На концах отрезка берем односторонние производные.

## Таблица разделенных разностей

**Определение 1 (разделенные разности).** Пусть $x_0<\cdots<x_n$ — различные узлы с заданными значениями $y_j$, где $n\geqslant0$. Положим $f[x_i]=y_i$. Разделенная разность первого порядка равна

$$
f[x_i,x_{i+1}]=\frac{y_{i+1}-y_i}{x_{i+1}-x_i}.\tag{1}
$$

При $k\geqslant2$ определим разделенные разности рекуррентно

$$
f[x_i,\ldots,x_{i+k}]=\frac{f[x_{i+1},\ldots,x_{i+k}]-f[x_i,\ldots,x_{i+k-1}]}{x_{i+k}-x_i}.\tag{2}
$$

Здесь $1\leqslant k\leqslant n$ и $0\leqslant i\leqslant n-k$. Если данные получены из функции $f$, то $y_j=f(x_j)$. Для рекурсии нужны только узлы и заданные в них значения.

Каждая разность помещается между двумя значениями, из которых она вычисляется. В таблице 1 показано расположение для четырех узлов.

***Таблица 1.*** *Разделенные разности для четырех различных узлов*

$$
\begin{array}{c|c|c|c|c}
x & y & k=1 & k=2 & k=3 \\
\hline
x_0 & y_0 & & & \\
& & f[x_0,x_1] & & \\
x_1 & y_1 & & f[x_0,x_1,x_2] & \\
& & f[x_1,x_2] & & f[x_0,x_1,x_2,x_3] \\
x_2 & y_2 & & f[x_1,x_2,x_3] & \\
& & f[x_2,x_3] & & \\
x_3 & y_3 & & &
\end{array}
$$

Для $n+1$ узлов пронумеруем строки данных как $0,2,\ldots,2n$. Тогда разность $f[x_i,\ldots,x_{i+k}]$ стоит в строке $2i+k$. В последнем столбце есть только $f[x_0,\ldots,x_n]$, в строке $n$. Полная таблица содержит $n(n+1)/2$ разностей положительного порядка. Если расстояния между узлами уже известны, для каждой новой разности нужны одно вычитание и одно деление.

**Пример 1 (неравноотстоящие узлы).** Для $f(t)=t^2$ возьмем узлы $0,1,3,4$ со значениями $0,1,9,16$. Разности первого порядка равны $1,4,7$. В следующих двух столбцах получаем

$$
f[0,1,3]=\frac{4-1}{3-0}=1,\quad f[1,3,4]=\frac{7-4}{4-1}=1,\quad f[0,1,3,4]=\frac{1-1}{4-0}=0.\tag{3}
$$

Разности второго порядка равны старшему коэффициенту $t^2$, а разность третьего порядка равна нулю. Следующее утверждение объясняет это свойство для многочленов.

## Явная формула и интерполяция

**Утверждение 1 (явная формула и симметрия).** Для различных узлов $x_i,\ldots,x_{i+k}$

$$
f[x_i,\ldots,x_{i+k}]=\sum_{j=i}^{i+k}
\frac{y_j}{\displaystyle\prod_{\substack{i\leqslant l\leqslant i+k\\l\ne j}}(x_j-x_l)}.\tag{4}
$$

где пустое произведение равно $1$. Результат не меняется при перестановках пар «узел — значение».

**Доказательство.** При $k=0$ это равенство $f[x_i]=y_i$. В шаге индукции подставим в (2) формулы для двух разностей порядка $k-1$ (при $k=1$ используем (1)). После приведения к общему произведению в знаменателе в (4) числитель коэффициента при внутреннем значении $y_j$ равен $(x_j-x_i)-(x_j-x_{i+k})=x_{i+k}-x_i$. Для каждого крайнего узла вклад дает только одна из двух разностей с тем же множителем. Он сокращается со знаменателем рекуррентной формулы. Полученное выражение симметрично относительно пар «узел — значение», поэтому рекурсия дает ту же разность при любой перестановке узлов. $\square$

**Следствие 1 (коэффициент интерполяционного многочлена).** Разделенная разность $f[x_i,\ldots,x_{i+k}]$ равна коэффициенту при $t^k$ в интерполяционном многочлене

$$
Q(t)=\sum_{j=i}^{i+k}y_j
\prod_{\substack{i\leqslant l\leqslant i+k\\l\ne j}}\frac{t-x_l}{x_j-x_l}.\tag{5}
$$

**Доказательство.** В точке $t=x_j$ ненулевой вклад может дать только $j$-е слагаемое, и $Q(x_j)=y_j$. Каждое произведение в числителе имеет степень $k$ и старший коэффициент $1$. Поэтому его вклад в коэффициент при $t^k$ совпадает с соответствующим слагаемым в (4). $\square$

**Замечание 1 (многочлены и форма Ньютона).** Для данных многочлена степени не выше $k$ многочлен (5) воспроизводит его точно. Тогда разделенная разность порядка $k$ равна коэффициенту при $t^k$. В форме Ньютона эта разность стоит перед произведением $(t-x_i)\cdots(t-x_{i+k-1})$, старший коэффициент которого равен $1$.

## Теорема о среднем и совпадающие узлы

**Теорема 1 (теорема о среднем для разделенной разности).** Пусть $k\geqslant1$, узлы $x_i<\cdots<x_{i+k}$ различны и $f\in C^k([x_i,x_{i+k}])$. Разделенные разности берутся от точных значений $y_j=f(x_j)$. Существует точка $\xi$, для которой

$$
f[x_i,\ldots,x_{i+k}]=\frac{f^{(k)}(\xi)}{k!},\quad \xi\in(x_i,x_{i+k}).\tag{6}
$$

**Доказательство.** Пусть $Q$ — интерполяционный многочлен (5). Функция $f-Q$ имеет $k+1$ различных нулей в узлах. Применяя теорему Ролля $k$ раз, получаем $(f-Q)^{(k)}(\xi)=0$ во внутренней точке. По следствию 1 имеем $Q^{(k)}(\xi)=k!f[x_i,\ldots,x_{i+k}]$, откуда следует (6). $\square$

**Замечание 2 (совпадающие узлы).** Пусть $f\in C^k(J)$ и все $k+1$ различных узлов из $J$ стремятся к одной точке $x\in J$. Точка $\xi$ из (6) стремится к $x$, поэтому в пределе получаем

$$
f[\underbrace{x,\ldots,x}_{k+1}]=\frac{f^{(k)}(x)}{k!}.\tag{7}
$$

При $k=0$ это означает $f[x]=f(x)$.
