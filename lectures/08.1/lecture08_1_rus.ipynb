# Краевые задачи для одномерного стационарного уравнения теплопроводности

## Дифференциальное уравнение и краевые условия

Рассмотрим дифференциальное уравнение второго порядка

$$
-\frac{d}{dx}\left(k(x)\frac{du}{dx}\right)+q(x)u(x)=f(x),\quad a<x<b. \tag{1}
$$

Его называют *одномерным стационарным уравнением теплопроводности*. Например, оно описывает стационарное распределение температуры $u(x)$ в теплопроводящем стержне длины $l=b-a$. Здесь $k(x)$ — коэффициент теплопроводности, $w(x)=-k(x)u'(x)$ — плотность теплового потока в направлении возрастания $x$, а $q(x)u(x)$ описывает отвод тепла, пропорциональный температуре относительно опорного уровня. Функция $f(x)$ описывает внешний источник тепла при $f(x)>0$ и сток при $f(x)<0$.

Уравнение (1) описывает также стационарную диффузию в пористых средах, где $u(x)$ — концентрация диффундирующего вещества. В одногрупповой модели диффузии нейтронов $u(x)$ — скалярный поток нейтронов, $k(x)$ — коэффициент диффузии, $q(x)$ — коэффициент поглощения, а $f(x)$ — заданный источник. Поэтому (1) называют также *одномерным стационарным уравнением диффузии*.

Предположим, что заданные коэффициенты удовлетворяют условиям

$$
k(x)\geqslant k_0>0,\quad q(x)\geqslant0,\quad x\in[a,b]. \tag{2}
$$

**Определение 1 (краевые условия Дирихле).** Краевые условия первого рода задают значения решения на концах отрезка,

$$
u(a)=u_a,\quad u(b)=u_b. \tag{3}
$$

В задаче теплопроводности это означает, что на концах стержня поддерживаются заданные температуры.

**Определение 2 (краевые условия Неймана).** Краевые условия второго рода задают нормальную производную или соответствующий поток на границе. Если $w_a,w_b$ — плотности теплового потока со знаком в направлении возрастания $x$, то условия на левом и правом концах имеют вид

$$
-k(a)u'(a)=w_a, \tag{4}
$$

$$
-k(b)u'(b)=w_b. \tag{5}
$$

Плотность потока, выходящего из стержня, равна $-w_a$ на левом конце и $w_b$ на правом.

Далее будем рассматривать краевую задачу Дирихле

$$
L[u](x)=f(x),\quad a<x<b, \tag{6}
$$

$$
u(a)=u_a,\quad u(b)=u_b. \tag{7}
$$

Здесь $L$ — дифференциальный оператор

$$
L[u](x)=-\bigl(k(x)u'(x)\bigr)'+q(x)u(x). \tag{8}
$$

## Разрешимость краевой задачи

Предположим, что $q,f\in C[a,b]$, $k\in C^1[a,b]$ и выполнены условия (2).

**Определение 3 (классическое решение).** Функция $u\in C^2[a,b]$ называется классическим решением задачи (6), (7), если она удовлетворяет дифференциальному уравнению (6) в $(a,b)$ и обоим краевым условиям (7).

Приведем два стандартных результата о разрешимости и гладкости.

**Теорема 1 (существование и единственность).** Краевая задача (6), (7) имеет единственное классическое решение.

**Теорема 2 (гладкость).** Если $q,f\in C^m[a,b]$ и $k\in C^{m+1}[a,b]$ для целого $m\geqslant0$, то решение принадлежит $C^{m+2}[a,b]$.

## Принцип максимума для дифференциального уравнения

**Теорема 3 (принцип максимума).** Пусть $u$ — решение (6), (7). Если $f(x)\leqslant0$, $u_a\leqslant0$ и $u_b\leqslant0$, то $u(x)\leqslant0$ на $[a,b]$.

**Доказательство.** Если $u$ где-либо положительна, возьмем интервал $(s,t)\subset(a,b)$, на котором $u>0$, причем $u(s)=u(t)=0$. Его концы могут совпадать с $a$ или $b$. Умножим (6) на $u$ и проинтегрируем по частям. Получим $\int_s^t(k(u')^2+qu^2)\,dx=\int_s^t fu\,dx\leqslant0$. Левая часть неотрицательна и обращается в нуль лишь при $u'=0$ на $(s,t)$. Это противоречит положительности $u$ внутри интервала и нулевым значениям на его концах. $\square$

Если положительных источников тепла нет и температура на концах стержня неположительна, то внутри стержня она не может стать положительной.

Любую функцию $u\in C^2[a,b]$ можно рассматривать как решение (6), (7) с данными $f=L[u]$, $u_a=u(a)$, $u_b=u(b)$. Поэтому теорему 3 можно сформулировать в следующем равносильном виде.

**Теорема 4 (операторная форма принципа максимума).** Если $u\in C^2[a,b]$ удовлетворяет $L[u]\leqslant0$ в $(a,b)$ и $u(a),u(b)\leqslant0$, то $u\leqslant0$ на $[a,b]$.

**Теорема 5 (теорема сравнения).** Пусть $u,v\in C^2[a,b]$ и

$$
L[u]\leqslant L[v]\quad\text{в }(a,b),\quad u(a)\leqslant v(a),\quad u(b)\leqslant v(b). \tag{9}
$$

Тогда $u\leqslant v$ на $[a,b]$.

**Доказательство.** Применим теорему 4 к разности $u-v$. $\square$

## Априорная оценка и устойчивость решения

Теорема сравнения позволяет оценить максимальное по модулю значение $u$ через данные задачи.

**Теорема 6 (априорная оценка).** Решение (6), (7) удовлетворяет оценке

$$
\max_{[a,b]}|u|\leqslant \max\{|u_a|,|u_b|\}+K\max_{[a,b]}|f|, \tag{10}
$$

где

$$
K=\frac{Rl}{4},\quad R=\int_a^b\frac{dx}{k(x)},\quad l=b-a. \tag{11}
$$

**Доказательство.** Обозначим $r(x)=\int_a^x ds/k(s)$ и $A=\max_{[a,b]}|f|$. Решение $w$ уравнения $-(kw')'=A$ с нулевыми краевыми условиями равно интегралу от $A G(x,s)$ по $s\in[a,b]$, где функция Грина имеет вид

$$
G(x,s)=\frac{r(\min\{x,s\})\bigl(R-r(\max\{x,s\})\bigr)}{R}. \tag{12}
$$

Она равна нулю на концах отрезка, а скачок $kG_x$ слева направо в точке $x=s$ равен $-1$. Кроме того, $0\leqslant G\leqslant R/4$, поскольку $r$ возрастает и $t(R-t)\leqslant R^2/4$ при $0\leqslant t\leqslant R$. Поэтому $0\leqslant w\leqslant ARl/4$. При $M=\max\{|u_a|,|u_b|\}$ имеем $L[M+w]=A+q(M+w)\geqslant |f|$, а на концах $M+w$ не меньше $u$ и $-u$. По теореме 5 получаем $|u|\leqslant M+w$. $\square$

**Замечание 1.** Неравенства вида (10) называют *априорными оценками*. Они выражаются через данные задачи и могут быть вычислены до нахождения решения.

**Замечание 2.** Если $k$ — коэффициент теплопроводности, то $1/k$ — удельное тепловое сопротивление, а $R$ из (11) — тепловое сопротивление стержня при единичной площади поперечного сечения.

**Замечание 3.** При $k(x)\equiv1$ уравнение (6) принимает вид

$$
-u''(x)+q(x)u(x)=f(x),\quad a<x<b. \tag{13}
$$

В этом случае квадратичная функция $w(x)=\tfrac12\max_{[a,b]}|f|\,(x-a)(b-x)$ дает более точную оценку

$$
\max_{[a,b]}|u|\leqslant\max\{|u_a|,|u_b|\}+\frac{l^2}{8}\max_{[a,b]}|f|. \tag{14}
$$

Теперь рассмотрим, как погрешности в краевых значениях и правой части влияют на решение. Пусть $u$ — решение (6), (7), а $u^*$ — решение задачи

$$
L[u^*](x)=f^*(x),\quad a<x<b, \tag{15}
$$

$$
u^*(a)=u_a^*,\quad u^*(b)=u_b^*. \tag{16}
$$

Здесь $f^*$ непрерывна. Обозначим погрешности данных через $\delta f=f-f^*$, $\varepsilon_a=u_a-u_a^*$ и $\varepsilon_b=u_b-u_b^*$.

**Теорема 7 (непрерывная зависимость от данных).** Для двух решений выполняется оценка

$$
\max_{[a,b]}|u-u^*|\leqslant\max\{|\varepsilon_a|,|\varepsilon_b|\}+K\max_{[a,b]}|\delta f|, \tag{17}
$$

где $K$ задано в (11). При $k\equiv1$ можно взять $K=l^2/8$.

**Доказательство.** Разность

$$
\varepsilon(x)=u(x)-u^*(x) \tag{18}
$$

удовлетворяет задаче

$$
L[\varepsilon](x)=\delta f(x),\quad a<x<b, \tag{19}
$$

$$
\varepsilon(a)=\varepsilon_a,\quad \varepsilon(b)=\varepsilon_b. \tag{20}
$$

Применим к $\varepsilon$ оценку (10), а при $k\equiv1$ — оценку (14). $\square$

Оценка показывает, насколько могут усиливаться погрешности данных. Умеренное значение $K$ в выбранных единицах дает оценку влияния ошибок в $f$. Одна лишь большая верхняя граница еще не свидетельствует о плохой обусловленности.

## Метод конечных разностей

**Метод конечных разностей (метод сеток)** широко применяется для решения краевых задач. Его популярность во многом объясняется простотой замены дифференциальных уравнений алгебраическими.

Область изменения непрерывного аргумента заменяют конечным множеством точек — *узлов сетки*. Используют функции, заданные в этих узлах, а производные заменяют разностями, то есть линейными комбинациями значений в выбранных узлах. В результате получают конечную систему уравнений — *дискретную краевую задачу*. Ее решение дает приближение в узлах.

Для одной краевой задачи можно построить разные разностные схемы. Покажем применение метода для задачи (6), (7) с $k(x)\equiv1$,

$$
-u''(x)+q(x)u(x)=f(x),\quad a<x<b, \tag{21}
$$

$$
u(a)=u_a,\quad u(b)=u_b. \tag{22}
$$

### Построение сетки и введение сеточных функций

**Определение 4 (сетка и сеточная функция).** Сетка $\bar{\omega}^h$ на $[a,b]$ — конечное множество упорядоченных узлов

$$
a=x_0<x_1<\dots<x_N=b. \tag{23}
$$

*Сеточная функция* задает значение в каждом узле своей области определения. Для простоты используем равномерную сетку с $N\geqslant2$ и шагом

$$
h=\frac{b-a}{N},\quad x_i=a+ih. \tag{24}
$$

При $N\to\infty$ шаг $h$ стремится к нулю. Сетка разбивается на два множества,

$$
\bar{\omega}^h=\omega^h\cup\gamma^h, \tag{25}
$$

где $\omega^h=\{x_1,\ldots,x_{N-1}\}$ — множество внутренних узлов, а $\gamma^h=\{x_0,x_N\}$ — множество граничных узлов.

Неизвестной является сеточная функция $u^h$. Ее значения $u_i=u^h(x_i)$ приближают $u(x_i)$. Введем также сеточные функции $q^h$ и $f^h$, принимающие в узлах значения заданных функций,

$$
q_i=q^h(x_i)=q(x_i),\quad f_i=f^h(x_i)=f(x_i). \tag{26}
$$

### Построение разностной схемы

Вспомним центральную разностную формулу для $u''(x)$,

$$
D_h^2u(x)=\frac{u(x-h)-2u(x)+u(x+h)}{h^2}=u''(x)+r_h(x). \tag{27}
$$

Если $u\in C^4[x-h,x+h]$, то $r_h(x)=u^{(7)}(\xi)h^2/12$ для некоторого $\xi\in(x-h,x+h)$. Заменим $u''$ в (21) этой разностью во внутренних узлах и получим приближенное соотношение

$$
-\frac{u(x_{i-1})-2u(x_i)+u(x_{i+1})}{h^2}+q_i u(x_i)\approx f_i,\quad 1\leqslant i\leqslant N-1. \tag{28}
$$

Потребуем, чтобы неизвестная сеточная функция $u^h$ удовлетворяла разностному уравнению точно,

$$
-\frac{u_{i-1}-2u_i+u_{i+1}}{h^2}+q_i u_i=f_i,\quad 1\leqslant i\leqslant N-1. \tag{29}
$$

В граничных узлах зададим значения

$$
u^h(x_0)=u_a,\quad u^h(x_N)=u_b. \tag{30}
$$

Уравнения (29), (30) образуют систему из $N+1$ линейных алгебраических уравнений для $N+1$ неизвестных

$$
u_i=u^h(x_i),\quad i=0,1,\ldots,N. \tag{31}
$$

Определим линейный разностный оператор $L^h$ на произвольных сеточных функциях $v^h$,

$$
L^h[v^h](x_i)=-\frac{v_{i-1}-2v_i+v_{i+1}}{h^2}+q_i v_i,\quad x_i\in\omega^h. \tag{32}
$$

**Определение 5 (разностная схема).** Разностной схемой для (21), (22) называется следующая дискретная задача, зависящая от $h$,

$$
L^h[u^h](x_i)=f_i,\quad x_i\in\omega^h, \tag{33}
$$

$$
u^h(x_0)=u_a,\quad u^h(x_N)=u_b. \tag{34}
$$

### Решение разностной схемы методом прогонки

Перепишем систему сеточных уравнений в виде

$$
-u_{i-1}+(2+h^2q_i)u_i-u_{i+1}=h^2f_i,\quad 1\leqslant i\leqslant N-1, \tag{35}
$$

$$
u_0=u_a,\quad u_N=u_b. \tag{36}
$$

Это трехдиагональная система вида

$$
\begin{aligned}
b_0u_0+c_0u_1&=d_0,\\
a_iu_{i-1}+b_iu_i+c_iu_{i+1}&=d_i,\quad 1\leqslant i\leqslant N-1,\\
a_Nu_{N-1}+b_Nu_N&=d_N .
\end{aligned} \tag{37}
$$

В нашем случае коэффициенты равны

$$
a_i=c_i=-1,\quad b_i=2+h^2q_i,\quad d_i=h^2f_i,\quad 1\leqslant i\leqslant N-1, \tag{38}
$$

$$
b_0=b_N=1,\quad c_0=a_N=0,\quad d_0=u_a,\quad d_N=u_b. \tag{39}
$$

**Метод прогонки** — вариант метода Гаусса для трехдиагональных систем. Он требует $O(N)$ операций и состоит из прямого и обратного ходов. Для его применения ведущие элементы $\gamma_i$ должны быть ненулевыми.

На прямом ходе выразим каждое $u_i$ через $u_{i+1}$. При $i=0$ положим

$$
\gamma_0=b_0,\quad \alpha_0=-\frac{c_0}{\gamma_0},\quad \beta_0=\frac{d_0}{\gamma_0}. \tag{40}
$$

Для $i=1,\ldots,N-1$ подставим предыдущее соотношение $u_{i-1}=\alpha_{i-1}u_i+\beta_{i-1}$ в строку $i$ и получим

$$
\gamma_i=b_i+a_i\alpha_{i-1},\quad
\alpha_i=-\frac{c_i}{\gamma_i},\quad \beta_i=\frac{d_i-a_i\beta_{i-1}}{\gamma_i}. \tag{41}
$$

В последней строке системы прямой ход заканчивается вычислением

$$
\gamma_N=b_N+a_N\alpha_{N-1},\quad \beta_N=\frac{d_N-a_N\beta_{N-1}}{\gamma_N}. \tag{42}
$$

Обратный ход начинается с

$$
u_N=\beta_N \tag{43}
$$

и восстанавливает остальные значения в порядке убывания индекса,

$$
u_i=\alpha_i u_{i+1}+\beta_i,\quad i=N-1,N-2,\ldots,0. \tag{44}
$$

Для системы (35), (36) эти соотношения упрощаются,

$$
\begin{aligned}
\alpha_0&=0,\quad \beta_0=u_a,\\
\gamma_i&=2+h^2q_i-\alpha_{i-1},\\
\alpha_i&=\frac{1}{\gamma_i},\quad
\beta_i=\frac{h^2f_i+\beta_{i-1}}{\gamma_i},\quad 1\leqslant i\leqslant N-1 .
\end{aligned} \tag{45}
$$

При обратном ходе полагаем

$$
u_N=u_b \tag{46}
$$

и находим значения во внутренних узлах по формуле

$$
u_i=\alpha_i u_{i+1}+\beta_i,\quad i=N-1,N-2,\ldots,1. \tag{47}
$$

**Замечание 4 (ведущие элементы и диагональное преобладание).** Коэффициенты (35) во внутренних строках удовлетворяют условиям

$$
a_i\leqslant0,\quad b_i>0,\quad c_i<0,\quad a_i+b_i+c_i\geqslant0,\quad 1\leqslant i<N. \tag{48}
$$

В системе есть диагональное преобладание, причем в граничных строках оно строгое,

$$
|b_0|>|c_0|,\quad |b_i|\geqslant |a_i|+|c_i|\quad(1\leqslant i<N),\quad |b_N|>|a_N|. \tag{49}
$$

Для этих коэффициентов из (45) по индукции следует, что $\gamma_i>1$ и $0<\alpha_i<1$ во внутренних узлах, поскольку $\alpha_0=0$ и $q_i\geqslant0$. Оба граничных ведущих элемента равны единице. Поэтому на обратном ходе погрешность, переданная из одного следующего узла, не усиливается. Равномерная оценка устойчивости по данным задачи доказана ниже.

### Существование и единственность решения

**Теорема 8 (разрешимость схемы).** Разностная схема (33), (34) имеет единственное решение.

**Доказательство.** По замечанию 4 все ведущие элементы ненулевые. Равносильные формулы исключения (45) однозначно определяют прогоночные коэффициенты, а формулы (46), (47) — решение. $\square$

### Принцип максимума для разностной схемы

Для разностной схемы выполняется дискретный аналог принципа максимума.

**Лемма 1 (принцип максимума для системы сеточных уравнений).** Пусть коэффициенты системы

$$
a_i u_{i-1}+b_i u_i+c_i u_{i+1}=d_i,\quad 1\leqslant i\leqslant N-1,\quad u_0=u_a,\quad u_N=u_b \tag{50}
$$

удовлетворяют (48). Если $u_a,u_b\leqslant0$ и $d_i\leqslant0$ во всех внутренних узлах, то $u_i\leqslant0$ во всех узлах.

**Доказательство.** Предположим, что

$$
u_{\max}=\max_{0\leqslant i\leqslant N}u_i>0. \tag{51}
$$

Пусть $j$ — наибольший индекс, в котором достигается максимум. Краевые значения неположительны, поэтому узел $j$ внутренний и

$$
u_{j-1}\leqslant u_j,\quad u_{j+1}<u_j. \tag{52}
$$

Так как $a_j\leqslant0$ и $c_j<0$, то

$$
a_j u_{j-1}\geqslant a_j u_j,\quad c_j u_{j+1}>c_j u_j. \tag{53}
$$

Тогда из строки $j$ системы (50) получаем противоречие

$$
0\leqslant (a_j+b_j+c_j)u_j<a_j u_{j-1}+b_j u_j+c_j u_{j+1}=d_j\leqslant0. \tag{54}
$$

Следовательно, все $u_i$ неположительны. $\square$

**Теорема 9 (принцип максимума для схемы).** Пусть $u^h$ — решение (33), (34). Если

$$
f^h\leqslant0\quad\text{на }\omega^h,\quad u_a\leqslant0,\quad u_b\leqslant0, \tag{55}
$$

то $u^h\leqslant0$ на $\bar{\omega}^h$.

**Доказательство.** Коэффициенты (35), (36) удовлетворяют (48) и $d_i=h^2f_i\leqslant0$. Применим лемму 1. $\square$

Любую сеточную функцию $u^h$ можно рассматривать как решение (33), (34) с данными

$$
f^h=L^h[u^h],\quad u_a=u^h(x_0),\quad u_b=u^h(x_N). \tag{56}
$$

**Теорема 10 (операторная форма дискретного принципа максимума).** Если

$$
L^h[u^h]\leqslant0\quad\text{на }\omega^h,\quad u^h(x_0)\leqslant0,\quad u^h(x_N)\leqslant0, \tag{57}
$$

то

$$
u^h\leqslant0\quad\text{на }\bar{\omega}^h. \tag{58}
$$

Это теорема 9 с данными (56).

**Теорема 11 (дискретная теорема сравнения).** Пусть $u^h$ и $v^h$ удовлетворяют условиям

$$
L^h[u^h]\leqslant L^h[v^h]\quad\text{на }\omega^h,\quad u^h(x_0)\leqslant v^h(x_0),\quad u^h(x_N)\leqslant v^h(x_N). \tag{59}
$$

Тогда

$$
u^h\leqslant v^h\quad\text{на }\bar{\omega}^h. \tag{60}
$$

**Доказательство.** Разность

$$
y^h=u^h-v^h \tag{61}
$$

удовлетворяет условиям

$$
L^h[y^h]=L^h[u^h]-L^h[v^h]\leqslant0, \tag{62}
$$

$$
y^h(x_0)\leqslant0,\quad y^h(x_N)\leqslant0. \tag{63}
$$

По теореме 10 имеем $y^h\leqslant0$. $\square$

**Следствие 1 (двустороннее сравнение).** Пусть

$$
|L^h[u^h]|\leqslant L^h[v^h]\quad\text{на }\omega^h, \tag{64}
$$

$$
|u^h(x_0)|\leqslant v^h(x_0),\quad |u^h(x_N)|\leqslant v^h(x_N). \tag{65}
$$

Тогда

$$
-v^h\leqslant u^h\leqslant v^h\quad\text{на }\bar{\omega}^h. \tag{66}
$$

**Доказательство.** Применим теорему 11 к $u^h$ и $v^h$, а затем к $-u^h$ и $v^h$. $\square$

### Априорная оценка решения разностной схемы

Чтобы оценить максимальное по модулю значение $u^h$ через данные, сначала установим два вспомогательных результата.

**Лемма 2 (влияние краевых значений).** Решение задачи

$$
L^h[v^h]=0\quad\text{на }\omega^h, \tag{67}
$$

$$
v^h(x_0)=u_a,\quad v^h(x_N)=u_b \tag{68}
$$

удовлетворяет оценке

$$
\max_{0\leqslant i\leqslant N}|v_i|\leqslant\max\{|u_a|,|u_b|\}. \tag{69}
$$

**Доказательство.** Возьмем постоянную сеточную функцию $w^h\equiv M$, где

$$
M=\max\{|u_a|,|u_b|\}. \tag{70}
$$

Во внутренних узлах

$$
|L^h[v^h]|=0\leqslant L^h[w^h]=q^hM, \tag{71}
$$

а в граничных узлах

$$
|v^h(x_0)|=|u_a|\leqslant M,\quad |v^h(x_N)|=|u_b|\leqslant M. \tag{72}
$$

Тогда из (66) следует, что $|v^h|\leqslant M$. $\square$

**Лемма 3 (влияние правой части).** Решение задачи

$$
L^h[z^h]=f^h\quad\text{на }\omega^h, \tag{73}
$$

$$
z^h(x_0)=0,\quad z^h(x_N)=0 \tag{74}
$$

удовлетворяет оценке

$$
\max_{0\leqslant i\leqslant N}|z_i|\leqslant\frac{l^2}{8}\max_{0<i<N}|f_i|. \tag{75}
$$

**Доказательство.** Определим сеточную функцию

$$
w^h(x_i)=\frac A2(x_i-a)(b-x_i), \tag{76}
$$

где

$$
A=\max_{0<i<N}|f_i|. \tag{77}
$$

Имеем $w^h\geqslant0$, а центральная вторая разность точна для квадратичных функций. Поэтому

$$
L^h[w^h]=A+q^hw^h. \tag{78}
$$

Условия (66) выполняются, поскольку

$$
|L^h[z^h]|=|f^h|\leqslant A\leqslant L^h[w^h], \tag{79}
$$

$$
|z^h(x_0)|=0=w^h(x_0),\quad |z^h(x_N)|=0=w^h(x_N). \tag{80}
$$

Следовательно, $|z^h|\leqslant w^h$. Квадратичная функция

$$
\frac A2(x-a)(b-x) \tag{81}
$$

принимает максимальное значение $Al^2/8$ в точке $x=(a+b)/2$. Ее значения в узлах удовлетворяют неравенствам

$$
0\leqslant w^h\leqslant\frac{Al^2}{8}. \tag{82}
$$

Это доказывает (75). $\square$

**Теорема 12 (априорная оценка для схемы).** Решение (33), (34) удовлетворяет оценке

$$
\max_{0\leqslant i\leqslant N}|u_i|\leqslant \max\{|u_a|,|u_b|\}+\frac{l^2}{8}\max_{0<i<N}|f_i|. \tag{83}
$$

**Доказательство.** Из линейности и единственности решения следует, что

$$
u^h=v^h+z^h, \tag{84}
$$

где $v^h$ — решение (67), (68), а $z^h$ — решение (73), (74). Воспользуемся неравенством

$$
\max_{0\leqslant i\leqslant N}|u_i|\leqslant\max_{0\leqslant i\leqslant N}|v_i|+\max_{0\leqslant i\leqslant N}|z_i| \tag{85}
$$

и оценками (69) и (75). Получаем (83). $\square$

### Устойчивость

Пусть $u^h$ — решение (33), (34), а $u^{*h}$ — решение той же схемы с возмущенными данными,

$$
L^h[u^{*h}]=f^{*h}\quad\text{на }\omega^h, \tag{86}
$$

$$
u^{*h}(x_0)=u_a^*,\quad u^{*h}(x_N)=u_b^*, \tag{87}
$$

где

$$
f^{*h}=f^h-\delta f^h,\quad u_a^*=u_a-\varepsilon_a,\quad u_b^*=u_b-\varepsilon_b. \tag{88}
$$

**Определение 6 (устойчивость).** Разностная схема (33), (34) устойчива в норме максимума, если для любых возмущений данных

$$
\max_{0\leqslant i\leqslant N}|u_i-u_i^*|\leqslant\max\{|\varepsilon_a|,|\varepsilon_b|\}+K\max_{0<i<N}|\delta f_i|, \tag{89}
$$

где $u_i^*=u^{*h}(x_i)$, а $K$ не зависит от $h$. Это дискретный аналог (17).

**Теорема 13 (устойчивость разностной схемы).** Для схемы (33), (34) оценка (89) справедлива с постоянной

$$
K=\frac{l^2}{8}. \tag{90}
$$

**Доказательство.** Разность

$$
\varepsilon^h=u^h-u^{*h} \tag{91}
$$

удовлетворяет задаче

$$
L^h[\varepsilon^h]=\delta f^h\quad\text{на }\omega^h, \tag{92}
$$

$$
\varepsilon^h(x_0)=\varepsilon_a,\quad \varepsilon^h(x_N)=\varepsilon_b. \tag{93}
$$

Применим теорему 12 к $\varepsilon^h$. $\square$

**Замечание 5.** При $k\equiv1$ получена та же верхняя граница усиления погрешностей данных, что и в непрерывной оценке (17).

### Аппроксимация

Пусть $u$ — решение дифференциального уравнения (21), а $U^h(x_i)=u(x_i)$ — его ограничение на сетку.

**Определение 7 (погрешность и порядок аппроксимации).** *Погрешность аппроксимации* разностного уравнения (33) — это невязка, полученная подстановкой точных значений,

$$
\psi^h=L^h[U^h]-f^h\quad\text{на }\omega^h. \tag{94}
$$

Равносильная запись имеет вид

$$
L^h[U^h]=f^h+\psi^h\quad\text{на }\omega^h. \tag{95}
$$

Разностное уравнение аппроксимирует (21), если

$$
\max_{0<i<N}|\psi_i|\to0\quad\text{при }h\to0,\quad \psi_i=\psi^h(x_i). \tag{96}
$$

Оно аппроксимирует (21) с порядком $m>0$, если при всех достаточно малых $h$

$$
\max_{0<i<N}|\psi_i|\leqslant Ch^m \tag{97}
$$

с постоянной $C$, не зависящей от $h$.

**Лемма 4 (аппроксимация второго порядка).** Если $q,f\in C^2[a,b]$, то

$$
\max_{0<i<N}|\psi_i|\leqslant\frac{M_4}{12}h^2,\quad M_4=\max_{[a,b]}|u^{(7)}|. \tag{98}
$$

**Доказательство.** По теореме 2 имеем $u\in C^4[a,b]$. Подставив точные значения в (94), получим

$$
\begin{aligned}
\psi_i&=-D_h^2u(x_i)+q_i u(x_i)-f_i\\
&=-u''(x_i)-r_h(x_i)+q_i u(x_i)-f_i=-r_h(x_i).
\end{aligned} \tag{99}
$$

Согласно (27),

$$
r_h(x_i)=\frac{u^{(7)}(\xi_i)}{12}h^2,\quad \xi_i\in(x_{i-1},x_{i+1}). \tag{100}
$$

Следовательно,

$$
\max_{0<i<N}|\psi_i|=\max_{0<i<N}|r_h(x_i)|\leqslant\frac{M_4}{12}h^2. \tag{101}
$$

$\square$

Краевые условия в этой задаче удовлетворяются точно, поскольку $U^h$ выполняет (34) без погрешности.

### Сходимость

**Определение 8 (погрешность и сходимость схемы).** Погрешность схемы (33), (34) — сеточная функция $\varepsilon^h$ со значениями

$$
\varepsilon_i=\varepsilon^h(x_i)=u(x_i)-u_i,\quad 0\leqslant i\leqslant N. \tag{102}
$$

Схема сходится, если

$$
\max_{0\leqslant i\leqslant N}|\varepsilon_i|\to0\quad\text{при }h\to0, \tag{103}
$$

и сходится с порядком $m>0$, если

$$
\max_{0\leqslant i\leqslant N}|\varepsilon_i|\leqslant Ch^m \tag{104}
$$

при всех достаточно малых $h$ с постоянной $C$, не зависящей от $h$.

**Теорема 14 (сходимость второго порядка).** Если $q,f\in C^2[a,b]$, то погрешность (33), (34) удовлетворяет оценке

$$
\max_{0\leqslant i\leqslant N}|u(x_i)-u_i|\leqslant Ch^2, \tag{105}
$$

где

$$
C=\frac{l^2}{96}\max_{[a,b]}|u^{(7)}|. \tag{106}
$$

**Доказательство.** Снова рассмотрим ограничение точного решения на сетку,

$$
U^h(x_i)=u(x_i),\quad 0\leqslant i\leqslant N. \tag{107}
$$

Из (95) следует, что $U^h$ удовлетворяет схеме вида (86), (87) с данными

$$
f^{*h}=f^h+\psi^h,\quad u_a^*=u_a,\quad u_b^*=u_b. \tag{108}
$$

Для погрешности

$$
\varepsilon^h=U^h-u^h \tag{109}
$$

теорема 13 дает оценку

$$
\max_{0\leqslant i\leqslant N}|\varepsilon_i|\leqslant\frac{l^2}{8}\max_{0<i<N}|\psi_i|. \tag{110}
$$

Подстановка (98) дает (105). $\square$

Таким образом, сходимость второго порядка следует из устойчивости схемы и аппроксимации второго порядка.

### Оценка погрешности по правилу Рунге

Априорная оценка (105) зависит от производной точного решения. Для практической оценки погрешности сравним решения на двух вложенных сетках с шагами $h$ и $2h$.

**Утверждение 1 (оценка Рунге).** Пусть погрешность со знаком в общих узлах обеих сеток имеет главный член $c(x)h^2$ с одним и тем же коэффициентом $c(x)$ и остатком $O(h^4)$. Вычитая разложения для двух сеток, получаем

$$
\varepsilon^h(x)=u(x)-u^h(x)=\frac{u^h(x)-u^{2h}(x)}{3}+O(h^4),\quad x\in\bar{\omega}^{2h}. \tag{111}
$$

Дробь в правой части — *оценка Рунге* для погрешности со знаком на мелкой сетке. Она доступна в общих узлах, включая концы отрезка, где погрешность равна нулю. Одной оценки $O(h^2)$ для обоснования этого разложения недостаточно.

**Замечание 6.** Для данной схемы достаточное условие разложения имеет вид $u\in C^6[a,b]$, $q\in C^2[a,b]$. Действительно, во внутренних узлах по формуле Тейлора $L^h[U^h]-f^h=-h^2u^{(7)}/12+O(h^4)$. Пусть $c$ решает задачу $L[c]=-u^{(7)}/12$ с нулевыми краевыми значениями при $k\equiv1$. По теореме 2, $c\in C^4[a,b]$. Применив (89) к остатку после вычитания сеточных значений $h^2c$ из $\varepsilon^h$, получаем равномерную оценку $O(h^4)$.

### Численный пример

**Пример 1.** Применим схему (33), (34) с мелким шагом

$$
h_1=\frac18 \tag{112}
$$

для решения краевой задачи

$$
-u''(x)+x^2u(x)=\left(\frac{\pi^2}{4}+x^2\right)\cos\frac{\pi x}{2},\quad 0<x<1, \tag{113}
$$

$$
u(0)=1,\quad u(1)=0. \tag{114}
$$

Коэффициент и правая часть равны

$$
q(x)=x^2,\quad f(x)=\left(\frac{\pi^2}{4}+x^2\right)\cos\frac{\pi x}{2}. \tag{115}
$$

Краевые уравнения в (35), (36) принимают вид

$$
u_0=1,\quad u_N=0, \tag{116}
$$

а уравнения во внутренних узлах имеют вид

$$
\begin{aligned}
-u_{i-1}+(2+h^2x_i^2)u_i-u_{i+1}
&=h^2\left(\frac{\pi^2}{4}+x_i^2\right)\cos\frac{\pi x_i}{2},\\
&\quad i=1,\ldots,N-1.
\end{aligned} \tag{117}
$$

При $h_1=1/8$ получаем следующую систему. Коэффициенты и правые части округлены только при записи.

$$
\begin{aligned}
u_0 &= 1,\\
-u_0+2.000244\,u_1-u_2 &= 0.03805180,\\
-u_1+2.000977\,u_2-u_3 &= 0.03652069,\\
-u_2+2.002197\,u_3-u_4 &= 0.03388273,\\
-u_3+2.003906\,u_4-u_5 &= 0.03002332,\\
-u_4+2.006104\,u_5-u_6 &= 0.02480991,\\
-u_5+2.008789\,u_6-u_7 &= 0.01811708,\\
-u_6+2.011963\,u_7-u_8 &= 0.00985519,\\
u_8 &= 0.
\end{aligned} \tag{118}
$$

Точное решение имеет вид

$$
u(x)=\cos\frac{\pi x}{2}. \tag{119}
$$

Действительно, $-u''=(\pi^2/4)u$, а значения на концах удовлетворяют (114). Для оценки Рунге решим ту же задачу с крупным шагом

$$
h_2=2h_1=\frac14. \tag{120}
$$

Получим систему

$$
\begin{aligned}
u_0 &= 1,\\
-u_0+2.003906\,u_1-u_2 &= 0.14608274,\\
-u_1+2.015625\,u_2-u_3 &= 0.12009330,\\
-u_2+2.035156\,u_3-u_4 &= 0.07246831,\\
u_4 &= 0.
\end{aligned} \tag{121}
$$

Обе системы решим методом прогонки, округляя числа лишь при выводе. В таблице приведены точные значения и значения на мелкой сетке, а также их разность со знаком. В последних двух столбцах для совпадающих узлов даны решение на крупной сетке и оценка (111).

```python
import numpy as np
```

```python
def u_exact(x):
    return np.cos(np.pi * x / 2)

def q(x):
    return x**2

def f(x):
    return (np.pi**2 / 4 + x**2) * u_exact(x)
```

```python
def solve_tridiag(a, b, c, d):
    n = len(d)
    alpha = np.zeros(n - 1)
    beta = np.zeros(n)

    alpha[0] = -c[0] / b[0]
    beta[0] = d[0] / b[0]

    for i in range(1, n - 1):
        gamma = b[i] + a[i] * alpha[i - 1]
        alpha[i] = -c[i] / gamma
        beta[i] = (d[i] - a[i] * beta[i - 1]) / gamma

    gamma = b[-1] + a[-1] * alpha[-1]
    beta[-1] = (d[-1] - a[-1] * beta[-2]) / gamma

    u = np.zeros(n)
    u[-1] = beta[-1]
    for i in range(n - 2, -1, -1):
        u[i] = alpha[i] * u[i + 1] + beta[i]

    return u

def solve_fd(n):
    h = 1 / n
    x = np.linspace(0, 1, n + 1)
    a = np.zeros(n + 1)
    b = np.ones(n + 1)
    c = np.zeros(n + 1)
    d = np.zeros(n + 1)
    d[0] = 1

    for i in range(1, n):
        a[i] = -1
        b[i] = 2 + h**2 * q(x[i])
        c[i] = -1
        d[i] = h**2 * f(x[i])

    return x, solve_tridiag(a, b, c, d)
```

```python
n_fine, n_coarse = 8, 4
x_fine, u_fine = solve_fd(n_fine)
x_coarse, u_coarse = solve_fd(n_coarse)
exact_fine = u_exact(x_fine)
error_fine = exact_fine - u_fine
runge = (u_fine[::2] - u_coarse) / 3

print(f"{'x_i':>6} {'u_exact':>12} {'u^h (1/8)':>12} {'eps_h':>12} {'u^(2h)':>12} {'Runge':>12}")
print("-" * 71)
for i, x in enumerate(x_fine):
    row = f"{x:6.3f} {exact_fine[i]:12.6f} {u_fine[i]:12.6f} {error_fine[i]:12.1e}"
    if i % 2 == 0:
        j = i // 2
        print(f"{row} {u_coarse[j]:12.6f} {runge[j]:12.1e}")
    else:
        print(f"{row} {'-':>12} {'-':>12}")
```

В общих внутренних узлах оценка Рунге имеет тот же знак, что и фактическая погрешность, и близка к ней по величине. Проверим также максимальную невязку во внутренних узлах мелкой сетки и оценим порядок сходимости по максимальным узловым погрешностям на двух сетках.

```python
h = 1 / n_fine
x = x_fine[1:-1]
second_difference = (u_fine[:-2] - 2 * u_fine[1:-1] + u_fine[2:]) / h**2
residual = -second_difference + q(x) * u_fine[1:-1] - f(x)
error_coarse = u_exact(x_coarse) - u_coarse
observed_order = np.log2(np.max(np.abs(error_coarse)) / np.max(np.abs(error_fine)))
print(f"Максимальная невязка во внутренних узлах = {np.max(np.abs(residual)):.2e}")
print(f"Наблюдаемый порядок = {observed_order:.3f}")
```
