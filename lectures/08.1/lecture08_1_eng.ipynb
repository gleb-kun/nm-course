# Boundary value problems for the one-dimensional steady-state heat equation

## Differential equation and boundary conditions

Consider the second-order differential equation

$$
-\frac{d}{dx}\left(k(x)\frac{du}{dx}\right)+q(x)u(x)=f(x),\quad a<x<b. \tag{1}
$$

This equation is called the *one-dimensional steady-state heat equation*. For example, it describes the steady temperature distribution $u(x)$ in a heat-conducting rod of length $l=b-a$. Here $k(x)$ is the thermal conductivity, $w(x)=-k(x)u'(x)$ is the heat flux density in the positive $x$ direction, and $q(x)u(x)$ represents heat removal proportional to the temperature relative to a reference level. The term $f(x)$ represents an external heat source when $f(x)>0$ and a sink when $f(x)<0$.

Equation (1) also describes steady diffusion in porous media, where $u(x)$ is the concentration of the diffusing substance. In a one-group diffusion model for neutrons, $u(x)$ is the scalar neutron flux, $k(x)$ is the diffusion coefficient, $q(x)$ is the absorption coefficient, and $f(x)$ is a prescribed source. For this reason, (1) is also called a *one-dimensional steady diffusion equation*.

Assume the given coefficients satisfy

$$
k(x)\geq k_0>0,\quad q(x)\geq0,\quad x\in[a,b]. \tag{2}
$$

**Definition 1 (Dirichlet boundary conditions).** Boundary conditions of the first kind prescribe the solution values at the two endpoints,

$$
u(a)=u_a,\quad u(b)=u_b. \tag{3}
$$

In the heat conduction problem, this means maintaining fixed temperatures at the ends of the rod.

**Definition 2 (Neumann boundary conditions).** Boundary conditions of the second kind prescribe the normal derivative or the corresponding flux at an endpoint. If $w_a,w_b$ are the signed heat flux densities in the positive $x$ direction, the conditions at the left and right ends, respectively, are

$$
-k(a)u'(a)=w_a, \tag{4}
$$

$$
-k(b)u'(b)=w_b. \tag{5}
$$

The outward heat fluxes are $-w_a$ at the left end and $w_b$ at the right end.

We focus on the Dirichlet boundary value problem

$$
L[u](x)=f(x),\quad a<x<b, \tag{6}
$$

$$
u(a)=u_a,\quad u(b)=u_b. \tag{7}
$$

Here $L$ denotes the differential operator

$$
L[u](x)=-\bigl(k(x)u'(x)\bigr)'+q(x)u(x). \tag{8}
$$

## Solvability of the boundary value problem

Assume $q,f\in C[a,b]$, $k\in C^1[a,b]$, and that (2) holds.

**Definition 3 (classical solution).** A function $u\in C^2[a,b]$ is a classical solution of (6), (7) if it satisfies the differential equation (6) in $(a,b)$ and both boundary conditions (7).

We state two standard results about solvability and smoothness.

**Theorem 1 (existence and uniqueness).** The boundary value problem (6), (7) has a unique classical solution.

**Theorem 2 (regularity).** If $q,f\in C^m[a,b]$ and $k\in C^{m+1}[a,b]$ for an integer $m\geq0$, then the solution belongs to $C^{m+2}[a,b]$.

## Maximum principle for the differential equation

**Theorem 3 (maximum principle).** Let $u$ solve (6), (7). If $f(x)\leq0$, $u_a\leq0$, and $u_b\leq0$, then $u(x)\leq0$ on $[a,b]$.

**Proof.** If $u$ is positive somewhere, let $(s,t)\subset(a,b)$ be an interval on which $u>0$, with $u(s)=u(t)=0$. The endpoints can be $a$ or $b$. Multiplying (6) by $u$ and integrating by parts yields $\int_s^t(k(u')^2+qu^2)\,dx=\int_s^t fu\,dx\leq0$. The left side is nonnegative and can vanish only if $u'=0$ throughout $(s,t)$. This contradicts $u>0$ and the zero endpoint values. $\square$

If there are no positive heat sources and the temperatures at the ends of the rod are nonpositive, the temperature cannot become positive inside the rod.

Any $u\in C^2[a,b]$ is a solution of (6), (7) for the data $f=L[u]$, $u_a=u(a)$, and $u_b=u(b)$. Thus Theorem 3 has the following equivalent form.

**Theorem 4 (operator form of the maximum principle).** If $u\in C^2[a,b]$ satisfies $L[u]\leq0$ in $(a,b)$ and $u(a),u(b)\leq0$, then $u\leq0$ on $[a,b]$.

**Theorem 5 (comparison theorem).** Let $u,v\in C^2[a,b]$ and suppose that

$$
L[u]\leq L[v]\quad\text{in }(a,b),\quad u(a)\leq v(a),\quad u(b)\leq v(b). \tag{9}
$$

Then $u\leq v$ on $[a,b]$.

**Proof.** Apply Theorem 4 to $u-v$. $\square$

## A priori estimate and stability of the solution

The comparison theorem gives an estimate of the maximum absolute value of $u$ in terms of the data.

**Theorem 6 (a priori estimate).** The solution of (6), (7) satisfies

$$
\max_{[a,b]}|u|\leq \max\{|u_a|,|u_b|\}+K\max_{[a,b]}|f|, \tag{10}
$$

where

$$
K=\frac{Rl}{4},\quad R=\int_a^b\frac{dx}{k(x)},\quad l=b-a. \tag{11}
$$

**Proof.** Write $r(x)=\int_a^x ds/k(s)$ and $A=\max_{[a,b]}|f|$. The solution $w$ of $-(kw')'=A$ with zero endpoint values is the integral of $A G(x,s)$ over $s\in[a,b]$, where the Green function is

$$
G(x,s)=\frac{r(\min\{x,s\})\bigl(R-r(\max\{x,s\})\bigr)}{R}. \tag{12}
$$

It vanishes at the endpoints and the jump of $kG_x$ from left to right at $x=s$ is $-1$. Moreover, $0\leq G\leq R/4$ since $r$ is increasing and $t(R-t)\leq R^2/4$ for $0\leq t\leq R$. Thus $0\leq w\leq ARl/4$. With $M=\max\{|u_a|,|u_b|\}$, we have $L[M+w]=A+q(M+w)\geq |f|$, and $M+w$ dominates both $u$ and $-u$ at the endpoints. Theorem 5 gives $|u|\leq M+w$. $\square$

**Remark 1.** Inequalities of the form (10) are called *a priori estimates*. They are expressed in terms of the data and can be evaluated before the solution is found.

**Remark 2.** When $k$ is the thermal conductivity, $1/k$ is the thermal resistivity, and $R$ in (11) is the thermal resistance of the rod for unit cross-sectional area.

**Remark 3.** For $k(x)\equiv1$, equation (6) becomes

$$
-u''(x)+q(x)u(x)=f(x),\quad a<x<b. \tag{13}
$$

In this case the quadratic barrier $w(x)=\tfrac12\max_{[a,b]}|f|\,(x-a)(b-x)$ yields the sharper estimate

$$
\max_{[a,b]}|u|\leq\max\{|u_a|,|u_b|\}+\frac{l^2}{8}\max_{[a,b]}|f|. \tag{14}
$$

Now consider how errors in the boundary values and the right-hand side affect the solution. Let $u$ solve (6), (7), and let $u^*$ solve

$$
L[u^*](x)=f^*(x),\quad a<x<b, \tag{15}
$$

$$
u^*(a)=u_a^*,\quad u^*(b)=u_b^*. \tag{16}
$$

Here $f^*$ is continuous. Denote the data errors by $\delta f=f-f^*$, $\varepsilon_a=u_a-u_a^*$, and $\varepsilon_b=u_b-u_b^*$.

**Theorem 7 (continuous dependence on the data).** The two solutions satisfy

$$
\max_{[a,b]}|u-u^*|\leq\max\{|\varepsilon_a|,|\varepsilon_b|\}+K\max_{[a,b]}|\delta f|, \tag{17}
$$

with $K$ as in (11). For $k\equiv1$ one may use $K=l^2/8$.

**Proof.** The difference

$$
\varepsilon(x)=u(x)-u^*(x) \tag{18}
$$

solves the problem

$$
L[\varepsilon](x)=\delta f(x),\quad a<x<b, \tag{19}
$$

$$
\varepsilon(a)=\varepsilon_a,\quad \varepsilon(b)=\varepsilon_b. \tag{20}
$$

Apply (10), or (14) for $k\equiv1$, to $\varepsilon$. $\square$

The estimate controls how much errors in the data can be amplified. A moderate value of $K$ in the chosen units gives a useful bound on the effect of errors in $f$. A large upper bound alone does not establish ill-conditioning.

## Finite difference method

The **finite difference method (or grid method)** is a widely used approach to solving boundary value problems. Its popularity is largely due to the simplicity of replacing differential equations with algebraic equations.

The domain of the continuous variable is replaced by a finite set of points, called *grid nodes*. We use functions defined at those nodes and replace derivatives by differences, which are linear combinations of values at selected nodes. The resulting finite system is a *discrete boundary value problem*. Its solution gives an approximation at the nodes.

Many difference schemes can be constructed for a single boundary value problem. Here we illustrate the method for the problem (6), (7) with $k(x)\equiv1$,

$$
-u''(x)+q(x)u(x)=f(x),\quad a<x<b, \tag{21}
$$

$$
u(a)=u_a,\quad u(b)=u_b. \tag{22}
$$

### Construction of the grid and introduction of grid functions

**Definition 4 (grid and grid function).** A grid $\bar{\omega}^h$ on $[a,b]$ is a finite set of nodes ordered by

$$
a=x_0<x_1<\dots<x_N=b. \tag{23}
$$

A *grid function* assigns a value to each node of its domain. For simplicity we use a uniform grid with $N\geq2$ and step

$$
h=\frac{b-a}{N},\quad x_i=a+ih. \tag{24}
$$

As $N\to\infty$, the step $h$ tends to zero. The grid splits into two subsets,

$$
\bar{\omega}^h=\omega^h\cup\gamma^h, \tag{25}
$$

where $\omega^h=\{x_1,\ldots,x_{N-1}\}$ is the set of interior nodes and $\gamma^h=\{x_0,x_N\}$ is the set of boundary nodes.

The unknown is the grid function $u^h$. We write $u_i=u^h(x_i)$ for its values, which approximate $u(x_i)$. We also introduce the grid functions $q^h$ and $f^h$ by sampling the given data,

$$
q_i=q^h(x_i)=q(x_i),\quad f_i=f^h(x_i)=f(x_i). \tag{26}
$$

### Construction of the finite difference scheme

Recall the centered difference formula for $u''(x)$,

$$
D_h^2u(x)=\frac{u(x-h)-2u(x)+u(x+h)}{h^2}=u''(x)+r_h(x). \tag{27}
$$

If $u\in C^4[x-h,x+h]$, then $r_h(x)=u^{(7)}(\xi)h^2/12$ for some $\xi\in(x-h,x+h)$. Replacing $u''$ in (21) by this difference at the interior nodes gives the approximate relation

$$
-\frac{u(x_{i-1})-2u(x_i)+u(x_{i+1})}{h^2}+q_i u(x_i)\approx f_i,\quad 1\leq i\leq N-1. \tag{28}
$$

We require the unknown grid function $u^h$ to satisfy the difference equation exactly,

$$
-\frac{u_{i-1}-2u_i+u_{i+1}}{h^2}+q_i u_i=f_i,\quad 1\leq i\leq N-1. \tag{29}
$$

At the boundary nodes we impose the given values,

$$
u^h(x_0)=u_a,\quad u^h(x_N)=u_b. \tag{30}
$$

Equations (29), (30) are a system of $N+1$ linear algebraic equations for the $N+1$ unknowns

$$
u_i=u^h(x_i),\quad i=0,1,\ldots,N. \tag{31}
$$

Define the linear finite difference operator $L^h$ on arbitrary grid functions $v^h$ by

$$
L^h[v^h](x_i)=-\frac{v_{i-1}-2v_i+v_{i+1}}{h^2}+q_i v_i,\quad x_i\in\omega^h. \tag{32}
$$

**Definition 5 (finite difference scheme).** The $h$-dependent discrete problem for (21), (22) is

$$
L^h[u^h](x_i)=f_i,\quad x_i\in\omega^h, \tag{33}
$$

$$
u^h(x_0)=u_a,\quad u^h(x_N)=u_b. \tag{34}
$$

### Computing the solution of the difference scheme using the Thomas algorithm

Rewrite the system of grid equations as

$$
-u_{i-1}+(2+h^2q_i)u_i-u_{i+1}=h^2f_i,\quad 1\leq i\leq N-1, \tag{35}
$$

$$
u_0=u_a,\quad u_N=u_b. \tag{36}
$$

This is a tridiagonal system of the form

$$
\begin{aligned}
b_0u_0+c_0u_1&=d_0,\\
a_iu_{i-1}+b_iu_i+c_iu_{i+1}&=d_i,\quad 1\leq i\leq N-1,\\
a_Nu_{N-1}+b_Nu_N&=d_N .
\end{aligned} \tag{37}
$$

In our case the coefficients are

$$
a_i=c_i=-1,\quad b_i=2+h^2q_i,\quad d_i=h^2f_i,\quad 1\leq i\leq N-1, \tag{38}
$$

$$
b_0=b_N=1,\quad c_0=a_N=0,\quad d_0=u_a,\quad d_N=u_b. \tag{39}
$$

The **Thomas algorithm** is a specialized form of Gaussian elimination for tridiagonal systems. It takes $O(N)$ operations and consists of a forward sweep and a backward sweep. It requires the pivots $\gamma_i$ below to be nonzero.

The forward sweep expresses each $u_i$ in terms of $u_{i+1}$. At $i=0$ set

$$
\gamma_0=b_0,\quad \alpha_0=-\frac{c_0}{\gamma_0},\quad \beta_0=\frac{d_0}{\gamma_0}. \tag{40}
$$

For $i=1,\ldots,N-1$, substitute the preceding relation $u_{i-1}=\alpha_{i-1}u_i+\beta_{i-1}$ into row $i$ to obtain

$$
\gamma_i=b_i+a_i\alpha_{i-1},\quad
\alpha_i=-\frac{c_i}{\gamma_i},\quad \beta_i=\frac{d_i-a_i\beta_{i-1}}{\gamma_i}. \tag{41}
$$

At the last row the forward sweep ends with

$$
\gamma_N=b_N+a_N\alpha_{N-1},\quad \beta_N=\frac{d_N-a_N\beta_{N-1}}{\gamma_N}. \tag{42}
$$

The backward sweep starts with

$$
u_N=\beta_N \tag{43}
$$

and recovers the other values in decreasing order,

$$
u_i=\alpha_i u_{i+1}+\beta_i,\quad i=N-1,N-2,\ldots,0. \tag{44}
$$

For system (35), (36) these relations simplify to

$$
\begin{aligned}
\alpha_0&=0,\quad \beta_0=u_a,\\
\gamma_i&=2+h^2q_i-\alpha_{i-1},\\
\alpha_i&=\frac{1}{\gamma_i},\quad
\beta_i=\frac{h^2f_i+\beta_{i-1}}{\gamma_i},\quad 1\leq i\leq N-1 .
\end{aligned} \tag{45}
$$

In the backward sweep set

$$
u_N=u_b \tag{46}
$$

and recover the interior values by

$$
u_i=\alpha_i u_{i+1}+\beta_i,\quad i=N-1,N-2,\ldots,1. \tag{47}
$$

**Remark 4 (pivots and diagonal dominance).** The interior coefficients of (35) satisfy

$$
a_i\leq0,\quad b_i>0,\quad c_i<0,\quad a_i+b_i+c_i\geq0,\quad 1\leq i<N. \tag{48}
$$

The system is diagonally dominant with strict dominance in the boundary rows,

$$
|b_0|>|c_0|,\quad |b_i|\geq |a_i|+|c_i|\quad(1\leq i<N),\quad |b_N|>|a_N|. \tag{49}
$$

For these coefficients, (45) shows directly by induction that $\gamma_i>1$ and $0<\alpha_i<1$ at interior nodes, since $\alpha_0=0$ and $q_i\geq0$. The boundary pivots are both one. The backward sweep therefore does not amplify a perturbation passed from a single following node. A uniform stability estimate for perturbations in the problem data is proved below.

### Existence and uniqueness of the solution

**Theorem 8 (solvability of the scheme).** The finite difference scheme (33), (34) has a unique solution.

**Proof.** By Remark 4 all pivots are nonzero. The equivalent elimination formulas (45) determine the sweep coefficients uniquely, and (46), (47) determine the solution uniquely. $\square$

### Maximum principle for the finite difference scheme

The finite difference scheme satisfies a discrete analogue of the maximum principle.

**Lemma 1 (maximum principle for the system of grid equations).** Let the coefficients of the system

$$
a_i u_{i-1}+b_i u_i+c_i u_{i+1}=d_i,\quad 1\leq i\leq N-1,\quad u_0=u_a,\quad u_N=u_b \tag{50}
$$

satisfy (48). If $u_a,u_b\leq0$ and $d_i\leq0$ at every interior node, then $u_i\leq0$ at every node.

**Proof.** Suppose that

$$
u_{\max}=\max_{0\leq i\leq N}u_i>0. \tag{51}
$$

Let $j$ be the largest index where the maximum is attained. The boundary values are nonpositive, so $j$ is interior and

$$
u_{j-1}\leq u_j,\quad u_{j+1}<u_j. \tag{52}
$$

Since $a_j\leq0$ and $c_j<0$, it follows that

$$
a_j u_{j-1}\geq a_j u_j,\quad c_j u_{j+1}>c_j u_j. \tag{53}
$$

Hence row $j$ of (50) gives the contradiction

$$
0\leq (a_j+b_j+c_j)u_j<a_j u_{j-1}+b_j u_j+c_j u_{j+1}=d_j\leq0. \tag{54}
$$

Thus every $u_i$ is nonpositive. $\square$

**Theorem 9 (maximum principle for the scheme).** Let $u^h$ solve (33), (34). If

$$
f^h\leq0\quad\text{on }\omega^h,\quad u_a\leq0,\quad u_b\leq0, \tag{55}
$$

then $u^h\leq0$ on $\bar{\omega}^h$.

**Proof.** The coefficients of (35), (36) satisfy (48) and $d_i=h^2f_i\leq0$. Apply Lemma 1. $\square$

Any grid function $u^h$ may be viewed as a solution of (33), (34) with the data

$$
f^h=L^h[u^h],\quad u_a=u^h(x_0),\quad u_b=u^h(x_N). \tag{56}
$$

**Theorem 10 (operator form of the discrete maximum principle).** If

$$
L^h[u^h]\leq0\quad\text{on }\omega^h,\quad u^h(x_0)\leq0,\quad u^h(x_N)\leq0, \tag{57}
$$

then

$$
u^h\leq0\quad\text{on }\bar{\omega}^h. \tag{58}
$$

This is Theorem 9 with the data (56).

**Theorem 11 (discrete comparison theorem).** Let $u^h$ and $v^h$ satisfy

$$
L^h[u^h]\leq L^h[v^h]\quad\text{on }\omega^h,\quad u^h(x_0)\leq v^h(x_0),\quad u^h(x_N)\leq v^h(x_N). \tag{59}
$$

Then

$$
u^h\leq v^h\quad\text{on }\bar{\omega}^h. \tag{60}
$$

**Proof.** The difference

$$
y^h=u^h-v^h \tag{61}
$$

satisfies

$$
L^h[y^h]=L^h[u^h]-L^h[v^h]\leq0, \tag{62}
$$

$$
y^h(x_0)\leq0,\quad y^h(x_N)\leq0. \tag{63}
$$

By Theorem 10, $y^h\leq0$. $\square$

**Corollary 1 (two-sided comparison).** Suppose that

$$
|L^h[u^h]|\leq L^h[v^h]\quad\text{on }\omega^h, \tag{64}
$$

$$
|u^h(x_0)|\leq v^h(x_0),\quad |u^h(x_N)|\leq v^h(x_N). \tag{65}
$$

Then

$$
-v^h\leq u^h\leq v^h\quad\text{on }\bar{\omega}^h. \tag{66}
$$

**Proof.** Apply Theorem 11 to $u^h$ and $v^h$, then to $-u^h$ and $v^h$. $\square$

### A priori estimate for the finite difference solution

To estimate the maximum absolute value of $u^h$ in terms of the data, we first establish two auxiliary results.

**Lemma 2 (effect of the boundary values).** The solution of

$$
L^h[v^h]=0\quad\text{on }\omega^h, \tag{67}
$$

$$
v^h(x_0)=u_a,\quad v^h(x_N)=u_b \tag{68}
$$

satisfies

$$
\max_{0\leq i\leq N}|v_i|\leq\max\{|u_a|,|u_b|\}. \tag{69}
$$

**Proof.** Use the constant grid function $w^h\equiv M$, where

$$
M=\max\{|u_a|,|u_b|\}. \tag{70}
$$

At interior nodes,

$$
|L^h[v^h]|=0\leq L^h[w^h]=q^hM, \tag{71}
$$

and at the boundary nodes,

$$
|v^h(x_0)|=|u_a|\leq M,\quad |v^h(x_N)|=|u_b|\leq M. \tag{72}
$$

Thus (66) gives $|v^h|\leq M$. $\square$

**Lemma 3 (effect of the right-hand side).** The solution of

$$
L^h[z^h]=f^h\quad\text{on }\omega^h, \tag{73}
$$

$$
z^h(x_0)=0,\quad z^h(x_N)=0 \tag{74}
$$

satisfies

$$
\max_{0\leq i\leq N}|z_i|\leq\frac{l^2}{8}\max_{0<i<N}|f_i|. \tag{75}
$$

**Proof.** Define the grid function

$$
w^h(x_i)=\frac A2(x_i-a)(b-x_i), \tag{76}
$$

where

$$
A=\max_{0<i<N}|f_i|. \tag{77}
$$

We have $w^h\geq0$, and the centered second difference is exact on quadratics. Hence

$$
L^h[w^h]=A+q^hw^h. \tag{78}
$$

The inequalities for (66) hold because

$$
|L^h[z^h]|=|f^h|\leq A\leq L^h[w^h], \tag{79}
$$

$$
|z^h(x_0)|=0=w^h(x_0),\quad |z^h(x_N)|=0=w^h(x_N). \tag{80}
$$

It follows that $|z^h|\leq w^h$. The continuous quadratic

$$
\frac A2(x-a)(b-x) \tag{81}
$$

attains its maximum $Al^2/8$ at $x=(a+b)/2$. Its nodal values satisfy

$$
0\leq w^h\leq\frac{Al^2}{8}. \tag{82}
$$

This proves (75). $\square$

**Theorem 12 (a priori estimate for the scheme).** The solution of (33), (34) satisfies

$$
\max_{0\leq i\leq N}|u_i|\leq \max\{|u_a|,|u_b|\}+\frac{l^2}{8}\max_{0<i<N}|f_i|. \tag{83}
$$

**Proof.** By linearity and uniqueness,

$$
u^h=v^h+z^h, \tag{84}
$$

where $v^h$ solves (67), (68) and $z^h$ solves (73), (74). Using

$$
\max_{0\leq i\leq N}|u_i|\leq\max_{0\leq i\leq N}|v_i|+\max_{0\leq i\leq N}|z_i| \tag{85}
$$

and applying (69) and (75) gives (83). $\square$

### Stability

Let $u^h$ solve (33), (34) and let $u^{*h}$ solve the same scheme with perturbed data,

$$
L^h[u^{*h}]=f^{*h}\quad\text{on }\omega^h, \tag{86}
$$

$$
u^{*h}(x_0)=u_a^*,\quad u^{*h}(x_N)=u_b^*, \tag{87}
$$

where

$$
f^{*h}=f^h-\delta f^h,\quad u_a^*=u_a-\varepsilon_a,\quad u_b^*=u_b-\varepsilon_b. \tag{88}
$$

**Definition 6 (stability).** The finite difference scheme (33), (34) is stable in the maximum norm if for arbitrary data perturbations

$$
\max_{0\leq i\leq N}|u_i-u_i^*|\leq\max\{|\varepsilon_a|,|\varepsilon_b|\}+K\max_{0<i<N}|\delta f_i|, \tag{89}
$$

where $u_i^*=u^{*h}(x_i)$ and $K$ is independent of $h$. This is the discrete analogue of (17).

**Theorem 13 (stability of the finite difference scheme).** For (33), (34), estimate (89) holds with

$$
K=\frac{l^2}{8}. \tag{90}
$$

**Proof.** The difference

$$
\varepsilon^h=u^h-u^{*h} \tag{91}
$$

solves

$$
L^h[\varepsilon^h]=\delta f^h\quad\text{on }\omega^h, \tag{92}
$$

$$
\varepsilon^h(x_0)=\varepsilon_a,\quad \varepsilon^h(x_N)=\varepsilon_b. \tag{93}
$$

Apply Theorem 12 to $\varepsilon^h$. $\square$

**Remark 5.** For $k\equiv1$ this is the same upper bound on amplification of input errors as in the continuous estimate (17).

### Approximation

Let $u$ be the solution of the differential equation (21) and let $U^h(x_i)=u(x_i)$ denote its restriction to the grid.

**Definition 7 (approximation error and order).** The *approximation error* of the difference equation (33) is the residual obtained by substituting the exact values,

$$
\psi^h=L^h[U^h]-f^h\quad\text{on }\omega^h. \tag{94}
$$

Equivalently,

$$
L^h[U^h]=f^h+\psi^h\quad\text{on }\omega^h. \tag{95}
$$

The difference equation approximates (21) if

$$
\max_{0<i<N}|\psi_i|\to0\quad\text{as }h\to0,\quad \psi_i=\psi^h(x_i). \tag{96}
$$

It approximates (21) with order $m>0$ if, for all sufficiently small $h$,

$$
\max_{0<i<N}|\psi_i|\leq Ch^m \tag{97}
$$

with $C$ independent of $h$.

**Lemma 4 (second-order approximation).** If $q,f\in C^2[a,b]$, then

$$
\max_{0<i<N}|\psi_i|\leq\frac{M_4}{12}h^2,\quad M_4=\max_{[a,b]}|u^{(7)}|. \tag{98}
$$

**Proof.** Theorem 2 gives $u\in C^4[a,b]$. Substituting the exact values in (94) gives

$$
\begin{aligned}
\psi_i&=-D_h^2u(x_i)+q_i u(x_i)-f_i\\
&=-u''(x_i)-r_h(x_i)+q_i u(x_i)-f_i=-r_h(x_i).
\end{aligned} \tag{99}
$$

By (27),

$$
r_h(x_i)=\frac{u^{(7)}(\xi_i)}{12}h^2,\quad \xi_i\in(x_{i-1},x_{i+1}). \tag{100}
$$

Consequently,

$$
\max_{0<i<N}|\psi_i|=\max_{0<i<N}|r_h(x_i)|\leq\frac{M_4}{12}h^2. \tag{101}
$$

$\square$

The boundary values introduce no approximation error in this problem because $U^h$ satisfies (34) exactly.

### Convergence

**Definition 8 (error and convergence of the scheme).** The nodal error of (33), (34) is the grid function $\varepsilon^h$ with values

$$
\varepsilon_i=\varepsilon^h(x_i)=u(x_i)-u_i,\quad 0\leq i\leq N. \tag{102}
$$

The scheme converges if

$$
\max_{0\leq i\leq N}|\varepsilon_i|\to0\quad\text{as }h\to0, \tag{103}
$$

and converges with order $m>0$ if

$$
\max_{0\leq i\leq N}|\varepsilon_i|\leq Ch^m \tag{104}
$$

for all sufficiently small $h$, with $C$ independent of $h$.

**Theorem 14 (second-order convergence).** If $q,f\in C^2[a,b]$, the error of (33), (34) satisfies

$$
\max_{0\leq i\leq N}|u(x_i)-u_i|\leq Ch^2, \tag{105}
$$

where

$$
C=\frac{l^2}{96}\max_{[a,b]}|u^{(7)}|. \tag{106}
$$

**Proof.** Consider again the restriction of the exact solution to the grid,

$$
U^h(x_i)=u(x_i),\quad 0\leq i\leq N. \tag{107}
$$

By (95), $U^h$ solves a scheme of the form (86), (87) with

$$
f^{*h}=f^h+\psi^h,\quad u_a^*=u_a,\quad u_b^*=u_b. \tag{108}
$$

For the error

$$
\varepsilon^h=U^h-u^h, \tag{109}
$$

Theorem 13 gives

$$
\max_{0\leq i\leq N}|\varepsilon_i|\leq\frac{l^2}{8}\max_{0<i<N}|\psi_i|. \tag{110}
$$

Substitute (98) to obtain (105). $\square$

This argument explains how stability and second-order approximation give second-order convergence here.

### Runge error estimate

The a priori estimate (105) depends on a derivative of the exact solution. For practical error estimation, compare the solutions on two nested grids, one with step $h$ and the other with step $2h$.

**Proposition 1 (Runge estimate).** Suppose the signed nodal error has a leading term $c(x)h^2$ on both grids, with the same $c(x)$ at common nodes and a remainder $O(h^4)$. Subtracting the expansions on the two grids gives

$$
\varepsilon^h(x)=u(x)-u^h(x)=\frac{u^h(x)-u^{2h}(x)}{3}+O(h^4),\quad x\in\bar{\omega}^{2h}. \tag{111}
$$

The fraction is the *Runge estimate* for the signed error on the fine grid. It is available at the common nodes, including the two endpoints, where the error is zero. The global bound $O(h^2)$ alone does not imply the needed expansion.

**Remark 6.** For this scheme a sufficient condition for the expansion is $u\in C^6[a,b]$ and $q\in C^2[a,b]$. Indeed, $L^h[U^h]-f^h=-h^2u^{(7)}/12+O(h^4)$ at the interior nodes by Taylor's formula. Let $c$ solve $L[c]=-u^{(7)}/12$ with zero boundary values, where $k\equiv1$. Then $c\in C^4[a,b]$ by Theorem 2. Applying (89) to the remainder after subtracting the nodal values of $h^2c$ from $\varepsilon^h$ gives a uniform $O(h^4)$ bound.

### Numerical example

**Example 1.** Use (33), (34) with the fine step

$$
h_1=\frac18 \tag{112}
$$

to solve the boundary value problem

$$
-u''(x)+x^2u(x)=\left(\frac{\pi^2}{4}+x^2\right)\cos\frac{\pi x}{2},\quad 0<x<1, \tag{113}
$$

$$
u(0)=1,\quad u(1)=0. \tag{114}
$$

The right-hand side and coefficient are

$$
q(x)=x^2,\quad f(x)=\left(\frac{\pi^2}{4}+x^2\right)\cos\frac{\pi x}{2}. \tag{115}
$$

In (35), (36) the boundary rows become

$$
u_0=1,\quad u_N=0, \tag{116}
$$

and the interior rows become

$$
\begin{aligned}
-u_{i-1}+(2+h^2x_i^2)u_i-u_{i+1}
&=h^2\left(\frac{\pi^2}{4}+x_i^2\right)\cos\frac{\pi x_i}{2},\\
&\quad i=1,\ldots,N-1.
\end{aligned} \tag{117}
$$

For $h_1=1/8$ we obtain the following system. The coefficients and right-hand sides below are rounded for display.

$$
\begin{aligned}
u_0 &= 1,\\
-u_0+2.000244\,u_1-u_2 &= 0.03805180,\\
-u_1+2.000977\,u_2-u_3 &= 0.03652069,\\
-u_2+2.002197\,u_3-u_4 &= 0.03388273,\\
-u_3+2.003906\,u_4-u_5 &= 0.03002332,\\
-u_4+2.006104\,u_5-u_6 &= 0.02480991,\\
-u_5+2.008789\,u_6-u_7 &= 0.01811708,\\
-u_6+2.011963\,u_7-u_8 &= 0.00985519,\\
u_8 &= 0.
\end{aligned} \tag{118}
$$

The exact solution is

$$
u(x)=\cos\frac{\pi x}{2}. \tag{119}
$$

Indeed, $-u''=(\pi^2/4)u$ and the function has the boundary values (114). For the Runge estimate, solve the same problem with the coarse step

$$
h_2=2h_1=\frac14. \tag{120}
$$

The corresponding system is

$$
\begin{aligned}
u_0 &= 1,\\
-u_0+2.003906\,u_1-u_2 &= 0.14608274,\\
-u_1+2.015625\,u_2-u_3 &= 0.12009330,\\
-u_2+2.035156\,u_3-u_4 &= 0.07246831,\\
u_4 &= 0.
\end{aligned} \tag{121}
$$

We solve both systems by the Thomas algorithm using floating-point values before display rounding. The output table gives the exact and fine-grid values and their signed difference. At common nodes the last two columns give the coarse-grid solution and the estimate (111).

```python
import numpy as np
```

```python
def u_exact(x):
    return np.cos(np.pi * x / 2)

def q(x):
    return x**2

def f(x):
    return (np.pi**2 / 4 + x**2) * u_exact(x)
```

```python
def solve_tridiag(a, b, c, d):
    n = len(d)
    alpha = np.zeros(n - 1)
    beta = np.zeros(n)

    alpha[0] = -c[0] / b[0]
    beta[0] = d[0] / b[0]

    for i in range(1, n - 1):
        gamma = b[i] + a[i] * alpha[i - 1]
        alpha[i] = -c[i] / gamma
        beta[i] = (d[i] - a[i] * beta[i - 1]) / gamma

    gamma = b[-1] + a[-1] * alpha[-1]
    beta[-1] = (d[-1] - a[-1] * beta[-2]) / gamma

    u = np.zeros(n)
    u[-1] = beta[-1]
    for i in range(n - 2, -1, -1):
        u[i] = alpha[i] * u[i + 1] + beta[i]

    return u

def solve_fd(n):
    h = 1 / n
    x = np.linspace(0, 1, n + 1)
    a = np.zeros(n + 1)
    b = np.ones(n + 1)
    c = np.zeros(n + 1)
    d = np.zeros(n + 1)
    d[0] = 1

    for i in range(1, n):
        a[i] = -1
        b[i] = 2 + h**2 * q(x[i])
        c[i] = -1
        d[i] = h**2 * f(x[i])

    return x, solve_tridiag(a, b, c, d)
```

```python
n_fine, n_coarse = 8, 4
x_fine, u_fine = solve_fd(n_fine)
x_coarse, u_coarse = solve_fd(n_coarse)
exact_fine = u_exact(x_fine)
error_fine = exact_fine - u_fine
runge = (u_fine[::2] - u_coarse) / 3

print(f"{'x_i':>6} {'u_exact':>12} {'u^h (1/8)':>12} {'eps_h':>12} {'u^(2h)':>12} {'Runge':>12}")
print("-" * 71)
for i, x in enumerate(x_fine):
    row = f"{x:6.3f} {exact_fine[i]:12.6f} {u_fine[i]:12.6f} {error_fine[i]:12.1e}"
    if i % 2 == 0:
        j = i // 2
        print(f"{row} {u_coarse[j]:12.6f} {runge[j]:12.1e}")
    else:
        print(f"{row} {'-':>12} {'-':>12}")
```

The table shows that the Runge estimate has the same sign as the error at the common interior nodes and is close to it in magnitude. We can also check the maximum interior residual on the fine grid and the convergence order from the maximum nodal errors on the two grids.

```python
h = 1 / n_fine
x = x_fine[1:-1]
second_difference = (u_fine[:-2] - 2 * u_fine[1:-1] + u_fine[2:]) / h**2
residual = -second_difference + q(x) * u_fine[1:-1] - f(x)
error_coarse = u_exact(x_coarse) - u_coarse
observed_order = np.log2(np.max(np.abs(error_coarse)) / np.max(np.abs(error_fine)))
print(f"Maximum interior residual = {np.max(np.abs(residual)):.2e}")
print(f"Observed order = {observed_order:.3f}")
```
