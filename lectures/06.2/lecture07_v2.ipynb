# Numerical methods for solving the Cauchy problem

## Using the Taylor Formula

One of the simplest approaches to understanding and solving the Cauchy problem is based on the Taylor formula

$$
\begin{gathered}
y(t+h) = y(t) + y'(t)h + \frac{y''(t)}{2}h^2 + \ldots\\\ldots + \frac{y^{(p)}(t)}{p!}h^p + R_{p+1}(t,h).
\end{gathered}
\tag{39}
$$

Here  
$$
R_{p+1}(t,h) = \frac{y^{(p+1)}(\xi)}{(p+1)!} h^{p+1}
\tag{f1}
$$  
is the remainder term of the Taylor formula, and $\xi$ is some point belonging to the interval $[t, t+h]$.

By discarding the remainder term, we obtain the approximate equality

$$
y(t+h) \approx y(t) + y'(t)h + \frac{y''(t)}{2}h^2 + \ldots + \frac{y^{(p)}(t)}{p!}h^p.
\tag{40}
$$

If the value of the solution $y$ at the point $t$ is known, then by virtue of the equality

$$
y'(t) = f(t, y(t))
\tag{41}
$$

the value of the derivative $y'(t)$ can also be considered known. To compute the higher-order derivatives $y''$, $y'''$, … that appear in formula (40), we differentiate equality (41) with respect to $t$, using the chain rule. Then we obtain

$$
y'' = f_t' + f_y' y' = f_t' + f_y' f,
\tag{42}
$$

$$
y''' = f_{tt}^{(2)} + f_{ty}^{(2)} y' + (f_{yt}^{(2)} + f_{yy}^{(2)} y') y' + f_y' (f_t' + f_y' y')
= f_{tt}^{(2)} + 2 f_{ty}^{(2)} f + f_{yy}^{(2)} f^2 + f_y' f_t' + (f_y')^2 f,
\tag{43}
$$

and so on. As is easy to see, the expressions for the derivatives $y^{(k)}$ become more complicated as the order $k$ increases.

Using the approximate formula (40) leads to the following explicit one-step method:

$$
y_{n+1} = y_n + y_n' h + \frac{y_n''}{2!} h^2 + \ldots + \frac{y_n^{(p)}}{p!} h^p.
\tag{44}
$$


Here $y_n' = f(t_n, y_n)$, and the values of $y_n''$ and $y_n^{(k)}$ are obtained by substituting $t = t_n$ and $y = y_n$ into formulas (42) and (43); similarly, the values of $y_n^{(k)}$ for $k > 3$ are computed.

The local error of this method $l_n$ coincides with the value  
$R_{p+1}(t_n, h)$ of the remainder term of the Taylor formula, which is proportional to $h^{p+1}$. Using this fact, one can prove that method (44) converges and has an accuracy order equal to $p$.

Despite the fact that this method theoretically makes it possible to obtain a solution with any desired order of accuracy, in practice it is used quite rarely. The reason is that the use of formula (44) leads to the necessity of computing a large number of partial derivatives  
$$
\frac{\partial^{l+s} f}{\partial t^l \, \partial y^s},
\tag{f2}
$$  
which is most often a very laborious and sometimes analytically infeasible operation.

A more significant argument against using method (44) is that by now effective numerical methods for solving the Cauchy problem have been developed (for example, Runge–Kutta and Adams methods), which require only the evaluation of the function $f$ and do not use its partial derivatives. It is these methods, implemented in standard programs and application software packages, that we will focus on further.

Nevertheless, for solving certain special classes of problems, the approximate method described above can be useful. In particular, it is used in solving certain problems of celestial mechanics, where the computation of the derivatives $y^{(k)}$ does not require significant additional effort due to the special structure of the right-hand sides $f(t, y)$.

## Euler's Method

### Geometric Interpretation of Euler's Method

The simplest and historically the first numerical method for solving the Cauchy problem (1), (2) is Euler's method. It can be obtained by keeping only the first two terms in the approximate equality (44) (i.e., by taking $p = 1$). Then formula (44) takes the form

$$
y_{n+1} = y_n + h f(t_n, y_n).
\tag{46}
$$

The geometric interpretation of one step of Euler's method consists in approximating the solution on the interval $[t_n, t_{n+1}]$ by the tangent  
$y = y_n + y'(t_n)(t - t_n)$ drawn at the point $(t_n, y_n)$ to the integral curve passing through this point (Fig. 14.7).

![](pictures/img01.png)

Thus, after performing $N$ steps, the unknown integral curve is replaced by a polyline (Euler polyline), in which the slope $k_n$ of each $n$-th segment is equal to the value $f(t_n, y_n)$ (Fig. 14.8).

Euler’s method is an explicit one-step method. For it, the approximation error has the form

$$
\Psi_n = \frac{h}{2} y''(\xi_n).
\tag{f3}
$$

### Stability

We will prove that Euler’s method is stable on a finite interval. First, we establish the validity of the following auxiliary statement.

**Lemma 1.** Let $z^n$ be a nonnegative grid function satisfying, for all $n \ge 0$, the inequality  
$$
z_{n+1} \le (1 + \alpha) z_n + \beta_n,
\tag{f4}
$$
where $\alpha \ge 0$, $\beta_n \ge 0$. Then for all $n \ge 0$ the estimate

$$
z_n \le e^{n\alpha} \left( z_0 + \sum_{k=0}^{n-1} \beta_k \right)
\tag{47}
$$

holds.

**Proof.** We establish the validity of inequality (47) by induction. For $n = 0$ it becomes the obvious $z_0 \le z_0$.

Assume now that inequality (47) holds for some $n = m$. Then, using the estimates  
$1 + \alpha \le e^\alpha$ and $1 \le e^{(m+1)\alpha} e^{-m\alpha}$, we obtain the following chain of inequalities

$$
\begin{gathered}
z_{m+1} \le (1 + \alpha) z_m + \beta_m \le e^\alpha z_m + e^{(m+1)\alpha} \beta_m \le\\
\le e^{\alpha} e^{m\alpha} \left( z_0 + \sum_{k=0}^{m-1} \beta_k \right)+\\
+ e^{(m+1)\alpha} \beta_m
= e^{(m+1)\alpha} \left( z_0 + \sum_{k=0}^{m} \beta_k \right),
\end{gathered}
\tag{f5}
$$

i.e., inequality (47) also holds for $n = m + 1$. Thus, it holds for all $n$. ■

Now let $y^{*\,h}$ be the solution of the perturbed discrete Cauchy problem

$$
y_{n+1}^{*\,h} = y_n^{*\,h} + h \bigl( f(t_n, y_n^{*\,h}) + \psi_n \bigr),
\tag{48}
$$

$$
y^{*\,h}(t_0) = y_0^{*}.
\tag{49}
$$

**Theorem 5.** Suppose the function $f$ satisfies the condition $|f_y'| \le L$. Then the inequality

$$
\max_{0 \le n \le N} |y_n^{*\,h} - y_n|
\le e^{L(T - t_0)} \left( |y_0^{*} - y_0| + h \sum_{k=0}^{N-1} |\psi_k| \right)
\tag{50}
$$

holds, which means that Euler’s method is stable on a finite interval.

**Proof.** Subtracting equation (46) from equation (48)  
and using Lagrange’s finite increment formula

$$
f(t_n, y_n^{*\,h}) - f(t_n, y_n) = f_y'(t_n, \widetilde{y}_n) (y_n^{*\,h} - y_n),
\tag{f6}
$$

we obtain

$$
y_{n+1}^{*\,h} - y_{n+1}
= (1 + h f_y'(t_n, \widetilde{y}_n)) (y_n^{*\,h} - y_n) + h \psi_n,
\tag{f7}
$$

from which it follows that

$$
|y_{n+1}^{*\,h} - y_{n+1}|
\le (1 + hL)\, |y_n^{*\,h} - y_n| + h |\psi_n|.
\tag{51}
$$

Now note that for $z_n = |y_n^{*\,h} - y_n|$, $\alpha = hL$, $\beta_n = h|\psi_n|$,  
inequality (51) implies

$$
z_{n+1} \le (1 + \alpha) z_n + \beta_n.
\tag{f8}
$$

According to Lemma 1, we have

$$
|y_n^{*\,h} - y_n|
\le e^{nhL} \left( |y_0^{*} - y_0| + h \sum_{k=0}^{n-1} |\psi_k| \right).
\tag{f9}
$$

Taking into account that $nh \le N h = T - t_0$, we arrive at inequality (50). ■


### Error Estimate

Since Euler’s method is stable on a finite interval and has first-order accuracy, it follows from Theorem 4 that it converges with first-order accuracy. More precisely, the following theorem holds.

**Theorem 6.** Suppose the function $f$ satisfies the condition $|f_y'| \le L$. Then for Euler’s method the following estimate of the global error holds:

$$
\max_{0 \le n \le N} |y(t_n) - y_n| \le C(T)\,h,
\tag{52}
$$

where  
$$
C(T) = e^{L(T - t_0)} (T - t_0) M_2 / 2,\qquad
M_2 = \max_{[t_0,\,T]} |y''(t)|.
\tag{f10}
$$

We present a proof of the theorem that does not use Theorem 4.

**Proof.** Let $\psi_n$ be the approximation error. Rewrite its definition (33) in the form

$$
y(t_{n+1}) = y(t_n) + h \bigl( f(t_n, y(t_n)) + \psi_n \bigr).
\tag{f11}
$$

Setting $y_n^{*} = y(t_n)$, we observe that the grid function $y^{*\,h}$ is the solution of the discrete Cauchy problem (48), (49), where $y_0^{*} = y_0$. Then, by Theorem 5, the estimate

$$
\max_{0 \le n \le N} |y(t_n) - y_n|
\le e^{L(T - t_0)} \sum_{n=0}^{N-1} |\psi_n|\, h
\tag{f12}
$$

holds.

Taking into account that  
$$
\sum_{n=0}^{N-1} |\psi_n|\, h \le \max_{0 \le n \le N} |\psi_n|\,(T - t_0),
\tag{f13}
$$  
and using estimate (34) for the approximation error, we obtain inequality (52). ■


## Runge–Kutta Methods

Among the classical explicit one-step methods, the most popular are the Runge–Kutta methods. The Euler, Euler–Cauchy, and improved Euler methods can be regarded as the simplest representatives of this class of methods.

### Derivation of Computational Formulas

Let us explain the derivation of the computational formulas of the Runge–Kutta method. Let $y(t) = y(t_n) = y_n$ be the solution of the differential equation $y' = f(t, y)$ satisfying the condition $y(t_n) = y_n$. Write equality (58) in the following form:

$$
y(t_{n+1}) = y(t_n) + \int\limits_{t_n}^{t_{n+1}} f(t, y(t))\, dt.
\tag{66}
$$

If the integral in this equality could be computed exactly, we would obtain a simple formula allowing sequential computation of solution values at grid nodes. Since this is impossible in practice, we attempt to obtain an approximate formula by replacing the integral with a quadrature sum.

On the interval $[t_n, t_{n+1}]$ introduce $m$ auxiliary nodes

$$
t_n^{(1)} = t_n + \alpha_1 h,\quad
t_n^{(2)} = t_n + \alpha_2 h,\quad \ldots,\quad
t_n^{(m)} = t_n + \alpha_m h,
\tag{f14}
$$

where $0 = \alpha_1 \le \alpha_2 \le \ldots \le \alpha_m \le 1$.  
Note that $t_n^{(1)} = t_n$, $t_n^{(m)} = t_{n+1}$. Replacing the integral in equality (66) with a quadrature sum with nodes $t_n^{(1)}, \ldots, t_n^{(m)}$, we obtain the approximate equality

$$
y(t_{n+1}) \approx y(t_n) + h \sum_{i=1}^{m} c_i f(t_n^{(i)},\, y(t_n^{(i)})).
\tag{67}
$$

However, equality (67) cannot be used to compute $y(t_{n+1})$, since the values of the function $y$ at the points $t_n^{(i)}$ for $i = 2, 3, \ldots, m$ are unknown. To find these values, write equalities analogous to (66):

$$
y(t_n^{(i)}) = y(t_n) + \int\limits_{t_n}^{t_n^{(i)}} f(t, y(t))\, dt,
\qquad (i = 2, 3, \ldots, m),
\tag{68}
$$

Replacing, for each $i$, the integral in formula (68) with the corresponding quadrature formula having nodes  
$t_n^{(1)}, t_n^{(2)}, \ldots, t_n^{(i)}$, we arrive at the approximate equalities

$$
y(t_n^{(2)}) \approx y(t_n) + h\, \beta_{21} f(t_n^{(1)}, y(t_n^{(1)})),
\tag{f15}
$$

$$
y(t_n^{(3)}) \approx y(t_n) + h(\beta_{31} f(t_n^{(1)}, y(t_n^{(1)})) + \beta_{32} f(t_n^{(2)}, y(t_n^{(2)}))),
\tag{f16}
$$

$$
\ldots\ldots\ldots
$$

$$
y(t_n^{(i)}) \approx y(t_n) + h \sum_{j=1}^{i-1} \beta_{ij} f(t_n^{(j)}, y(t_n^{(j)})),
\tag{f17}
$$

$$
\ldots\ldots\ldots
$$

$$
y(t_n^{(m)}) \approx y(t_n) + h \sum_{j=1}^{m-1} \beta_{mj} f(t_n^{(j)}, y(t_n^{(j)})),
\tag{f18}
$$

which allow sequential computation of approximations to  
$y(t_n^{(2)}), \ldots, y(t_n^{(m)})$.

Now denote by $y_n^{(i)}$ the auxiliary quantities representing approximations to the values $y(t_n^{(i)})$; let  
$k_n^{(i)} = f(t_n^{(i)}, y_n^{(i)})$ be the approximation to the slope $k$ at the point $t_n^{(i)}$. Then the computational formulas take the form:

$$
y_{n+1} = y_n + h k_n,\qquad k_n = \sum_{i=1}^{m} c_i k_n^{(i)},
\tag{f19}
$$

$$
k_n^{(1)} = f(t_n^{(1)}, y_n^{(1)}),\qquad y_n^{(1)} = y_n,
\tag{f20}
$$

$$
k_n^{(2)} = f(t_n^{(2)}, y_n^{(2)}),\qquad y_n^{(2)} = y_n + h \beta_{21} k_n^{(1)},
\tag{f21}
$$

$$
k_n^{(3)} = f(t_n^{(3)}, y_n^{(3)}),\qquad
y_n^{(3)} = y_n + h (\beta_{31} k_n^{(1)} + \beta_{32} k_n^{(2)}),
\tag{f22}
$$

$$
\ldots\ldots\ldots
$$

$$
k_n^{(m)} = f(t_n^{(m)}, y_n^{(m)}),\qquad
y_n^{(m)} = y_n + h \sum_{j=1}^{m-1} \beta_{mj} k_n^{(j)}.
\tag{f23}
$$

Often the auxiliary quantities $y_n^{(i)}$ are eliminated from these formulas, and the formulas are written as:

$$
y_{n+1} = y_n + h k_n,\qquad k_n = \sum_{i=1}^{m} c_i k_n^{(i)}.
\tag{f24}
$$
$$
k_n^{(1)} = f(t_n, y_n),
\tag{f25}
$$

$$
k_n^{(2)} = f(t_n + \alpha_2 h,\; y_n + h \beta_{21} k_n^{(1)}),
\tag{f26}
$$

$$
k_n^{(3)} = f(t_n + \alpha_3 h,\; y_n + h (\beta_{31} k_n^{(1)} + \beta_{32} k_n^{(2)})),
\tag{f27}
$$

$$
\ldots\ldots\ldots
$$

$$
k_n^{(m)} = f\!\left(t_n + \alpha_m h,\; y_n + h \sum_{j=1}^{m-1} \beta_{mj} k_n^{(j)}\right).
\tag{f28}
$$

Note that the derived formulas define an explicit one-step method of the form  
$$
y_{n+1} = y_n + h\, \Phi(t_n, y_n, h),
\tag{f29}
$$
where the computation of  
$$
\Phi(t_n, y_n, h) = \sum_{i=1}^{m} c_i k_n^{(i)}
\tag{f30}
$$
uses values of the right-hand side $f$ at $m$ auxiliary points. Therefore this method is called an **explicit $m$-stage Runge–Kutta method**.

The choice of specific values of the parameters $c_i$, $\alpha_i$, $\beta_{ij}$ is made on the basis of various considerations. Naturally, one of the main goals is to make the approximation order $p$ as high as possible.

---


### Stability and Convergence

The following theorem allows us to refer to Runge–Kutta methods having approximation order $p$ as methods of $p$-th order accuracy.

**Theorem 7.** Suppose the right-hand side of the differential equation satisfies the condition $|f_y'| \le L$. Then every explicit $m$-stage Runge–Kutta method is stable on a finite interval.

**Corollary.** Suppose the condition $|f_y'| \le L$ holds. Then if an explicit $m$-stage Runge–Kutta method has approximation order $p$, it converges with $p$-th order accuracy.

The validity of the corollary follows from Theorem 4.


### Family of Explicit Two-Stage Methods

We derive the computational formulas for a family of explicit two-stage Runge–Kutta methods of second-order accuracy. Write the formulas of an explicit two-stage method:

$$
y_{n+1} = y_n + h (c_1 k_n^{(1)} + c_2 k_n^{(2)}),
\tag{f31}
$$

$$
k_n^{(1)} = f(t_n, y_n), \qquad
k_n^{(2)} = f(t_n + \alpha h,\; y_n + h \beta k_n^{(1)}).
\tag{f32}
$$

in the form

$$
\frac{y_{n+1} - y_n}{h}
= c_1 f(t_n, y_n) + c_2 f(t_n + \alpha h,\; y_n + h\beta f(t_n, y_n)).
\tag{f33}
$$

The parameters of this method are the quantities $c_1$, $c_2$, $\alpha$, $\beta$.

Represent the approximation error

$$
\begin{gathered}
\psi = \frac{y(t+h) + y(t)}{h}-\\
    - c_1 f(t, y) - c_2 f(t+\alpha h,\; y + h\beta f(t,y)),
    \end{gathered}
\tag{f34}
$$

(where $t = t_n$, $y = y(t_n)$, $y(t_{n+1})$ is the solution of the differential equation  
$y' = f(t,y)$) in the form of an expansion in powers of $h$.

The Taylor formula:

$$
y(t+h) = y(t) + y'(t)h + \frac{y''(t)}{2}h^2 + O(h^3),
\tag{f35}
$$

taking into account the equalities $y' = f$, $y'' = f_t' + f_y' f$ (see (42)), gives the formula

$$
\frac{y(t+h) - y(t)}{h}
= f + \frac{1}{2}(f_t' + f_y'f)h + O(h^2)
\tag{f36}
$$

(the arguments $t$, $y$ of the function $f$ and its partial derivatives $f_t'$, $f_y'$ are omitted).

Represent the value of the function $f$ at the point $(t+\alpha h,\; y + h\beta f)$ using the Taylor formula for a function of two variables with center at the point $(t,y)$:

$$
f(t+\alpha h,\; y + h\beta f)
= f(t,y) + f_t' \alpha h + f_y' f\, \beta h + O(h^2).
\tag{f37}
$$

Thus,

$$
\begin{gathered}
\psi = (1 - c_1 - c_2) f
      + \{[(1/2 - c_2 \alpha) f_t' + (1/2 - c_2 \beta) f_y' f] h\}
      + O(h^2).
\end{gathered}
\tag{f38}
$$

If we require that the conditions  
$c_1 + c_2 = 1$, $c_2 \alpha = 1/2$, $c_2 \beta = 1/2$  
(which is equivalent to choosing $c_1 = 1 - 1/(2\alpha)$, $c_2 = 1/(2\alpha)$, $\beta = \alpha$)  
be satisfied, then the first two terms in the formula for $\psi$ vanish, and therefore the method will have second-order accuracy.

Thus (taking into account the corollary from Theorem 7), we can assert that for any  
$\alpha \in (0, 1]$ the method

$$
y_{n+1} = y_n + h\left[
\left(1 - \frac{1}{2\alpha}\right) f(t_n, y_n)
+ \frac{1}{2\alpha} f(t_n + \alpha h,\; y_n + \alpha h f(t_n, y_n))
\right]
\tag{69}
$$

has second-order accuracy.

Note that for $\alpha = 1$ formula (69) gives the Euler–Cauchy method, and for  
$\alpha = 1/2$ it gives the improved Euler method.


### Runge–Kutta Method of Fourth-Order Accuracy

The most well-known of the Runge–Kutta methods is the classical four-stage method of fourth-order accuracy:

$$
y_{n+1} = y_n + h k_n,\qquad
k_n = \frac{1}{6}\left( k_n^{(1)} + 2 k_n^{(2)} + 2 k_n^{(3)} + k_n^{(4)} \right);
$$

$$
k_n^{(1)} = f(t_n, y_n),\qquad
k_n^{(2)} = f\!\left(t_n + \frac{h}{2},\, y_n + \frac{h}{2} k_n^{(1)}\right),
$$

$$
k_n^{(3)} = f\!\left(t_n + \frac{h}{2},\, y_n + \frac{h}{2} k_n^{(2)}\right),\qquad
k_n^{(4)} = f(t_n + h,\, y_n + h k_n^{(3)}).
\tag{70}
$$

This method is very simple and, as practice shows, quite effective in ordinary computations when the interval $[t_0, T]$ is not very large and only relatively low accuracy is required.

**Remark.** Applying method (70) to the problem of computing the integral (36) yields Simpson’s formula:

$$
y_{n+1} = y_n + \frac{h}{6} \left( f(t_n) + 4 f(t_{n+1/2}) + f(t_{n+1}) \right).
\tag{f39}
$$

Thus, the classical fourth-order Runge–Kutta method (70) can be regarded as an analogue of Simpson’s formula corresponding to the solution of the Cauchy problem.


## Linear Multistep Methods. Adams Methods

### Adams Methods

In one-step methods, once the next value $y_n$ at the point $t_n$ has been found, the value $y_{n-1}$ is discarded and is no longer used in subsequent computations. It is natural, however, to try to extract some benefit from the information about the values of the solution $y_{n-k+1}, \ldots, y_{n-1}, y_n$ not at a single point, but at the previous points $t_{n-k+1}, \ldots, t_{n-1}, t_n$, i.e., to apply a multistep method.

Among multistep methods, the **Adams methods** are the most widely used in computational practice:

$$
\frac{y_{n+1} - y_n}{h}
= \sum_{j=0}^{k} \beta_j f_{n+1-j};
\tag{75}
$$

here $\beta_0, \beta_1, \ldots, \beta_k$ are numerical coefficients,  
$f_{n+1-j} = f(t_{n+1-j},\, y_{n+1-j})$.

Equation (75) allows one to find the new value $y_{n+1}$ using the previously found values $y_n, y_{n-1}, \ldots, y_{n-k+1}$. Therefore, one must first specify $k$ initial values $y_0, y_1, \ldots, y_{k-1}$.

If $\beta_0 = 0$, the Adams method is **explicit**, since the value $y_{n+1}$ is expressed through the previously computed values by the formula

$$
y_{n+1} = y_n + h \sum_{j=1}^{k} \beta_j f_{n+1-j}.
\tag{76}
$$

If $\beta_0 \ne 0$, then to find $y_{n+1}$ one must solve the nonlinear equation

$$
y_{n+1} = y_n + h \beta_0 f(t_{n+1},\, y_{n+1}) + g_n,
\tag{77}
$$

where  

$$
g_n = y_n + h \sum_{j=1}^{k} \beta_j f_{n+1-j}
\tag{f40}
$$

is a known value. Therefore, when $\beta_0 \ne 0$, the Adams method (75) is **implicit**.

There exist various ways to derive formulas (75). We present two of them. As in § 6, we use the equality

$$
y(t_{n+1}) = y(t_n) + \int_{t_n}^{t_{n+1}} f(t, y(t))\, dt.
\tag{78}
$$

Approximate the function $F(t) = f(t, y(t))$ by the interpolation polynomial $P_k(t)$ of degree $(k-1)$ that takes the values  
$f_{n+1-j}$ at the nodes $t_{n+1-k}, \ldots, t_n, t_{n+1}$, where the values of the grid function $y_n$ are already known. Integrating this polynomial gives the approximate equality

$$
\int_{t_n}^{t_{n+1}} f(t, y(t))\, dt
\approx \int_{t_n}^{t_{n+1}} P_k(t)\, dt
= h \sum_{j=1}^{k} \beta_j f_{n+1-j}.
\tag{79}
$$


As a result of (78), we arrive at formula (76), corresponding to the
explicit $k$-step Adams–Bashforth method.

**Remark.** Since the polynomial $P_{k-1}$ is used to approximate
the function $F$ outside the interval on which its values are known,  
equality (79) is in fact based on extrapolation. Therefore, the
corresponding method is also called the *extrapolation Adams method*.

If, however, in the integral appearing in equality (78) we replace
the integrand by the interpolation polynomial of degree $k$,  
$Q_k(t)$, matching the values  
$f_{n+1-k}, \ldots, f_n, f_{n+1}$  
at the nodes  
$t_{n+1-k}, \ldots, t_n, t_{n+1}$,  
then we obtain the formula

$$
y_{n+1} = y_n + h \sum_{j=0}^{k} \beta_j f_{n+1-j},
\tag{80}
$$

corresponding to the $k$-step **Adams–Moulton method**. Note that this
method is implicit.

**Remark.** Method (80) is also commonly referred to as the *interpolation Adams method*.

We now derive the formulas of the two-step Adams–Bashforth method and the
one-step Adams–Moulton method.

The interpolation polynomials $P_1(t)$ and $Q_1(t)$ are:

$$
P_1(t) = f_n - \frac{t_n - t}{h} f_n + \frac{t_n - t}{h} f_{n-1},
\tag{f41}
$$

$$
Q_1(t) = f_n \frac{t_{n+1}-t}{h} + f_{n+1} \frac{t - t_n}{h}.
\tag{f42}
$$

Integrating them with respect to $t$ gives:

$$
\int_{t_n}^{t_{n+1}} P_1(t)\,dt
    = h\left(\frac{3}{2}f_n - \frac12 f_{n-1}\right),
\tag{f43}
$$

$$
\int_{t_n}^{t_{n+1}} Q_1(t)\,dt
    = h\left(\frac12 f_n + \frac12 f_{n+1}\right).
\tag{f44}
$$

Thus, the two-step Adams–Bashforth formula has the form

$$
y_{n+1} = y_n + \frac{h}{2}\left(3 f_n - f_{n-1}\right),
\tag{f45}
$$

and the one-step Adams–Moulton formula has the form

$$
y_{n+1} = y_n + \frac{h}{2}\left(f_n + f_{n+1}\right).
\tag{f46}
$$


**Proposition 1.** Let the solution of the Cauchy problem $y(t)$ be $k$ times continuously differentiable on the interval $[t_0, T]$. Then the $k$-step Adams–Bashforth method and the $(k-1)$-step Adams–Moulton method have approximation order equal to $k$.

The following theorem gives grounds to call Adams methods that have approximation order $p$ methods of $p$-th order accuracy.

**Theorem 8.** Suppose the condition $|f_y'| \le L$ holds. Then explicit Adams methods are stable on a finite interval. Moreover, if the condition
$$
h \le h_0 = \frac{1}{2|\beta_0|L}
\tag{f47}
$$
is satisfied, then implicit Adams methods are also stable on a finite interval. ■

**Remark.** If the condition $f_y' \le 0$ holds, then implicit Adams methods are stable for any $h$.

**Corollary.** Suppose the condition $|f_y'| \le L$ holds. Then if a $k$-step Adams method has approximation order $p$, and the initial values $y_1, y_2, \ldots, y_{k-1}$ are determined with $p$-th order accuracy, the method also converges with $p$-th order accuracy.

The corollary holds by virtue of Theorem 4.

\tag{f48}
$$

\tag{f49}
