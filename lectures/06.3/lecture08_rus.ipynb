# Устойчивость численных методов решения задачи Коши

Как в теории численных методов решения задачи Коши, так и в практическом плане вопросы об устойчивости методов к малым погрешностям задания начальных данных и правой части уравнения, а также об устойчивости к погрешностям вычислений являются одними из центральных. Рассмотрим некоторые естественные требования устойчивости, которые накладываются на дискретные методы

$$
\frac{1}{h} \sum_{j=0}^{k} \alpha_j\, y_{n+1-j}
    = \Phi(t_n, y_{n+1-k}, \ldots, y_n, y_{n+1}, h).
\tag{83}
$$

Уделим сначала основное внимание исследованию устойчивости дискретной задачи Коши для уравнения (83) к малым погрешностям в начальных данных. Пусть $y^h$ — решение дискретной задачи, соответствующее начальным значениям $y_0, y_1, \ldots, y_{k-1}$, а $y_n^{*h}$ — решение той же задачи, соответствующее начальным значениям $y_0^*, y_1^*, \ldots, y_{k-1}^*$.

Если отрезок $[t_0, T]$, на котором ищется решение, и шаг $h$ фиксированы, то для всякого приемлемого метода решения задачи Коши его решение непрерывным образом зависит от начальных значений. Более того, если погрешности $\varepsilon_i = y_i - y_i^*$ ($i = 0, 1, \ldots, k-1$) задания начальных данных достаточно малы, то погрешности значений $y_n^{*}$ можно оценить следующим образом:

$$
\max_{0 \le n \le N} |y_n^{*} - y_n|
    \le K^{h}(T)\,
       \max_{0 \le i \le k-1} |\varepsilon_i|.
\tag{84}
$$

Величина $K^{h}(T)$, входящая в правую часть неравенства (84), играет роль числа обусловленности метода. Подчеркнем, что в общем случае она зависит как от $T$, так и от $h$.

## Нуль-устойчивость.

Будем стремиться к тому, чтобы при сколь угодно малых значениях шага $h$ дискретная задача Коши не только имела близкое к $y(t)$ решение $y^{h}$, но и обладала другими важными свойствами.

свойствами, аналогичными свойствам исходной задачи. В силу неравенства (11) погрешность, внесенная в начальное значение задачи Коши, на отрезке $[t_0, T]$ возрастает не более, чем в $K(T)$ раз (где $K(T)$, вообще говоря, растет с ростом $T$), поэтому в общем случае рост величины $K^{h}(T)$ с ростом $T$ также допустим. Однако если коэффициент $K^{h}(T)$ может неограниченно возрастать при $h \to 0$, то уменьшение шага $h$ приводит не к уточнению решения, а, наоборот, к неограниченному росту погрешности. Таким образом, следует требовать, чтобы для дискретной задачи Коши при всех достаточно малых $h$ было выполнено неравенство

$$
\max_{0 \le n \le N} |y_n^{*} - y_n|
\le \widetilde{K}(T)\,
    \max_{0 \le i \le k-1} |\varepsilon_i|,
\tag{85}
$$

где $\widetilde{K}(T)$ не зависит от $h$.

Методы, для которых неравенство (85) выполнено в случае, когда решается задача Коши для однородного уравнения $y' = 0$, будем называть **нуль-устойчивыми**.

Чтобы отбросить те из методов (83), которые заведомо не обладают свойством нуль-устойчивости, применим метод (83) к решению задачи Коши для уравнения $y' = 0$. Тогда $f \equiv 0$, $\Phi \equiv 0$ и уравнение (83) принимает вид

$$
\alpha_0 y_{n+1} + \alpha_1 y_n + \ldots + \alpha_k y_{n-k+1} = 0.
\tag{86}
$$

Будем считать, что $\alpha_0 \ne 0$ и $\alpha_k \ne 0$. Такие уравнения называют **линейными однородными разностными уравнениями $k$-го порядка с постоянными коэффициентами**.

Пусть $y^{*h}$ — решение того же уравнения, соответствующее возмущенным начальным значениям $y_0^*, y_1^*, \ldots, y_{k-1}^*$. Тогда в силу линейности этих уравнений погрешность $\varepsilon_n = y_n - y_n^{*}$ также является его решением:

$$
\alpha_0 \varepsilon_{n+1} + \alpha_1 \varepsilon_n + \ldots + \alpha_k \varepsilon_{n-k+1} = 0.
\tag{87}
$$

Будем искать частное решение уравнения (87) в виде $\varepsilon_n = q^n$. Подставляя $\varepsilon_{n+1-j} = q^{n+1-j}$ ($j = 0, 1, \ldots, k$) в (87), и сокращая на общий множитель $q^{n+1-k}$, видим, что величина $q$ должна удовлетворять уравнению

$$
\alpha_0 q^{k} + \alpha_1 q^{k-1} + \ldots + \alpha_k = 0,
\tag{88}
$$

которое называют **характеристическим уравнением**, соответствующим методу (83). Многочлен  
$P(q) = \alpha_0 q^k + \alpha_1 q^{k-1} + \ldots + \alpha_k$  
называется **характеристическим многочленом**.


Приведем некоторые факты, известные из теории линейных разностных уравнений. Пусть $q$ — корень уравнения (88) (вообще говоря, комплексный). Тогда сеточная функция $\varepsilon_n = q^n$ является решением разностного уравнения (87). Если же $q$ — кратный корень кратности $m \ge 2$, то ему отвечают частные решения  
$q^n,\; n q^n,\; n^2 q^n,\; \ldots,\; n^{m-1} q^n$.  

Опишем теперь структуру общего решения разностного уравнения. Пусть $q_1, q_2, \ldots, q_r$ — корни характеристического уравнения, а $m_1, m_2, \ldots, m_r$ — их кратности ($m_1 + m_2 + \ldots + m_r = k$). Тогда всякое решение уравнения (87) может быть представлено в виде¹

$$
\varepsilon_n = 
\sum_{l=0}^{m_1-1} c_{1l} n^{l} q_1^{\,n}
\;+\;
\sum_{l=0}^{m_2-1} c_{2l} n^{l} q_2^{\,n}
\;+\;\ldots+\;
\sum_{l=0}^{m_r-1} c_{rl} n^{l} q_r^{\,n}.
\tag{89}
$$

В частности, если все корни $q_1, q_2, \ldots, q_k$ — простые, то для всякого решения уравнения (87) справедливо представление

$$
\varepsilon_n = c_1 q_1^{\,n} + c_2 q_2^{\,n} + \ldots + c_k q_k^{\,n}.\tag{f0}
$$

Оказывается, что наличие или отсутствие у метода (83) нуль-устойчивости определяется исключительно расположением корней характеристического уравнения на комплексной плоскости.

Будем говорить, что выполнено **корневое условие**, если все корни  
$q_1, q_2, \ldots, q_r$ характеристического уравнения лежат внутри или на границе единичного круга комплексной плоскости (т. е. удовлетворяют условию $|q_i| \le 1,\; i = 1, 2, \ldots, r$), причем на границе единичного круга нет кратных корней.  

Заметим, что в силу равенства (88) число $q = 1$ всегда является корнем характеристического уравнения.

**Теорема 9.**
Для того чтобы метод (83) обладал нуль-устойчивостью, необходимо и достаточно, чтобы выполнялось корневое условие.

**Доказательство.** Ограничимся доказательством необходимости выполнения корневого условия. Предположим, что метод обладает свойством нуль-устойчивости, а корневое условие не выполнено. Тогда характеристическое уравнение имеет либо корень $q$ такой, что $|q| > 1$, либо корень $q$ кратности $m \ge 2$ такой, что $|q| = 1$. В первом случае корень $q$ вносит в решение $\varepsilon_n$ (соответствующее заданным начальным значениям $\varepsilon_i= \varepsilon\, q^{\,i}$ ($i = 0, 1, \ldots, k-1$).  
Во втором случае решением, соответствующим заданию начальных значений  
$\varepsilon_i = \varepsilon\, i\, q^{\,i}$ ($i = 0, 1, \ldots, k-1$), является функция  
$\varepsilon^h$ со значениями $\varepsilon_n = \varepsilon\, n\, q^{\,n}$.  
И в том, и в другом случае благодаря выбору $\varepsilon$ начальные погрешности могут быть сделаны сколь угодно малыми, но в то же время $|\varepsilon_n| \to \infty$ при $n \to \infty$.

Учитывая, что  
$N = (T - t_0)/h \to \infty$ при $h \to 0$,  
получаем $|y_N^{*} - y_N| = |\varepsilon_N| \to \infty$ при $h \to 0$,  
т. е. неравенство (85) не может выполняться для всех $h$.  
Итак, необходимость корневого условия доказана. ■


**Теорема 10.** Методы Рунге—Кутты и Адамса обладают свойством нуль-устойчивости.

**Доказательство.**  
Методам Рунге—Кутты соответствует однородное разностное уравнение $y_{n+1} - y_n = 0$,  
а ему, в свою очередь, соответствует характеристическое уравнение 
$q - 1 = 0$. 
Последнее имеет один простой корень $q = 1$, 
т. е. корневое условие выполнено.

Аналогично, $k$-шаговому методу Адамса соответствует разностное уравнение 
$$
y_{n+1} - y_n + 0 \cdot y_{n-1} + \ldots + 0 \cdot y_{n-k+1} = 0,\tag{f1}
$$

а характеристическое уравнение 
$q^k - q^{k-1} = 0$.  

Последнее имеет два корня: 
$q_1 = 1$ и корень $q_2 = 0$ кратности $k-1$, 
т. е. корневое условие здесь также выполняется. ■


## Абсолютная устойчивость

Как уже отмечалось ранее, необходимость решения задачи Коши на больших временных отрезках $[t_0, T]$ возникает в самых различных областях науки и техники. Наибольший интерес при этом представляет изучение устойчивых решений. Отметим, что нуль-устойчивость гарантирует устойчивое развитие погрешностей при $h \to 0$ только в том случае, когда отрезок интегрирования $[t_0, T]$ фиксирован. Однако наличие нуль-устойчивости вовсе не исключает того, что сколь угодно малая погрешность в начальных значенияx при неограниченном росте $t_n$ (при $T \to \infty$) может приводить к сколь угодно большой погрешности решения. Значительную часть таких заведомо непригодных для решения задачи Коши на больших временных отрезках методов можно отбросить, если исследовать результат их применения к решению модельной задачи

$$
y' = \lambda y,\qquad y(t_0) = y_0.
\tag{91}
$$

Напомним, что решение этой задачи устойчиво по Ляпунову,  
если комплексный параметр $\lambda$ удовлетворяет условию $\text{Re} \lambda \le 0$,  
и асимптотически устойчиво, если $\text{Re} \lambda < 0$.

Большинство используемых дискретных методов (в том числе и методы Рунге—Кутты и Адамса) в применении к задаче (91) становятся линейными и приобретают вид

$$
\frac{1}{h}\sum_{j=0}^{k} \alpha_j\, y_{n+1-j}
    = \lambda \sum_{j=0}^{k} \beta_j(h\lambda)\, y_{n+1-j};
\tag{92}
$$

здесь $\beta_j(h\lambda)$ — некоторые зависящие от величины $z = \lambda h$ функции.

В силу линейности уравнения (92) ошибка 
$\varepsilon_n = y_n - y_n^{*}$, возникающая из-за погрешности в начальных значениях, удовлетворяет уравнению:
$$
\frac{1}{h} \sum_{j=0}^{k} \alpha_j\, \varepsilon_{n+1-j}
    = \lambda \sum_{j=0}^{k} \beta_j(h\lambda)\, \varepsilon_{n+1-j}.\tag{f2}
$$


Перепишем это уравнение в виде

$$
\sum_{j=0}^{k} \gamma_j(z)\,\varepsilon_{n+1-j}=0,
\tag{93}
$$

где 
$
\gamma_j(z)=\alpha_j - z\,\beta_j(z),\qquad z=\lambda h.
$

Заметим, что (93) — это линейное однородное разностное уравнение.  
Поэтому для того, чтобы при фиксированном $z$ погрешность $\varepsilon_n$ оставалась ограниченной при $n\to\infty$, **необходимо и достаточно** выполнение корневого условия для отвечающего уравнению (93) полинома.

Назовем метод (83) **абсолютно устойчивым для данного** $z = h\lambda$,  
если при этом $z$ **все корни полинома устойчивости**

$$
P(q, z)=\sum_{j=0}^{k}\gamma_j(z)\,q^{k-j}
\tag{94}
$$

лежат в комплексной плоскости **внутри единичного круга и на границе**  
этого круга **нет кратных корней**.

Множество $D$ точек комплексной плоскости, состоящее из тех $z$, для которых метод абсолютно устойчив, называют **областью абсолютной устойчивости метода**.


Предположим, что параметр $\lambda$, входящий в модельное уравнение (91), отрицателен. Тогда условие абсолютной устойчивости метода Эйлера $|1 + h\lambda| \le 1$ оказывается эквивалентным неравенству

$$
h \le h_0 = \frac{2}{|\lambda|}.\tag{96}
$$

Такое же ограничение на шаг возникает при использовании метода Эйлера–Коши и усовершенствованного метода Эйлера. В то же время метод Рунге–Кутты четвёртого порядка точности оказывается абсолютно устойчивым при выполнении чуть менее ограничительного условия

$$
h \le h_0 \approx \frac{2{,}8}{|\lambda|}.\tag{97}
$$

Отметим, что при $\lambda < 0$ неявный метод Эйлера и правило трапеций оказываются абсолютно устойчивыми при любых $h$.

### $A$-устойчивость

Для того чтобы исключить ограничение на шаг $h$ при решении устойчивой по Ляпунову модельной задачи (91), необходимо потребовать, чтобы область абсолютной устойчивости метода включала в себя полуплоскость $\operatorname{Re} z < 0$. Численный метод, обладающий таким свойством, называют **$A$-устойчивым**.

Примерами A-устойчивых методов служат неявный метод Эйлера и правило трапеций. В то же время метод Эйлера и метод Рунге–Кутты четвёртого порядка точности не являются $A$-устойчивыми.

# Решение задачи Коши для систем обыкновенных дифференциальных уравнений и дифференциальных уравнений *m*-го порядка

## Задача Коши для систем дифференциальных уравнений первого порядка. 
Как правило, возникающие на практике проблемы приводят к необходимости решать задачу Коши не для одного дифференциального уравнения, а для систем дифференциальных уравнений вида

$$
y_1'(t) = f_1\bigl(t,\, y_1(t), y_2(t), \ldots, y_m(t)\bigr),
$$

$$
y_2'(t) = f_2\bigl(t,\, y_1(t), y_2(t), \ldots, y_m(t)\bigr),
$$

$$
\vdots
$$

$$
y_m'(t) = f_m\bigl(t,\, y_1(t), y_2(t), \ldots, y_m(t)\bigr),
\tag{106}
$$

где $y_1(t), y_2(t), \ldots, y_m(t)$ — искомые функции, значения которых подлежат определению при $t \in [t_0, T]$.

В момент времени $t = t_0$ задаются **начальные условия**.


$$
y_1(t_0) = y_{10}, \quad
y_2(t_0) = y_{20}, \ldots,\,
y_m(t_0) = y_{m0},
\tag{107}
$$

определяющие начальное состояние физической системы, развитие которой описывается уравнениями (106).

Введем вектор-функции  
$y(t) = (y_1(t), y_2(t), \ldots, y_m(t))^T$,  
$f(t, y) = (f_1(t, y), f_2(t, y), \ldots, f_m(t, y))^T$  
и вектор $y_0 = (y_{10}, y_{20}, \ldots, y_{m0})^T$.  

Тогда задачу Коши (106), (107) можно записать в компактной форме:

$$
y'(t) = f(t, y(t)),
\tag{108}
$$

$$
y(t_0) = y_0.
\tag{109}
$$

Для того чтобы охватить ряд важных для технических областей задач (электротехника, радиотехника и др.), будем считать, что функции $y_i$ и $f_i$ ($i = 1,2,\ldots,m$) могут принимать комплексные значения.

## Разрешимость задачи Коши

Пусть $\Pi_T$ — множество таких точек $(t, y)$, для которых  
$t \in [t_0, T]$,  
а $y_1, y_2, \ldots, y_m$ — произвольные комплексные числа.  
Это множество будем называть *слоем*.

Будем, как и ранее, использовать обозначения:

$$
(y, z) = \sum_{i=1}^m y_i \,\overline{z_i},\tag{f4}
$$

$$
\|y\| = \left( \sum_{i=1}^m |y_i|^2 \right)^{1/2}\tag{f5}
$$

для скалярного произведения и евклидовой нормы $m$-мерных комплексных векторов.

Сформулируем аналог теоремы о разрешимости задачи Коши, но для многомерного случая.

---

**Теорема 15.**
Пусть вектор-функция $f(t, y)$ определена и непрерывна в слое $\Pi_T$. Предположим также, что она удовлетворяет условию Липшица:

$$
\|f(t, y_1) - f(t, y_2)\| \le L \|y_1 - y_2\|
\tag{110}
$$

для всех $t_0 \le t \le T$ и произвольных $y_1, y_2$, где $L > 0$ — некоторая постоянная (постоянная Липшица).

Тогда для каждого начального значения $y_0$ существует единственное решение $y(t)$ задачи Коши (108), (109), определенное на отрезке $[t_0, T]$.

---

**Замечание 1.** Можно показать, что если функции $f_1, f_2, \ldots, f_m$ непрерывно дифференцируемы по $y_1, y_2, \ldots, y_m$, то условие Липшица (110) выполняется с постоянной $L$ тогда и только тогда, когда матрица Якоби
$$
f_y'(t, y) =
\begin{bmatrix}
f'_{1\,y_1}(t, y) & f'_{1\,y_2}(t, y) & \cdots & f'_{1\,y_m}(t, y) \\
f'_{2\,y_1}(t, y) & f'_{2\,y_2}(t, y) & \cdots & f'_{2\,y_m}(t, y) \\
\vdots & \vdots & \ddots & \vdots \\
f'_{m\,y_1}(t, y) & f'_{m\,y_2}(t, y) & \cdots & f'_{m\,y_m}(t, y)
\end{bmatrix}\tag{f6}
$$

удовлетворяет неравенству  
$$\|f_y'(t, y)\| \le L.\tag{f7}$$

---

**Замечание 2.** 
Теорема 15 остаётся справедливой, если в её формулировке условие Липшица (110) заменить менее ограничительным **односторонним условием Липшица**:

$$
\operatorname{Re}\bigl(f(t, y_1) - f(t, y_2),\, y_1 - y_2 \bigr)
\le \sigma \|y_1 - y_2\|^2,
\tag{111}
$$

для некоторых $t_0 \le t \le T$ и любых $y_1, y_2$, где $\sigma$ — некоторая постоянная.

---

Назовём систему дифференциальных уравнений **диссипативной**,  
если вектор-функция удовлетворяет неравенству

$$
\operatorname{Re}\bigl(f(t, y_1) - f(t, y_2),\, y_1 - y_2 \bigr) \le 0\tag{f8}
$$

(т.е. если $f$ удовлетворяет одностороннему условию Липшица с постоянной $\sigma = 0$).


## Устойчивость решения задачи Коши  
Приведём аналог теоремы 3.

**Теорема 16.** Пусть выполнены условия теоремы 15. Далее, пусть $y(t)$ — решение задачи (108), (109), а $y^{*}(t)$ — решение задачи

$$
(y^{*})'(t) = f\bigl(t, y^{*}(t)\bigr) + \psi(t),
\tag{112}
$$

$$
y^{*}(t_0) = y_0^{*}.
\tag{113}
$$

Тогда справедлива оценка

$$
\max_{t_0 \le t \le T} \|y(t) - y^{*}(t)\|
\le
K(T)\!\left(
\|y_0 - y_0^{*}\|
+ \int_{t_0}^{T} \|\psi(t)\|\, dt
\right),
\tag{114}
$$

выражающая устойчивость на конечном отрезке $[t_0, T]$ решения задачи Коши по начальным значениям и правой части. Здесь

$$
K(T) = e^{L(T - t_0)}.\tag{f9}
$$

Если в теореме 15 условие Липшица (110) заменить односторонним условием (111), то оценка (114) будет выполнена с постоянной

$$
K(T) = e^{\sigma (T - t_0)} \quad \text{при } \sigma > 0,\tag{f10}
$$

и с постоянной

$$
K(T) = 1 \quad \text{при } \sigma \le 0.\tag{f11}
$$


**Следствие.** 
Если система (108) диссипативна, то справедлива оценка

$$
\max_{t_0 \le t \le T} \|y(t) - y^{*}(t)\|
\le
\|y_0 - y_0^{*}\| + \int_{t_0}^{T} \|\psi(t)\|\, dt.
\tag{115}
$$

**Замечание.** 
Можно показать, что если условие (111) выполнено с постоянной $\sigma < 0$, то для всех $t \in [t_0, T]$ справедлива оценка

$$
\|y(t) - y^{*}(t)\|
\le
e^{\sigma (t - t_0)} \|y_0 - y_0^{*}\|
+
\frac{1}{|\sigma|}
\max_{t_0 \le t' \le t} \|\psi(t')\|.
\tag{116}
$$

По аналогии со случаем одного дифференциального уравнения рассмотрим вопрос об устойчивости решения задачи Коши к возмущениям начальных данных при $T \to \infty$. 
Будем считать, что на каждом отрезке $[t_0, T]$ ($t_0 < T$ произвольно) неравенство (111) выполнено с некоторой постоянной $\sigma = \sigma(T)$. Тогда решение $y(t)$ определено для всех $t_0 \le t < \infty$.  
Пусть $y^{*}(t)$ — решение задачи (112), (113), отвечающее произвольному начальному значению $y_0^{*}$ и $\varphi(t)\equiv 0$. 
Будем называть решение задачи Коши (108), (109) **устойчивым по Ляпунову**, если справедлива оценка

$$
\max_{t_0 \le t' \le T} \|y(t) - y^{*}(t)\|
\le
K \|y_0 - y_0^{*}\|,\tag{f12}
$$

где постоянная $K$ не зависит от $T$.
Если дополнительно известно, что
$$
\|y(t) - y^{*}(t)\| \to 0 \quad \text{при } t \to \infty,\tag{f13}
$$

то решение называется **асимптотически устойчивым**.


**Замечание 1.** 
При $\psi(t) \equiv 0$ из неравенства (115) следует оценка

$$
\max_{t_0 \le t \le T} \|y(t) - y^{*}(t)\|
\le
\|y_0 - y_0^{*}\|,\tag{f14}
$$

поэтому всякое решение диссипативной системы является устойчивым по Ляпунову.

**Замечание 2.** 
Предположим, что одностороннее условие Липшица (111) для всех $t_0 \le t < \infty$ выполняется с одной и той же постоянной $\sigma < 0$. Тогда при $\psi(t) \equiv 0$ из неравенства (116) получаем оценку

$$
\|y(t) - y^{*}(t)\|
\le
e^{\sigma (t - t_0)} \|y_0 - y_0^{*}\|,\tag{f15}
$$

откуда следует, что  

$$
\|y(t) - y^{*}(t)\| \to 0, \quad t \to \infty,\tag{f16}
$$

т.е. решение $y(t)$ асимптотически устойчиво.


## Система линейных дифференциальных уравнений с постоянными коэффициентами

Рассмотрим систему

$$
y'(t) = A y(t),
\tag{117}
$$

являющуюся простейшим примером системы дифференциальных уравнений первого порядка; здесь $A$ — квадратная матрица порядка $m$.  
В теории численных методов решения систем дифференциальных уравнений система (117) играет роль, аналогичную той, которую при исследовании методов численного интегрирования одного уравнения $y' = f(t, y)$ выполняет модельное уравнение $y'(t) = \lambda y(t)$.

Напомним структуру решения системы линейных уравнений с постоянными коэффициентами в наиболее простом и важном случае, когда матрица $A$ имеет простую структуру. В этом случае существует набор $e_1, e_2, \ldots, e_m$ собственных векторов матрицы $A$, соответствующих собственным значениям $\lambda_1, \lambda_2, \ldots, \lambda_m$, который образует базис в пространстве $m$-мерных векторов. Матрица $P$, столбцами которой служат векторы $e_1, e_2, \ldots, e_m$, не вырождена и такова, что

$$
P^{-1} A P = \Lambda =
\begin{bmatrix}
\lambda_1 & 0        & \cdots & 0 \\
0         & \lambda_2 & \cdots & 0 \\
\vdots    & \vdots    & \ddots & \vdots \\
0         & 0         & \cdots & \lambda_m
\end{bmatrix}.
\tag{118}
$$

Обозначим через $z_1(t), z_2(t), \ldots, z_m(t)$ координаты вектора $y(t)$ в базисе $e_1, \ldots, e_m$. Так как

$$
y(t) = z_1(t) e_1 + z_2(t) e_2 + \cdots + z_m(t) e_m,
\tag{119}
$$

то вектор $y(t)$ связан с вектором $z(t) = (z_1(t), z_2(t), \ldots, z_m(t))^T$ равенством  
$y(t) = P z(t)$. Умножив обе части системы (117) слева на $P^{-1}$, получим для вектор-функции $z(t)$ систему уравнений

$$
z'(t) = \Lambda z(t).\tag{f17}
$$

В силу диагональной структуры матрицы $\Lambda$ эта система распадается на $m$ независимых дифференциальных уравнений

$$
z_i'(t) = \lambda_i z_i(t), \qquad i = 1, 2, \ldots, m.
\tag{120}
$$

Заметим, что уравнение (120) для $i$-й компоненты вектора $z$ есть модельное уравнение с параметром $\lambda = \lambda_i$.

Таким образом, в рассматриваемом нелинейном случае интегрирование систем линейных дифференциальных уравнений эквивалентно интегрированию $m$ модельных уравнений (120). Решая каждое из уравнений, получаем  
$$
z_i(t) = c_i\, e^{\lambda_i (t - t_0)}, 
\qquad c_i = z_i(t_0),\tag{f18}
$$
и в силу равенства (119) находим

$$
y(t) = \sum_{i=1}^m c_i\, e^{\lambda_i (t - t_0)} e_i,\tag{f19}
$$

где  
$$
c = (c_1, c_2, \ldots, c_m)^T = P^{-1} y_0.\tag{f20}
$$

Пусть  
$$
\varepsilon(t) = y(t) - y^{*}(t)\tag{f21}
$$
— погрешность решения, вызванная погрешностью начальных значений  
$\varepsilon_0 = y_0 - y_0^{*}$.  

В силу линейности системы (119) погрешность является решением той же системы:

$$
\varepsilon'(t) = A \varepsilon(t).\tag{f22}
$$

Следовательно,

$$
e(t) = \sum_{i=1}^m \alpha_i\, e^{\lambda_i (t - t_0)} e_i,
\tag{121}
$$

где  
$$
\alpha = (\alpha_1, \alpha_2, \ldots, \alpha_m)^T = P^{-1} \varepsilon_0.\tag{f23}
$$

Формула (121) позволяет сделать ряд важных выводов.  
В частности, из неё следует, что решение системы (117) с постоянной матрицей $A$ и простой структурой устойчиво по Ляпунову **тогда и только тогда**, когда  
$$
\operatorname{Re}\lambda_i \le 0 \quad \text{для всех } i = 1, 2, \ldots, m,\tag{f24}
$$
и **асимптотически устойчиво тогда и только тогда**, когда  
$$
\operatorname{Re}\lambda_i < 0 \quad \text{для всех } i = 1, 2, \ldots, m.\tag{f25}
$$

Отметим также, что в отличие от случая одного дифференциального уравнения для систем характерно наличие *m* временных постоянных  

$$
\tau_i = \frac{1}{|\operatorname{Re}\lambda_i|},
\qquad i = 1, 2, \ldots, m.\tag{f26}
$$

Наличие в задаче физических компонент с существенно различными временными постоянными может привести к серьёзным затруднениям при численном решении соответствующих задач. Более подробно этот вопрос рассматривается в §11.

---

Пусть теперь $y^{*}(t)$ — решение нелинейной системы

$$
(y^{*})'(t) = f(t, y^{*}(t)),
\tag{122}
$$

отвечающее возмущенному начальному условию $y^{*}(t_0) = y_0^{*}$.  

Вычитая из уравнения $y'(t) = f(t, y(t))$ уравнение (122) и используя приближённое равенство

$$
f(t, y(t)) - f(t, y^{*}(t))
\approx
f_y'(t, y(t)) \bigl( y(t) - y^{*}(t) \bigr),\tag{f27}
$$

получаем, что погрешность  
$$
\varepsilon(t) = y(t) - y^{*}(t)\tag{f28}
$$
удовлетворяет приближённому уравнению

$$
\varepsilon'(t) \approx A \varepsilon(t),\tag{f29}
$$

где  
$$
A = f_y'(t, y(t)).\tag{f30}
$$

Таким образом, можно предположить, что в малой окрестности точки $(\tilde t, y(\tilde t))$ эволюция погрешности $\varepsilon(t)$ происходит примерно так, как и для системы (119), т.е.

$$
e(t) \approx \sum_{i=1}^m \alpha_i\, e^{\lambda_i (t - \tilde t)} e_i.\tag{f31}
$$

Здесь $\lambda_i$ и $e_i$ ($i = 1, \ldots, m$) — собственные значения и собственные векторы матрицы  
$$
A = f_y'(\tilde t, y(\tilde t)).\tag{f32}
$$

## Понятие о численных методах решения задачи Коши для систем уравнений первого порядка.

Описанные выше применительно к решению задачи Коши для одного уравнения методы можно использовать и для систем уравнений первого порядка, причем форма их записи претерпевает минимальные изменения. Следует лишь заменить в расчетных формулах числа $y_n$ на векторы  

$$
y_n = (y_{1n}, y_{2n}, \ldots, y_{mn})^T,\tag{f33}
$$

функцию $f$ — на вектор-функцию $f$, и т.д.

В результате дискретное уравнение (18) преобразуется в систему дискретных уравнений:

$$
\frac{1}{h} \sum_{j=0}^{k} \alpha_j\, y_{n+1-j}
= 
\Phi(t_n, y_{n+1-k}, \ldots, y_n, y_{n+1}, h).\tag{f34}
$$

Например, расчетная формула метода Эйлера  
$$
y_{n+1} = y_n + h f(t_n, y_n)\tag{f35}
$$

применительно к решению системы (108) принимает вид

$$
y_{n+1} = y_n + h\, f(t_n, y_n).\tag{f36}
$$

Покоординатная запись этого соотношения выглядит так:

$$
\begin{aligned}
y_{1,n+1} &= y_{1n} + h f_1(t_n, y_{1n}, y_{2n}, \ldots, y_{mn}), \\
y_{2,n+1} &= y_{2n} + h f_2(t_n, y_{1n}, y_{2n}, \ldots, y_{mn}), \\
&\ \vdots \\
y_{m,n+1} &= y_{mn} + h f_m(t_n, y_{1n}, y_{2n}, \ldots, y_{mn}).
\end{aligned}\tag{f37}
$$

---

Аналогично, метод Рунге–Кутты четвёртого порядка точности (70) порождает для систем дифференциальных уравнений первого порядка следующий метод:

$$
y_{n+1} = y_n + h k_n,\tag{f38}
$$

где

$$
k_n = \frac{1}{6}\left(k_n^{(1)} + 2k_n^{(2)} + 2k_n^{(3)} + k_n^{(4)}\right),\tag{f39}
$$

а промежуточные векторы вычисляются по формулам:

$$
k_n^{(1)} = f(t_n, y_n),\tag{f40}
$$

$$
k_n^{(2)} = f\!\left(t_n + \frac{h}{2},\, y_n + \frac{h}{2} k_n^{(1)}\right),\tag{f41}
$$

$$
k_n^{(3)} = f\!\left(t_n + \frac{h}{2},\, y_n + \frac{h}{2} k_n^{(2)}\right),\tag{f42}
$$

$$
k_n^{(4)} = f\!\left(t_n + h,\, y_n + h k_n^{(3)}\right).\tag{f43}
$$


Теория численных методов решения задачи Коши для систем дифференциальных уравнений имеет много общего с соответствующей теорией решения задачи Коши для одного дифференциального уравнения.  
В частности, справедливы аналоги всех изложенных выше результатов, касающихся устойчивости и сходимости дискретных методов на конечном отрезке. Однако имеют место и существенные новые явления. Одним из таких эффектов — жёсткость.

Прежде чем переходить к его изложению, выясним, какие изменения появляются при применении дискретных методов к решению задачи Коши для системы уравнений с постоянными коэффициентами (117).

---

Рассмотрим линейный многошаговый метод

$$
\frac{1}{h} \sum_{j=0}^{k} \alpha_j\, y_{n+1-j}
=
\sum_{j=0}^{k} \beta_j\, A\, y_{n+1-j}.
\tag{123}
$$

Предположим для упрощения, что матрица $A$ имеет простую структуру.  

Положим  
$$
z_n = P^{-1} y_n,\tag{f44}
$$
где $P$ — матрица, удовлетворяющая равенству (118).  

Умножая обе части уравнения (123) на матрицу $P^{-1}$ слева и учитывая, что  

$$
y_{n+1-j} = P z_{n+1-j}, 
\qquad
P^{-1} A y_{n+1-j} = P^{-1} A P\, z_{n+1-j} = \Lambda z_{n+1-j},\tag{f45}
$$

получаем соотношение

$$
\frac{1}{h} \sum_{j=0}^{k} \alpha_j\, z_{n+1-j}
=
\sum_{j=0}^{k} \beta_j\, \Lambda z_{n+1-j}.\tag{f46}
$$

Так как матрица $\Lambda$ — диагональная, то это эквивалентно следующей системе уравнений для компонент вектора  
$$
z_n = (z_{1n}, z_{2n}, \ldots, z_{mn})^T:\tag{f47}
$$

$$
\frac{1}{h} \sum_{j=0}^{k} \alpha_j\, z_{i,\,n+1-j}
=
\sum_{j=0}^{k} \beta_j\, \lambda_i\, z_{i,\,n+1-j},
\qquad i = 1, 2, \ldots, m.
\tag{124}
$$

Заметим, что (124) есть то же самое, что результат применения линейного многошагового метода к решению уравнений (120).

---

Таким образом, если решение системы (117) устойчиво по Ляпунову, то для чтобы погрешности $\varepsilon_n$ оставались ограниченными при $n \to \infty$, необходимо потребовать, чтобы для всех $i = 1, \ldots, m$ величина  
$$
h \lambda_i\tag{f48}
$$
принадлежала области $D$ абсолютной устойчивости применяемого метода.

Можно показать, что такой же вывод справедлив и для метода Рунге–Кутты. Требование, чтобы  

$$
h \lambda_i \in D \quad \text{при всех } i = 1, \ldots, m,\tag{f49}
$$

для задач, допускающих собственные значения $A$-устойчивого типа, приводит к существенным ограничениям на длину шага $h$.

---

Предположим, например, что все собственные значения матрицы $A$ отрицательны. Тогда условие (96) абсолютной устойчивости метода Эйлера приводит к следующему ограничению на длину шага интегрирования:

$$
h \le h_0 = \min_{1 \le i \le m} \frac{2}{|\lambda_i|}.
\tag{125}
$$

Такое же ограничение на шаг возникает при использовании метода Эйлера–Коши и усовершенствованного метода Эйлера.  
Метод Рунге–Кутты четвёртого порядка точности, как следует из неравенства (97), оказывается абсолютно устойчивым при таком ограничении на длину шага:

$$
h \le h_0 = \min_{1 \le i \le m} \frac{2.8}{|\lambda_i|}.\tag{f50}
$$

Следовательно, для явных методов шаг интегрирования должен не превышать значения $h_0$, пропорционального наименьшей из временных постоянных системы.

## Сведение задачи Коши для уравнений *m*-го порядка к задаче Коши для системы уравнений первого порядка

Задача Коши для дифференциального уравнения *m*-го порядка состоит в нахождении функции $y(t)$, удовлетворяющей при $t \ge t_0$ дифференциальному уравнению

$$
y^{(m)}(t)
=
f\!\bigl(t, y(t), y'(t), \ldots, y^{(m-1)}(t)\bigr),
\tag{126}
$$

а при $t = t_0$ — начальным условиям

$$
y^{(0)}(t_0) = y_{10},
\quad
y^{(1)}(t_0) = y_{20},
\quad
\ldots,
\quad
y^{(m-1)}(t_0) = y_{m0}.
\tag{127}
$$

Рассмотрим функции  

$$
y_1(t) = y(t), \quad
y_2(t) = y'(t), \quad
\ldots, \quad
y_m(t) = y^{(m-1)}(t).\tag{f51}
$$

Заметим, что $y_k'(t) = y^{(k)}(t)$.  
Поэтому введённые функции удовлетворяют системе дифференциальных уравнений первого порядка:

$$
y_1' = y_2,\tag{f52}
$$

$$
y_2' = y_3,\tag{f53}
$$

$$
\vdots
$$

$$
y_{m-1}' = y_m,\tag{f54}
$$

$$
y_m' = f\!\bigl(t, y_1, y_2, \ldots, y_m\bigr).\tag{f55}
$$

Начальные условия (127) в новых обозначениях принимают вид

$$
y_1(t_0) = y_{10},\quad
y_2(t_0) = y_{20},\quad
\ldots,\quad
y_m(t_0) = y_{m0}.
\tag{129}
$$

**Пример 21.**
Для задачи Коши для дифференциального уравнения второго порядка  
$$
y'' = -25y + \sin t,
\qquad
y(0) = 0,\quad
y'(0) = 1\tag{f56}
$$

введённые новых искомых функций  
$$
y_1(t) = y(t), \qquad y_2(t) = y'(t)\tag{f57}
$$
сводится к эквивалентной задаче Коши для системы дифференциальных уравнений первого порядка

$$
y_1' = y_2, \qquad
y_2' = -25 y_1 + \sin t,
\qquad
y_1(0) = 0, \quad y_2(0) = 1.\tag{f58}
$$

---

Для решения задачи Коши (126), (127), приведённой к виду (128), (129), можно воспользоваться известными методами или даже готовыми программами.  
Часто именно так и поступают. Следует всё же иметь в виду, что вычисления можно организовать и так, что сведение уравнения (126) к системе (128) не потребуется.

Например, для решения дифференциального уравнения второго порядка  
$$
y'' = f(t, y)\tag{f59}
$$
используется ряд специальных методов. Одним из наиболее популярных среди них является **метод Нумерова** четвёртого порядка точности:

$$
\frac{y_{n+1} - 2y_n + y_{n-1}}{h^2}
=
\frac{1}{12}
\Bigl(
f(t_{n-1}, y_{n-1})
+ 10 f(t_n, y_n)
+ f(t_{n+1}, y_{n+1})
\Bigr).\tag{f60}
$$
