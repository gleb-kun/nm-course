# Composite Quadrature Formulas

## Main Information

Let us consider the quadrature formula

$$
\int_{c}^{d} f(x)\, dx \approx \sum_{k=1}^{m} A_k f(a_k),
\tag{1}
$$

where the interval $[c, d]$ is finite, and the nodes $a_k$ belong to $[c, d]$.  
In (1) and in the other quadrature formulas of this section, the weight function is assumed to be equal to one.

Suppose we need to evaluate the integral

$$
\int_{\gamma}^{\delta} \varphi(\xi)\, d\xi\tag{f1}
$$

using the quadrature formula (1), where the interval $[\gamma, \delta]$ is finite and does not coincide with $[c, d]$.  
Applying the linear substitution of the integration variable
$\xi = \gamma + q(x - c)$, where

$$
q = \frac{\delta - \gamma}{d - c},\tag{f2}
$$

we obtain

$$
\int_{\gamma}^{\delta} \varphi(\xi)\, d\xi
= \int_{c}^{d} \varphi\!\bigl(\gamma + q(x - c)\bigr)\, q\, dx.
\tag{2}
$$

Now the quadrature formula (1) can be applied to the integral on the right-hand side:

$$
\int_{c}^{d} \varphi\!\bigl(\gamma + q(x - c)\bigr)\, q\, dx
\approx
\sum_{k=1}^{m} A_k\, q \, \varphi\!\bigl(\gamma + q(a_k - c)\bigr).
\tag{3}
$$

From (2) and (3) we obtain the approximate equality

$$
\int_{\gamma}^{\delta} \varphi(\xi)\, d\xi \approx \sum_{k=1}^{m} \alpha_k \varphi(\xi_k),
\tag{4}
$$

where

$$
\alpha_k = A_k q,\qquad \xi_k = \gamma + q(a_k - c), \quad k = 1, 2, \dots, m,
\tag{5}
$$

and $q = (\delta - \gamma)/(d - c)$. Since $a_k \in [c, d]$, it follows that $\xi_k \in [\gamma, \delta]$.

---

### Definition
Quadrature formulas of the form (1) are called **similar** if the following equalities hold for their coefficients and nodes:

$$
\sum_{k=1}^{m} A_k = d - c, \qquad \sum_{k=1}^{m} A_k a_k = \frac{d^2 - c^2}{2}.
\tag{5'}
$$

---

### Composite quadrature formula
A composite quadrature formula for computing the integral

$$
\int_{a}^{b} f(x)\, dx\tag{f3}
$$

over a finite interval $[a, b]$ is constructed as follows.  
The interval $[a, b]$ is divided into $n$ subintervals of equal length, and the integral over each subinterval is approximated by a quadrature formula similar to formula (1). Composite quadrature formulas are sometimes called **extended** or **refined formulas**.

We obtain an explicit representation of the composite quadrature formula.  
Let $n$ be a natural number. We set

$$
h = \frac{b - a}{n}, \qquad x_i = a + ih, \quad i = 0, 1, \dots, n.
\tag{6}
$$

By the additivity property of the integral,

$$
\int_{a}^{b} f(x)\, dx = \sum_{i=1}^{n} \int_{x_{i-1}}^{x_i} f(x)\, dx.
\tag{7}
$$

To evaluate the integrals on the right-hand side of (7), we apply quadrature formulas similar to (1):

$$
\int_{0}^{1} \varphi(t)\, dt = \sum_{k=1}^{m} A_k \varphi(a_k) + S(\varphi),
\tag{8}
$$

where $a_k \in [0, 1]$ and $S(\varphi)$ is the error term of the formula.  
We assume that the subintervals $[x_{i-1}, x_i]$ are arranged in such a way that they do not overlap.  


To evaluate the integral

$$
h \int_{0}^{1} f(x_{i-1} + h t)\, dt,\tag{f4}
$$

we make use of the quadrature formula (8). We obtain the quadrature formula

$$
\int_{x_{i-1}}^{x_i} f(x)\, dx = h\sum_{k=1}^{m} A_k\, f(x_{i-1} + h a_k) + h\, S[f(x_{i-1} + h t)].
\tag{9}
$$

This is a quadrature formula of the same type as (8).

To obtain the composite quadrature formula, we replace the integrals on the right-hand side of equality (7) by the right-hand side of (9):

$$
\int_{a}^{b} f(x)\, dx = 
\sum_{i=1}^{n} \sum_{k=1}^{m} h\, A_k\, f(x_{i-1} + h a_k)
\;+\;
h \sum_{i=1}^{n} S[f(x_{i-1} + h t)].
\tag{10}
$$

This is the **composite quadrature formula** for computing the integral  
$\displaystyle \int_{a}^{b} f(x)\, dx$.

Let us introduce the notation for the quadrature sum of this formula:

$$
\sigma_{m,n}(f) = \sum_{i=1}^{n} \sum_{k=1}^{m} h\, A_k\, f(x_{i-1} + h a_k).
\tag{11}
$$

Then the quadrature formula (10) can be written in the form

$$
\int_{a}^{b} f(x)\, dx = \sigma_{m,n}(f) + R_n(f),
\tag{12}
$$

where

$$
R_n(f) = h \sum_{i=1}^{n} S[f(x_{i-1} + h t)]\tag{13}
$$

is the error term of the composite quadrature formula.

---

We now give an example of constructing a composite quadrature formula.  
As the basic quadrature formula, we take the trapezoidal rule:

$$
\int_{0}^{1} \varphi(t)\, dt \approx \frac{1}{2} \bigl(\varphi(0) + \varphi(1)\bigr).\tag{f5}
$$


We have

$$
\int_{a}^{b} f(x)\, dx = \sum_{i=1}^{n} \int_{x_{i-1}}^{x_i} f(x)\, dx
= \sum_{i=1}^{n} \int_{0}^{1} f(x_{i-1} + h t)\, dt
\approx
\sum_{i=1}^{n} \frac{h}{2}\,[f(x_{i-1}) + f(x_i)].\tag{f6}
$$

Combining like terms in the last sum, we obtain the **composite trapezoidal rule**:

$$
\int_{a}^{b} f(x)\, dx \approx \frac{b - a}{n}
\left[
\frac{f(a)}{2} + f(x_1) + \dots + f(x_{n-1}) + \frac{f(b)}{2}
\right],
\tag{14}
$$

where $x_i$ is defined by (6). The number of nodes is $n + 1$.

In the integral $\displaystyle \int_{x_{i-1}}^{x_i} f(x)\, dx$ we used the substitution  
$x = x_{i-1} + h t$, although we could have applied the trapezoidal rule directly to this integral:

$$
\int_{x_{i-1}}^{x_i} f(x)\, dx \approx \frac{h}{2}\,[f(x_{i-1}) + f(x_i)].\tag{f7}
$$

---

Next, we derive the **composite Simpson rule**.

Quadrature formula of Newton-Cotes for $n = 2$:

$$
\int_{a}^{b} f(x)\, dx \;\approx\; \frac{b - a}{6}\,\Bigl[f(a) + 4 f\!\left(\frac{a + b}{2}\right) + f(b)\Bigr].\tag{f8}
$$

We derive the composite Simpson quadrature formula.  
In the right-hand side of equality (7), apply Quadrature Simpson’s formula to the integral over $[x_{i-1}, x_i]$:

$$
\begin{gathered}
    \int_{a}^{b} f(x)\,dx
    = \sum_{i=1}^{n} \int_{x_{i-1}}^{x_i} f(x)\,dx\\
    \approx
    \sum_{i=1}^{n} \frac{h}{6}\Bigl[f(x_{i-1})
    + 4 f\!\left(x_{i-1}+\frac{h}{2}\right)
    + f(x_i)\Bigr].
    \end{gathered}\tag{f9}
$$

Collecting like terms in the last sum, we obtain
$$
\begin{gathered}
    \int_{a}^{b} f(x)\,dx \approx \frac{h}{6}\Bigl[
    f(a)
    + 4\bigl(f(x_{0}+\tfrac{h}{2})+\dots+f(x_{n-1}+\tfrac{h}{2})\bigr)\\
    + 2\bigl(f(x_{1})+\dots+f(x_{n-1})\bigr)
    + f(b)
\Bigr].
\end{gathered}\tag{f10}
$$

If we denote the midpoint of the subinterval $[x_{i-1}, x_i]$ by $x_{i-\tfrac12}$, then the nodes of the composite Simpson formula can be written as

$$
x_j = a + \frac{j h}{2} = a + \frac{j(b-a)}{2n}, \qquad j=0,1,\dots,2n.
\tag{15}
$$

The composite Simpson formula takes the form

$$
    \begin{gathered}
        \int_a^b f(x)\,dx \approx \frac{b-a}{6n}\!\left[
        f(a)
        + 4\bigl(f(x_{1/2})+f(x_{3/2})+\dots+f(x_{n-1/2})\bigr)\right.+\\+\left.
        + 2\bigl(f(x_1)+\dots+f(x_{n-1})\bigr)
        + f(b)
        \right],
    \end{gathered}\tag{16}
$$

where the nodes $x_j$ are defined by (15).  
The number of nodes is $2n+1$. In the same way one can obtain other composite Newton–Cotes formulas.

---

### Theorem 1
The algebraic degree of accuracy of the basic quadrature formula (8) coincides with that of the composite quadrature formula (12).

*Proof.* Let $d$ be the algebraic degree of accuracy of (8). This means that $S(p)=0$ when $\varphi(t)=p(t)$ is a polynomial of degree $\le d$, and $S(t^{d+1})\ne 0$.  
If $f(x)$ is a polynomial of degree $\le d$, then from (9) and (13) we get $R_n(f)=0$, so the composite quadrature is exact for all polynomials of degree $\le d$.  
It remains to show that $R_n(x^{d+1})\ne 0$. Setting $f(x)=x^{d+1}$ in (13), we obtain

$$
R_n(x^{d+1}) = h \sum_{i=1}^{n} S\!\left[(x_{i-1}+ht)^{d+1}\right]
= h \sum_{i=1}^{n} S\!\bigl(h^{d+1} t^{\,d+1}\bigr)
= n h^{d+2} S(t^{d+1}) \ne 0.\tag{f11}
$$

Поскольку рассматривается многочлен точно степени $d+1$, его разложение сдвигается, но базовая формула 
$S(⋅)$ точна только до степени $d$. Поэтому в разложении остается только "старший" член, создающий ошибку:
$S\!\left[(x_{i-1}+ht)^{d+1}\right] = S\!\bigl(h^{d+1} t^{\,d+1}\bigr)$. Функционал $S(⋅)$ линейный, соответственно получаем, выносим константу и получаем $n h^{d+2} S(t^{d+1})$.


Hence the two algebraic degrees of accuracy are the same. ∎

### Theorem 2
For the limit relation
$$
\lim_{n\to\infty} \sigma_{m,n}(f) \;=\; \int_a^b f(x)\,dx
\tag{17}
$$
to hold for every Riemann–integrable function $f$ on $[a,b]$, it is necessary and sufficient that the underlying quadrature formula (8) be exact for constants; equivalently,
$$
S(1)=0 \quad\Longleftrightarrow\quad \sum_{k=1}^{m} A_k = 1 .\tag{f12}
$$


*Proof.* In the quadrature sum (11) we change the order of summation:

$$
\sigma_{m,n}(f) = \sum_{k=1}^{m} A_k \sum_{i=1}^{n} \frac{b-a}{n}\,
f\!\left(x_{i-1} + \frac{b-a}{n}\, a_k \right).\tag{f13}
$$

Since $0 \le a_k \le 1$, the inner sum for fixed $k$ is a Riemann sum for the integral  
$\displaystyle \int_{a}^{b} f(x)\, dx$. Therefore, assuming $f(x)$ is Riemann integrable, we have

$$
\lim_{n\to\infty} \sigma_{m,n}(f)
= \sum_{k=1}^{m} A_k \lim_{n\to\infty} 
\sum_{i=1}^{n} \frac{b-a}{n}\,
f\!\left(x_{i-1} + \frac{b-a}{n}\, a_k \right)
= \sum_{k=1}^{m} A_k \int_{a}^{b} f(x)\, dx.\tag{f14}
$$

Thus, the limiting relation (17) holds **if and only if**
$$
\sum_{k=1}^{m} A_k = 1,\tag{f15}
$$
which is equivalent to the fact that the basic quadrature formula (8) is exact for the constant function 1.

We now state a theorem describing the error term of the composite quadrature formula.

### Theorem 3
Let the error of the basic quadrature formula (8) admit the representation

$$
S(\varphi) = c_r\, \varphi^{(r)}(\eta), \qquad \eta \in [0,1],
\tag{18}
$$

under the assumption that $\varphi^{(r)}(t)$ exists and is continuous on $[a,b]$ and  
$\varphi^{(r)}(t) \ne 0$. Then the error of the composite quadrature formula (12) admits the representation

$$
R_n(f) = c_r\, \frac{(b-a)^{\,r+1}}{n^{\,r}}\, f^{(r)}(\xi), \qquad \xi \in [a,b].
\tag{19}
$$


*Proof.* Using formula (13), we obtain

$$
R_n(f) = h \sum_{i=1}^{n} S\!\bigl(f(x_{i-1} + h t)\bigr).\tag{f16}
$$

From (18), each summand involves the derivative $(f(x_{i-1} + h t))^{(r)}$.  
Since $f(x)$ has a continuous $r$-th derivative on $[a,b]$ and using the chain rule
we obtain
$$
\varphi_i^{(r)}(t) = f^{(r)}(x_{i-1} + h t)\, h^{r},\tag{f17}
$$

By condition (18) we have

$$
S\!\bigl(f(x_{i-1} + h t)\bigr) = c_r\, h^{\,r}\, f^{(r)}(x_{i-1} + \eta_i h), 
\qquad \eta_i \in [0,1].\tag{f18}
$$

Using this relation in formula (13), we obtain

$$
R_n(f) = c_r\, h^{\,r+1} \sum_{i=1}^{n} f^{(r)}(x_{i-1} + \eta_i h).
\tag{20}
$$

Let

$$
m_r = \min_{x \in [a,b]} f^{(r)}(x), 
\qquad
M_r = \max_{x \in [a,b]} f^{(r)}(x).\tag{f19}
$$

Since $\eta_i \in [0,1]$, the points $x_{i-1} + \eta_i h$ $(i = 1,2,\dots,n)$ belong to $[a,b]$, and therefore

$$
m_r \;\le\; \frac{1}{n} \sum_{i=1}^{n} f^{(r)}(x_{i-1} + \eta_i h) \;\le\; M_r .
$$

By continuity of $f^{(r)}(x)$, the Cauchy mean value theorem implies that there exists a point $\xi \in [a,b]$
such that

$$
\frac{1}{n} \sum_{i=1}^{n} f^{(r)}(x_{i-1} + \eta_i h) = f^{(r)}(\xi).\tag{f20}
$$

Substituting this into (20), we arrive at representation (19).  
Thus, the theorem is proved. ∎

**Remark.** In the error representation (18) the constant $c_{r}$ is nonzero.  
At the nodes of the quadrature formula (8) we construct the polynomial

$$
\omega_{m}(t) = (t - a_{1})(t - a_{2}) \cdots (t - a_{m}).\tag{f21}
$$

For the polynomial $\omega_{m}^{2}(t)$ of degree $2m$, which has derivatives of all orders, representation (18) takes the form

$$
S(\omega_{m}^{2}) = c_{r}\,[\omega_{m}^{2}]^{(r)}(\eta), \qquad \eta \in [0,1].\tag{f22}
$$

Since $S(\omega_{m}^{2}) \ne 0$, it follows from this equality that $c_{r} \ne 0$ and $r \le 2m$.

Using Theorem 3, we obtain the error representations of the composite trapezoidal and Simpson quadrature formulas.  
We begin with the trapezoidal quadrature formula (14).  
The basic quadrature formula is the trapezoidal rule
$$
\int_{a}^{b} f(x)\,dx \;\approx\; \frac{b-a}{2}\,\bigl[f(a) + f(b)\bigr]
\tag{*.18}
$$
on the interval $[0,1]$, and its error has the representation
$$
R^{(1)}(f) = -\,\frac{(b-a)^{3}}{12}\, f''(\eta), 
\qquad \eta \in [a,b]\tag{*.25}
$$
for $a = 0, b = 1$:

$$
R^{(1)}(f) = -\frac{1}{12}\, f''(\eta), \qquad \eta \in [0,1],\tag{f23}
$$

so in (18) we have $r = 2, \; c_{2} = -1/12$.  
Assuming that the function $f(x)$ is twice continuously differentiable on $[a,b]$, we obtain from (19) the error formula for the composite trapezoidal rule:

$$
R_{n}^{(1)}(f) = -\frac{(b - a)^{3}}{12 n^{2}}\, f''(\xi), 
\qquad \xi \in [a,b].
\tag{21}
$$

Now we turn to the composite Simpson formula (16).  
The basic quadrature formula is Simpson’s rule on the interval $[0,1]$,
and its error is given by
$$
R^{(2)}(f) = -\,\frac{(b-a)^{5}}{2880}\, f^{(IV)}(\eta), 
\qquad \eta \in [a,b]\tag{*.30}
$$
for $a = 0, b = 1$:

$$
R_{1}^{(2)}(\varphi) = -\frac{1}{2880}\, \varphi^{(4)}(\eta), \qquad \eta \in [0,1].\tag{f24}
$$

In representation (18) we must take $r = 4, \; c_{4} = -1/2880$.  
If the function $f(x)$ has a continuous fourth derivative on $[a,b]$, then from (19) we obtain the error formula for the composite Simpson rule:

$$
R_{n}^{(2)}(f)
= -\frac{(b - a)^{5}}{2880\, n^{4}}\, f^{(4)}(\xi),
\qquad \xi \in [a,b].
\tag{22}
$$

Usually, in the error representation of the composite Simpson formula, instead of $n$ — the number of subintervals of length
$h = (b - a)/n$ — the number $m = 2n$ is used, corresponding to subintervals of length $h/2$.  
In this case, formula (22) takes the form

$$
R_{m}^{(2)}(f)
= -\frac{(b - a)^{5}}{180\, m^{4}}\, f^{(4)}(\xi),
\qquad \xi \in [a,b].
\tag{23}
$$


## On the Error of Quadrature Formulas

Consider the quadrature formula

$$
\int_{a}^{b} w(x)\, f(x)\, dx \;\approx\; \sum_{j=1}^{N} C_j\, f(x_j),
\tag{1}
$$

with real coefficients and nodes belonging to the interval of integration $[a,b]$.  
Theorem 1 below gives a representation of the error term of formula (1) in the Peano form.

---

### Theorem 1
Let the quadrature formula (1) be exact for all polynomials of degree not higher than $n$, and let the function $f(x)$ have a continuous derivative $f^{(n+1)}(x)$ on the interval $[a,b]$.  
Then the error of the quadrature formula (1) can be written as

$$
R(f) = \int_{a}^{b} w(x)\, f(x)\, dx \;-\; \sum_{j=1}^{N} C_j\, f(x_j).
\tag{2}
$$

The following representation holds:

$$
R(f) = \frac{1}{m!} \int_{a}^{b} K_{m+1}(t)\, f^{(m+1)}(t)\, dt,
\tag{3}
$$

where

$$
K_{m+1}(t) = R\!\left[(x - t)_{+}^{m}\right]
= \int_{a}^{b} w(x)\, (x - t)_{+}^{m}\, dx \;-\; \sum_{j=1}^{N} C_j\, (x_j - t)_{+}^{m},\tag{f1}
$$

and $(x - t)_{+}^{m}$ is the truncated power function.

---

### Proof
Since by the hypothesis of the theorem the function $f(x)$ has a continuous $(m+1)$-st derivative on $[a,b]$, it satisfies the Taylor representation (1.11.4) with replacement of $m$ by $m+1$:

$$
f(x) = \sum_{k=0}^{m} \frac{f^{(k)}(a)}{k!}\,(x-a)^{k}
+ \frac{1}{m!} \int_{a}^{x} (x - t)^{m} f^{(m+1)}(t)\, dt.\tag{f2}
$$

Here $(x - t)_{+}^{m}$ is the truncated power function defined in (1.9.4).

The error term (2) of the quadrature formula is a linear functional.  
Applying the functional $R$ to both sides of the Taylor expansion and using additivity and linearity, we obtain

$$
R(f) = R\!\left(\sum_{k=0}^{m} \frac{f^{(k)}(a)}{k!}\,(x-a)^{k}\right)
+ \frac{1}{m!}\, R\!\left( \int_{a}^{x} (x - t)^{m} f^{(m+1)}(t)\, dt \right).\tag{f3}
$$

Since the quadrature formula is exact for all polynomials of degree $\le m$,  
the first term on the right-hand side is zero, and we obtain

$$
R(f) = \frac{1}{m!}\, R\!\left( \int_{a}^{x} (x - t)^{m} f^{(m+1)}(t)\, dt \right).\tag{f4}
$$

The functional $R$ may be moved inside the integral:

$$
R(f)= \frac{1}{m!}\int_{a}^{b} R\!\left[(x - t)_{+}^{m}\right] f^{(m+1)}(t)\, dt.\tag{f5}
$$

Carrying $R$ under the integral sign is equivalent to interchanging the order of integration:  
from a double integral to a single integral squared, which yields representation (3) with

$$
K_{m+1}(t) = R\!\left[(x - t)_{+}^{m}\right].\tag{f6}
$$

Since the function $w(x)$ is integrable (we assume the existence of all moments of $w(x)$),  
the truncated power $(x - t)_{+}^{m}$ is piecewise continuous for $m \ge 1$, and $f^{(m+1)}(t)$ is continuous.  
For $m = 0$ the function $(x - t)_{+}^{0}$ is equal to 1 for $x > t$ and zero for $x \le t$.  
The interchange of the order of integration is therefore valid.  
Equality (6) completes the proof of the theorem.


**Definition.** The function $K_{m+1}(t)$ defined by (4) is called the **Peano kernel**.  
It is continuous for $m \ge 1$, but in general may be discontinuous for $m = 0$.

If $K_{m+1}(t)$ does not change sign on $[a,b]$, then we may apply the mean value theorem (Theorem 2.1) to integral (3), obtaining

$$
R(f) = \frac{f^{(m+1)}(\xi)}{m!} \int_{a}^{b} K_{m+1}(t)\, dt,
\qquad \xi \in [a,b].
\tag{7}
$$

Substituting $f(x) = x^{m+1}$ into (3), we get

$$
R(x^{m+1}) = (m+1) \int_{a}^{b} K_{m+1}(t)\, dt,\tag{f7}
$$

and equality (7) may be written in the form

$$
R(f) = \frac{f^{(m+1)}(\xi)}{(m+1)!}\, R(x^{m+1}).
\tag{8}
$$

Equality (8), like (7), is valid provided that $K_{m+1}(t)$ does not change sign on $[a,b]$.

---

Let $d$ be the algebraic degree of accuracy of the quadrature formula (1).  
By the assumption of Theorem 1, the formula is exact for all polynomials of degree $\le m$, where $m \le d$.  
If the kernel has constant sign (does not change sign) for $m+1 \le x^{m+1}$, then  

$$
R(x^{m+1}) = 0 \quad \Longrightarrow \quad \int_{a}^{b} K_{m+1}(t)\, dt = 0.\tag{f8}
$$

Hence, $K_{m+1}(t)$ must change sign on $[a,b]$.  
Thus, the sign–definiteness property of the Peano kernel can be used only for $m < d$.

---

### Example 1
We now illustrate Theorem 1.

Let us find the error term of the Newton–Cotes quadrature formulas of degree $d = 1$.  
The total number of nodes is $n = d + 1 = 2$.  
The algebraic degree of accuracy of this quadrature is $d = 1$, therefore $m = 1$, so the Peano kernel is of order $m+1 = 2$ and can be written as

$$
K_2(t) = R\!\left[(x - t)_{+}^{1}\right].\tag{f9}
$$

Assuming $f(x)$ has a continuous second derivative on $[a,b]$, we obtain a Peano representation for the trapezoidal rule.


According to formula (4), we find the Peano kernel:

$$
\begin{gathered}
K_{2}(t) = \int_{a}^{b} (x - t)_{+}\, dx = \frac{b - a}{2}\,\bigl[(a - t)_{+} + (b - t)_{+}\bigr]=\\
= \int_{t}^{b} (x - t)\, dx = \frac{b - a}{2}\,(b - t) = \frac{1}{2}(b - t)(a - t).
\end{gathered}\tag{f10}
$$

Here we used the definition of the truncated power function. The kernel $K_{2}(t)$ is negative on $(a, b)$, therefore we may apply formula (8). We have

$$
R^{(1)}(x^{2}) = \int_{a}^{b} x^{2}\, dx = \frac{b - a}{2}\,(a^{2} + b^{2})
= -\frac{(b - a)^{3}}{6},\tag{f11}
$$

and therefore,

$$
R^{(1)}(f) = -\frac{(b - a)^{3}}{12}\, f''(\xi), \qquad \xi \in [a, b].\tag{f12}
$$

### Example 2

Find the error formula for Simpson’s quadrature formula

$$
R^{(2)}(f) = \int_{a}^{b} f(x)\, dx - \frac{b - a}{6} \bigl[f(a) + 4 f(c) + f(b)\bigr],
\qquad c = (a + b)/2.\tag{f13}
$$

The algebraic degree of accuracy of Simpson’s formula is $d = 3$, so $0 \le m \le 3$. We take $m = 3$ and assume that $f(x)$ has a continuous fourth derivative on $[a, b]$.

Using formula (4):

$$
K_{4}(t) = R\bigl[(x - t)_{+}^{3}\bigr]
= \int_{a}^{b} (x - t)_{+}^{3}\, dx - \frac{b - a}{6}\,\bigl[(a - t)_{+}^{3} + 4(c - t)_{+}^{3} + (b - t)_{+}^{3}\bigr].\tag{f14}
$$

As a result of the calculations we get

$$
K_{4}(t) = \frac{1}{2}
\Bigl[
3(t - a)_{+}^{4} - 2(b - a)(t - a)_{+}^{3}
\Bigr], \qquad a \le t \le b.\tag{f15}
$$
which is the same as:
$$
K_{4}(t) =
\begin{cases}
\dfrac{1}{12}\,\bigl[3(t - a)^{4} - 2(b - a)(t - a)^{3}\bigr], & a \le t \le c, \\[6pt]
\dfrac{1}{12}\,\bigl[3(b - t)^{4} - 2(b - a)(b - t)^{3}\bigr], & c \le t \le b.
\end{cases}\tag{f16}
$$

It is easy to verify that $K_{4}(t) \le 0$ on $[a, b]$. We check this assertion for $a \le t \le c$. We have

$$
K_{4}(t) = \frac{(t - a)^{3}}{4} \Bigl(t - a - \frac{a + 2b}{3}\Bigr),\tag{f17}
$$

and since the factor

$$
t - \frac{a + 2b}{3} = t - c - \frac{b - a}{6}\tag{f18}
$$

is negative for $t \le c$, we have $K_{4}(t) < 0$ for $a < t < c$.  
By the same argument we verify that $K_{4}(t) < 0$ for $c < t < b$.

The kernel $K_{4}(t)$ does not change sign on $[a, b]$, therefore we may use formula (7). Indeed,

$$
\int_{a}^{b} K_{4}(t)\, dt = -\,\frac{(b - a)^{5}}{480},\tag{f19}
$$

so from (7) we obtain

$$
R^{(2)}(f) = -\,\frac{(b - a)^{5}}{2880}\, f^{(4)}(\xi), 
\qquad \xi \in [a, b],\tag{f20}
$$

---

We now give a theorem for estimating the quadrature error through the best uniform approximation of the function $f(x)$.

### Theorem 2
If the quadrature formula (1) is exact for all polynomials of degree not higher than $m$, and if $f(x)$ is continuous on $[a, b]$, then the error of the quadrature formula (2) satisfies the inequality

$$
|R(f)| \le \left( \int_{a}^{b} |w(x)|\, dx + \sum_{j=1}^{N} |C_{j}| \right) E_{m}(f),
\tag{9}
$$

where $E_{m}(f)$ is the best uniform approximation error of $f(x)$ on $[a, b]$.

---

**Proof.**
Let $P_{m}^{*}(x)$ be the polynomial of best uniform approximation of $f(x)$ on $[a, b]$. Then

$$
E_{m}(f) = \|f - P_{m}^{*}\|_{\infty}.\tag{f21}
$$

By the linearity of the functional $R(f)$ and the fact that the quadrature formula is exact for polynomials of degree not higher than $m$, we have

$$
R(f) = R(f - P_{m}^{*}) = R(f) - R(P_{m}^{*})
= \int_{a}^{b} w(x)\,(f - P_{m}^{*})(x)\, dx
- \sum_{j=1}^{N} C_{j}\,(f(x_{j}) - P_{m}^{*}(x_{j})).\tag{f22}
$$


Estimating the absolute values of the integral and the quadrature sum in this equality, we obtain

$$
|R(f)| \le
\max_{x \in [a,b]} |f(x) - P_{m}^{*}(x)| \cdot \int_{a}^{b} |w(x)|\, dx
\;+\;
\max_{1 \le j \le N} |f(x_{j}) - P_{m}^{*}(x_{j})|
\cdot \sum_{j=1}^{N} |C_{j}|.\tag{f21}
$$

From this it follows that inequality (9) holds, since by the assumption $x_{j} \in [a,b]$ we have

$$
\max_{1 \le j \le N} |f(x_{j}) - P_{m}^{*}(x_{j})|
\;\le\;
\|f - P_{m}^{*}\|_{C[a,b]}
= E_{m}(f).\tag{f22}
$$

The theorem is proved.

## Highest Algebraic Degree of Accuracy

For a weight $w(x)\ge 0$ on $[a,b]$ and $n$ nodes,
the Gaussian rule
$$
\int_a^b f(x)\,w(x)\,dx \approx \sum_{i=1}^n w_i\, f(x_i)\tag{f27}
$$
with $x_i$ the zeros of the orthogonal polynomial $p_n$ (w.r.t. $w$) has
$$
\deg_{\text{exact}} = 2n-1,\tag{f28}
$$
and no $n$-node rule can have a higher algebraic degree.  
Error for sufficiently smooth $f$:
$$
\int_a^b f(x)\,w(x)\,dx-\sum_{i=1}^n w_i f(x_i)
= \frac{f^{(2n)}(\xi)}{(2n)!}\int_a^b \bigl[p_n(x)\bigr]^2
\frac{w(x)}{\left(p_n'(x)\right)^2}\,dx.\tag{f29}
$$
