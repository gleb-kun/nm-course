# Composite quadrature formulas

We construct composite rules from a basic quadrature formula, study their convergence and remainders, and then examine Peano kernels, best-approximation bounds, and Gaussian quadrature.

All functions are real-valued, and all nodes and weights are real.

## Main information

Let us consider the quadrature formula

$$
\int_c^d f(x)\,dx\approx\sum_{k=1}^m A_kf(a_k).\tag{1}
$$

Here $c<d$ are finite, the $m\geq1$ real nodes $a_k$ belong to $[c,d]$, and the weights $A_k$ are real. In this section the weight function is one, and all integrands are real-valued and Riemann integrable.

Suppose we need to evaluate

$$
\int_\gamma^\delta\varphi(\xi)\,d\xi\tag{2}
$$

on a finite interval $\gamma<\delta$, possibly different from $[c,d]$. Apply the linear substitution

$$
\xi=\gamma+q(x-c),\quad q=\frac{\delta-\gamma}{d-c}>0.\tag{3}
$$

Then

$$
\int_\gamma^\delta\varphi(\xi)\,d\xi=\int_c^d\varphi\!\bigl(\gamma+q(x-c)\bigr)q\,dx.\tag{4}
$$

Applying (1) to the right-hand side gives

$$
\int_c^d\varphi\!\bigl(\gamma+q(x-c)\bigr)q\,dx\approx\sum_{k=1}^m qA_k\varphi\!\bigl(\gamma+q(a_k-c)\bigr).\tag{5}
$$

Therefore,

$$
\int_\gamma^\delta\varphi(\xi)\,d\xi\approx\sum_{k=1}^m\alpha_k\varphi(\xi_k),\tag{6}
$$

where

$$
\alpha_k=qA_k,\quad \xi_k=\gamma+q(a_k-c),\quad k=1,\ldots,m.\tag{7}
$$

The transformed nodes belong to $[\gamma,\delta]$.

**Definition 1 (affinely similar quadrature formulas).** The formulas (1) and (6) are called **affinely similar** if their nodes and weights are related by (7). The transformation preserves the algebraic degree of accuracy, because composition with an invertible affine map preserves polynomial degree.

**Proposition 1 (moment checks).** The formula (1) is exact for all linear polynomials if and only if

$$
\sum_{k=1}^m A_k=d-c,\quad\sum_{k=1}^m A_ka_k=\frac{d^2-c^2}{2}.\tag{8}
$$

**Proof.** These are the exactness conditions for $f(x)=1$ and $f(x)=x$. By linearity, together they are equivalent to exactness for every polynomial $\alpha+\beta x$. They describe polynomial exactness; affine similarity is specified by the node and weight transformation (7).

### Composite quadrature formulas

For the composite construction, first map the basic rule to the reference interval $[0,1]$. From now on, $A_k,a_k$ denote the weights and nodes of this rule:

$$
\int_0^1\varphi(t)\,dt=\sum_{k=1}^m A_k\varphi(a_k)+S(\varphi),\quad a_k\in[0,1].\tag{9}
$$

Here $S(\varphi)$ is the error, defined as the integral minus the quadrature sum. For fixed nodes and weights, $S$ is a linear functional.

**Definition 2 (composite quadrature formula).** Let $a<b$ be finite and $n\geq1$ an integer. Divide $[a,b]$ into $n$ subintervals with disjoint interiors:

$$
h=\frac{b-a}{n}>0,\quad x_i=a+ih,\quad i=0,1,\ldots,n.\tag{10}
$$

By additivity,

$$
\int_a^b f(x)\,dx=\sum_{i=1}^n\int_{x_{i-1}}^{x_i}f(x)\,dx.\tag{11}
$$

On each subinterval, set

$$
\varphi_i(t)=f(x_{i-1}+ht),\quad\int_{x_{i-1}}^{x_i}f(x)\,dx=h\int_0^1\varphi_i(t)\,dt.\tag{12}
$$

Using (9), we obtain

$$
\int_{x_{i-1}}^{x_i}f(x)\,dx=h\sum_{k=1}^m A_kf(x_{i-1}+ha_k)+hS(\varphi_i).\tag{13}
$$

Summing these identities gives

$$
\int_a^b f(x)\,dx=\sum_{i=1}^n\sum_{k=1}^m hA_kf(x_{i-1}+ha_k)+h\sum_{i=1}^n S(\varphi_i).\tag{14}
$$

The corresponding **composite quadrature sum** is

$$
\sigma_{m,n}(f)=\sum_{i=1}^n\sum_{k=1}^m hA_kf(x_{i-1}+ha_k).\tag{15}
$$

Composite rules are also called extended or refined rules. Their error is

$$
\int_a^b f(x)\,dx=\sigma_{m,n}(f)+R_n(f),\quad R_n(f)=h\sum_{i=1}^nS(\varphi_i).\tag{16}
$$

Adjacent subintervals share endpoints; repeated node contributions are combined by adding their weights.

**Example 1 (composite trapezoidal rule).** As the basic formula, take

$$
\int_0^1\varphi(t)\,dt\approx\frac{\varphi(0)+\varphi(1)}2.\tag{17}
$$

Then

$$
\begin{aligned}
\int_a^b f(x)\,dx
&=h\sum_{i=1}^n\int_0^1 f(x_{i-1}+ht)\,dt\\
&\approx\sum_{i=1}^n\frac h2\,[f(x_{i-1})+f(x_i)].
\end{aligned}\tag{18}
$$

Combining repeated endpoint contributions yields

$$
T_n(f)=h\left[\frac{f(a)}2+\sum_{i=1}^{n-1}f(x_i)+\frac{f(b)}2\right].\tag{19}
$$

There are $n+1$ distinct nodes. The same result follows by applying the trapezoidal rule directly on each $[x_{i-1},x_i]$.

**Example 2 (composite Simpson rule).** The single-panel Simpson formula is

$$
\int_\alpha^\beta f(x)\,dx\approx\frac{\beta-\alpha}{6}\left[f(\alpha)+4f\!\left(\frac{\alpha+\beta}{2}\right)+f(\beta)\right].\tag{20}
$$

Apply it to every interval $[x_{i-1},x_i]$ of width $h$:

$$
\int_a^b f(x)\,dx\approx\sum_{i=1}^n\frac h6\left[f(x_{i-1})+4f\!\left(x_{i-1}+\frac h2\right)+f(x_i)\right].\tag{21}
$$

Let $x_{i-1/2}=x_{i-1}+h/2$. Collecting the common endpoints gives

$$
Q_n^{\mathrm S}(f)=\frac h6\left[f(a)+4\sum_{i=1}^nf(x_{i-1/2})+2\sum_{i=1}^{n-1}f(x_i)+f(b)\right].\tag{22}
$$

To enumerate all nodes without changing the meaning of $x_i$, introduce the finer grid

$$
\begin{aligned}
z_j&=a+\frac{jh}{2},\quad j=0,1,\ldots,2n,\\
z_{2i}&=x_i,\quad i=0,1,\ldots,n,\\
z_{2i-1}&=x_{i-1/2},\quad i=1,\ldots,n.
\end{aligned}\tag{23}
$$

Thus

$$
Q_n^{\mathrm S}(f)=\frac h6\left[f(z_0)+4\sum_{i=1}^nf(z_{2i-1})+2\sum_{i=1}^{n-1}f(z_{2i})+f(z_{2n})\right].\tag{24}
$$

There are $2n+1$ distinct nodes. Here $n$ counts Simpson panels, so any positive integer $n$ is admissible. The number of fine subintervals is $2n$, which is even. Other composite Newton–Cotes rules are constructed in the same way.

**Theorem 1 (preservation of algebraic degree).** If the basic rule (9) on $[0,1]$ has finite algebraic degree of accuracy $d\geq0$, then the composite rule (15) has exactly the same degree for every $n\geq1$.

**Proof.** If $f$ is a polynomial of degree at most $d$, then every $\varphi_i(t)=f(x_{i-1}+ht)$ has degree at most $d$. Hence $S(\varphi_i)=0$ and $R_n(f)=0$.

For $f(x)=x^{d+1}$, expand $(x_{i-1}+ht)^{d+1}$. All terms except $h^{d+1}t^{d+1}$ have degree at most $d$ and are annihilated by $S$. Therefore

$$
\begin{aligned}
R_n(x^{d+1})
&=h\sum_{i=1}^nS\!\left[(x_{i-1}+ht)^{d+1}\right]\\
&=nh^{d+2}S(t^{d+1})\ne0.
\end{aligned}\tag{25}
$$

The last inequality follows from $S(t^{d+1})\ne0$ and $n,h>0$. Thus the composite degree is $d$.

**Theorem 2 (convergence for Riemann-integrable functions).** Fix the finite number $m$, nodes $a_k\in[0,1]$, and real weights $A_k$, independently of $n$. Then

$$
\lim_{n\to\infty}\sigma_{m,n}(f)=\int_a^b f(x)\,dx\tag{26}
$$

holds for every Riemann-integrable $f$ on $[a,b]$ if and only if the basic formula is exact for constants:

$$
S(1)=0\quad\Longleftrightarrow\quad\sum_{k=1}^m A_k=1.\tag{27}
$$

**Proof.** Changing the order of summation in (15), we obtain

$$
\sigma_{m,n}(f)=\sum_{k=1}^m A_k\left[h\sum_{i=1}^nf(x_{i-1}+ha_k)\right].\tag{28}
$$

For each fixed $k$, the bracket is a tagged Riemann sum with one point in every subinterval. Since $h\to0$, it converges to the integral of $f$. The outer sum is finite, so

$$
\lim_{n\to\infty}\sigma_{m,n}(f)=\left(\sum_{k=1}^m A_k\right)\int_a^b f(x)\,dx.\tag{29}
$$

If (27) holds, this is (26). Conversely, apply the required convergence to $f=1$. For every $n$,

$$
\sigma_{m,n}(1)=(b-a)\sum_{k=1}^m A_k,\tag{30}
$$

so convergence to $b-a>0$ forces $\sum A_k=1$. This also gives a useful normalization check for the composite weights.

**Theorem 3 (transfer of a single-point remainder).** Let $r\geq1$ be an integer. Suppose that, for every $\varphi\in C^r([0,1])$, the basic error admits the representation

$$
S(\varphi)=c_r\varphi^{(r)}(\eta_\varphi),\quad \eta_\varphi\in[0,1],\tag{31}
$$

where $c_r$ is independent of $\varphi$, while the point $\eta_\varphi$ may depend on it. If $f\in C^r([a,b])$, then

$$
R_n(f)=c_r\frac{(b-a)^{r+1}}{n^r}f^{(r)}(\xi)=c_r(b-a)h^rf^{(r)}(\xi),\quad \xi\in[a,b].\tag{32}
$$

**Proof.** The chain rule gives

$$
\varphi_i^{(r)}(t)=h^rf^{(r)}(x_{i-1}+ht).\tag{33}
$$

Apply (31) to each $\varphi_i$ and substitute into (16):

$$
R_n(f)=c_rh^{r+1}\sum_{i=1}^nf^{(r)}(x_{i-1}+h\eta_i),\quad \eta_i\in[0,1].\tag{34}
$$

All evaluation points belong to $[a,b]$, hence

$$
\min_{[a,b]}f^{(r)}\leq\frac1n\sum_{i=1}^nf^{(r)}(x_{i-1}+h\eta_i)\leq\max_{[a,b]}f^{(r)}.\tag{35}
$$

Since $f^{(r)}$ is continuous, the intermediate value theorem yields a point $\xi\in[a,b]$ such that

$$
\frac1n\sum_{i=1}^nf^{(r)}(x_{i-1}+h\eta_i)=f^{(r)}(\xi),\quad \xi\in[a,b].\tag{36}
$$

Using $nh=b-a$ in (34) gives (32). In particular,

$$
|R_n(f)|\leq |c_r|(b-a)h^r\max_{x\in[a,b]}|f^{(r)}(x)|.\tag{37}
$$

Thus the global error is $O(h^r)$ for a fixed function and interval. The theorem requires the single-point remainder hypothesis (31); no assumption that $f^{(r)}$ is nonzero is needed.

**Proposition 2 (the coefficient and derivative order).** Under the hypothesis (31), a finite $m$-node basic rule satisfies

$$
c_r=\frac{S(t^r)}{r!}\ne0,\quad r\leq2m,\quad d=r-1,\tag{38}
$$

where $d$ is its algebraic degree of accuracy.

**Proof.** Construct the nodal polynomial

$$
\omega_m(t)=\prod_{k=1}^m(t-a_k).\tag{39}
$$

Its square vanishes at every quadrature node, but is positive except at finitely many points. Therefore

$$
S(\omega_m^2)=\int_0^1\omega_m(t)^2\,dt>0.\tag{40}
$$

If $c_r=0$, (31) would give zero for this error. If $r>2m$, the $r$-th derivative of $\omega_m^2$ would vanish identically, leading to the same contradiction. Thus $c_r\ne0$ and $r\leq2m$.

For every polynomial of degree below $r$, (31) gives zero error, whereas $S(t^r)=c_rr!\ne0$. Hence $d=r-1$ and the coefficient formula in (38) follows.

**Example 3 (composite trapezoidal error).** For the trapezoidal rule on $[0,1]$, the single-panel remainder from the lecture on numerical integration gives

$$
S_{\mathrm T}(\varphi)=-\frac1{12}\varphi''(\eta),\quad \eta\in[0,1].\tag{41}
$$

Thus $r=2$ and $c_2=-1/12$. If $f\in C^2([a,b])$, Theorem 3 yields

$$
R_n^{\mathrm T}(f)=-\frac{(b-a)^3}{12n^2}f''(\xi)=-\frac{(b-a)h^2}{12}f''(\xi),\quad \xi\in[a,b].\tag{42}
$$

This is the error of (19); its absolute-value bound follows by replacing $f''(\xi)$ with $\max|f''|$.

**Example 4 (composite Simpson error).** For Simpson's basic rule on $[0,1]$,

$$
S_{\mathrm S}(\varphi)=-\frac1{2880}\varphi^{(4)}(\eta),\quad \eta\in[0,1].\tag{43}
$$

Take $r=4$ and $c_4=-1/2880$. If $f\in C^4([a,b])$, the error of (22), with $n$ panels of width $h$, is

$$
R_n^{\mathrm S}(f)=-\frac{(b-a)^5}{2880n^4}f^{(4)}(\xi)=-\frac{(b-a)h^4}{2880}f^{(4)}(\xi),\quad \xi\in[a,b].\tag{44}
$$

It is also common to count the fine subintervals instead of panels. Set

$$
N=2n,\quad s=\frac{b-a}{N}=\frac h2.\tag{45}
$$

The same error becomes

$$
R_n^{\mathrm S}(f)=-\frac{(b-a)^5}{180N^4}f^{(4)}(\xi)=-\frac{(b-a)s^4}{180}f^{(4)}(\xi).\tag{46}
$$

Thus the coefficient $1/2880$ accompanies the panel width $h$, whereas $1/180$ accompanies the fine spacing $s$. The fine-grid count $N$ must be even. This explains the two usual forms of the composite Simpson remainder.

## On the error of quadrature formulas

Now consider a weighted formula

$$
\int_a^b w(x)f(x)\,dx\approx\sum_{j=1}^NC_jf(x_j)\tag{47}
$$

on a finite interval $a<b$. Assume that $w$ is real and measurable, $\int_a^b|w(x)|\,dx<\infty$, the $N\geq1$ real nodes $x_j$ belong to $[a,b]$, and the coefficients $C_j$ are real. The weight may change sign in this section.

**Definition 3 (quadrature error functional).** For bounded measurable functions $g$, define

$$
R(g)=\int_a^b w(x)g(x)\,dx-\sum_{j=1}^NC_jg(x_j).\tag{48}
$$

The functional $R$ is linear. We will apply it to continuous functions, polynomials, and the truncated powers defined below.

**Definition 4 (truncated power).** For an integer $p\geq0$, set

$$
(x-t)_+^p=
\begin{cases}
(x-t)^p,&x>t,\\
0,&x\leq t,
\end{cases}\quad p=0,1,2,\ldots.\tag{49}
$$

In particular, $(x-t)_+^0$ is one for $x>t$ and zero for $x\leq t$. This convention specifies its value at the jump without using $0^0$.

**Definition 5 (Peano kernel).** The function

$$
K_{p+1}(t)=R[(x-t)_+^p]=\int_a^bw(x)(x-t)_+^p\,dx-\sum_{j=1}^NC_j(x_j-t)_+^p\tag{50}
$$

is the **Peano kernel** of order $p+1$. In this formula, $R$ acts on the variable $x$, while $t$ is fixed. For $p\geq1$, the kernel is continuous. For $p=0$, it is bounded and may have jumps at the quadrature nodes; it is still integrable.

**Theorem 4 (Peano representation).** Let the formula (47) be exact for every polynomial of degree at most $p$, where $p\geq0$. If $f\in C^{p+1}([a,b])$, then

$$
R(f)=\frac1{p!}\int_a^bK_{p+1}(t)f^{(p+1)}(t)\,dt.\tag{51}
$$

**Proof.** Taylor's formula with integral remainder gives

$$
f(x)=\sum_{k=0}^p\frac{f^{(k)}(a)}{k!}(x-a)^k+\frac1{p!}\int_a^x(x-t)^pf^{(p+1)}(t)\,dt.\tag{52}
$$

The last integral can be written over $[a,b]$ using $(x-t)_+^p$. Apply $R$ to the identity. The polynomial part is annihilated by exactness, so

$$
R(f)=\frac1{p!}R\!\left[\int_a^b(x-t)_+^pf^{(p+1)}(t)\,dt\right].\tag{53}
$$

We may interchange the integrations and the finite sum:

$$
\begin{aligned}
R(f)&=\frac1{p!}\int_a^b
\left[\int_a^bw(x)(x-t)_+^p\,dx-\sum_{j=1}^NC_j(x_j-t)_+^p\right]
f^{(p+1)}(t)\,dt\\
&=\frac1{p!}\int_a^bK_{p+1}(t)f^{(p+1)}(t)\,dt.
\end{aligned}\tag{54}
$$

Indeed, $|(x-t)_+^p|\leq(b-a)^p$ on the integration square, and $f^{(p+1)}$ is bounded. The double integral is therefore absolutely integrable because $w$ is absolutely integrable. The point-evaluation sum has finitely many terms. This also justifies the interchange for $p=0$, where the truncated power has a jump. The result is (51).

**Corollary 1 (a general derivative bound).** Under the assumptions of Theorem 4,

$$
|R(f)|\leq\frac{\|f^{(p+1)}\|_\infty}{p!}\int_a^b|K_{p+1}(t)|\,dt.\tag{55}
$$

**Proof.** Take absolute values in (51) and bound $|f^{(p+1)}(t)|$ by its maximum. This estimate is valid whether or not the kernel changes sign.

**Corollary 2 (a fixed-sign kernel).** Under the assumptions of Theorem 4, if $K_{p+1}$ does not change sign almost everywhere on $[a,b]$, then there exists $\xi\in[a,b]$ such that

$$
R(f)=\frac{f^{(p+1)}(\xi)}{p!}\int_a^bK_{p+1}(t)\,dt,\quad \xi\in[a,b].\tag{56}
$$

Equivalently,

$$
R(f)=\frac{f^{(p+1)}(\xi)}{(p+1)!}R(x^{p+1}),\quad \xi\in[a,b].\tag{57}
$$

**Proof.** Apply the weighted mean value theorem to (51), using continuity of $f^{(p+1)}$. If the kernel is zero almost everywhere, both sides are zero. To obtain the second form, substitute $f(x)=x^{p+1}$ into (51):

$$
R(x^{p+1})=(p+1)\int_a^bK_{p+1}(t)\,dt.\tag{58}
$$

Combining this identity with (56) gives (57).

**Proposition 3 (kernel sign and algebraic degree).** Let the quadrature formula have finite algebraic degree $d$. If $0\leq p<d$, its kernel $K_{p+1}$ must change sign. Among the admissible choices $0\leq p\leq d$, a nontrivial fixed-sign kernel can therefore occur only when $p=d$; its sign must still be checked.

**Proof.** Since $p+1\leq d$, exactness and (58) give

$$
p<d\quad\Longrightarrow\quad R(x^{p+1})=0\quad\Longrightarrow\quad\int_a^bK_{p+1}(t)\,dt=0.\tag{59}
$$

If this kernel had one sign almost everywhere, it would be zero almost everywhere. Theorem 4 would then give $R(f)=0$ for every $f\in C^{p+1}([a,b])$, including $x^{d+1}$. This contradicts the definition of degree $d$. Thus for $p<d$ the kernel has both positive and negative parts. At $p=d$, (58) gives a nonzero integral, but that alone does not establish a fixed sign.

**Example 5 (the trapezoidal Peano kernel).** Consider the trapezoidal rule with two endpoint nodes. Its algebraic degree is $d=1$, so take $p=1$ and assume $f\in C^2([a,b])$. From (50),

$$
\begin{aligned}
K_2(t)&=\int_a^b(x-t)_+\,dx-\frac{b-a}{2}\bigl[(a-t)_++(b-t)_+\bigr]\\
&=\frac{(b-t)^2}{2}-\frac{b-a}{2}(b-t)
=\frac12(b-t)(a-t),\quad a\leq t\leq b.
\end{aligned}\tag{60}
$$

The kernel is negative for $a<t<b$. We also have

$$
R_{\mathrm T}(x^2)=\int_a^bx^2\,dx-\frac{b-a}{2}(a^2+b^2)=-\frac{(b-a)^3}{6}.\tag{61}
$$

Using (57) yields

$$
R_{\mathrm T}(f)=-\frac{(b-a)^3}{12}f''(\xi),\quad\xi\in[a,b].\tag{62}
$$

This agrees with the single-panel trapezoidal remainder and with (42) at $n=1$. The node count here follows from the chosen trapezoidal rule; algebraic degree one alone does not determine a node count.

**Example 6 (the Simpson Peano kernel).** Consider

$$
R_{\mathrm S}(f)=\int_a^bf(x)\,dx-\frac H6[f(a)+4f(c)+f(b)],\quad H=b-a,\quad c=\frac{a+b}{2}.\tag{63}
$$

The algebraic degree is $d=3$. Take $p=3$ and $f\in C^4([a,b])$. Since $t\in[a,b]$, the term $(a-t)_+^3$ vanishes, and (50) gives

$$
K_4(t)=\frac{(b-t)^4}{4}-\frac H6\bigl[4(c-t)_+^3+(b-t)^3\bigr],\quad a\leq t\leq b.\tag{64}
$$

Expanding on the two half-intervals gives the symmetric expression

$$
K_4(t)=
\begin{cases}
\dfrac{(t-a)^3}{12}\,[3(t-a)-2H],&a\leq t\leq c,\\[5pt]
\dfrac{(b-t)^3}{12}\,[3(b-t)-2H],&c\leq t\leq b.
\end{cases}\tag{65}
$$

For the left half, put $u=t-a$; for the right half, put $u=b-t$. In either case,

$$
0<u\leq\frac H2\quad\Longrightarrow\quad 3u-2H\leq-\frac H2<0.\tag{66}
$$

Hence $K_4(t)<0$ throughout $(a,b)$, including the midpoint, and the two pieces agree at $c$. Direct integration gives

$$
\int_a^bK_4(t)\,dt=2\int_0^{H/2}\frac{u^3}{12}(3u-2H)\,du=-\frac{H^5}{480}.\tag{67}
$$

Substituting this into (56), with $p!=3!=6$, we obtain

$$
R_{\mathrm S}(f)=-\frac{(b-a)^5}{2880}f^{(4)}(\xi),\quad\xi\in[a,b].\tag{68}
$$

This recovers the Simpson remainder used in Example 4 and checks both the kernel normalization and the sign.

### Bounds through best uniform approximation

**Definition 6 (best uniform approximation error).** Let $\mathbb P_p$ be the polynomials of degree at most $p\geq0$. For $f\in C([a,b])$, set

$$
E_p(f)=\inf_{P\in\mathbb P_p}\|f-P\|_\infty=\|f-P_p^*\|_\infty.\tag{69}
$$

Here $\|g\|_\infty=\max_{x\in[a,b]}|g(x)|$, and $P_p^*$ is a best-approximating polynomial. Such a polynomial exists on a compact interval.

**Theorem 5 (best-approximation error bound).** If the weighted formula (47) is exact for all polynomials of degree at most $p$ and $f\in C([a,b])$, then

$$
|R(f)|\leq\left(\int_a^b|w(x)|\,dx+\sum_{j=1}^N|C_j|\right)E_p(f).\tag{70}
$$

**Proof.** Since $R(P_p^*)=0$, linearity gives

$$
R(f)=R(f-P_p^*)=\int_a^bw(x)(f-P_p^*)(x)\,dx-\sum_{j=1}^NC_j(f-P_p^*)(x_j).\tag{71}
$$

Estimate the integral and the finite sum separately. Because every $x_j\in[a,b]$,

$$
\begin{aligned}
|R(f)|&\leq\|f-P_p^*\|_\infty\int_a^b|w(x)|\,dx
+\max_{1\leq j\leq N}|f(x_j)-P_p^*(x_j)|\sum_{j=1}^N|C_j|\\
&\leq E_p(f)\left(\int_a^b|w(x)|\,dx+\sum_{j=1}^N|C_j|\right).
\end{aligned}\tag{72}
$$

This proves (70). Unlike the derivative estimates, this bound requires only continuity of $f$.

**Corollary 3 (nonnegative weights).** If additionally $w\geq0$, $C_j\geq0$, $\mu_0=\int_a^bw>0$, and the formula is exact for constants, then

$$
|R(f)|\leq2\mu_0E_p(f),\quad \mu_0=\int_a^bw(x)\,dx>0.\tag{73}
$$

**Proof.** Exactness for constants gives $\sum C_j=\mu_0$, so the coefficient in (70) equals $2\mu_0$.

## Highest algebraic degree of accuracy

Assume now that $w\geq0$ almost everywhere on a finite interval $[a,b]$, $\int_a^b|w|<\infty$, and $\mu_0=\int_a^bw(x)\,dx>0$. These assumptions make the polynomial inner product positive definite.

**Definition 7 (Gaussian quadrature).** For $n\geq1$, let $p_n$ be the **monic** orthogonal polynomial of degree $n$ for the weight $w$:

$$
p_n(x)=\prod_{i=1}^n(x-x_i),\quad\int_a^bw(x)p_n(x)q(x)\,dx=0\quad(q\in\mathbb P_{n-1}).\tag{74}
$$

Its $n$ roots $x_i$ are distinct and belong to $(a,b)$; these standard properties of orthogonal polynomials follow under the stated weight assumptions [1, 3]. Take them as the nodes and define

$$
\lambda_i=\int_a^b\ell_i(x)w(x)\,dx,\quad\ell_i(x)=\prod_{j\ne i}\frac{x-x_j}{x_i-x_j}.\tag{75}
$$

Then

$$
\int_a^bf(x)w(x)\,dx\approx G_n(f)=\sum_{i=1}^n\lambda_i f(x_i)\tag{76}
$$

is the **Gaussian quadrature formula**. For $w\equiv1$ on $[-1,1]$, the nodes are the zeros of the Legendre polynomial; its monic normalization is used in the error formula below.

**Theorem 6 (Gaussian degree and positivity).** Under the assumptions above, the Gaussian rule has

$$
d=2n-1,\quad\lambda_i>0,\quad\sum_{i=1}^n\lambda_i=\mu_0.\tag{77}
$$

No quadrature formula with $n$ real nodes can have a higher algebraic degree for this weight.

**Proof.** For $P\in\mathbb P_{2n-1}$, polynomial division gives

$$
P=qp_n+r,\quad q,r\in\mathbb P_{n-1}\quad(P\in\mathbb P_{2n-1}).\tag{78}
$$

Orthogonality gives $\int_a^bwqp_n=0$, and the nodal sum of $qp_n$ is zero. The interpolatory weights (75) integrate $r$ exactly, so $G_n$ integrates every such $P$ exactly.

In particular, exactness for $\ell_i^2$ gives

$$
\lambda_i=G_n(\ell_i^2)=\int_a^b\ell_i(x)^2w(x)\,dx>0,\tag{79}
$$

and exactness for constants gives $\sum\lambda_i=\mu_0$. To prove the upper bound for any real nodes $\zeta_i$, use

$$
\omega(x)=\prod_{i=1}^n(x-\zeta_i),\quad\int_a^b\omega(x)^2w(x)\,dx>0,\quad\sum_{i=1}^n C_i\omega(\zeta_i)^2=0.\tag{80}
$$

Thus a rule with $n$ nodes cannot be exact for every polynomial of degree $2n$. The Gaussian degree is exactly $2n-1$.

**Proposition 4 (Gaussian remainder).** Under the same weight assumptions, if $f\in C^{2n}([a,b])$, then

$$
R_{\mathrm G,n}(f)=\int_a^bf(x)w(x)\,dx-G_n(f)=\frac{f^{(2n)}(\xi)}{(2n)!}\int_a^bp_n(x)^2w(x)\,dx,\quad\xi\in[a,b].\tag{81}
$$

Here $p_n$ is monic. For an orthogonal polynomial with leading coefficient $\kappa_n$, its squared norm must be divided by $\kappa_n^2$.

**Proof.** Let $H_{2n-1}$ be the Hermite interpolating polynomial matching $f$ and $f'$ at all $n$ Gaussian nodes. Its degree is at most $2n-1$, so $\int wH_{2n-1}=G_n(H_{2n-1})=G_n(f)$. The Hermite remainder gives, with $\eta_x\in[a,b]$,

$$
f(x)-H_{2n-1}(x)=\frac{f^{(2n)}(\eta_x)}{(2n)!}p_n(x)^2\quad(x\ne x_i).\tag{82}
$$

Away from the nodes, define $q(x)=(2n)![f(x)-H_{2n-1}(x)]/p_n(x)^2$. It is continuous there and bounded between the minimum and maximum of $f^{(2n)}$. Assign values within the same bounds at the nodes. Then

$$
R_{\mathrm G,n}(f)=\frac1{(2n)!}\int_a^bq(x)p_n(x)^2w(x)\,dx.\tag{83}
$$

The weight $p_n^2w$ is nonnegative and has a positive integral. Integrating the bounds on $q$ and using continuity of $f^{(2n)}$ yields (81). No continuous choice of the intermediate point $\eta_x$ is required.

As a normalization check, $f(x)=x^{2n}$ gives

$$
R_{\mathrm G,n}(x^{2n})=\int_a^bp_n(x)^2w(x)\,dx,\tag{84}
$$

because the monic polynomial $x^{2n}-H_{2n-1}(x)$ has double zeros at all $x_i$ and therefore equals $p_n(x)^2$.

**Example 7 (two-node Gauss–Legendre check).** On $[-1,1]$ with $w=1$,

$$
p_2(x)=x^2-\frac13,\quad x_1=-\frac1{\sqrt3},\quad x_2=\frac1{\sqrt3},\quad \lambda_1=\lambda_2=1.\tag{85}
$$

The squared norm of the monic polynomial is

$$
\int_{-1}^1\left(x^2-\frac13\right)^2\,dx=\frac8{45}.\tag{86}
$$

Thus for $f\in C^4([-1,1])$,

$$
\int_{-1}^1f(x)\,dx-f\!\left(-\frac1{\sqrt3}\right)-f\!\left(\frac1{\sqrt3}\right)=\frac1{135}f^{(4)}(\xi),\quad\xi\in[-1,1].\tag{87}
$$

For $f(x)=x^4$, the exact error is

$$
R_{\mathrm G,2}(x^4)=\frac25-\frac29=\frac8{45}=\frac{24}{135},\tag{88}
$$

which agrees with $f^{(4)}=24$ in (87).
