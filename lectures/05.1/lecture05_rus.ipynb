# Составные квадратурные формулы

Построим составные формулы на основе исходной квадратуры, рассмотрим их сходимость и остаточные члены, а затем изучим ядра Пеано, оценки через наилучшее приближение и квадратуры Гаусса.

Все функции считаются вещественнозначными, узлы и веса — вещественными.

## Основные сведения

Рассмотрим квадратурную формулу

$$
\int_c^d f(x)\,dx\approx\sum_{k=1}^m A_kf(a_k).\tag{1}
$$

Здесь $c<d$ — конечные числа, $m\geqslant1$ вещественных узлов $a_k$ принадлежат $[c,d]$, а веса $A_k$ вещественны. В этом разделе весовая функция равна единице, а все подынтегральные функции вещественнозначны и интегрируемы по Риману.

Пусть требуется вычислить

$$
\int_\gamma^\delta\varphi(\xi)\,d\xi\tag{2}
$$

на конечном отрезке $\gamma<\delta$, который может отличаться от $[c,d]$. Применим линейную замену

$$
\xi=\gamma+q(x-c),\quad q=\frac{\delta-\gamma}{d-c}>0.\tag{3}
$$

Тогда

$$
\int_\gamma^\delta\varphi(\xi)\,d\xi=\int_c^d\varphi\!\bigl(\gamma+q(x-c)\bigr)q\,dx.\tag{4}
$$

Применяя (1) к правой части, получаем

$$
\int_c^d\varphi\!\bigl(\gamma+q(x-c)\bigr)q\,dx\approx\sum_{k=1}^m qA_k\varphi\!\bigl(\gamma+q(a_k-c)\bigr).\tag{5}
$$

Следовательно,

$$
\int_\gamma^\delta\varphi(\xi)\,d\xi\approx\sum_{k=1}^m\alpha_k\varphi(\xi_k),\tag{6}
$$

где

$$
\alpha_k=qA_k,\quad \xi_k=\gamma+q(a_k-c),\quad k=1,\ldots,m.\tag{7}
$$

Преобразованные узлы принадлежат $[\gamma,\delta]$.

**Определение 1 (аффинно подобные квадратурные формулы).** Формулы (1) и (6) называются **аффинно подобными**, если их узлы и веса связаны соотношениями (7). Преобразование сохраняет алгебраическую степень точности, поскольку композиция с обратимым аффинным отображением сохраняет степень многочлена.

**Утверждение 1 (проверка моментов).** Формула (1) точна для всех линейных многочленов тогда и только тогда, когда

$$
\sum_{k=1}^m A_k=d-c,\quad\sum_{k=1}^m A_ka_k=\frac{d^2-c^2}{2}.\tag{8}
$$

**Доказательство.** Это условия точности для $f(x)=1$ и $f(x)=x$. В силу линейности вместе они равносильны точности для любого многочлена $\alpha+\beta x$. Эти условия характеризуют точность на многочленах; аффинное подобие задается преобразованием узлов и весов (7).

### Составные квадратурные формулы

Для построения составной формулы сначала приведем исходную квадратуру к стандартному отрезку $[0,1]$. Далее $A_k,a_k$ обозначают веса и узлы этой формулы:

$$
\int_0^1\varphi(t)\,dt=\sum_{k=1}^m A_k\varphi(a_k)+S(\varphi),\quad a_k\in[0,1].\tag{9}
$$

Здесь $S(\varphi)$ — погрешность, определенная как интеграл минус квадратурная сумма. При фиксированных узлах и весах $S$ является линейным функционалом.

**Определение 2 (составная квадратурная формула).** Пусть $a<b$ — конечные числа, а $n\geqslant1$ — целое. Разобьем $[a,b]$ на $n$ отрезков с непересекающимися внутренностями:

$$
h=\frac{b-a}{n}>0,\quad x_i=a+ih,\quad i=0,1,\ldots,n.\tag{10}
$$

В силу аддитивности

$$
\int_a^b f(x)\,dx=\sum_{i=1}^n\int_{x_{i-1}}^{x_i}f(x)\,dx.\tag{11}
$$

На каждом отрезке положим

$$
\varphi_i(t)=f(x_{i-1}+ht),\quad\int_{x_{i-1}}^{x_i}f(x)\,dx=h\int_0^1\varphi_i(t)\,dt.\tag{12}
$$

Применяя (9), получаем

$$
\int_{x_{i-1}}^{x_i}f(x)\,dx=h\sum_{k=1}^m A_kf(x_{i-1}+ha_k)+hS(\varphi_i).\tag{13}
$$

Суммирование этих равенств дает

$$
\int_a^b f(x)\,dx=\sum_{i=1}^n\sum_{k=1}^m hA_kf(x_{i-1}+ha_k)+h\sum_{i=1}^n S(\varphi_i).\tag{14}
$$

Соответствующая **составная квадратурная сумма** имеет вид

$$
\sigma_{m,n}(f)=\sum_{i=1}^n\sum_{k=1}^m hA_kf(x_{i-1}+ha_k).\tag{15}
$$

Составные формулы также называют расширенными или уточненными. Их погрешность равна

$$
\int_a^b f(x)\,dx=\sigma_{m,n}(f)+R_n(f),\quad R_n(f)=h\sum_{i=1}^nS(\varphi_i).\tag{16}
$$

Соседние отрезки имеют общие концы; вклады повторяющихся узлов объединяются сложением их весов.

**Пример 1 (составная формула трапеций).** Возьмем в качестве исходной формулы

$$
\int_0^1\varphi(t)\,dt\approx\frac{\varphi(0)+\varphi(1)}2.\tag{17}
$$

Тогда

$$
\begin{aligned}
\int_a^b f(x)\,dx
&=h\sum_{i=1}^n\int_0^1 f(x_{i-1}+ht)\,dt\\
&\approx\sum_{i=1}^n\frac h2\,[f(x_{i-1})+f(x_i)].
\end{aligned}\tag{18}
$$

Объединяя вклады общих концов, получаем

$$
T_n(f)=h\left[\frac{f(a)}2+\sum_{i=1}^{n-1}f(x_i)+\frac{f(b)}2\right].\tag{19}
$$

Число различных узлов равно $n+1$. Тот же результат получается при непосредственном применении формулы трапеций на каждом $[x_{i-1},x_i]$.

**Пример 2 (составная формула Симпсона).** Формула Симпсона на одном отрезке имеет вид

$$
\int_\alpha^\beta f(x)\,dx\approx\frac{\beta-\alpha}{6}\left[f(\alpha)+4f\!\left(\frac{\alpha+\beta}{2}\right)+f(\beta)\right].\tag{20}
$$

Применим ее на каждом отрезке $[x_{i-1},x_i]$ ширины $h$:

$$
\int_a^b f(x)\,dx\approx\sum_{i=1}^n\frac h6\left[f(x_{i-1})+4f\!\left(x_{i-1}+\frac h2\right)+f(x_i)\right].\tag{21}
$$

Положим $x_{i-1/2}=x_{i-1}+h/2$. Собирая общие концы, получаем

$$
Q_n^{\mathrm S}(f)=\frac h6\left[f(a)+4\sum_{i=1}^nf(x_{i-1/2})+2\sum_{i=1}^{n-1}f(x_i)+f(b)\right].\tag{22}
$$

Чтобы перечислить все узлы, сохранив смысл $x_i$, введем более мелкую сетку

$$
\begin{aligned}
z_j&=a+\frac{jh}{2},\quad j=0,1,\ldots,2n,\\
z_{2i}&=x_i,\quad i=0,1,\ldots,n,\\
z_{2i-1}&=x_{i-1/2},\quad i=1,\ldots,n.
\end{aligned}\tag{23}
$$

Следовательно,

$$
Q_n^{\mathrm S}(f)=\frac h6\left[f(z_0)+4\sum_{i=1}^nf(z_{2i-1})+2\sum_{i=1}^{n-1}f(z_{2i})+f(z_{2n})\right].\tag{24}
$$

Число различных узлов равно $2n+1$. Здесь $n$ — число частей, на каждой из которых применяется формула Симпсона, поэтому допустимо любое целое положительное $n$. Число мелких промежутков равно $2n$ и является четным. Аналогично строятся другие составные формулы Ньютона–Котеса.

**Теорема 1 (сохранение алгебраической степени точности).** Если исходная формула (9) на $[0,1]$ имеет конечную алгебраическую степень точности $d\geqslant0$, то составная формула (15) имеет в точности ту же степень при любом $n\geqslant1$.

**Доказательство.** Если $f$ — многочлен степени не выше $d$, то каждый $\varphi_i(t)=f(x_{i-1}+ht)$ также имеет степень не выше $d$. Поэтому $S(\varphi_i)=0$ и $R_n(f)=0$.

Для $f(x)=x^{d+1}$ разложим $(x_{i-1}+ht)^{d+1}$. Все слагаемые, кроме $h^{d+1}t^{d+1}$, имеют степень не выше $d$ и обнуляются функционалом $S$. Следовательно,

$$
\begin{aligned}
R_n(x^{d+1})
&=h\sum_{i=1}^nS\!\left[(x_{i-1}+ht)^{d+1}\right]\\
&=nh^{d+2}S(t^{d+1})\ne0.
\end{aligned}\tag{25}
$$

Последнее неравенство следует из $S(t^{d+1})\ne0$ и $n,h>0$. Значит, степень точности составной формулы равна $d$.

**Теорема 2 (сходимость для интегрируемых по Риману функций).** Зафиксируем конечное число $m$, узлы $a_k\in[0,1]$ и вещественные веса $A_k$, не зависящие от $n$. Тогда соотношение

$$
\lim_{n\to\infty}\sigma_{m,n}(f)=\int_a^b f(x)\,dx\tag{26}
$$

выполнено для каждой интегрируемой по Риману функции $f$ на $[a,b]$ тогда и только тогда, когда исходная формула точна для постоянных:

$$
S(1)=0\quad\Longleftrightarrow\quad\sum_{k=1}^m A_k=1.\tag{27}
$$

**Доказательство.** Изменяя порядок суммирования в (15), получаем

$$
\sigma_{m,n}(f)=\sum_{k=1}^m A_k\left[h\sum_{i=1}^nf(x_{i-1}+ha_k)\right].\tag{28}
$$

При каждом фиксированном $k$ выражение в скобках является интегральной суммой с одной отмеченной точкой на каждом отрезке. Так как $h\to0$, она стремится к интегралу от $f$. Внешняя сумма конечна, поэтому

$$
\lim_{n\to\infty}\sigma_{m,n}(f)=\left(\sum_{k=1}^m A_k\right)\int_a^b f(x)\,dx.\tag{29}
$$

Если выполнено (27), получаем (26). Обратно, применим требуемую сходимость к $f=1$. При каждом $n$

$$
\sigma_{m,n}(1)=(b-a)\sum_{k=1}^m A_k,\tag{30}
$$

поэтому сходимость к $b-a>0$ требует $\sum A_k=1$. Это также дает удобную проверку нормировки составных весов.

**Теорема 3 (перенос остаточного члена с одной промежуточной точкой).** Пусть $r\geqslant1$ — целое. Предположим, что для каждой $\varphi\in C^r([0,1])$ погрешность исходной формулы допускает представление

$$
S(\varphi)=c_r\varphi^{(r)}(\eta_\varphi),\quad \eta_\varphi\in[0,1],\tag{31}
$$

где $c_r$ не зависит от $\varphi$, а точка $\eta_\varphi$ может от нее зависеть. Если $f\in C^r([a,b])$, то

$$
R_n(f)=c_r\frac{(b-a)^{r+1}}{n^r}f^{(r)}(\xi)=c_r(b-a)h^rf^{(r)}(\xi),\quad \xi\in[a,b].\tag{32}
$$

**Доказательство.** По правилу дифференцирования сложной функции

$$
\varphi_i^{(r)}(t)=h^rf^{(r)}(x_{i-1}+ht).\tag{33}
$$

Применим (31) к каждому $\varphi_i$ и подставим в (16):

$$
R_n(f)=c_rh^{r+1}\sum_{i=1}^nf^{(r)}(x_{i-1}+h\eta_i),\quad \eta_i\in[0,1].\tag{34}
$$

Все точки вычисления принадлежат $[a,b]$, поэтому

$$
\min_{[a,b]}f^{(r)}\leqslant\frac1n\sum_{i=1}^nf^{(r)}(x_{i-1}+h\eta_i)\leqslant\max_{[a,b]}f^{(r)}.\tag{35}
$$

Из непрерывности $f^{(r)}$ и теоремы о промежуточном значении следует существование $\xi\in[a,b]$, для которой

$$
\frac1n\sum_{i=1}^nf^{(r)}(x_{i-1}+h\eta_i)=f^{(r)}(\xi),\quad \xi\in[a,b].\tag{36}
$$

Подстановка $nh=b-a$ в (34) дает (32). В частности,

$$
|R_n(f)|\leqslant |c_r|(b-a)h^r\max_{x\in[a,b]}|f^{(r)}(x)|.\tag{37}
$$

Таким образом, при фиксированных функции и отрезке глобальная погрешность равна $O(h^r)$. Теорема требует представления (31); условие ненулевой производной $f^{(r)}$ не требуется.

**Утверждение 2 (коэффициент и порядок производной).** При условии (31) для исходной формулы с конечным числом узлов $m$ справедливо

$$
c_r=\frac{S(t^r)}{r!}\ne0,\quad r\leqslant2m,\quad d=r-1,\tag{38}
$$

где $d$ — ее алгебраическая степень точности.

**Доказательство.** Построим узловой многочлен

$$
\omega_m(t)=\prod_{k=1}^m(t-a_k).\tag{39}
$$

Его квадрат обращается в нуль во всех узлах квадратуры, но положителен вне конечного множества точек. Поэтому

$$
S(\omega_m^2)=\int_0^1\omega_m(t)^2\,dt>0.\tag{40}
$$

При $c_r=0$ представление (31) дало бы нулевую погрешность. При $r>2m$ производная порядка $r$ от $\omega_m^2$ тождественно равнялась бы нулю, что приводит к тому же противоречию. Значит, $c_r\ne0$ и $r\leqslant2m$.

Для любого многочлена степени ниже $r$ представление (31) дает нулевую погрешность, а $S(t^r)=c_rr!\ne0$. Поэтому $d=r-1$ и выполнена формула коэффициента из (38).

**Пример 3 (погрешность составной формулы трапеций).** Для формулы трапеций на $[0,1]$ остаточный член из лекции по численному интегрированию имеет вид

$$
S_{\mathrm T}(\varphi)=-\frac1{12}\varphi''(\eta),\quad \eta\in[0,1].\tag{41}
$$

Следовательно, $r=2$ и $c_2=-1/12$. Если $f\in C^2([a,b])$, то теорема 3 дает

$$
R_n^{\mathrm T}(f)=-\frac{(b-a)^3}{12n^2}f''(\xi)=-\frac{(b-a)h^2}{12}f''(\xi),\quad \xi\in[a,b].\tag{42}
$$

Это погрешность формулы (19); оценка модуля получается заменой $f''(\xi)$ на $\max|f''|$.

**Пример 4 (погрешность составной формулы Симпсона).** Для исходной формулы Симпсона на $[0,1]$

$$
S_{\mathrm S}(\varphi)=-\frac1{2880}\varphi^{(4)}(\eta),\quad \eta\in[0,1].\tag{43}
$$

Возьмем $r=4$ и $c_4=-1/2880$. Если $f\in C^4([a,b])$, то погрешность (22) с $n$ частями ширины $h$ равна

$$
R_n^{\mathrm S}(f)=-\frac{(b-a)^5}{2880n^4}f^{(4)}(\xi)=-\frac{(b-a)h^4}{2880}f^{(4)}(\xi),\quad \xi\in[a,b].\tag{44}
$$

Также принято считать мелкие промежутки вместо частей применения формулы. Положим

$$
N=2n,\quad s=\frac{b-a}{N}=\frac h2.\tag{45}
$$

Тогда та же погрешность принимает вид

$$
R_n^{\mathrm S}(f)=-\frac{(b-a)^5}{180N^4}f^{(4)}(\xi)=-\frac{(b-a)s^4}{180}f^{(4)}(\xi).\tag{46}
$$

Таким образом, коэффициент $1/2880$ соответствует ширине части $h$, а $1/180$ — мелкому шагу $s$. Число мелких промежутков $N$ должно быть четным. Это объясняет две обычные формы остаточного члена составной формулы Симпсона.

## О погрешности квадратурных формул

Теперь рассмотрим формулу с весом

$$
\int_a^b w(x)f(x)\,dx\approx\sum_{j=1}^NC_jf(x_j)\tag{47}
$$

на конечном отрезке $a<b$. Предположим, что $w$ — вещественная измеримая функция, $\int_a^b|w(x)|\,dx<\infty$, вещественные узлы $x_j$ при $N\geqslant1$ принадлежат $[a,b]$, а коэффициенты $C_j$ вещественны. В этом разделе вес может менять знак.

**Определение 3 (функционал погрешности квадратуры).** Для ограниченных измеримых функций $g$ положим

$$
R(g)=\int_a^b w(x)g(x)\,dx-\sum_{j=1}^NC_jg(x_j).\tag{48}
$$

Функционал $R$ линеен. Далее он применяется к непрерывным функциям, многочленам и определенным ниже усеченным степеням.

**Определение 4 (усеченная степень).** Для целого $p\geqslant0$ положим

$$
(x-t)_+^p=
\begin{cases}
(x-t)^p,&x>t,\\
0,&x\leqslant t,
\end{cases}\quad p=0,1,2,\ldots.\tag{49}
$$

В частности, $(x-t)_+^0$ равна единице при $x>t$ и нулю при $x\leqslant t$. Это соглашение определяет значение в точке скачка без использования $0^0$.

**Определение 5 (ядро Пеано).** Функция

$$
K_{p+1}(t)=R[(x-t)_+^p]=\int_a^bw(x)(x-t)_+^p\,dx-\sum_{j=1}^NC_j(x_j-t)_+^p\tag{50}
$$

называется **ядром Пеано** порядка $p+1$. Здесь $R$ действует по переменной $x$, а $t$ фиксировано. При $p\geqslant1$ ядро непрерывно. При $p=0$ оно ограничено и может иметь скачки в узлах квадратуры, но остается интегрируемым.

**Теорема 4 (представление Пеано).** Пусть формула (47) точна для любого многочлена степени не выше $p$, где $p\geqslant0$. Если $f\in C^{p+1}([a,b])$, то

$$
R(f)=\frac1{p!}\int_a^bK_{p+1}(t)f^{(p+1)}(t)\,dt.\tag{51}
$$

**Доказательство.** Формула Тейлора с интегральным остаточным членом дает

$$
f(x)=\sum_{k=0}^p\frac{f^{(k)}(a)}{k!}(x-a)^k+\frac1{p!}\int_a^x(x-t)^pf^{(p+1)}(t)\,dt.\tag{52}
$$

Последний интеграл можно записать по $[a,b]$ с помощью $(x-t)_+^p$. Применим $R$ к этому равенству. Многочленная часть обнуляется в силу точности, поэтому

$$
R(f)=\frac1{p!}R\!\left[\int_a^b(x-t)_+^pf^{(p+1)}(t)\,dt\right].\tag{53}
$$

Можно переставить интегралы и конечную сумму:

$$
\begin{aligned}
R(f)&=\frac1{p!}\int_a^b
\left[\int_a^bw(x)(x-t)_+^p\,dx-\sum_{j=1}^NC_j(x_j-t)_+^p\right]
f^{(p+1)}(t)\,dt\\
&=\frac1{p!}\int_a^bK_{p+1}(t)f^{(p+1)}(t)\,dt.
\end{aligned}\tag{54}
$$

Действительно, на квадрате интегрирования $|(x-t)_+^p|\leqslant(b-a)^p$, а $f^{(p+1)}$ ограничена. Поэтому двойной интеграл абсолютно интегрируем в силу абсолютной интегрируемости $w$. Сумма значений в узлах содержит конечное число слагаемых. Это обосновывает перестановку и при $p=0$, когда усеченная степень имеет скачок. Получено (51).

**Следствие 1 (общая оценка через производную).** При условиях теоремы 4 справедливо

$$
|R(f)|\leqslant\frac{\|f^{(p+1)}\|_\infty}{p!}\int_a^b|K_{p+1}(t)|\,dt.\tag{55}
$$

**Доказательство.** Возьмем модули в (51) и оценим $|f^{(p+1)}(t)|$ ее максимумом. Оценка применима независимо от того, меняет ли ядро знак.

**Следствие 2 (ядро постоянного знака).** При условиях теоремы 4, если $K_{p+1}$ почти всюду на $[a,b]$ не меняет знак, существует $\xi\in[a,b]$, для которой

$$
R(f)=\frac{f^{(p+1)}(\xi)}{p!}\int_a^bK_{p+1}(t)\,dt,\quad \xi\in[a,b].\tag{56}
$$

Равносильная форма этого представления имеет вид

$$
R(f)=\frac{f^{(p+1)}(\xi)}{(p+1)!}R(x^{p+1}),\quad \xi\in[a,b].\tag{57}
$$

**Доказательство.** Применим взвешенную теорему о среднем к (51), используя непрерывность $f^{(p+1)}$. Если ядро почти всюду равно нулю, обе части нулевые. Для получения второй формы подставим $f(x)=x^{p+1}$ в (51):

$$
R(x^{p+1})=(p+1)\int_a^bK_{p+1}(t)\,dt.\tag{58}
$$

Сочетание этого равенства с (56) дает (57).

**Утверждение 3 (знак ядра и алгебраическая степень точности).** Пусть квадратура имеет конечную алгебраическую степень точности $d$. При $0\leqslant p<d$ ее ядро $K_{p+1}$ обязано менять знак. Поэтому среди допустимых значений $0\leqslant p\leqslant d$ нетривиальное ядро постоянного знака возможно только при $p=d$; при этом его знак все равно нужно проверить.

**Доказательство.** Так как $p+1\leqslant d$, из точности и (58) следует

$$
p<d\quad\Longrightarrow\quad R(x^{p+1})=0\quad\Longrightarrow\quad\int_a^bK_{p+1}(t)\,dt=0.\tag{59}
$$

Если бы это ядро почти всюду имело один знак, оно почти всюду равнялось бы нулю. Тогда теорема 4 давала бы $R(f)=0$ для каждой $f\in C^{p+1}([a,b])$, в том числе для $x^{d+1}$. Это противоречит определению степени $d$. Следовательно, при $p<d$ у ядра есть положительная и отрицательная части. При $p=d$ формула (58) дает ненулевой интеграл, однако сама по себе не гарантирует постоянства знака.

**Пример 5 (ядро Пеано для формулы трапеций).** Рассмотрим формулу трапеций с двумя концевыми узлами. Ее алгебраическая степень точности равна $d=1$, поэтому возьмем $p=1$ и предположим $f\in C^2([a,b])$. По (50)

$$
\begin{aligned}
K_2(t)&=\int_a^b(x-t)_+\,dx-\frac{b-a}{2}\bigl[(a-t)_++(b-t)_+\bigr]\\
&=\frac{(b-t)^2}{2}-\frac{b-a}{2}(b-t)
=\frac12(b-t)(a-t),\quad a\leqslant t\leqslant b.
\end{aligned}\tag{60}
$$

Ядро отрицательно при $a<t<b$. Кроме того,

$$
R_{\mathrm T}(x^2)=\int_a^bx^2\,dx-\frac{b-a}{2}(a^2+b^2)=-\frac{(b-a)^3}{6}.\tag{61}
$$

Применение (57) дает

$$
R_{\mathrm T}(f)=-\frac{(b-a)^3}{12}f''(\xi),\quad\xi\in[a,b].\tag{62}
$$

Это совпадает с остаточным членом формулы трапеций на одном отрезке и с (42) при $n=1$. Число узлов здесь определяется выбором формулы трапеций; одна лишь алгебраическая степень точности один не определяет число узлов.

**Пример 6 (ядро Пеано для формулы Симпсона).** Рассмотрим

$$
R_{\mathrm S}(f)=\int_a^bf(x)\,dx-\frac H6[f(a)+4f(c)+f(b)],\quad H=b-a,\quad c=\frac{a+b}{2}.\tag{63}
$$

Алгебраическая степень точности равна $d=3$. Возьмем $p=3$ и $f\in C^4([a,b])$. Поскольку $t\in[a,b]$, слагаемое $(a-t)_+^3$ равно нулю, и (50) дает

$$
K_4(t)=\frac{(b-t)^4}{4}-\frac H6\bigl[4(c-t)_+^3+(b-t)^3\bigr],\quad a\leqslant t\leqslant b.\tag{64}
$$

Раскрывая выражение на двух половинах отрезка, получаем симметричную формулу

$$
K_4(t)=
\begin{cases}
\dfrac{(t-a)^3}{12}\,[3(t-a)-2H],&a\leqslant t\leqslant c,\\[5pt]
\dfrac{(b-t)^3}{12}\,[3(b-t)-2H],&c\leqslant t\leqslant b.
\end{cases}\tag{65}
$$

На левой половине положим $u=t-a$, на правой — $u=b-t$. В обоих случаях

$$
0<u\leqslant\frac H2\quad\Longrightarrow\quad 3u-2H\leqslant-\frac H2<0.\tag{66}
$$

Следовательно, $K_4(t)<0$ на всем $(a,b)$, в том числе в середине, а обе части согласованы в $c$. Непосредственное интегрирование дает

$$
\int_a^bK_4(t)\,dt=2\int_0^{H/2}\frac{u^3}{12}(3u-2H)\,du=-\frac{H^5}{480}.\tag{67}
$$

Подставляя это в (56) при $p!=3!=6$, получаем

$$
R_{\mathrm S}(f)=-\frac{(b-a)^5}{2880}f^{(4)}(\xi),\quad\xi\in[a,b].\tag{68}
$$

Мы восстановили остаточный член Симпсона из примера 4 и проверили нормировку ядра и знак.

### Оценки через наилучшее равномерное приближение

**Определение 6 (погрешность наилучшего равномерного приближения).** Обозначим через $\mathbb P_p$ многочлены степени не выше $p\geqslant0$. Для $f\in C([a,b])$ положим

$$
E_p(f)=\inf_{P\in\mathbb P_p}\|f-P\|_\infty=\|f-P_p^*\|_\infty.\tag{69}
$$

Здесь $\|g\|_\infty=\max_{x\in[a,b]}|g(x)|$, а $P_p^*$ — многочлен наилучшего приближения. На компактном отрезке такой многочлен существует.

**Теорема 5 (оценка через наилучшее приближение).** Если формула с весом (47) точна для всех многочленов степени не выше $p$ и $f\in C([a,b])$, то

$$
|R(f)|\leqslant\left(\int_a^b|w(x)|\,dx+\sum_{j=1}^N|C_j|\right)E_p(f).\tag{70}
$$

**Доказательство.** Так как $R(P_p^*)=0$, по линейности

$$
R(f)=R(f-P_p^*)=\int_a^bw(x)(f-P_p^*)(x)\,dx-\sum_{j=1}^NC_j(f-P_p^*)(x_j).\tag{71}
$$

Оценим интеграл и конечную сумму отдельно. Поскольку каждый $x_j\in[a,b]$,

$$
\begin{aligned}
|R(f)|&\leqslant\|f-P_p^*\|_\infty\int_a^b|w(x)|\,dx
+\max_{1\leqslant j\leqslant N}|f(x_j)-P_p^*(x_j)|\sum_{j=1}^N|C_j|\\
&\leqslant E_p(f)\left(\int_a^b|w(x)|\,dx+\sum_{j=1}^N|C_j|\right).
\end{aligned}\tag{72}
$$

Получено (70). В отличие от оценок через производные, здесь требуется только непрерывность $f$.

**Следствие 3 (неотрицательные веса).** Если дополнительно $w\geqslant0$, $C_j\geqslant0$, $\mu_0=\int_a^bw>0$ и формула точна для постоянных, то

$$
|R(f)|\leqslant2\mu_0E_p(f),\quad \mu_0=\int_a^bw(x)\,dx>0.\tag{73}
$$

**Доказательство.** Из точности для постоянных следует $\sum C_j=\mu_0$, поэтому коэффициент в (70) равен $2\mu_0$.

## Наивысшая алгебраическая степень точности

Теперь предположим, что $w\geqslant0$ почти всюду на конечном отрезке $[a,b]$, $\int_a^b|w|<\infty$ и $\mu_0=\int_a^bw(x)\,dx>0$. Эти условия обеспечивают положительную определенность скалярного произведения на многочленах.

**Определение 7 (квадратура Гаусса).** При $n\geqslant1$ пусть $p_n$ — ортогональный многочлен степени $n$ с **единичным старшим коэффициентом** для веса $w$:

$$
p_n(x)=\prod_{i=1}^n(x-x_i),\quad\int_a^bw(x)p_n(x)q(x)\,dx=0\quad(q\in\mathbb P_{n-1}).\tag{74}
$$

Его $n$ корней $x_i$ различны и принадлежат $(a,b)$; эти стандартные свойства ортогональных многочленов следуют из указанных условий на вес [1, 3]. Выберем их узлами и определим

$$
\lambda_i=\int_a^b\ell_i(x)w(x)\,dx,\quad\ell_i(x)=\prod_{j\ne i}\frac{x-x_j}{x_i-x_j}.\tag{75}
$$

Тогда

$$
\int_a^bf(x)w(x)\,dx\approx G_n(f)=\sum_{i=1}^n\lambda_i f(x_i)\tag{76}
$$

называется **квадратурной формулой Гаусса**. При $w\equiv1$ на $[-1,1]$ узлы являются нулями многочлена Лежандра; в формуле погрешности ниже используется его нормировка с единичным старшим коэффициентом.

**Теорема 6 (степень точности и положительность весов Гаусса).** При указанных выше условиях квадратура Гаусса имеет свойства

$$
d=2n-1,\quad\lambda_i>0,\quad\sum_{i=1}^n\lambda_i=\mu_0.\tag{77}
$$

Для этого веса никакая формула с $n$ вещественными узлами не может иметь более высокую алгебраическую степень точности.

**Доказательство.** Для $P\in\mathbb P_{2n-1}$ деление многочленов дает

$$
P=qp_n+r,\quad q,r\in\mathbb P_{n-1}\quad(P\in\mathbb P_{2n-1}).\tag{78}
$$

По ортогональности $\int_a^bwqp_n=0$, а сумма значений $qp_n$ в узлах равна нулю. Интерполяционные веса (75) точно интегрируют $r$, поэтому $G_n$ точно интегрирует каждый такой $P$.

В частности, из точности для $\ell_i^2$ получаем

$$
\lambda_i=G_n(\ell_i^2)=\int_a^b\ell_i(x)^2w(x)\,dx>0,\tag{79}
$$

а из точности для постоянных — $\sum\lambda_i=\mu_0$. Для доказательства верхней границы при произвольных вещественных узлах $\zeta_i$ используем

$$
\omega(x)=\prod_{i=1}^n(x-\zeta_i),\quad\int_a^b\omega(x)^2w(x)\,dx>0,\quad\sum_{i=1}^n C_i\omega(\zeta_i)^2=0.\tag{80}
$$

Следовательно, формула с $n$ узлами не может быть точной для всех многочленов степени $2n$. Степень точности Гаусса равна в точности $2n-1$.

**Утверждение 4 (остаточный член Гаусса).** При тех же условиях на вес, если $f\in C^{2n}([a,b])$, то

$$
R_{\mathrm G,n}(f)=\int_a^bf(x)w(x)\,dx-G_n(f)=\frac{f^{(2n)}(\xi)}{(2n)!}\int_a^bp_n(x)^2w(x)\,dx,\quad\xi\in[a,b].\tag{81}
$$

Здесь старший коэффициент $p_n$ равен единице. Для ортогонального многочлена со старшим коэффициентом $\kappa_n$ его квадрат нормы нужно разделить на $\kappa_n^2$.

**Доказательство.** Пусть $H_{2n-1}$ — интерполяционный многочлен Эрмита, совпадающий с $f$ и $f'$ во всех $n$ узлах Гаусса. Его степень не выше $2n-1$, поэтому $\int wH_{2n-1}=G_n(H_{2n-1})=G_n(f)$. Остаточный член Эрмита дает для некоторой $\eta_x\in[a,b]$

$$
f(x)-H_{2n-1}(x)=\frac{f^{(2n)}(\eta_x)}{(2n)!}p_n(x)^2\quad(x\ne x_i).\tag{82}
$$

Вне узлов определим $q(x)=(2n)![f(x)-H_{2n-1}(x)]/p_n(x)^2$. Это частное непрерывно вне узлов и лежит между минимумом и максимумом $f^{(2n)}$. Зададим в узлах значения в тех же границах. Тогда

$$
R_{\mathrm G,n}(f)=\frac1{(2n)!}\int_a^bq(x)p_n(x)^2w(x)\,dx.\tag{83}
$$

Вес $p_n^2w$ неотрицателен и имеет положительный интеграл. Интегрирование ограничений для $q$ и непрерывность $f^{(2n)}$ дают (81). Непрерывный выбор промежуточной точки $\eta_x$ не требуется.

Для проверки нормировки положим $f(x)=x^{2n}$. Получаем

$$
R_{\mathrm G,n}(x^{2n})=\int_a^bp_n(x)^2w(x)\,dx,\tag{84}
$$

поскольку многочлен $x^{2n}-H_{2n-1}(x)$ имеет старший коэффициент один и двойные нули во всех $x_i$, а значит, равен $p_n(x)^2$.

**Пример 7 (проверка двухузловой квадратуры Гаусса–Лежандра).** На $[-1,1]$ при $w=1$

$$
p_2(x)=x^2-\frac13,\quad x_1=-\frac1{\sqrt3},\quad x_2=\frac1{\sqrt3},\quad \lambda_1=\lambda_2=1.\tag{85}
$$

Квадрат нормы многочлена с единичным старшим коэффициентом равен

$$
\int_{-1}^1\left(x^2-\frac13\right)^2\,dx=\frac8{45}.\tag{86}
$$

Поэтому при $f\in C^4([-1,1])$

$$
\int_{-1}^1f(x)\,dx-f\!\left(-\frac1{\sqrt3}\right)-f\!\left(\frac1{\sqrt3}\right)=\frac1{135}f^{(4)}(\xi),\quad\xi\in[-1,1].\tag{87}
$$

Для $f(x)=x^4$ точная погрешность составляет

$$
R_{\mathrm G,2}(x^4)=\frac25-\frac29=\frac8{45}=\frac{24}{135},\tag{88}
$$

что согласуется с $f^{(4)}=24$ в (87).
