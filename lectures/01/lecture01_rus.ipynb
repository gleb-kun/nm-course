# Численное дифференцирование

## Простейшие формулы численного дифференцирования

### Вычисление первой производной

**Определение производной функции через предел.** Пусть в некоторой окрестности точки $x_0 \in \mathbb{R}$ определена функция ${\displaystyle f\colon U(x_{0})\subset \mathbb {R} \to \mathbb {R} .}$ Производной функции ${\displaystyle f}$ в точке ${\displaystyle x_{0}}$ называется предел, если он существует
$$
\begin{equation}
{\displaystyle f'(x_{0})=\lim \limits _{x\to x_{0}}{\frac {f(x)-f(x_{0})}{x-x_{0}}}=\lim _{\Delta x\to 0}{\frac {f(x_{0}+\Delta x)-f(x_{0})}{\Delta x}}=\lim \limits _{{\Delta x}\to 0}{\frac {\Delta {f(x)}}{\Delta x}}.}\tag{1}
\end{equation}
$$


Исходя из опредеения первой производной запишем производную в точке $x$:
$$
f'(x)=\lim_{\Delta x\rightarrow 0}\frac{f(x+\Delta x)-f(x)}{h},\tag{2}
$$
отсюда можно записать две простейшие приближенные формулы

$$
f'(x)\approx \frac{f(x+h)-f(x)}{h},\tag{3}
$$

$$
f'(x)\approx \frac{f(x)-f(x-h)}{h},\tag{4}
$$
соответствующие выбору фиксированных значений $\Delta x= h$ и $\Delta x = -h$, $h>0$ — шаг.


**Определение 1**
Формулу (3) называют правой разностаной схемой.


**Определение 2**
Формулу (4) называют левой разностаной схемой.

Для оценки погрешностей

$$
r_{+}(x,h)=f'(x)-\frac{f(x+h)-f(x)}{h},\tag{5}
$$

$$
r_{-}(x,h)=f'(x)-\frac{f(x)-f(x-h)}{h}\tag{6}
$$
введенных формул численного дифференцирования (погрешностей аппроксимации) воспользуемся формулой Тейлора
$$
f(x\pm h)=f(x)\pm f'(x)h+\frac{f''(\xi_{\pm})}{2}h^2.\tag{7}
$$
Здесь и ниже $\xi_{+}$ и $\xi_{-}$ — некоторые точки, расположенные на интервалах $(x,x+h)$ и $(x-h, x)$ соответственно.
Подставляя разложения в выражения для $r_{\pm}$, получаем $r_{+}(x,h)=-\frac{1}{2}f''(\xi_{+})h$, $r_{-}(x,h)=\frac{1}{2}f''(\xi_{-})h$. Следовательно
$$
|r_{+}(x,h)|\leqslant\frac{1}{2}M_1 h,\quad M_1=\max_{\left[x,x+h\right]}|f''(\xi)|,\tag{8}
$$

$$
|r_{-}(x,h)|\leqslant\frac{1}{2}M_2 h,\quad M_2=\max_{\left[x-h,x\right]}|f''(\xi)|.\tag{9}
$$

#### Геометрическая интерпретация

Приведенные формулы имеют геометрическую интерпретацию. Пусть $N_0$, $N_{-}$, $N_{+}$ — расположенные на графике функции $y=f(x)$ точки с координатами $(x, f(x))$, $(x-h, f(x-h))$ и $(x+h, f(x+h))$.

**Определение 3.** Центральная разностная производная

$$
f'(x)\approx\frac{f(x+h)-f(x-h)}{2h}.\tag{10}
$$


Далее оценим погрешность. Подставляя в выражение для погрешности
$$
r_0(x,h)=f'(x)-\frac{f(x+h)-f(x-h)}{2h}\tag{11}
$$
соответствующие разложения по формуле Тейлора
$$
f(x\pm h)=f(x)\pm f'(x)h+\frac{f''(x)}{2}h^2\pm \frac{f^{(3)}(\xi_{\pm})}{6}h^3,\tag{12}
$$
получаем
$$
r_0(x,h)=-\frac{f^{(3)}(\xi_{+})+f^{(3)}(\xi_{-})}{12}h^2.\tag{13}
$$

Следовательно, справедлива оценка погрешности
$$
|r_0(x,h)|\leqslant\frac{M_3}{6}h^2,\quad M_3=\max_{\left[x-h,x+h\right]}|f^{(3)}(\xi)|.\tag{14}
$$

Для вычисления $f'(x)$ можно получить формулы любого порядка точности. Однако в таких формулах с ростом порядка точности возрастает и число используемых значений функции. В качестве примера приведем формулу
$$
f'(x)\approx\frac{f(x-2h)-8f(x-h)+8f(x+h)-f(x+2h)}{12h}\tag{15}
$$
имеющую четвертый порядок точности.

**Пример 1.** Пусть функция $f(x)=e^x$ задана на отрезке $[0,1]$ таблицей значений с шагом $h=0.2$.

```python
import math

# параметры
a = 0       # начало отрезка
b = 1       # конец отрезка
h = 0.2     # шаг

print(" x\t|\t e^x")
print("--------------------")

x_values = []
x = a
while x <= b:
    fx = math.exp(x)
    print(f"{x:.1f}\t|\t{fx:.6f}")
    x_values.append(x)
    x += h
```

```python
f_values = [math.exp(x) for x in x_values]

derivatives = []
n = len(x_values)

for i in range(n):
    if i == 0:
        d = (f_values[i + 1] - f_values[i]) / h
    elif i == n - 1:
        d = (f_values[i] - f_values[i - 1]) / h
    else:
        d = (f_values[i + 1] - f_values[i - 1]) / (2 * h)
    derivatives.append(d)

print(" x\t|\t e^x\t\t|\t f'(x)_num\t|\t f'(x)_exact\t|\t Error")
print("--------------------------------------------------------------------------")

for x, f, d in zip(x_values, f_values, derivatives):
    exact = math.exp(x)
    error = abs(d - exact)
    print(f"{x:.1f}\t|\t{f:.6f}\t|\t{d:.6f}\t|\t{exact:.6f}\t|\t{error:.6e}")
```

Можно для оценки погрешностей использовать априорные оценки погрешностей. Например, ранее рассмотренное неравенство $|r_{+}(x,h)|\leqslant \frac{1}{2}M_1 h$, где $M_1=\max_{\left[x,x+h\right]}|f''(\xi)|$ (8), дает в точке $x=0$ оценку:
$$
|r|\leqslant \frac{1}{2} M_1 h = \frac{1}{2}\max_{\left[0,0.2\right]}|\exp(\xi)|h=\frac{1}{2}\cdot 1.22140\cdot 0.2\approx 0.13.\tag{16}
$$
В точке $x=0.2$ из неравенства $|r_0(x,h)|\leqslant\frac{M_3}{6}h^2$, где $M_3=\max_{[x-h,x+h]}|f^{(3)}(\xi)|$ (14),  следует, что
$$
|r|\leqslant \frac{1}{6} M_3 h^2 = \frac{1}{6}\max_{[0,0.4]}|\exp(\xi)|h^2=\frac{1}{6}\cdot 1.49182\cdot 0.04\approx 0.01, \text{и т.д.}\tag{17}
$$

Для вычисления $f'(x)$ при $x=0.4$ и $0.6$ можно также применить формулу
$$
f'(x)\approx\frac{f(x-2h)-8f(x-h)+8f(x+h)-f(x+2h)}{12h}\tag{18}
$$
и получить значения $f'(x)\approx 1.49176$ и $f'(x)\approx 1.82203$ с погрешностями, приближенно равными $6\cdot 10^{-5}$ и $9\cdot 10^{-5}$ соответственно.

### Вычисление второй производной

Часто используется формула показанная в следующем определении.

**Определение.** Вторая разностная производная
$$
f''(x)\approx \frac{f(x-h)-2f(x)+f(x+h)}{h^2}.\tag{19}
$$

Оценим погрешность. Подставляя в выражение для погрешности
$$
r(x,h)=f''(x)-\frac{f(x-h)-2f(x)+f(x+h)}{h^2}\tag{20}
$$
соответствующие разложения по формуле Тейлора
$$
f(x\pm h)=f(x)\pm f'(x)h+\frac{f''(x)}{2}h^2\pm \frac{f^{(3)}(x)}{6}h^3+\frac{f^{(4)}(\xi_{\pm})}{24}h^4,\tag{21}
$$
получаем
$$
r(x,h)=-\frac{f^{(4)}(\xi_{+})+f^{(4)}(\xi_{-})}{24}h^2\tag{22}
$$

Следовательно, 
$$
|r(x,h)|\leqslant \frac{M_4}{12}h^2,\quad M_4=\max_{[x-h,x+h]}|f^{(4)}(\xi)|.\tag{23}
$$


Аналогично случаю с первой производной, для вычисления $f''(x)$ можно получить формулы любого порядка точности. Например, формула
$$
f''(x)\approx \frac{-f(x-2h)+16f(x-h)-30f(x)+16f(x+h)-f(x+2h)}{12h^2}\tag{24}
$$
имеет четвертый порядок точности.

**Пример 2** Использовав ранее полученыые значения функции $f(x)=\exp(x)$, найдем с помощью формул численного дифференцирования значения $f''(x)$ во внутренних узлах таблицы.

Вычисление по формуле $f''(x)\approx \frac{f(x-h)-2f(x)+f(x+h)}{h^2}$ дает следующие результаты

```python
print(" x\t|\t e^x")
print("--------------------")
for x, f in zip(x_values, f_values):
    print(f"{x:.1f}\t|\t{f:.6f}")

second_derivatives = []
n = len(x_values)

for i in range(n):
    if i == 0 or i == n - 1:
        second_derivatives.append(None)
    else:
        d2 = (f_values[i - 1] - 2 * f_values[i] + f_values[i + 1]) / (h ** 2)
        second_derivatives.append(d2)

print("\n x\t|\t f(x)\t\t|\t f''(x)_num\t|\t f''(x)_exact\t|\t Error")
print("--------------------------------------------------------------------------")

for x, f, d2 in zip(x_values, f_values, second_derivatives):
    exact = math.exp(x)  # для exp(x), f''(x) = exp(x)
    if d2 is None:
        print(f"{x:.1f}\t|\t{f:.6f}\t|\t   ---   \t|\t{exact:.6f}\t|\t   ---")
    else:
        error = abs(d2 - exact)
        print(f"{x:.1f}\t|\t{f:.6f}\t|\t{d2:.6f}\t|\t{exact:.6f}\t|\t{error:.6e}")
```

Вычисление по формуле (24):
$$
f''(x)\approx \frac{-f(x-2h)+16f(x-h)-30f(x)+16f(x+h)-f(x+2h)}{12h^2}\tag{25}
$$
дает следующие результаты:

```python
print(" x\t|\t e^x")
print("--------------------")
for x, f in zip(x_values, f_values):
    print(f"{x:.1f}\t|\t{f:.6f}")

second_derivatives = []
n = len(x_values)

for i in range(n):
    if i < 2 or i > n - 3:
        second_derivatives.append(None)
    else:
        d2 = (-f_values[i - 2] + 16 * f_values[i - 1] - 30 * f_values[i]
              + 16 * f_values[i + 1] - f_values[i + 2]) / (12 * h ** 2)
        second_derivatives.append(d2)

print("\n x\t|\t f(x)\t\t|\t f''(x)_num\t|\t f''(x)_exact\t|\t Error")
print("--------------------------------------------------------------------------")

for x, f, d2 in zip(x_values, f_values, second_derivatives):
    exact = math.exp(x)  # для exp(x), f''(x) = exp(x)
    if d2 is None:
        print(f"{x:.1f}\t|\t{f:.6f}\t|\t   ---   \t|\t{exact:.6f}\t|\t   ---")
    else:
        error = abs(d2 - exact)
        print(f"{x:.1f}\t|\t{f:.6f}\t|\t{d2:.6f}\t|\t{exact:.6f}\t|\t{error:.6e}")
```

Применение формулы (24) позволяет получить значения $f''(0.4)\approx 1.49204$, $f''(0.6)\approx 1.82183$ с погрешностями, приближенно равными $-2\cdot 10^{-5}$, $-3\cdot 10^{-5}$.

## О выводе формул численного дифференцирования

Предположим, что в окрестности точки $x$ функция аппроксимируется некоторой другой функцией $g$, причем производная $g^{(k)}$ в точке $x$ легко вычисляется. Естественно в такой ситуации попытаться воспользоваться приближенной формулой:
$$
f^{(k)}(x)\approx g^{(k)}(x).\tag{26}
$$

### Формулы численного дифференцирования, основанные на интерполяции алгебраическими полиномами.

Пусть $P_n(x)$ —  интерполяционный полином степени $n$ с узлами интерполяции $x_0 < x_1 < x_2 < ... < x_n$ и $x\in [x_0, x_n]$. В этом случае формула (26) принимает вид
$$
\begin{equation}
f^{(k)}(x)\approx P_n^{(k)}(x),\quad 0\leqslant k \leqslant n. \tag{27}
\end{equation}
$$


Оценка погрешности формулы (27):
$$
|f^{(k)}(x)-P_n^{(k)(x)}|\leqslant C_{n,k}M_{n+1}h^{n+1-k}_{\max}, \quad 0\leqslant k\leqslant n,\tag{28}
$$
где $C_{n,k}$ — положительные числа, $M_{n+1}=\max\limits_{[x_0, x_n]}|f^{n+1}(x)|$.


**Замечание 1.** Порядок точности формулы (27) относительно $h_{\text{max}}$ равен разности между числом узлов интерполяции и порядком вычисляемой производной.

**Замечание 2.** Если формула (27) применяется для вычисления производной в точке, относительно которой узлы таблицы расположены симметрично, и число $n-k$ четно, то порядок точности формулы повышается на единицу по сравнению с порядком $n+1-k$, гарантируемым оценкой (28). Таковы, например формулы (10), (15), (19), (24).

**Напоминание**

Напомним, что интерполяционный многочлен Ньютона с раздельными разностями строится следующим образом:
$$
\begin{equation}
\begin{gathered}
    P_n(x)=f(x_0)+f(x_0; x_1)(x-x_0) + f(x_0;x_1;x_2)(x-x_0)(x-x_1)+...\\
    ...+f(x_0;x_1;...;x_n)(x-x_0)(x-x_1)...(x-x_{n-1})=\\
    =\sum_{k=0}^{n}f(x_0;x_1;...;x_k)\omega_k(x).
\end{gathered}
\end{equation}\tag{29}
$$
Здесь $\omega_0\equiv 1$, $\omega_k(x)=(x-x_0)(x-x_1) \cdots (x-x_{k-1})$ при $k\geqslant 1$.

Также напомним, как строится таблица разделенных разностей. Пусть функция $f$ задана на таблице $x_0$, $x_1$, $...$, $x_n$ значений аргумента с произвольным (не обязательно постоянным) шагом, причем точки таблицы занумернованы в произвольном (не обязательно возрастающем) порядке. Величины
$$
f(x_i; x_{i+1})=\frac{f(x_{i+1})-f(x_i)}{x_{i+1}-x_i}\tag{30}
$$
принято называть резделенными разностями первого порядка функции $f$.
Разделенные разности второго порядка опредляются формулой
$$
f(x_i;x_{i+1};x_{i+2}=\frac{f(x_{i+1};x_{i+2})-f(x_i;x_{i+1})}{x_{i+2}-x_i}\tag{31}
$$
Аналогично определяются разделенные разности третьего и более высоких порядков. Общее определение разделенной разности порядка $k\geqslant 2$ таково:

$$
f(x_i, x_{i+1}, ... x_{i+k}=\frac{f(x_{i+1};...;x_{i+k})-f(x_i);...;x_{i+k-1})}{x_{i+k}-x_i}.\tag{32}
$$

Таблицу разделенных разностей обычно располагают следующим образом:

$$
\begin{array}{c|c|c|c|c}
x & f(x) & \text{1-го порядка} & \text{2-го порядка} & \cdots & \text{$n$-го} \\
\hline
x_0 & f(x_0) &  & &  \\
 &  & f(x_0;x_1) &  &  \\
x_1 & f(x_1) &   & f(x_0;x_1;x_2) & & \\
 &  &  f(x_1;x_2) & \vdots & \text{⋱} &\\
x_2 & f(x_2) & \vdots & \vdots & \cdots & f(x_0; x_1;...; x_n)\\
\vdots & \vdots & \vdots & \vdots & \text{⋰} \\
\vdots & \vdots& \vdots&  f(x_{n-2};x_{n-1};x_n) & \\
\vdots & \vdots & f(x_{n-1};x_n) & \\
x_n & f(x_n) &
\end{array}
$$


Обозначим через $P_{(k,k+1,...,m)}(x)$ интерполяционный полином степени $m-k$ с узлами интерполяции $x_k, x_{k+1}, ..., x_{m}$. В частности обозначим $P_{(k)}(x)=y_k$. В этих обозначениях справедливо равенство
$$
P_{(k,k+1,...,m+1)}(x)=\frac{P_{(k+1,...,m+1)}(x)(x-x_k)-P_{(k,...,m)}(x)(x-x_{m+1})}{x_{m+1}-x_k}.\tag{33}
$$
Действительно, правая часть это полином степени $m+1-k$.

Удобный способ вычисления значений полинома $P_n(x)=P_{(0,1,\ldots, n)}(x)$ это схема Эйткена.

В таблице представлены значения интерполяционных полиномов, вычисляемых по схеме Эйткена, различных порядков:

$$
\begin{array}{c|c|c|c|c}
 \text{нулевого} & \text{первого} & \text{второго} & \cdots & \text{$n$-го} \\
\hline
 P_{(0)}(x)=y_0 &  & &  \\
 & P_{(0,1)}(x) &  &  \\
 P_{(1)}(x)=y_1 &   & P_{(0,1,2)}(x) & & \\
  &  P_{(1,2)}(x) & \vdots & \text{⋱} &\\
 P_{(2)}(x)=y_2 & \vdots & \vdots & \cdots & P_{(0,1,...,n)}(x)\\
 \vdots & \vdots & \vdots & \text{⋰} \\
 \vdots& \vdots&  P_{(n-2,n-1,n)}(x) & \\
 \vdots & P_{(n-1,n)}(x) & \\
 P_{(n)}(x)=y_n &
\end{array}
$$

**Свойства разделенных разностей**

1. $f(x_i;x_{i+1};...;x_{i+k})$ является симметричной функцией своих аргументов $x_i$, $x_{i+1}$, $...$, $x_{i+k}$.
2. Справедливо равенство
   $$
    f(x_i;x_{i+1};...;x_{i+k})=\frac{f^{(k)}(\xi)}{k!},\tag{34}
   $$
   где $\xi$ — некоторая точка на $(a,b)$.
3. $f(x_i;x_{i+1};...;x_{i+k})=\frac{\Delta^k y_i}{h^k k!}$.

**Возвращаемся к формулам численного дифференцирования**

Заметим, что $P_n^{(n)}(x)=n! f(x_0; x_1; ... , x_n)$ (это следует, в частности, из формулы (29)). Таким образом, справедлива приближенная формула (она вытекает и из (34))
$$
f^{(n)}(x)\approx n! f(x_0;x_1;...;x_n),\tag{35}
$$
имеющая как минимум первый порядок точности. Частные случаи:
$$
f'(x)\approx f(x_0;x_1)=\frac{f(x_1)-f(x_0)}{x_1-x_0},\tag{36}
$$
$$
f''(x)\approx 2f(x_0;x_1;x_2)=\frac{2}{x_2-x_0}\left[\frac{f(x_2)-f(x_1)}{x_2-x_1}-\frac{f(x_1)-f(x_0)}{x_1-x_0}\right].\tag{37}
$$

### Использование таблиц с постоянным шагом


$$
f^{(n)}(x)\approx\frac{\Delta^n y_0}{h^n}.\tag{38}
$$

$$
f'(x_0)\approx \frac{1}{h}\sum_{j=1}^n\frac{(-1)^{j-1}}{j}\Delta^{j}y_0,\tag{39}
$$

$$
f'(x_n)\approx \frac{1}{h}\sum_{j=1}^{n}\frac{1}{j}\nabla^{j}y_n.\tag{40}
$$


При $n=2$ из (39) и (40) получаются формулы
$$
f'(x_0)\approx\frac{1}{2h}(-3f(x_0)+4f(x_1)-(x_2)),
$$

$$
f'(x_n)\approx\frac{1}{2h}(f(x_{n-2})-4f(x_{n-1})+3f(x_n)),
$$
имеющие второй порядок точности.
