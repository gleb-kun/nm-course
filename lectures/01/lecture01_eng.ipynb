# Numerical differentiation

## The simplest formulas for numerical differentiation

### Calculating the First Derivative

**Defining the Derivative of a Function through a Limit.**

Let $f : U(x_{0})\subset \mathbb{R}\to \mathbb{R}$
be defined in some neighborhood of a point
$x_0 \in \mathbb {R}$.

The derivative of ${\displaystyle f}$ at ${\displaystyle x_{0}}$ is called its limit if it exists.
$$
{\displaystyle f'(x_{0})=\lim \limits _{x\to x_{0}}{\frac {f(x)-f(x_{0})}{x-x_{0}}}=\lim _{\Delta x\to 0}{\frac {f(x_{0}+\Delta x)-f(x_{0})}{\Delta x}}=\lim \limits _{{\Delta x}\to 0}{\frac {\Delta {f(x)}}{\Delta x}}.}\tag{1}
$$

Based on the definition of the first derivative, we write the derivative at point $x$:
$$
f'(x)=\lim_{\Delta x\rightarrow 0}\frac{f(x+\Delta x)-f(x)}{h},\tag{2}
$$
From this, we can write two simple approximate formulas

$$
f'(x)\approx \frac{f(x+h)-f(x)}{h},\tag{3}
$$

$$
f'(x)\approx \frac{f(x)-f(x-h)}{h},\tag{4}
$$
corresponding to the choice of fixed values ​​$\Delta x= h$ and $\Delta x = -h$, $h>0$ is the step.

**Definition 1**
Formula (3) is called a right-difference scheme.

**Definition 2**
Formula (4) is called a left-difference scheme.

To estimate the errors

$$
r_{+}(x,h)=f'(x)-\frac{f(x+h)-f(x)}{h},\tag{5}
$$

$$
r_{-}(x,h)=f'(x)-\frac{f(x)-f(x-h)}{h}\tag{6}
$$
of the introduced numerical differentiation formulas (approximation errors), we can use Taylor's formula
$$
f(x\pm h)=f(x)\pm f'(x)h+\frac{f''(\xi_{\pm})}{2}h^2.\tag{7}
$$
Here and below, $\xi_{+}$ and $\xi_{-}$ are points located on the intervals $(x,x+h)$ and $(x-h, x)$, respectively.
Substituting the expansions into the expressions for $r_{\pm}$, we obtain $r_{+}(x,h)=-\frac{1}{2}f''(\xi_{+)}h$, $r_{-}(x,h)=\frac{1}{2}f''(\xi_{-})h$. Hence
$$
|r_{+}(x,h)|\leqslant\frac{1}{2}M_1 h,\quad M_1=\max_{\left[x,x+h\right]}|f''(\xi)|,\tag{8}
$$

$$
|r_{-}(x,h)|\leqslant\frac{1}{2}M_2 h,\quad M_2=\max_{\left[x-h,x\right]}|f''(\xi)|.\tag{9}
$$

#### Geometric Interpretation

The formulas given have a geometric interpretation. Let $N_0$, $N_{-}$, $N_{+}$ be points on the graph of the function $y=f(x)$ with coordinates $(x, f(x))$, $(x-h, f(x-h))$, and $(x+h, f(x+h))$.

**Definition 3.** Central difference derivative

$$
f'(x)\approx\frac{f(x+h)-f(x-h)}{2h}.\tag{10}
$$

Lets see the error. Substituting into the error expression
$$
r_0(x,h)=f'(x)-\frac{f(x+h)-f(x-h)}{2h}\tag{11}
$$
the corresponding Taylor expansions
$$
f(x\pm h)=f(x)\pm f'(x)h+\frac{f''(x)}{2}h^2\pm \frac{f^{(3)}(\xi_{\pm})}{6}h^3,\tag{12}
$$
we obtain
$$
r_0(x,h)=-\frac{f^{(3)}(\xi_{+})+f^{(3)}(\xi_{-})}{12}h^2.\tag{13}
$$

Therefore, the error estimate is valid
$$
|r_0(x,h)|\leqslant\frac{M_3}{6}h^2,\quad M_3=\max_{\left[x-h,x+h\right]}|f^{(3)}(\xi)|.\tag{14}
$$

Formulas of any order of accuracy can be obtained for calculating $f'(x)$. However, in such formulas, the number of function values ​​used increases with increasing order of accuracy. As an example, we give the formula
$$
f'(x)\approx\frac{f(x-2h)-8f(x-h)+8f(x+h)-f(x+2h)}{12h}\tag{15}
$$
which has fourth-order accuracy.

**Example 1.** Let the function $f(x)=e^x$ be defined on the interval $[0,1]$ by a table of values ​​with a step $h=0.2$.

```python
import math

# parameters
a = 0       # start of interval
b = 1       # end of interval
h = 0.2     # step

print(" x\t|\t e^x")
print("--------------------")

x_values = []
x = a
while x <= b:
    fx = math.exp(x)
    print(f"{x:.1f}\t|\t{fx:.6f}")
    x_values.append(x)
    x += h
```

```python
f_values = [math.exp(x) for x in x_values]

derivatives = []
n = len(x_values)

for i in range(n):
    if i == 0:
        d = (f_values[i + 1] - f_values[i]) / h
    elif i == n - 1:
        d = (f_values[i] - f_values[i - 1]) / h
    else:
        d = (f_values[i + 1] - f_values[i - 1]) / (2 * h)
    derivatives.append(d)

print(" x\t|\t e^x\t\t|\t f'(x)_num\t|\t f'(x)_exact\t|\t Error")
print("--------------------------------------------------------------------------")

for x, f, d in zip(x_values, f_values, derivatives):
    exact = math.exp(x)
    error = abs(d - exact)
    print(f"{x:.1f}\t|\t{f:.6f}\t|\t{d:.6f}\t|\t{exact:.6f}\t|\t{error:.6e}")
```

A priori error estimates can be used to estimate the errors. For example, the previously considered inequality $|r_{+}(x,h)|\leqslant \frac{1}{2}M_1 h$, where $M_1=\max_{\left[x,x+h\right]}|f''(\xi)|$ (8), gets the estimate at the point $x=0$:
$$
|r|\leqslant \frac{1}{2} M_1 h = \frac{1}{2}\max_{\left[0,0.2\right]}|\exp(\xi)|h=\frac{1}{2}\cdot 1.22140\cdot 0.2\approx 0.13.\tag{16}
$$
At the point $x=0.2$, from the inequality $|r_0(x,h)|\leqslant\frac{M_3}{6}h^2$, where $M_3=\max_{[x-h,x+h]}|f^{(3)}(\xi)|$ (14), it follows that
$$
|r|\leqslant \frac{1}{6} M_3 h^2 = \frac{1}{6}\max_{[0,0.4]}|\exp(\xi)|h^2=\frac{1}{6}\cdot 1.49182\cdot 0.04\approx 0.01, \text{etc.}\tag{17}
$$

To calculate $f'(x)$ for $x=0.4$ and $0.6$, we can also apply formula (15):
$$
f'(x)\approx\frac{f(x-2h)-8f(x-h)+8f(x+h)-f(x+2h)}{12h}\tag{18}
$$
and obtain the values ​​$f'(x)\approx 1.49176$ and $f'(x)\approx 1.82203$ with errors approximately equal to $6\cdot 10^{-5}$ and $9\cdot 10^{-5}$, respectively.

### Calculating the Second Derivative

The formula shown in the following definition is often used.

**Definition.** Second Difference Derivative
$$
f''(x)\approx \frac{f(x-h)-2f(x)+f(x+h)}{h^2}.\tag{19}
$$

Lets see the error. Substituting into the error expression
$$
r(x,h)=f''(x)-\frac{f(x-h)-2f(x)+f(x+h)}{h^2}\tag{20}
$$
the corresponding Taylor expansions
$$
f(x\pm h)=f(x)\pm f'(x)h+\frac{f''(x)}{2}h^2\pm \frac{f^{(3)}(x)}{6}h^3+\frac{f^{(4)}(\xi_{\pm})}{24}h^4,\tag{21}
$$
we obtain
$$
r(x,h)=-\frac{f^{(4)}(\xi_{+})+f^{(4)}(\xi_{-})}{24}h^2\tag{22}
$$

Therefore,
$$
|r(x,h)|\leqslant \frac{M_4}{12}h^2,\quad M_4=\max_{[x-h,x+h]}|f^{(4)}(\xi)|.\tag{23}
$$

Similarly to the case with the first derivative, formulas of any order of accuracy can be obtained for calculating $f''(x)$. For example, the formula
$$
f''(x)\approx \frac{-f(x-2h)+16f(x-h)-30f(x)+16f(x+h)-f(x+2h)}{12h^2}\tag{24}
$$
has fourth-order accuracy.

**Example 2** Using the previously obtained values ​​of the function $f(x)=\exp(x)$, we find the values ​​of $f''(x)$ at the internal nodes of the table using numerical differentiation formulas.

Calculating using the formula $f''(x)\approx \frac{f(x-h)-2f(x)+f(x+h)}{h^2}$ (19) yields the following results

```python
print(" x\t|\t e^x")
print("--------------------")
for x, f in zip(x_values, f_values):
    print(f"{x:.1f}\t|\t{f:.6f}")

second_derivatives = []
n = len(x_values)

for i in range(n):
    if i == 0 or i == n - 1:
        second_derivatives.append(None)
    else:
        d2 = (f_values[i - 1] - 2 * f_values[i] + f_values[i + 1]) / (h ** 2)
        second_derivatives.append(d2)

print("\n x\t|\t f(x)\t\t|\t f''(x)_num\t|\t f''(x)_exact\t|\t Error")
print("--------------------------------------------------------------------------")

for x, f, d2 in zip(x_values, f_values, second_derivatives):
    exact = math.exp(x)  # for exp(x), f''(x) = exp(x)
    if d2 is None:
        print(f"{x:.1f}\t|\t{f:.6f}\t|\t   ---   \t|\t{exact:.6f}\t|\t   ---")
    else:
        error = abs(d2 - exact)
        print(f"{x:.1f}\t|\t{f:.6f}\t|\t{d2:.6f}\t|\t{exact:.6f}\t|\t{error:.6e}")
```

**Example 2.1** Calculation using formula (24):
$$
f''(x)\approx \frac{-f(x-2h)+16f(x-h)-30f(x)+16f(x+h)-f(x+2h)}{12h^2}\tag{25}
$$
gives the following results:

```python
print(" x\t|\t e^x")
print("--------------------")
for x, f in zip(x_values, f_values):
    print(f"{x:.1f}\t|\t{f:.6f}")

second_derivatives = []
n = len(x_values)

for i in range(n):
    if i < 2 or i > n - 3:
        second_derivatives.append(None)
    else:
        d2 = (-f_values[i - 2] + 16 * f_values[i - 1] - 30 * f_values[i]
              + 16 * f_values[i + 1] - f_values[i + 2]) / (12 * h ** 2)
        second_derivatives.append(d2)

print("\n x\t|\t f(x)\t\t|\t f''(x)_num\t|\t f''(x)_exact\t|\t Error")
print("--------------------------------------------------------------------------")

for x, f, d2 in zip(x_values, f_values, second_derivatives):
    exact = math.exp(x)  # for exp(x), f''(x) = exp(x)
    if d2 is None:
        print(f"{x:.1f}\t|\t{f:.6f}\t|\t   ---   \t|\t{exact:.6f}\t|\t   ---")
    else:
        error = abs(d2 - exact)
        print(f"{x:.1f}\t|\t{f:.6f}\t|\t{d2:.6f}\t|\t{exact:.6f}\t|\t{error:.6e}")
```

Using formula (24) allows us to obtain the values ​​$f''(0.4)\approx 1.49204$, $f''(0.6)\approx 1.82183$ with errors approximately equal to $-2\cdot 10^{-5}$, $-3\cdot 10^{-5}$.

## On the Derivation of Numerical Differentiation Formulas

Suppose that in the neighborhood of a point $x$, a function is approximated by another function $g$, and the derivative $g^{(k)}$ at $x$ is easily calculated. In such a situation, it is natural to try to use the approximate formula:
$$
f^{(k)}(x)\approx g^{(k)}(x).\tag{26}
$$

### Numerical Differentiation Formulas Based on Interpolation by Algebraic Polynomials.

Let $P_n(x)$ be an interpolation polynomial of degree $n$ with interpolation nodes $x_0 < x_1 < x_2 < ... < x_n$ and $x\in [x_0, x_n]$. In this case, formula (26) takes the form
$$
f^{(k)}(x)\approx P_n^{(k)}(x),\quad 0\leqslant k \leqslant n. \tag{27}
$$

Error estimate for formula (27):
$$
|f^{(k)}(x)-P_n^{(k)(x)}|\leqslant C_{n,k}M_{n+1}h^{n+1-k}_{\max}, \quad 0\leqslant k\leqslant n,\tag{28}
$$
where $C_{n,k}$ are positive numbers, $M_{n+1}=\max\limits_{[x_0, x_n]}|f^{n+1}(x)|$.

**Remark 1.** The order of accuracy of formula (27) with respect to $h_{\text{max}}$ is equal to the difference between the number of interpolation nodes and the order of the calculated derivative.

**Remark 2.** If formula (27) is used to calculate the derivative at a point relative to which the nodes of the table are located symmetrically, and the number $n-k$ is even, then the order of accuracy of the formula increases by one compared to the order $n+1-k$ guaranteed by estimate (28). Such are, for example, formulas (10), (15), (19), (24).

**Reminder**

Recall that Newton's interpolation polynomial with divided differences is constructed as follows:
$$
\begin{gathered}
P_n(x)=f(x_0)+f(x_0; x_1)(x-x_0) + f(x_0;x_1;x_2)(x-x_0)(x-x_1)+...\\
...+f(x_0;x_1;...;x_n)(x-x_0)(x-x_1)...(x-x_{n-1})=\\
=\sum_{k=0}^{n}f(x_0;x_1;...;x_k)\omega_k(x).
\end{gathered}\tag{29}
$$
Here $\omega_0\equiv 1$, $\omega_k(x)=(x-x_0)(x-x_1) \cdots (x-x_{k-1})$ for $k\geqslant 1$.

Let us also recall how a table of divided differences is constructed. Let a function $f$ be defined on a table $x_0$, $x_1$, $...$, $x_n$ of argument values with an arbitrary (not necessarily constant) step, with the points of the table numbered in an arbitrary (not necessarily ascending) order. The quantities
$$
f(x_i; x_{i+1})=\frac{f(x_{i+1})-f(x_i)}{x_{i+1}-x_i}\tag{30}
$$
are usually called the divided differences of the first order of the function $f$.
Second-order divided differences are defined by the formula
$$
f(x_i;x_{i+1};x_{i+2})=\frac{f(x_{i+1};x_{i+2})-f(x_i;x_{i+1})}{x_{i+2}-x_i}\tag{31}
$$
Divided differences of the third and higher orders are defined similarly. The general definition of a divided difference of order $k\geqslant 2$ is:

$$
f(x_i, x_{i+1}, ... x_{i+k})=\frac{f(x_{i+1};...;x_{i+k})-f(x_i);...;x_{i+k-1})}{x_{i+k}-x_i}.\tag{32}
$$

Table of divided differences are usually arranged as follows:

**Table 1.**
$$
\begin{array}{c|c|c|c|c|c}
x & f(x) & \text{1st order} & \text{2nd order} & \cdots & \text{$n$-th} \\
\hline
x_0 & f(x_0) &  &  &  & \\
    &        & f(x_0;x_1) &  &  & \\
x_1 & f(x_1) &  & f(x_0;x_1;x_2) &  & \\
    &        & f(x_1;x_2) & \vdots & \text{⋱} & \\
x_2 & f(x_2) & \vdots & \vdots & \cdots & f(x_0;x_1;\ldots;x_n) \\
\vdots & \vdots & \vdots & \vdots & \text{⋰} & \\
\vdots & \vdots & \vdots & f(x_{n-2};x_{n-1};x_n) &  & \\
\vdots & \vdots & f(x_{n-1};x_n) &  &  & \\
x_n & f(x_n) &  &  &  & \\
\end{array}
$$

**Properties of Divided Differences**

1. $f(x_i;x_{i+1};...;x_{i+k})$ is a symmetric function of its arguments $x_i$, $x_{i+1}$, $...$, $x_{i+k}$.
2. The following equality holds:
$$
f(x_i;x_{i+1};...;x_{i+k})=\frac{f^{(k)}(\xi)}{k!},\tag{34}
$$
where $\xi$ is some point on $(a,b)$.
3. $f(x_i;x_{i+1};...;x_{i+k})=\frac{\Delta^k y_i}{h^k k!}$.

**Aitken's Method**

Denote by $P_{(k,k+1,...,m)}(x)$ the interpolation polynomial of degree $m-k$ with interpolation nodes $x_k, x_{k+1}, ..., x_{m}$. In particular, we denote $P_{(k)}(x)=y_k$. With this notation, equality
$$
P_{(k,k+1,...,m+1)}(x)=\frac{P_{(k+1,...,m+1)}(x)(x-x_k)-P_{(k,...,m)}(x)(x-x_{m+1})}{x_{m+1}-x_k}.\tag{33}
$$
Indeed, the right-hand side is a polynomial of degree $m+1-k$.

A convenient way to calculate the values ​​of the polynomial $P_n(x)=P_{(0,1,\ldots, n)}(x)$ is the Aitken scheme.

The table presents the values ​​of interpolation polynomials calculated using the Aitken scheme of various orders:

**Table 2.**
$$
\begin{array}{c|c|c|c|c}
\text{zeroth} & \text{first} & \text{second} & \cdots & \text{$n$th} \\
\hline
P_{(0)}(x)=y_0 &  &  &  & \\
               & P_{(0,1)}(x) &  &  & \\
P_{(1)}(x)=y_1 &  & P_{(0,1,2)}(x) &  & \\
               & P_{(1,2)}(x) & \vdots & \text{⋱} & \\
P_{(2)}(x)=y_2 & \vdots & \vdots & \cdots & P_{(0,1,\ldots,n)}(x) \\
\vdots & \vdots & \vdots & \text{⋰} & \\
\vdots & \vdots & P_{(n-2,n-1,n)}(x) &  & \\
\vdots & P_{(n-1,n)}(x) &  &  & \\
P_{(n)}(x)=y_n &  &  &  & \\
\end{array}
$$

**Returning to numerical differentiation formulas**

Note that $P_n^{(n)}(x)=n! f(x_0; x_1; ... , x_n)$ (this follows, in particular, from formula (29)). Thus, the approximate formula (which also follows from (34)) is valid:
$$
f^{(n)}(x)\approx n! f(x_0;x_1;...;x_n),\tag{35}
$$
having at least first-order accuracy. Special cases:
$$
f'(x)\approx f(x_0;x_1)=\frac{f(x_1)-f(x_0)}{x_1-x_0},\tag{36}
$$
$$
f''(x)\approx 2f(x_0;x_1;x_2)=\frac{2}{x_2-x_0}\left[\frac{f(x_2)-f(x_1)}{x_2-x_1}-\frac{f(x_1)-f(x_0)}{x_1-x_0}\right].\tag{37}
$$

### Using tables with a constant step

The simplest form of numerical differentiation formulas is obtained when using a table  
$y_i = f(x_i)$ with a constant step. For example, formula (35) in this case takes the form:

$$
f^{(n)}(x) \approx \frac{\Delta^{n} y_{0}}{h^{n}} .\tag{38}
$$

In situations where the value of the derivative must be computed at the boundary points of the table,
$x_{0}$ and $x_{n}$, one-sided numerical differentiation formulas are used:

$$
f^{(k)}(x_{0}) \approx P_{n}^{(k)}(x_{0})
\qquad \text{and} \qquad
f^{(k)}(x_{n}) \approx P_{n}^{(k)}(x_{n}).\tag{39}
$$

Below we give the one-sided formulas (they can be easily obtained by differentiating the Newton polynomials and) for computing the first derivative $ f'$:
$$
f'(x_{0}) \approx \frac{1}{h} \sum_{j=1}^{n} \frac{(-1)^{\,j-1}}{j} \, \Delta^{j} y_{0},
\tag{40}
$$

$$
f'(x_{n}) \approx \frac{1}{h} \sum_{j=1}^{n} \frac{1}{j} \, \nabla^{j} y_{n},
\tag{41}
$$
which have $n$-th order accuracy.


For $n = 2$, formula (40) becomes
$$
f'(x_{0}) \approx \frac{1}{h} \left( \Delta y_{0} - \frac{1}{2} \Delta^{2} y_{0} \right).\tag{42}
$$

The forward finite differences are:
- $\Delta y_{0} = y_{1} - y_{0}$,
- $\Delta^{2} y_{0} = \Delta(\Delta y_{0}) = (y_{2} - y_{1}) - (y_{1} - y_{0}) = y_{2} - 2y_{1} + y_{0}$.

Substituting into the expression:
$$
\begin{aligned}
f'(x_{0}) &\approx \frac{1}{h} \left[ (y_{1} - y_{0}) - \frac{1}{2}(y_{2} - 2y_{1} + y_{0}) \right] \\[4pt]
&= \frac{1}{h} \left( y_{1} - y_{0} - \frac{1}{2}y_{2} + y_{1} - \frac{1}{2}y_{0} \right) \\[4pt]
&= \frac{-3y_{0} + 4y_{1} - y_{2}}{2h}.
\end{aligned}\tag{43}
$$

Since $y_{k} = f(x_{k})$, we obtain formula (44):
$$
f'(x_{0}) \approx \frac{1}{2h}\left( -3f(x_{0}) + 4f(x_{1}) - f(x_{2}) \right).
\tag{44}
$$


Also for $n = 2$, formula (41) becomes
$$
f'(x_{n}) \approx \frac{1}{h} \left( \nabla y_{n} + \frac{1}{2} \nabla^{2} y_{n} \right).\tag{45}
$$

The backward finite differences are:
- $\nabla y_{n} = y_{n} - y_{n-1}$,
- $\nabla^{2} y_{n} = (y_{n} - y_{n-1}) - (y_{n-1} - y_{n-2}) = y_{n} - 2y_{n-1} + y_{n-2}$.

Substituting:
$$
\begin{aligned}
f'(x_{n}) &\approx \frac{1}{h} \left[ (y_{n} - y_{n-1}) + \frac{1}{2}(y_{n} - 2y_{n-1} + y_{n-2}) \right] \\[4pt]
&= \frac{1}{h} \left( \frac{3}{2}y_{n} - 2y_{n-1} + \frac{1}{2}y_{n-2} \right) \\[4pt]
&= \frac{y_{n-2} - 4y_{n-1} + 3y_{n}}{2h}.
\end{aligned}\tag{46}
$$

That gives formula (43):
$$
f'(x_{n}) \approx \frac{1}{2h}\left( f(x_{n-2}) - 4f(x_{n-1}) + 3f(x_{n}) \right).
\tag{47}
$$

---

Both formulas are second-order accurate (local truncation error $O(h^{2})$).

### Conditionality of numerical differentiation formulas

Total error
$$
\begin{aligned}
r^{*}(x,h) &= f'(x) - \frac{f^{*}(x+h) - f^{*}(x)}{h}= \\
&= \underbrace{f'(x) - \frac{f(x+h) - f(x)}{h}}_{r_{+}(x,h)}
+ \underbrace{\frac{1}{h} \Big( [f(x+h) - f^{*}(x+h)] - [f(x) - f^{*}(x)] \Big)}_{r_{F}(x,h)}.
\end{aligned}\tag{48}
$$

The total error $r^{*}(x,h)$ of the actual calculated value of the right difference derivative is the sum of the approximation error $r_{+}(x,h)=f'(x)-\frac{f(x+h)-f(x)}{h}$ and the unremovable error $r_{F}(x,h)$.

Let $\bar\Delta$ be the upper bound of the absolute error $\Delta(f^{*}(x))=|f(x)-f^{*}(x)|$ of the function values used. Then the non-removable error $r_{F}$ is estimated as follows:
$$
|r_{F}|\leqslant 2\bar\Delta / h.\tag{49}
$$

Estimate (49) means that the sensitivity of formula (3) to input data errors is characterized by the absolute condition number $v_{\Delta}=2/h$. Since $v_{\Delta}\rightarrow\infty$ for $h\rightarrow 0$, formula (3) becomes very ill-conditioned for $h$.
Therefore, despite the fact that the approximation error tends to zero as $h \to 0$ (see estimate (12.4)), one should expect the total error to increase indefinitely as $h \to 0$.

At least, this is the behavior of the upper bound of the total error:
$$
\bar{r}(h) = \frac{1}{2} M_2 h + \frac{2 \bar{\Delta}}{h}.\tag{50}
$$

We choose the optimal step size $h$ at which $\bar{r}(h)$ reaches its minimum value. Equating the derivative
$$
\bar{r}'(h) = \frac{1}{2} M_2 - \frac{2 \bar{\Delta}}{h^2}\tag{51}
$$
to zero, we obtain the value
$$
h_{\text{opt}} = 2 \sqrt{\frac{\bar{\Delta}}{M_2}},\tag{52}
$$
which corresponds to the value
$$
\bar{r}_{\min} = \bar{r}(h_{\text{opt}}) = 2 \sqrt{\bar{\Delta} M_2}.\tag{53}
$$

Thus, when using formula (3) to calculate the derivative of a function $f$ specified with an error, special attention should be paid to the choice of the step $h$. However, even with an optimal choice of step, the total error will be a value proportional only to $\sqrt{\bar{\Delta}}$.









**Example 3.** Consider the derivative of the function $f(x)=\exp(x)$ at the point $x=1$. We will use different values ​​of the step $h$.

```python
import math

# function and exact derivative at point x=1
f = math.exp
x = 1.0
fprime_true = math.e

# steps h = 10^-1 ... 10^-6
hs = [10**(-k) for k in range(1, 15)]

# calculations
approxs = []
errors = []
for h in hs:
    approx = (f(x + h) - f(x)) / h
    err = fprime_true - approx
    approxs.append(approx)
    errors.append(err)
    
print("\nTable:")
print(f"{'h':>8} {'(f\')*':>12} {'r*':>16}")
for h,a,e in zip(hs, approxs, errors):
    print(f"{h:8.0e} {a:12.5f} {e:16.5e}")
```
