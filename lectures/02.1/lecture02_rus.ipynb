# Численное интегрирование

Построим интерполяционные квадратурные формулы, изучим формулы Ньютона–Котеса и их погрешности, а затем обсудим чувствительность к ошибкам значений функции.

Все функции и веса далее считаются вещественнозначными.

## Интерполяционные квадратурные формулы

Если функция $f(x)$ непрерывна на отрезке $[a,b]$, где $a<b$, то у нее есть первообразная $F$ на $[a,b]$ и справедливо равенство

$$
\int_a^b f(x)\,dx = F(b) - F(a)\tag{1}
$$

Это основная формула интегрального исчисления — формула Ньютона–Лейбница.

**Определение 1 (квадратурная формула).** Квадратурная формула приближает интеграл взвешенной суммой значений функции:

$$
\int_a^b f(x)\,dx \approx \sum_{j=1}^N C_j f(x_j).\tag{2}
$$

Сумма в (2) называется **квадратурной суммой**, точки $x_j$ — **узлами**, а числа $C_j$ — **весами**.

Для некоторых формул квадратурная сумма совпадает с интегральной суммой Римана. Например, формула прямоугольников имеет вид

$$
\int_a^b f(x)\,dx \approx \frac{b - a}{N} \sum_{j=1}^N f(\alpha + (j - 1)h),\tag{3}
$$

где $h=(b-a)/N$, $\alpha\in[a,a+h]$.

Запишем квадратурную формулу в более общем виде по сравнению с (2):

$$
\int_a^b w(x) f(x)\,dx \approx \sum_{j=1}^N C_j f(x_j).\tag{4}
$$

В (4) функция $w(x)$ называется **весовой функцией** и считается фиксированной.

### Алгебраическая интерполяция

Пусть требуется вычислить интеграл

$$
\int_a^b w(x) f(x)\,dx.\tag{5}
$$

Промежуток интегрирования может быть конечным или бесконечным. Предполагается, что вес $w$ — вещественная измеримая функция и все используемые далее моменты существуют абсолютно, то есть $\int_a^b|w(x)||x|^k\,dx<\infty$. Предполагается также $\int_a^b|w(x)f(x)|\,dx<\infty$.

**Определение 2 (моменты веса).** Моментами называются интегралы

$$
\mu_k = \int_a^b w(x)x^k\,dx, \quad k = 0, 1, 2, \ldots,\tag{6}
$$

Выберем $N\ge1$ различных вещественных узлов из промежутка интегрирования.
Рассмотрим **интерполяционный многочлен** $P(x)$ функции $f(x)$, построенный по узлам $x_1,x_2,\ldots,x_N$. Его степень не превосходит $N-1$, и он удовлетворяет условиям

$$
P(x_j) = f(x_j), \quad j = 1, 2, \ldots, N.\tag{7}
$$

Запишем многочлен в форме Лагранжа:

$$
P(x) = \sum_{j=1}^N l_j(x) f(x_j),\tag{8}
$$

где $l_j(x)=\prod_{k\ne j}(x-x_k)/(x_j-x_k)$, $l_j(x_i)=\delta_{ij}$ — фундаментальные интерполяционные многочлены.

В интеграле (5) заменим функцию $f(x)$ многочленом $P(x)$. Получим приближенное равенство

$$
\int_a^b w(x) f(x)\,dx \approx \int_a^b w(x) P(x)\,dx.\tag{9}
$$

Подставим вместо $P(x)$ правую часть равенства (8) и проинтегрируем. Получим формулу вида (4), коэффициенты которой определяются равенствами

$$
C_j = \int_a^b w(x) l_j(x)\,dx, \quad j = 1, 2, \ldots, N.\tag{10}
$$

**Определение 3.** Квадратурная формула вида (4) с различными узлами $x_j$ и коэффициентами из (10) называется **интерполяционной квадратурной формулой**.

**Теорема 1.** Для того чтобы квадратурная формула (4) с $N$ различными узлами была интерполяционной, необходимо и достаточно, чтобы она была точной для всех многочленов степени не выше $N-1$, то есть для $f(x)=1,x,x^2,\ldots,x^{N-1}$.

**Доказательство.**

**Необходимость.** Пусть квадратурная формула (4) интерполяционная, а $f$ — многочлен степени не выше $N-1$. Интерполяционный многочлен по $N$ узлам совпадает с $f$:

$$
f(x) = \sum_{j=1}^N l_j(x) f(x_j).\tag{11}
$$

Умножая на $w(x)$ и интегрируя, получаем

$$
\int_a^b w(x) f(x)\,dx = \sum_{j=1}^N f(x_j) \int_a^b w(x) l_j(x)\,dx = \sum_{j=1}^N C_j f(x_j),\tag{12}
$$

Значит, формула точна для $f$. Здесь использовано равенство (10).

**Достаточность.** Пусть формула (4) точна для всех многочленов степени не выше $N-1$. В частности, она точна для фундаментальных многочленов $l_k$:

$$
\int_a^b w(x) l_k(x)\,dx = \sum_{j=1}^N C_j l_k(x_j) = C_k, \quad k = 1, 2, \ldots, N.\tag{13}
$$

Эти равенства означают, что формула (4) интерполяционная. Теорема доказана.

**Определение 4.** Неотрицательное целое число $d$ называется **алгебраической степенью точности** квадратурной формулы (4), если формула точна для всех многочленов степени не выше $d$, но не точна для $x^{d+1}$.

**Следствие 1.** Из теоремы 1 следует, что для интерполяционной формулы с $N$ узлами $d\ge N-1$.

**Теорема 2 (верхняя граница степени точности).** Если дополнительно выполнены условия

$$
w(x) \ge 0, \quad x \in [a, b], \quad \mu_0 = \int_a^b w(x)\,dx > 0,\tag{14}
$$

то алгебраическая степень точности формулы (4) с $N$ узлами удовлетворяет неравенству

$$
d \le 2N - 1.\tag{15}
$$

**Доказательство.** Для вещественных узлов $x_1,\ldots,x_N$ построим многочлен

$$
\omega(x)=\prod_{j=1}^N(x-x_j).\tag{16}
$$

Многочлен $\omega(x)^2$ имеет степень $2N$ и положителен всюду, кроме конечного числа вещественных точек. Поскольку $w\ge0$ и $\mu_0>0$, имеем

$$
\int_a^b w(x)\omega(x)^2\,dx>0,\quad
\sum_{j=1}^N C_j\omega(x_j)^2=0.\tag{17}
$$

Следовательно, (15) выполняется. Теорема доказана.

**Следствие 2.** Если формула, удовлетворяющая (14), точна для всех многочленов степени не выше $m$, то

$$
N\ge\left\lceil\frac{m+1}{2}\right\rceil.\tag{18}
$$

Действительно, $m\le d\le2N-1$, а $N$ — целое число.

## Квадратурные формулы Ньютона–Котеса

Рассмотрим интеграл

$$
\int_a^b f(x)\,dx\tag{19}
$$

с единичным весом. Из условия существования моментов (6) при $w(x)=1$ следует, что промежуток интегрирования должен быть конечным.

Пусть $n$ — положительное целое число. Выберем равноотстоящие узлы с шагом

$$
h_n = \frac{b - a}{n}\tag{20}
$$

по формуле

$$
x_k^{(n)} = a + k h_n, \quad k = 0, 1, \ldots, n.\tag{21}
$$

Концы отрезка включены в число узлов; всего узлов $N=n+1$.

**Определение 5 (замкнутая формула Ньютона–Котеса).** Интерполяционная формула с этими узлами

$$
\int_a^b f(x)\,dx \approx \sum_{k=0}^n C_k^{(n)} f(x_k^{(n)})\tag{22}
$$

называется **квадратурной формулой Ньютона–Котеса**. Значения $n=1,2,3,\ldots$ задают семейство таких формул.

Коэффициенты определяются по (10):

$$
C_k^{(n)} = \int_a^b \frac{\omega_{n+1}(x)\,dx}{(x - x_k^{(n)}) \omega_{n+1}'(x_k^{(n)})},\tag{23}
$$

где

$$
\omega_{n+1}(x) = (x - x_0^{(n)})(x - x_1^{(n)}) \cdots (x - x_n^{(n)}).\tag{24}
$$

После сокращения множителя $x-x_k^{(n)}$ частное в (23) является многочленом: кажущаяся особенность устранима. Выполним замену $x=a+h_nt$. Тогда

$$
x - x_k^{(n)} = h_n (t - k),\tag{25}
$$

и

$$
\omega_{n+1}(x) = h_n^{n+1} t(t - 1)(t - 2) \cdots (t - n),\tag{26}
$$

а также

$$
\omega_{n+1}'(x_k^{(n)}) = h_n^n (-1)^{n-k} k!(n - k)!.\tag{27}
$$

Следовательно,

$$
C_k^{(n)} = \frac{h_n (-1)^{n-k}}{k!(n - k)!} \int_0^n \frac{t(t - 1)(t - 2) \cdots (t - n)}{t - k}\,dt.\tag{28}
$$

Коэффициент $C_k^{(n)}$ зависит от длины отрезка через $h_n=(b-a)/n$. Запишем

$$
C_k^{(n)} = (b - a) B_k^{(n)},\tag{29}
$$

где

$$
B_k^{(n)} = \frac{(-1)^{n - k}}{k!(n - k)!n} \int_0^n \frac{t(t - 1) \cdots (t - n)}{t - k}\,dt.\tag{30}
$$

Числа $B_k^{(n)}$ не зависят от отрезка $[a,b]$, поэтому их можно табулировать. Формула Ньютона–Котеса принимает вид

$$
\int_a^b f(x)\,dx \approx (b - a) \sum_{k=0}^n B_k^{(n)} f(x_k^{(n)}).\tag{31}
$$

Веса для $n=1,2,\ldots,10$ впервые вычислил Котес. Ниже приведены значения для $n=1,2,\ldots,6$.

Для приведенных коэффициентов выполняется симметрия

$$
B_k^{(n)} = B_{n - k}^{(n)}, \quad (k = 0, 1, \ldots, n).\tag{32}
$$

Это равенство справедливо при любом $n$. Докажем лемму, из которой оно следует как частный случай.

**Лемма 1 (симметрия весов).** Пусть квадратурная формула

$$
\int_a^b w(x) f(x)\,dx \approx \sum_{k=1}^N C_k f(x_k)\tag{33}
$$

удовлетворяет следующим условиям:

1) отрезок $[a,b]$ конечен;

2) весовая функция принимает одинаковые значения в точках, симметричных относительно середины $c=(a+b)/2$, то есть

$$
w(a + b - x) = w(x), \quad x \in [a, b];\tag{34}
$$

3) узлы симметричны относительно середины: вместе с $x_k$ узлом является $a+b-x_k$ для $k=1,2,\ldots,N$.

Если формула (33) интерполяционная, то коэффициенты при симметричных узлах равны.

**Доказательство.** Обозначим $x_{-j}=a+b-x_j$, $j=1,2,\ldots,N$. По условию 3 это также узлы формулы. Соответствующие веса обозначим $C_{-j}$ и $C_j$. Нужно доказать $C_{-j}=C_j$.

Из интерполяционности следует

$$
C_{-j} = \int_a^b \frac{w(x)\,\omega_N(x)}{(x - x_{-j})\,\omega_N'(x_{-j})}\,dx.\tag{35}
$$

Сделаем замену $x=a+b-t$. По условию 3 множество отраженных узлов совпадает с исходным, поэтому

$$
\begin{aligned}
\omega_N(a+b-t)&=\prod_{k=1}^N(a+b-t-x_k)\\
&=(-1)^N\prod_{k=1}^N\bigl(t-(a+b-x_k)\bigr)=(-1)^N\omega_N(t).
\end{aligned}\tag{36}
$$

Таким образом,

$$
\omega_N(x) = (-1)^N \omega_N(t),\tag{37}
$$

где $t=a+b-x$. Дифференцирование по $x$ дает

$$
\omega_N'(x) = (-1)^{N+1} \omega_N'(t).\tag{38}
$$

Это равенство связывает значения производной в симметричных точках. В частности,

$$
\omega_N'(x_{-j}) = (-1)^{N+1} \omega_N'(x_j).\tag{39}
$$

Используя (37), (39) и симметрию веса из условия 2, перепишем интеграл (35):

$$
\begin{aligned}
C_{-j}&=\int_b^a\frac{w(a+b-t)(-1)^N\omega_N(t)}{(a+b-t-x_{-j})(-1)^{N+1}\omega_N'(x_j)}(-dt)\\
&=\int_a^b\frac{w(t)\omega_N(t)}{(t-x_j)\omega_N'(x_j)}\,dt=C_j.
\end{aligned}\tag{40}
$$

Лемма доказана.

**Следствие 3.** Формула Ньютона–Котеса (31) удовлетворяет условиям леммы, поэтому $B_k^{(n)}=B_{n-k}^{(n)}$ при $k=0,1,\ldots,n$.

Нам потребуется теорема о среднем значении.

**Теорема 3 (взвешенная теорема о среднем).** Пусть $f$ и $g$ интегрируемы по Риману на конечном отрезке $[a,b]$ и:

1) $m\le f(x)\le M$ при всех $x\in[a,b]$;

2) $g(x)$ не меняет знак на $[a,b]$.

Тогда произведение $fg$ интегрируемо и

$$
\int_a^b f(x) g(x)\,dx = \mu \int_a^b g(x)\,dx,\tag{41}
$$

где $m\le\mu\le M$. Если $\int_a^b g=0$, обе части равны нулю. Если $f$ непрерывна, найдется $\eta\in[a,b]$, для которой $\mu=f(\eta)$.

**Теорема 4 (степень точности формул Ньютона–Котеса).** Алгебраическая степень точности $d_n$ формулы (31) равна

$$
d_n =
\begin{cases}
n, & \text{если } n \text{ нечетно}, \\
n + 1, & \text{если } n \text{ четно}.
\end{cases}\tag{42}
$$

**Доказательство.** Докажем точность для многочленов до указанной степени, а затем неточность на следующей степени. По интерполяционности и теореме 1 формула точна для всех многочленов степени не выше $n$. Это пока дает только нижнюю границу $d_n\ge n$.

При четном $n$ формула также точна для $(x-c)^{n+1}$. Его интеграл равен нулю по нечетности относительно $c$. Квадратурная сумма тоже нулевая: значения в симметричных узлах противоположны, а веса равны.

Поскольку формула точна для степеней до $n$ и для $(x-c)^{n+1}$, она точна для всех многочленов степени не выше $n+1$; значит, $d_n\ge n+1$.

Пусть $n$ четно. Многочлен $\omega_{n+1}$ из (24) принимает противоположные значения в симметричных относительно $c$ точках, поскольку $N=n+1$ нечетно. Поэтому

$$
\int_a^b \omega_{n+1}(x)\,dx = 0.\tag{43}
$$

Определим многочлен степени $n+2$

$$
\Omega(x) = \int_a^x \omega_{n+1}(t)\,dt,\tag{44}
$$

Он удовлетворяет условиям

$$
\Omega(a) = \Omega(b) = 0.\tag{45}
$$

Рассмотрим интегралы

$$
I_j = \int_{a + jh_n}^{a + (j+1)h_n} \omega_{n+1}(x)\,dx, \quad j = 0, 1, \ldots, n - 1.\tag{46}
$$

Докажем, что при $n\ge4$

$$
|I_0| > |I_1| > \cdots > |I_{n/2 - 1}|.\tag{47}
$$

В интеграле $I_j$, где $1\le j\le n/2-1$, сделаем замену $x=y+h_n$:

$$
\begin{aligned}
I_j&=\int_{a+(j-1)h_n}^{a+jh_n}\prod_{s=0}^n(y+h_n-x_s^{(n)})\,dy\\
&=\int_{a+(j-1)h_n}^{a+jh_n}\frac{y+h_n-x_0^{(n)}}{y-x_n^{(n)}}\omega_{n+1}(y)\,dy.
\end{aligned}\tag{48}
$$

Поскольку $\omega_{n+1}(y)$ сохраняет знак на промежутке интегрирования, по теореме 3

$$
I_j = \frac{\eta_j + h_n - x_0^{(n)}}{\eta_j - x_n^{(n)}} I_{j-1},\tag{49}
$$

где

$$
a + (j - 1)h_n \le \eta_j \le a + jh_n.\tag{50}
$$

Множитель в (49) при $1\le j\le n/2-1$ отрицателен, и

$$
\frac{h_n + \eta_j - a}{h_n n - (\eta_j - a)} \le \frac{h_n + (n/2 - 1)h_n}{h_n n - (n/2 - 1)h_n} = \frac{n}{n + 2} < 1.\tag{51}
$$

Здесь использована правая граница из (50) при $j=n/2-1$. Отсюда $|I_j|<|I_{j-1}|$, то есть (47) доказано.

Знаки интегралов чередуются: $\operatorname{sgn}I_j=(-1)^j$, а $I_j=-I_{n-1-j}$. Благодаря (47) чередующиеся частичные интегралы на левой половине остаются положительными; по симметрии $\Omega(a+b-x)=\Omega(x)$. При $n=2$ это непосредственно следует из $I_0>0$. Значит,

$$
\Omega(x) = \int_a^x \omega_{n+1}(t)\,dt > 0, \quad a < x < b.\tag{52}
$$

Докажем неточность формулы (31) для многочлена $x\omega_{n+1}(x)$ степени $n+2$. Квадратурная сумма равна нулю, поскольку многочлен обращается в ноль во всех узлах. Его интеграл ненулевой: интегрируя по частям и учитывая (45), (52), получаем

$$
\int_a^b x \omega_{n+1}(x)\,dx = x \Omega(x)\big|_a^b - \int_a^b \Omega(x)\,dx = -\int_a^b \Omega(x)\,dx < 0.\tag{53}
$$

Следовательно, при четном $n$ имеем $d_n=n+1$.

При нечетном $n$ многочлен $\omega_{n+1}$ симметричен относительно $c$. Интегралы по промежуткам удовлетворяют $I_j=I_{n-1-j}$, первый из них отрицателен, а далее знаки чередуются. То же сравнение после сдвига показывает строгое убывание модулей до середины, включая центральный промежуток: при $n=2m+1$ модуль множителя меньше единицы во внутренних точках для $1\le j\le m$. Попарное объединение симметричных промежутков дает $\int_a^b\omega_{n+1}(x)\,dx<0$. При $n=1$ подынтегральная функция отрицательна на $(a,b)$ непосредственно. Квадратурная сумма для $\omega_{n+1}$ равна нулю, поэтому формула не точна на степени $n+1$. Значит, при нечетном $n$ имеем $d_n=n$.

Рассмотрим конкретные **формулы Ньютона–Котеса** при $n=1,2,3$.

**Пример 1 (формула трапеций).** При $n=1$ формула (31) принимает вид

$$
\int_a^b f(x)\,dx \approx \frac{b - a}{2} [f(a) + f(b)].\tag{54}
$$

Это **формула трапеций** (см. рисунок 1 ниже). Ее алгебраическая степень точности равна 1.

*Рисунок 1. Формула трапеций использует прямую, соединяющую значения функции в концах отрезка.*

```python
import numpy as np
import matplotlib.pyplot as plt

a, b = 0.0, 1.0

def f(x):
    t = (x - a) / (b - a)
    return 0.25 + 0.9 * (1.5 * t - 0.5 * t**2)

x = np.linspace(a, b, 200)
plt.figure(figsize=(5, 3))
plt.plot(x, f(x), label=r"$f(x)$")
plt.plot([a, b], [f(a), f(b)], label="Линейный интерполянт")
plt.fill_between([a, b], [f(a), f(b)], alpha=0.15)
plt.xticks([a, b], ["a", "b"])
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

**Пример 2 (формула Симпсона).**

При $n=2$ получаем

$$
\int_a^b f(x)\,dx \approx \frac{b - a}{6} \left[ f(a) + 4f\!\left( \frac{a + b}{2} \right) + f(b) \right].\tag{55}
$$

Это **формула Симпсона**. Ее алгебраическая степень точности равна 3.

**Пример 3 (правило трех восьмых).**

При $n=3$ формула (31) принимает вид

$$
\int_a^b f(x)\,dx \approx \frac{b - a}{8} \left[ f(a) + 3f(a + h_3) + 3f(a + 2h_3) + f(b) \right].\tag{56}
$$

Это **правило трех восьмых**: название связано с множителем $3/8$. Алгебраическая степень точности также равна 3.

**Определение 6 (погрешность квадратуры).** **Погрешность**, или **остаточный член**, — разность между точным интегралом и квадратурной суммой. Для формулы Ньютона–Котеса (31) обозначим ее

$$
R^{(n)}(f) = \int_a^b f(x)\,dx - (b - a)\sum_{k=0}^n B_k^{(n)} f(x_k^{(n)}).\tag{57}
$$

Представления остаточного члена требуют достаточной гладкости $f$. Для четного и нечетного $n$ используются разные приемы. Подробно рассмотрим два простейших случая: $n=1$ и $n=2$.

**Утверждение 1 (остаточный член формулы трапеций).** Если $f\in C^2([a,b])$, погрешность (54) допускает представление (65).

**Доказательство.** Пусть $P_1$ интерполирует $f$ в точках $a,b$.

По стандартной формуле остатка линейной интерполяции для каждого $x\in(a,b)$

$$
f(x)-P_1(x)=\frac{f''(\xi)}{2}\,(x-a)(x-b),\qquad \xi=\xi(x)\in(a,b).\tag{58}
$$

Для строгого перехода к интегралу введем $q_2(x)=2[f(x)-P_1(x)]/[(x-a)(x-b)]$ при $a<x<b$. Из остаточной формулы следует $m_2\le q_2(x)\le M_2$, где $m_2,M_2$ определены в (62). Это частное непрерывно внутри отрезка и ограничено; задав в концах любые конечные значения, получим интегрируемую по Риману функцию. Непрерывность выбора $\xi(x)$ не предполагается. Интегрирование дает

$$
\int_a^b f(x)\,dx-\int_a^b P_1(x)\,dx
= \frac12\int_a^b q_2(x)\,(x-a)(x-b)\,dx.\tag{59}
$$

Формула трапеций точна для $P_1$, поэтому

$$
\int_a^b P_1(x)\,dx=\frac{b-a}{2}\,[P_1(a)+P_1(b)]
=\frac{b-a}{2}\,[f(a)+f(b)].\tag{60}
$$

Таким образом, левая часть (59) — погрешность $R^{(1)}(f)$ и

$$
R^{(1)}(f)=\frac12\int_a^b q_2(x)\,(x-a)(x-b)\,dx.\tag{61}
$$

Многочлен $(x-a)(x-b)$ отрицателен на $(a,b)$. Используем оценки для частного, не предполагая непрерывности промежуточной точки $\xi$ по $x$.

Положим

$$
m_2=\min_{a\le x\le b} f''(x),\qquad
M_2=\max_{a\le x\le b} f''(x).\tag{62}
$$

Интегрирование оценок для $q_2$ с множителем постоянного знака дает число $\mu_2$, для которого

$$
R^{(1)}(f)=\frac{\mu_2}{2}\int_a^b (x-a)(x-b)\,dx,\tag{63}
$$

где $m_2\le\mu_2\le M_2$.

Непосредственное вычисление дает

$$
\int_a^b (x-a)(x-b)\,dx=-\frac{(b-a)^3}{6}.\tag{64}
$$

Поскольку $f''$ непрерывна, по теореме о промежуточных значениях существует $\eta\in[a,b]$, для которой $\mu_2=f''(\eta)$. Тогда

$$
R^{(1)}(f)=-\,\frac{(b-a)^3}{12}\,f''(\eta),\qquad \eta\in[a,b].\tag{65}
$$

Это **представление остаточного члена формулы трапеций**.

**Утверждение 2 (оценка через остаток интерполяции).** При нечетном $n$ (и, в действительности, при любом $n\ge1$) из остатка интерполяции следует общая оценка

$$
|R^{(n)}(f)|\le\frac{M_{n+1}}{(n+1)!}\int_a^b|\omega_{n+1}(x)|\,dx,\quad
M_{n+1}=\max_{[a,b]}|f^{(n+1)}|.\tag{66}
$$

Здесь $f\in C^{n+1}([a,b])$, а $\omega_{n+1}$ — многочлен со старшим коэффициентом 1 и корнями в узлах квадратуры. При $n>1$ он меняет знак, поэтому простой переход к одной промежуточной точке, использованный для трапеций, непосредственно неприменим.

Оценка получается интегрированием модуля остатка интерполяции.

**Утверждение 3 (остаточный член формулы Симпсона).** Если $f\in C^4([a,b])$, погрешность (55) допускает представление (75).

**Доказательство.** Как и раньше поступить нельзя: $\omega_3(x)=(x-a)(x-c)(x-b)$ меняет знак.

Введем **интерполяционный многочлен Эрмита** $P_3$, удовлетворяющий условиям

$$
P_3(a) = f(a), \qquad P_3(c) = f(c), \qquad P_3'(c) = f'(c), \qquad P_3(b) = f(b),\tag{67}
$$

где $c=(a+b)/2$.

По стандартной формуле остатка интерполяции Эрмита вне узлов

$$
f(x) - P_3(x) = \frac{f^{(IV)}(\xi)}{4!}(x - a)(x - c)^2(x - b), \qquad \xi = \xi(x) \in (a,b).\tag{68}
$$

Положим $q_4(x)=24[f(x)-P_3(x)]/[(x-a)(x-c)^2(x-b)]$ вне узлов. Остаточная формула ограничивает это частное минимумом и максимумом $f^{(4)}$. Оно непрерывно вне узлов и ограничено, поэтому интегрируемо после задания конечных значений в узлах. Интегрируя от $a$ до $b$ и учитывая

$$
\int_a^b P_3(x)\,dx = \frac{b - a}{6}\,[f(a) + 4f(c) + f(b)],\tag{69}
$$

получаем

$$
R^{(2)}(f) = \frac{1}{24}\int_a^b q_4(x)\,(x - a)(x - c)^2(x - b)\,dx,\tag{70}
$$

где $R^{(2)}(f)$ — **погрешность формулы Симпсона**.

Множитель $(x-a)(x-c)^2(x-b)$ неположителен и имеет отрицательный интеграл. Ограниченное частное $q_4$ удовлетворяет оценкам

$$
m_4 \le q_4(x) \le M_4,\tag{71}
$$

где

$$
m_4 = \min_{a \le x \le b} f^{(IV)}(x), \qquad M_4 = \max_{a \le x \le b} f^{(IV)}(x).\tag{72}
$$

Поэтому из (70) следует

$$
R^{(2)}(f) = \frac{\mu_4}{24} \int_a^b (x - a)(x - c)^2(x - b)\,dx,\tag{73}
$$

где $m_4\le\mu_4\le M_4$.

Вычислим интеграл в (73), сделав замену $x-c=t$:

$$
\begin{aligned}
\int_a^b(x-a)(x-c)^2(x-b)\,dx
&=\int_{-\alpha}^{\alpha}t^2(t^2-\alpha^2)\,dt\\
&=-\frac{4\alpha^5}{15}=-\frac{(b-a)^5}{120}.
\end{aligned}\tag{74}
$$

Здесь $\alpha=(b-a)/2$.

По непрерывности $f^{(IV)}$ и теореме о промежуточных значениях найдется $\eta\in[a,b]$, для которой $\mu_4=f^{(IV)}(\eta)$. Следовательно,

$$
R^{(2)}(f) = -\,\frac{(b - a)^5}{2880}\, f^{(IV)}(\eta), \qquad \eta \in [a,b].\tag{75}
$$

Получено **представление остаточного члена формулы Симпсона**.

**Утверждение 4 (оценка при четном $n$).** При четном $n\ge2$ используем многочлен Эрмита степени не выше $n+1$, совпадающий с $f$ во всех $n+1$ узлах и с $f'$ в среднем узле $c=(a+b)/2$. По точности квадратуры до степени $n+1$ получаем оценку

$$
|R^{(n)}(f)|\le\frac{M_{n+2}}{(n+2)!}\int_a^b|(x-c)\omega_{n+1}(x)|\,dx,\quad
M_{n+2}=\max_{[a,b]}|f^{(n+2)}|.\tag{76}
$$

Здесь требуется $f\in C^{n+2}([a,b])$. Остаток Эрмита содержит многочлен $(x-c)\omega_{n+1}(x)$ степени $n+2$. При $n=2$ оценка согласуется с формулой (75) после перехода к модулю.

**Замечание 1 (формулы Ньютона–Котеса высоких степеней).**

Рассмотрим поведение коэффициентов $B_k^{(n)}$ при увеличении $n$. Первые веса для $n=1,2,\ldots,10$ вычислил **Котес**. При $n=8$ и $n=10$ появляются отрицательные веса. Последующие вычисления показали наличие положительных и отрицательных весов и при больших $n$.

Для оценки чувствительности введем

$$
\gamma_n=\sum_{k=0}^n|B_k^{(n)}|.\tag{77}
$$

Для классических формул Ньютона–Котеса последовательность сумм модулей весов неограничена (см. [1, 2]). Из точности на константах $\sum_{k=0}^n B_k^{(n)}=1$, поэтому $\gamma_n>1$ указывает на наличие отрицательных весов. Впервые они появляются при $n=8$; при $n=9$ все веса снова положительны. Поведение при больших $n$ нельзя выводить из предположения о монотонности при малых $n$.

**Таблица 2.** Значения $\gamma_n$ показывают быстрый рост чувствительности:

| n  | 1–7 | 8   | 9   | 10  | 20  | 30  | 40  | 50  |
|-----|-----|-----|-----|-----|-----|-----|-----|-----|
| $\gamma_n$ | 1 | 1.45122 | 1 | 3.06479 | 5.44177·10² | 2.11964·10⁵ | 1.10196·10⁸ | 6.69208·10¹⁰ |

Вернемся к **квадратурной формуле (33)** общего вида и предположим, что она точна для $f(x)=1$:

$$
\sum_{k=1}^N C_k = \mu_0 = \int_a^b w(x)\,dx,\tag{78}
$$

Здесь предполагаются вещественные веса, $w\ge0$ и $\mu_0>0$.

**Определение 7 (абсолютное число обусловленности).** Для квадратурной суммы (33) число $\gamma=\sum_{k=1}^N|C_k|$ является абсолютным числом обусловленности, если ошибки в узлах измеряются максимальным модулем.

Вместо точных значений $f(x_k)$ используются приближения $\tilde f(x_k)$ с ошибками $\varepsilon_k=f(x_k)-\tilde f(x_k)$. Пусть $|\varepsilon_k|\le\varepsilon$ при $k=1,\ldots,N$. Для оценки влияния ошибок данных будем считать веса и суммирование точными.

Тогда внесенная ошибками данных погрешность удовлетворяет оценке

$$
\left|\sum_{k=1}^N C_k\varepsilon_k\right|\le\varepsilon\sum_{k=1}^N|C_k|=\varepsilon\gamma.\tag{79}
$$

Фактическая погрешность может быть меньше, но это верхняя граница, пропорциональная $\gamma$. Чем больше $\gamma$, тем сильнее возможное влияние ошибок значений функции на квадратурную сумму.

Полученная оценка касается только возмущений значений в узлах; погрешность аппроксимации, округление весов и ошибки суммирования следует оценивать отдельно.

Наихудшая оценка достигается при $\varepsilon_k=\varepsilon\,\operatorname{sgn}C_k$; для конкретного набора ошибок результат может быть меньше.

**Утверждение 5 (минимальная чувствительность).** Величина $\gamma$ принимает **минимальное** значение при неотрицательных коэффициентах $C_k\ge0$, $k=1,\ldots,N$. **Доказательство.** Из (78)

$$
\gamma=\sum_{k=1}^N|C_k|\ge\left|\sum_{k=1}^N C_k\right|=\mu_0,\tag{80}
$$

При неотрицательных весах достигается равенство $\gamma=\mu_0$.

Равенство достигается **тогда и только тогда**, когда все вещественные коэффициенты $C_k$ неотрицательны: из $\mu_0>0$ следует, что хотя бы один из них положителен. Это объясняет практическую важность **квадратурных формул с положительными коэффициентами**.

**Замечание 2 (составные квадратуры).**

У **формул Ньютона–Котеса** высоких степеней появляются отрицательные веса, а $\gamma_n$ быстро растет. Поэтому использование таких формул при большом $n$ непрактично.

Точность выгоднее повышать **разбиением отрезка на меньшие части** и применением на каждой из них формулы Ньютона–Котеса с малым $n$. Это приводит к **составным квадратурным формулам**, которые будут рассмотрены в следующей лекции.

**Пример 4 (сравнение на полиноме).** Для $f(x)=x^4$ на $[0,1]$ сравним приближения трапеций и Симпсона с точным интегралом $1/5$. Выведем погрешность со знаком: точный интеграл минус квадратурная сумма.

```python
a, b = 0.0, 1.0

def f(x):
    return x**4

c = (a + b) / 2
exact = (b**5 - a**5) / 5
trapezoid = (b - a) * (f(a) + f(b)) / 2
simpson = (b - a) * (f(a) + 4 * f(c) + f(b)) / 6

print(f"Точный интеграл: {exact:.8f}")
print(f"Формула трапеций: {trapezoid:.8f}, погрешность: {exact - trapezoid:.8f}")
print(f"Формула Симпсона: {simpson:.8f}, погрешность: {exact - simpson:.8f}")
```
