# Numerical integration

## Interpolatory Quadrature Formulas

If a function $f(x)$ is continuous on the interval $[a,b]$, then it has an antiderivative $F(x)$, and the following holds:
$$
\int_a^b f(x)\,dx = F(b) - F(a)\tag{f1}
$$
known as the fundamental formula of integral calculus.

Formulas for approximate calculation of integrals have the form
$$
\int_a^b f(x)\,dx \approx \sum_{j=1}^N C_j f(x_j).\tag{1}
$$

For some formulas, the quadrature sum coincides with the Riemann sum. For example, the quadrature formula for rectangles:
$$
\int_a^b f(x)\,dx \approx \frac{b - a}{N} \sum_{j=1}^N f(\alpha + (j - 1)h),\tag{f2}
$$
where $h = (b - a)/N, \alpha \in [a, a + h]$.

We will write the quadrature formula in a more general form compared to (1):
$$
\int_a^b w(x) f(x)\,dx \approx \sum_{j=1}^N C_j f(x_j).\tag{2}
$$

### Algebraic interpolation

Let us assume that we are concerned with evaluating the integral

$$
\int_a^b w(x) f(x)\,dx.\tag{3}
$$

The integration interval may be finite or infinite.  
The main requirement for the weight function $w(x)$ is the existence of the integrals

$$
\mu_k = \int_a^b w(x)x^k\,dx, \quad k = 0, 1, 2, \ldots,\tag{4}
$$


Consider the **interpolation polynomial** $P(x)$ of the function $f(x)$, constructed through the nodes $x_1, x_2, \ldots, x_N$.  
The degree of $P(x)$ does not exceed $N - 1$, and it satisfies the interpolation conditions:
$$
P(x_j) = f(x_j), \quad j = 1, 2, \ldots, N.\tag{f3}
$$

We can express the polynomial $P(x)$ in Lagrange form:
$$
P(x) = \sum_{j=1}^N l_j(x) f(x_j),\tag{5}
$$

where $l_j(x)$ are the fundamental interpolation polynomials.

In the integral (3), we replace the function $f(x)$ with the polynomial $ P(x)$.  
As a result, we obtain the following approximate equality:

$$
\int_a^b w(x) f(x)\,dx \approx \int_a^b w(x) P(x)\,dx.\tag{f4}
$$


Substitute the right-hand side of equality (5) for $P(x)$ and perform the integration.  
We obtain the approximate equality of the form (2), in which the coefficients $C_j$ are defined by the relations

$$
C_j = \int_a^b w(x) l_j(x)\,dx, \quad j = 1, 2, \ldots, N.\tag{6}
$$


---

### Definition 1
A quadrature formula of the form (2), where the nodes $x_j$ are distinct numbers, and the coefficients $C_j$ are given by expressions (6), is called an **interpolatory quadrature formula**.

---

### Theorem 1
For a quadrature formula (2) with $N$ distinct nodes to be interpolatory, it is **necessary and sufficient** that it be **exact** for all polynomials $f(x)$ of degree not higher than $N - 1$,  
that is, for $f(x) = 1, x, x^2, \ldots, x^{N-1}$.

---

**Proof.**  
**Necessity.** Let the quadrature formula (2) be interpolatory, and let $f(x)$ be a polynomial of degree not higher than $N - 1$.  
The interpolation polynomial constructed for $f(x)$ at the $N$ nodes of the quadrature formula (2) coincides with $f(x)$:

$$
f(x) = \sum_{j=1}^N l_j(x) f(x_j).\tag{f5}
$$

Multiply both sides of this equality by $w(x)$ and integrate over $[a, b]$. We get

$$
\int_a^b w(x) f(x)\,dx = \sum_{j=1}^N f(x_j) \int_a^b w(x) l_j(x)\,dx = \sum_{j=1}^N C_j f(x_j),\tag{f6}
$$

which means that the quadrature formula is exact for $f(x)$.  
Here, we used relation (6), which is valid since the formula (2) is interpolatory.

---

**Sufficiency.**  
Suppose the quadrature formula (2) is exact for any polynomial of degree not higher than $N - 1$.  
In particular, it must be exact for the fundamental interpolation polynomials $l_k(x)$ constructed at the nodes of the quadrature formula (2):

$$
\int_a^b w(x) l_k(x)\,dx = \sum_{j=1}^N C_j l_k(x_j) = C_k, \quad k = 1, 2, \ldots, N.\tag{f7}
$$


These equalities mean that the quadrature formula (2) is interpolatory.  
The theorem is proven.

---

### Definition 2
A non-negative integer $d$ is called the **algebraic degree of accuracy** of a quadrature formula (2) if this formula is exact for all polynomials of degree no higher than $d$, but not exact for $x^{d+1}$.

From Theorem 1 it follows that the algebraic degree of accuracy of an interpolatory quadrature formula with $N$ nodes satisfies the inequality $d \geq N - 1$.

---

### Theorem 2
Under the additional assumption that the weight function satisfies

$$
w(x) \ge 0, \quad x \in [a, b], \quad \mu_0 = \int_a^b w(x)\,dx > 0,\tag{7}
$$

the algebraic degree of accuracy $d$ of the quadrature formula (2) with $N$ nodes satisfies the inequality

$$
d \le 2N - 1.\tag{8}
$$

---

**Proof.**  
For the nodes $x_1, x_2, \ldots, x_N$ of the quadrature formula (2), which may include complex values, construct the polynomial of degree $N$:

$$
\omega(x) = (x - x_1)(x - x_2)\cdots(x - x_N), \quad \overline{\bar\omega(x)} = (x - \overline{\bar x}_1)\cdots(x - \overline{\bar x}_N),\tag{f8}
$$

where $\overline{x}_j$ denotes the complex conjugate of $x_j$.  
Obviously, the polynomial $\omega(x)\overline{\omega(x)}$ has degree $2N$ and, for $x \in [a, b]$, equals $|\omega(x)|^2$, and thus is nonnegative on $[a, b]$ and positive everywhere except possibly at $N$ points where it can vanish.

For the polynomial $\omega(x)\overline{\omega(x)}$, the quadrature formula (2) is not exact, because

$$
\int_a^b w(x)|\omega(x)|^2\,dx > 0, \quad \sum_{j=1}^N C_j w(x_j)\overline{\omega(x_j)} = 0.\tag{f9}
$$

From this it follows that (8) holds.  
The theorem is proven.

---

From this theorem, we obtain the following statement:  
If the quadrature formula (2) with a weight function satisfying condition (7) is exact for all polynomials of degree up to $m$, then the number of its nodes satisfies the inequality $N \ge \frac{m}{2} + 1$.

Indeed, if $N \le \lfloor \frac{m}{2} \rfloor$, then by Theorem 2 the algebraic degree of accuracy $d$ of this formula satisfies the inequality $d \le 2\lfloor \frac{m}{2} \rfloor - 1 < m - 1$.  
This is impossible, since $d \ge m$.


## § 2. Newton–Cotes Quadrature Formulas

We consider the integral  
$$
\int_a^b f(x)\,dx\tag{f1}
$$  
with a unit weight function.  
From the requirement of the existence of moments (1.4) (formula 4) with $w(x) = 1$, it follows that the interval $[a, b]$ must be finite.  

Let $n$ be a natural number.  
We choose as nodes the equally spaced points with step  
$$
h_n = \frac{b - a}{n}\tag{f2}
$$  
given by  
$$
x_k^{(n)} = a + k h_n, \quad k = 0, 1, \ldots, n.\tag{1}
$$  

The endpoints $a$ and $b$ are included among the nodes, so the total number of nodes is $N = n + 1$.

The interpolatory quadrature formula with nodes (1)

$$
\int_a^b f(x)\,dx \approx \sum_{k=0}^n C_k^{(n)} f(x_k^{(n)})\tag{2}
$$  

is called the **Newton–Cotes quadrature formula**.  
Since $n$ can take the values $1, 2, 3, \ldots$, we obtain a sequence of Newton–Cotes quadrature formulas.

The quadrature formula (2) is interpolatory, and its coefficients are defined by formula (1.6):

$$
C_k^{(n)} = \int_a^b \frac{\omega_{n+1}(x)\,dx}{(x - x_k^{(n)}) \omega_n'(x_k^{(n)})},\tag{3}
$$  

where  

$$
\omega_{n+1}(x) = (x - x_0^{(n)})(x - x_1^{(n)}) \cdots (x - x_n^{(n)}).\tag{f3}
$$

In the integral (3), let us make the change of variable $x = a + h_n t$. We then have

$$
x - x_k^{(n)} = h_n (t - k),\tag{f4}
$$

and

$$
\omega_{n+1}(x) = h_n^{n+1} t(t - 1)(t - 2) \cdots (t - n),\tag{f5}
$$

while

$$
\omega_n'(x_k^{(n)}) = h_n^n (-1)^{n-k} k!(n - k)!.\tag{f6}
$$

Therefore,

$$
C_k^{(n)} = \frac{h_n (-1)^{n-k}}{k!(n - k)!} \int_0^n \frac{t(t - 1)(t - 2) \cdots (t - n)}{t - k}\,dt.\tag{f7}
$$


The obtained value of the coefficient $C_k^{(n)}$ depends on the length of the interval $h_n = (b - a)/n$.  
Let us write

$$
C_k^{(n)} = (b - a) B_k^{(n)},\tag{4}
$$

where

$$
B_k^{(n)} = \frac{(-1)^{n - k}}{k!(n - k)!n} \int_0^n \frac{t(t - 1) \cdots (t - n)}{t - k}\,dt.\tag{5}
$$

The values $B_k^{(n)}$ do not depend on the interval $[a, b]$, and thus can be tabulated.  
The Newton–Cotes quadrature formula takes the form

$$
\int_a^b f(x)\,dx \approx (b - a) \sum_{k=0}^n B_k^{(n)} f(x_k^{(n)}).\tag{6}
$$

The coefficients $B_k^{(n)}$ for $n = 1, 2, \ldots, 10$ were first computed by Cotes.

For the coefficients listed in the table, the following symmetry relation holds:

$$
B_k^{(n)} = B_{n - k}^{(n)}, \quad (k = 0, 1, \ldots, n).\tag{f8}
$$

This equality holds for any $n$.  
We now prove a lemma, from which this result follows as a special case.

---

### Lemma 1
Let the quadrature formula

$$
\int_a^b w(x) f(x)\,dx \approx \sum_{k=1}^N C_k f(x_k)\tag{7}
$$
satisfy the following conditions:


1) the integration interval $[a, b]$ is finite;

2) the weight function $w(x)$ takes equal values at points symmetric with respect to  
the midpoint $c = (a + b)/2$ of the interval $[a, b]$, i.e.

$$
w(a + b - x) = w(x), \quad x \in [a, b];\tag{f9}
$$

3) the nodes are symmetric with respect to the midpoint, i.e., if $x_k$ is a node, then $a + b - x_k$ is also a node for $k = 1, 2, \ldots, N$.

If the quadrature formula (7) is interpolatory, then the coefficients corresponding to the nodes symmetric with respect to $c$ are equal.

---

### Proof
Let us set $x_{-j} = a + b - x_j$, $j = 1, 2, \ldots, N$.  
According to condition 3 of the lemma, the points $x_{-j}$ are also nodes, and the nodes $x_{-j}$ and $x_j$ are symmetric with respect to $c$.  
Let us denote the corresponding coefficients by $C_{-j}$ and $C_j$.  
We need to show that $C_{-j} = C_j$.

Since the formula (7) is interpolatory, we have

$$
C_{-j} = \int_a^b \frac{w(x)\,\omega_N(x)}{(x - x_{-j})\,\omega_N'(x_{-j})}\,dx.\tag{8}
$$

Let us make the change of variable in this integral: $x = a + b - t$.  
According to condition 3 of the lemma, the set $\{x_{-j}\,|\, j = 1, 2, \ldots, N\}$ coincides with the set of all nodes of the quadrature formula.  
Therefore,

$$
\omega_N(x) = \prod_{k=1}^N (x - x_k) =
$$
$$
\prod_{k=1}^N (a + b - t - x_k) = (-1)^N \prod_{k=1}^N (t - (a + b - x_k)) = (-1)^N \omega_N(t).\tag{f10}
$$

We have thus obtained the relation

$$
\omega_N(x) = (-1)^N \omega_N(t),\tag{9}
$$

where $x$ and $t$ are related by $t = a + b - x$.  
Differentiating both sides of (9) with respect to $x$, we get

$$
\omega_N'(x) = (-1)^{N+1} \omega_N'(t).\tag{10}
$$

This equality connects the values of the derivative $\omega_N'(x)$ at points symmetric with respect to $c$.  
In particular,

$$
\omega_N'(x_{-j}) = (-1)^{N+1} \omega_N'(x_j).\tag{11}
$$


Using equalities (9), (11), and condition 2) of the lemma, we rewrite the integral (8) in the new variable of integration:

$$
C_{-j} = \int_b^a \frac{w(a + b - t)(-1)^N \omega_N(t)}{(a + b - t - x_{-j})(-1)^{N+1} \omega_N'(x_j)}(-dt)=
$$
$$
= \int_a^b \frac{w(t)\, \omega_N(t)}{(t - x_j)\, \omega_N'(x_j)}\,dt = C_j.\tag{f11}
$$

Thus, the lemma is proven.

---

The Newton–Cotes quadrature formula (6) satisfies the conditions of the lemma; therefore, we obtain  
$B_k^{(n)} = B_{n - k}^{(n)}$, for $k = 0, 1, \ldots, n$.

We now need a theorem concerning the mean value. Let us state it as follows:

---

### Theorem 1
Let the functions $f(x)$ and $g(x)$ be integrable on the finite interval $[a, b]$, and assume that:

1) $f(x)$ is bounded, i.e. $m \le f(x) \le M$ for all $x \in [a, b]$;  
2) $g(x)$ does not change sign on $[a, b]$.

Then the function $f(x) g(x)$ is integrable, and

$$
\int_a^b f(x) g(x)\,dx = \mu \int_a^b g(x)\,dx,\tag{f12}
$$

where $m \le \mu \le M$.  
If $f(x)$ is continuous on $[a, b]$, then there exists a point $\eta \in [a, b]$ such that $\mu = f(\eta)$.

---

### Theorem 2
The **algebraic degree of accuracy** $d_n$ of the Newton–Cotes quadrature formula (6) is given by

$$
d_n =
\begin{cases}
n, & \text{if } n \text{ is odd}, \\
n + 1, & \text{if } n \text{ is even}.
\end{cases}\tag{f13}
$$

---

### Proof
We will prove that the quadrature formula (6) is exact for polynomials of degree not higher than $d_n$.  
Since the quadrature formula (6) is interpolatory, by Theorem 1.1 it is exact for all polynomials of degree not exceeding $n$.  
Thus, the statement for odd $n$ is already proven.

Let $n$ be even.  
We will show that for any polynomial of degree $n + 1$, of the form $(x - c)^{n+1}$, the quadrature formula (6) is exact.  
Indeed, the integral over $[a, b]$ of $(x - c)^{n+1}$ equals zero.  
The quadrature sum for it is also zero because for symmetric nodes with respect to the midpoint $c$, the function takes equal absolute values but opposite signs, and since the coefficients corresponding to symmetric nodes are equal, the positive and negative terms cancel each other.


Since the coefficients corresponding to symmetric nodes are equal,  
and since the formula (6) is exact for polynomials of degree not higher than $n$,  
it follows that it is also exact for $(x - c)^{n+1}$.  
Thus, the formula is exact for all polynomials of degree not higher than $n + 1 = d_n$.

Let $n$ be even.  
The polynomial $\omega_{n+1}(x)$ from (9) has the following property:  
at points symmetric with respect to $c = (a + b)/2$,  
it takes values that are equal in magnitude but opposite in sign ($N = n + 1$).  
Therefore,

$$
\int_a^b \omega_{n+1}(x)\,dx = 0.\tag{f13}
$$

Define the function

$$
\Omega(x) = \int_a^x \omega_{n+1}(t)\,dt,\tag{12}
$$

which is a polynomial of degree $n + 2$ satisfying the conditions

$$
\Omega(a) = \Omega(b) = 0.\tag{13}
$$

---

Consider the integrals

$$
I_j = \int_{a + jh_n}^{a + (j+1)h_n} \omega_{n+1}(x)\,dx, \quad j = 0, 1, \ldots, n - 1.\tag{f14}
$$

We will show that for $n \ge 4$, the following inequalities hold:

$$
|I_0| > |I_1| > \cdots > |I_{n/2 - 1}|.\tag{14}
$$

In the integral $I_j$ (where $1 \le j \le n/2 - 1$),  
we make the substitution $x = y + h_n$. Then we obtain:

$$
I_j = \int_{a + (j - 1)h_n}^{a + jh_n} \prod_{s = 0}^n (y + h_n - x_s^{(n)})\,dy=
$$
$$
=\int_{a + (j - 1)h_n}^{a + jh_n} \frac{y + h_n - x_0^{(n)}}{y - x_n^{(n)}} \omega_{n+1}(y)\,dy.\tag{f15}
$$

Since $\omega_{n+1}(y)$ preserves its sign over the interval of integration,  
we can apply the Mean Value Theorem (Theorem 1) to the last integral:

$$
I_j = \frac{\eta_j + h_n - x_0^{(n)}}{\eta_j - x_n^{(n)}} I_{j-1},\tag{15}
$$

where

$$
a + (j - 1)h_n \le \eta_j \le a + jh_n.\tag{16}
$$

The factor in equality (15), for all admissible values of $j$ ($1 \le j \le n/2 - 1$),  
is negative, and its absolute value satisfies the inequality

$$
\frac{h_n + \eta_j - a}{h_n n - (\eta_j - a)} \le \frac{h_n + (n/2 - 1)h_n}{h_n n - (n/2 - 1)h_n} = \frac{n}{n + 2} < 1.\tag{f16}
$$

Here, the second of inequalities (16) for $j = n/2 - 1$ was used.  
From (15) we obtain $|I_j| < |I_{j-1}|$. Thus, inequality (14) is proven.

Since  
$\operatorname{sgn} I_j = (-1)^j$ and $I_j = -I_{n - 1 - j}$ for $j = 0, 1, \ldots, n - 1$,  
we get from (14) that

$$
\Omega(x) = \int_a^x \omega_{n+1}(t)\,dt > 0, \quad a < x < b.\tag{17}
$$

---

We now prove that the quadrature formula (6) is **not exact** for the polynomial  
$x \omega_{n+1}(x)$ of degree $n + 2$.  
Since this polynomial vanishes at the nodes of formula (6),  
it suffices to show that its integral over $[a, b]$ is nonzero.  
Integrating by parts and taking into account (13) and (17), we have:

$$
\int_a^b x \omega_{n+1}(x)\,dx = x \Omega(x)\big|_a^b - \int_a^b \Omega(x)\,dx = -\int_a^b \Omega(x)\,dx < 0.\tag{f17}
$$

Thus, we have proved that $d_n = n + 1$ for even $n$.

---

In the case of odd $n$, the quadrature formula (6) is not exact for the polynomial $\omega_{n+1}(x)$,  
since $\int_a^b \omega_{n+1}(x)\,dx$ is negative.  
This follows from the fact that the function $\Omega(x)$, defined by (12),  
for odd $n$ satisfies the inequality $\Omega(x) < 0$ for $a < x \le b$.  
We will not go further into the proof of this statement.

---

Let us now consider specific **Newton–Cotes quadrature formulas** for $n = 1, 2, 3$.  

For $n = 1$, the quadrature formula (6) has the form

$$
\int_a^b f(x)\,dx \approx \frac{b - a}{2} [f(a) + f(b)].\tag{18}
$$

It is called the **trapezoidal rule** (see Fig. 6).  
The algebraic degree of accuracy of this formula is 1.

---

For $n = 2$, the quadrature formula (6) has the form

$$
\int_a^b f(x)\,dx \approx \frac{b - a}{6} \left[ f(a) + 4f\!\left( \frac{a + b}{2} \right) + f(b) \right].\tag{19}
$$


It is called **Simpson’s formula**.  
The algebraic degree of accuracy is 3.

---

For $n = 3$, the quadrature formula (6) takes the form

$$
\int_a^b f(x)\,dx \approx \frac{b - a}{8} \left[ f(a) + 3f(a + h_3) + 3f(a + 2h_3) + f(b) \right].\tag{f18}
$$

It is called the **three-eighths rule**, the name originating from the coefficient $3/8$.  
The algebraic degree of accuracy is also 3.

---

The **error** or **remainder term** of a quadrature formula is the difference between the true integral and the quadrature sum.  
We denote the error of the Newton–Cotes quadrature formula (6) by

$$
R^{(n)}(f) = \int_a^b f(x)\,dx - (b - a)\sum_{k=0}^n B_k^{(n)} f(x_k^{(n)}).\tag{f19}
$$

Representations for $R^{(n)}(f)$ exist under the assumption that the function $f(x)$  
is sufficiently smooth (i.e., has a sufficient number of derivatives).  
These representations differ for even and odd $n$, and are derived using different techniques.  
We will not consider the general case for arbitrary $n$,  
but will limit ourselves to the two simplest and most typical cases: $n = 1$ and $n = 2$.

---

For $n = 1$, we are dealing with the error representation for the **trapezoidal rule** (formula (18)).  
Let us assume that the function $f(x)$ has a continuous second derivative on $[a, b]$.  
Let $P_1(x)$ be the **interpolation polynomial** of the function $f(x)$ constructed through the endpoints $a$ and $b$.


```python
import numpy as np
import matplotlib.pyplot as plt

a, b = 0.0, 1.0
def f(x):
    t = (x - a) / (b - a)
    return 0.25 + 0.9*np.sqrt(t)

xs = np.linspace(a, b, 400)
ys = f(xs)

plt.figure(figsize=(4, 3))

ymax = float(ys.max()) * 1.15
plt.annotate('', xy=(b + 0.08, 0), xytext=(a - 0.12, 0),
             arrowprops=dict(arrowstyle='->', linewidth=1.2))
plt.annotate('', xy=(a, ymax), xytext=(a, -0.02*ymax),
             arrowprops=dict(arrowstyle='->', linewidth=1.2))

plt.plot([a, b], [0, 0], linewidth=1.2)
plt.plot(xs, ys, linewidth=2.0)
plt.plot([a, b], [f(a), f(b)], linewidth=2.0)
tick_h = ymax * 0.04
plt.plot([a, a], [0, tick_h], linewidth=1.2)
plt.plot([b, b], [0, tick_h], linewidth=1.2)
plt.text((a + b) * 0.58, f(0.58*(b-a)+a) + 0.03*ymax, r'$y=f(x)$')
plt.text(a, -0.08*ymax, r'$a$', ha='center', va='top')
plt.text(b, -0.08*ymax, r'$b$', ha='center', va='top')
plt.xlim(a - 0.12, b + 0.1)
plt.ylim(-0.1*ymax, ymax)
plt.axis('off')
plt.tight_layout()
plt.show()
```

For the interpolation error (see formula (1.3.11)) we have
$$
f(x)-P_1(x)=\frac{f''(\xi)}{2}\,(x-a)(x-b),\qquad \xi=\xi(x)\in(a,b).\tag{21}
$$

Integrate both sides from $x=a$ to $x=b$:
$$
\int_a^b f(x)\,dx-\int_a^b P_1(x)\,dx
= \frac12\int_a^b f''(\xi)\,(x-a)(x-b)\,dx.\tag{22}
$$

Since the trapezoidal rule is exact for $P_1(x)$,
$$
\int_a^b P_1(x)\,dx=\frac{b-a}{2}\,[P_1(a)+P_1(b)]
=\frac{b-a}{2}\,[f(a)+f(b)].\tag{f20}
$$

Thus, the left–hand side of (22) is precisely the error $R^{(1)}(f)$ of the trapezoidal rule, and (22) can be written as
$$
R^{(1)}(f)=\frac12\int_a^b f''(\xi)\,(x-a)(x-b)\,dx.\tag{23}
$$

The integrand on the right–hand side of (23) is the product of the function $f''(\xi)$ and the polynomial $(x-a)(x-b)$, which is negative on $(a,b)$.  
From (21) it follows that the function $f''(\xi)=f''(\xi(x))$ is continuous as a function of $x$ on $[a,b]$ (except, possibly, at the endpoints $a,b$) and therefore integrable on $[a,b]$.

Set
$$
m_2=\min_{a\le x\le b} f''(x),\qquad
M_2=\max_{a\le x\le b} f''(x).\tag{f21}
$$

We can apply the Mean Value Theorem for integrals to (23):
$$
R^{(1)}(f)=\frac{\mu_2}{2}\int_a^b (x-a)(x-b)\,dx,\tag{24}
$$
(24)
where $m_2\le \mu_2\le M_2$.

It is easy to check that
$$
\int_a^b (x-a)(x-b)\,dx=-\frac{(b-a)^3}{6}.\tag{f22}
$$

Since $f''(x)$ is continuous and $m_2\le \mu_2\le M_2$, by Cauchy’s theorem there exists a point $\eta\in[a,b]$ such that $\mu_2=f''(\eta)$.  
Therefore (24) becomes
$$
R^{(1)}(f)=-\,\frac{(b-a)^3}{12}\,f''(\eta),\qquad \eta\in[a,b].\tag{25}
$$


This is the **representation of the error term of the trapezoidal rule**.

If $n$ is any odd number and the function $f(x)$ has a continuous derivative of order $n+1$ on $[a,b]$,  
then for the error of the Newton–Cotes quadrature formula with $n+1$ nodes, the following representation holds:

$$
R^{(n)}(f) = \frac{f^{(n+1)}(\eta)}{(n+1)!} \int_a^b \omega_{n+1}(x)\,dx, \qquad \eta \in [a,b],\tag{26}
$$

where $\omega_{n+1}(x)$ is the reduced polynomial of degree $n+1$ whose roots are the nodes of the Newton–Cotes quadrature formula.

---

Let us now derive the expression for the **error of Simpson’s formula** (19).  
Assume that the function $f(x)$ has a continuous fourth derivative on $[a,b]$.  
We cannot proceed exactly as in the trapezoidal case, since  
$\omega_3(x) = (x - a)(x - c)(x - b)$ changes sign on $[a,b]$,  
and thus the Mean Value Theorem cannot be directly applied.

Let $P_3(x)$ denote the **interpolating Hermite polynomial** of $f(x)$ satisfying the conditions

$$
P_3(a) = f(a), \qquad P_3(c) = f(c), \qquad P_3'(c) = f'(c), \qquad P_3(b) = f(b),\tag{f23}
$$

where $c = \frac{a+b}{2}$.

Using formula (1.6.11) for the **Hermite interpolation error**, we have

$$
f(x) - P_3(x) = \frac{f^{(IV)}(\xi)}{4!}(x - a)(x - c)^2(x - b), \qquad \xi = \xi(x) \in (a,b).\tag{27}
$$

Integrating both sides of this equality from $a$ to $b$,  
and taking into account that

$$
\int_a^b P_3(x)\,dx = \frac{b - a}{6}\,[f(a) + 4f(c) + f(b)],\tag{f24}
$$

we obtain

$$
R^{(2)}(f) = \frac{1}{24}\int_a^b f^{(IV)}(\xi)\,(x - a)(x - c)^2(x - b)\,dx,\tag{28}
$$

where $R^{(2)}(f)$ is the **error of Simpson’s rule**.

---

We can apply the Mean Value Theorem for integrals to the right-hand side of (28).  
Indeed, the integrand is the product of $f^{(IV)}(\xi) = f^{(IV)}(\xi(x))$  
and the polynomial $(x - a)(x - c)^2(x - b)$, which is nonpositive on $(a,b)$.  
The preservation of sign is ensured by the fact that the node $x = c$ is double.  
Since $f^{(IV)}(\xi)$ is continuous on $[a,b]$, as seen from (27),  
the function $f^{(IV)}(\xi(x))$ is also continuous on $[a,b]$.


The function $f^{(IV)}(\xi(x))$ is continuous on $[a,b]$, except possibly at the points $a$, $c$, and $b$,  
and satisfies the inequalities
$$
m_4 \le f^{(IV)}(\xi(x)) \le M_4,\tag{f25}
$$
where
$$
m_4 = \min_{a \le x \le b} f^{(IV)}(x), \qquad M_4 = \max_{a \le x \le b} f^{(IV)}(x).\tag{f26}
$$

Thus, from (28) we obtain
$$
R^{(2)}(f) = \frac{\mu_4}{24} \int_a^b (x - a)(x - c)^2(x - b)\,dx,\tag{29}
$$
where $m_4 \le \mu_4 \le M_4$.

---

To compute the integral on the right-hand side of (29), make the substitution $x - c = t$.  
Then we have
$$
\int_a^b (x - a)(x - c)^2(x - b)\,dx
= \int_{-\alpha}^{\alpha} t^2 (t^2 - \alpha^2)\,dt=
$$
$$
= -\frac{4\alpha^5}{15}
= -\frac{(b - a)^5}{120},\tag{f27}
$$
where $\alpha = (b - a)/2$.

Since $f^{(IV)}(x)$ is continuous on $[a,b]$ and satisfies $m_4 \le \mu_4 \le M_4$,  
by the Cauchy Mean Value Theorem there exists a point $\eta \in [a,b]$ such that $\mu_4 = f^{(IV)}(\eta)$.  
Hence, equality (29) can be written as
$$
R^{(2)}(f) = -\,\frac{(b - a)^5}{2880}\, f^{(IV)}(\eta), \qquad \eta \in [a,b].\tag{30}
$$

---

We have thus obtained the desired **representation of the error term for Simpson’s rule**.

If $n$ is any even number and the function $f(x)$ has a continuous derivative of order $n+2$ on $[a,b]$,  
then the error of the Newton–Cotes quadrature formula (6) is given by
$$
R^{(n)}(f) = \frac{f^{(n+2)}(\eta)}{(n+2)!} \int_a^b \omega_{n+1}^2(x)\,dx, \qquad \eta \in [a,b].\tag{31}
$$

Now let us consider the behavior of the coefficients $B_k^{(n)}$ as $n$ increases without bound.  
The first table of coefficients $B_k^{(n)}$ for $n = 1, 2, \ldots, 10$ was computed by **Cotes**.  
It turned out that for $n = 8$ and $n = 10$, negative coefficients appear.  
Later computations confirmed that for $n > 10$, the coefficients $B_k^{(n)}$ can be both positive and negative.

R.O. Kuzmin established that the asymptotic formula
$$
\gamma_n = \sum_{k=0}^n |B_k^{(n)}|\tag{f28}
$$
grows indefinitely as $n \to \infty$.  
Since $\sum_{k=0}^n B_k^{(n)} = 1$ (which follows from the fact that the Newton–Cotes formula (6) is exact for $f(x)=1$),  
we can conclude that

for large $n$, among the coefficients there are both positive and negative values.  
Moreover, this statement holds for all $n \ge 10$ (see [26]).

Let us now consider the **quadrature formula (7)** in its general form, assuming that it is exact for $f(x) = 1$:

$$
\sum_{k=1}^N C_k = \mu_0 = \int_a^b w(x)\,dx,\tag{32}
$$

where $\mu_0 > 0$.

Let us explain the meaning of the **sum of the absolute values of the coefficients**,  
defined as $\gamma = \sum_{k=1}^N |C_k|$, in the context of integration using formula (7).

When evaluating an integral using (7), the quadrature sum is computed approximately.  
The values $f(x_k)$ are replaced by their approximate counterparts $\tilde{f}(x_k)$,  
so that the computational error is $\varepsilon_k = f(x_k) - \tilde{f}(x_k)$.  
Suppose that $|\varepsilon_k| \le \varepsilon$ for all $k = 1, 2, \ldots, N$,  
and that the sum $\sum_{k=1}^N C_k \tilde{f}(x_k)$ is computed exactly  
(apart from possible rounding errors in $C_k$).  

Then, the **quadrature sum error** will not exceed
$$
\varepsilon \sum_{k=1}^N |C_k| = \varepsilon \gamma.\tag{f29}
$$

This estimate is not necessarily exact, since the actual error might be smaller,  
but it gives a **strict upper bound** on the computational error,  
proportional to $\gamma$.  
Thus, the larger the value of $\gamma$,  
the greater the potential computational error of the quadrature sum.

---

The value of $\gamma$ reaches its **minimum** when all coefficients $C_k > 0$,  
for $k = 1, 2, \ldots, N$.  
Indeed, from (32) we have

$$
\sum_{k=1}^N C_k = \mu_0 = \sum_{k=1}^N |C_k|,\tag{f30}
$$

which means that $\mu_0 = \gamma$.


This equality is achieved **if and only if** all coefficients $C_k$ have the same sign  
and are therefore **positive** ($\mu_0 > 0$).  
This explains the importance, in practical applications, of using **quadrature formulas with positive coefficients**.

For the **Newton–Cotes formulas**, when $n \ge 10$, the property of coefficient positivity no longer holds,  
and the sum of their absolute values, $\gamma_n$, grows rapidly.  
Therefore, applying Newton–Cotes formulas for large $n$ is impractical.

It is more advantageous to improve accuracy by **dividing the integration interval into smaller subintervals**  
and applying the Newton–Cotes quadrature formula with small $n$ to each part separately.  
This approach leads to the so-called **composite quadrature formulas**,  
which will be discussed in the following section.
