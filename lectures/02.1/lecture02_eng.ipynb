# Numerical integration

We construct interpolatory quadrature formulas, study Newton–Cotes rules and their errors, and discuss sensitivity to errors in function values.

All functions and weights below are real-valued.

## Interpolatory quadrature formulas

If a function $f(x)$ is continuous on the interval $[a,b]$, with $a<b$, then it has an antiderivative $F$ on $[a,b]$, and the following holds:
$$
\int_a^b f(x)\,dx = F(b) - F(a)\tag{1}
$$
known as the fundamental formula of integral calculus.

Formulas for approximate calculation of integrals have the form
$$
\int_a^b f(x)\,dx \approx \sum_{j=1}^N C_j f(x_j).\tag{2}
$$

For some formulas, the quadrature sum coincides with the Riemann sum. For example, the quadrature formula for rectangles:
$$
\int_a^b f(x)\,dx \approx \frac{b - a}{N} \sum_{j=1}^N f(\alpha + (j - 1)h),\tag{3}
$$
where $h = (b - a)/N, \alpha \in [a, a + h]$.

We will write the quadrature formula in a more general form compared to (2):
$$
\int_a^b w(x) f(x)\,dx \approx \sum_{j=1}^N C_j f(x_j).\tag{4}
$$

### Algebraic interpolation

Let us assume that we are concerned with evaluating the integral

$$
\int_a^b w(x) f(x)\,dx.\tag{5}
$$

The integration interval may be finite or infinite.  
Assume that $w$ is a real measurable weight and that all moments used below exist absolutely, so $\int_a^b |w(x)||x|^k\,dx<\infty$. Also assume $\int_a^b |w(x)f(x)|\,dx<\infty$. The moments are

$$
\mu_k = \int_a^b w(x)x^k\,dx, \quad k = 0, 1, 2, \ldots,\tag{6}
$$

Choose $N\ge1$ distinct real nodes in the integration interval. Consider the **interpolation polynomial** $P(x)$ of the function $f(x)$, constructed through the nodes $x_1, x_2, \ldots, x_N$.  
The degree of $P(x)$ does not exceed $N - 1$, and it satisfies the interpolation conditions:
$$
P(x_j) = f(x_j), \quad j = 1, 2, \ldots, N.\tag{7}
$$

We can express the polynomial $P(x)$ in Lagrange form:
$$
P(x) = \sum_{j=1}^N l_j(x) f(x_j),\tag{8}
$$

where $l_j(x)=\prod_{k\ne j}(x-x_k)/(x_j-x_k)$ and $l_j(x_i)=\delta_{ij}$ are the fundamental interpolation polynomials.

In the integral (5), we replace the function $f(x)$ with the polynomial $ P(x)$.  
As a result, we obtain the following approximate equality:

$$
\int_a^b w(x) f(x)\,dx \approx \int_a^b w(x) P(x)\,dx.\tag{9}
$$

Substitute the right-hand side of equality (8) for $P(x)$ and perform the integration.  
We obtain the approximate equality of the form (4), in which the coefficients $C_j$ are defined by the relations

$$
C_j = \int_a^b w(x) l_j(x)\,dx, \quad j = 1, 2, \ldots, N.\tag{10}
$$

**Definition 1.** A quadrature formula of the form (4), where the nodes $x_j$ are distinct numbers, and the coefficients $C_j$ are given by expressions (10), is called an **interpolatory quadrature formula**.

**Theorem 1.** For a quadrature formula (4) with $N$ distinct nodes to be interpolatory, it is **necessary and sufficient** that it be **exact** for all polynomials $f(x)$ of degree not higher than $N - 1$,  
that is, for $f(x) = 1, x, x^2, \ldots, x^{N-1}$.

**Proof.**  
**1. Necessity.** Let the quadrature formula (4) be interpolatory, and let $f(x)$ be a polynomial of degree not higher than $N - 1$.  
The interpolation polynomial constructed for $f(x)$ at the $N$ nodes of the quadrature formula (4) coincides with $f(x)$:

$$
f(x) = \sum_{j=1}^N l_j(x) f(x_j).\tag{11}
$$

Multiply both sides of this equality by $w(x)$ and integrate over $[a, b]$. We get

$$
\int_a^b w(x) f(x)\,dx = \sum_{j=1}^N f(x_j) \int_a^b w(x) l_j(x)\,dx = \sum_{j=1}^N C_j f(x_j),\tag{12}
$$

which means that the quadrature formula is exact for $f(x)$.  
Here, we used relation (10), which is valid since the formula (4) is interpolatory.

**2. Sufficiency.** Suppose the quadrature formula (4) is exact for any polynomial of degree not higher than $N - 1$.  
In particular, it must be exact for the fundamental interpolation polynomials $l_k(x)$ constructed at the nodes of the quadrature formula (4):

$$
\int_a^b w(x) l_k(x)\,dx = \sum_{j=1}^N C_j l_k(x_j) = C_k, \quad k = 1, 2, \ldots, N.\tag{13}
$$

These equalities mean that the quadrature formula (4) is interpolatory.  
The theorem is proven.

**Definition 2.** A non-negative integer $d$ is called the **algebraic degree of accuracy** of a quadrature formula (4) if this formula is exact for all polynomials of degree no higher than $d$, but not exact for $x^{d+1}$.

From Theorem 1 it follows that the algebraic degree of accuracy of an interpolatory quadrature formula with $N$ nodes satisfies the inequality $d \geq N - 1$.

**Theorem 2 (upper bound on the degree).** Under the additional assumption that the weight function satisfies

$$
w(x) \ge 0, \quad x \in [a, b], \quad \mu_0 = \int_a^b w(x)\,dx > 0,\tag{14}
$$

the algebraic degree of accuracy $d$ of the quadrature formula (4) with $N$ nodes satisfies the inequality

$$
d \le 2N - 1.\tag{15}
$$

**Proof.** For the real nodes $x_1,\ldots,x_N$, construct the polynomial
$$
\omega(x)=\prod_{j=1}^N(x-x_j).\tag{16}
$$
The polynomial $\omega(x)^2$ has degree $2N$ and is positive except at finitely many real points. Since $w\ge0$ and $\mu_0>0$, it follows that
$$
\int_a^b w(x)\omega(x)^2\,dx>0,\quad
\sum_{j=1}^N C_j\omega(x_j)^2=0.\tag{17}
$$
From this it follows that (15) holds.  
The theorem is proven.

**Corollary.** If a formula satisfying (14) is exact for every polynomial of degree at most $m$, then
$$
N\ge\left\lceil\frac{m+1}{2}\right\rceil.\tag{18}
$$
Indeed, $m\le d\le2N-1$, and $N$ is an integer.

## Newton–Cotes quadrature formulas

We consider the integral  
$$
\int_a^b f(x)\,dx\tag{19}
$$  
with a unit weight function.  
From the requirement of the existence of moments (6) with $w(x) = 1$, it follows that the interval $[a, b]$ must be finite.  

Let $n$ be a positive integer.  
We choose as nodes the equally spaced points with step  
$$
h_n = \frac{b - a}{n}\tag{20}
$$  
given by  
$$
x_k^{(n)} = a + k h_n, \quad k = 0, 1, \ldots, n.\tag{21}
$$  

The endpoints $a$ and $b$ are included among the nodes, so the total number of nodes is $N = n + 1$.

The interpolatory quadrature formula with nodes (21)

$$
\int_a^b f(x)\,dx \approx \sum_{k=0}^n C_k^{(n)} f(x_k^{(n)})\tag{22}
$$  

is called the **Newton–Cotes quadrature formula**.  
Since $n$ can take the values $1, 2, 3, \ldots$, we obtain a sequence of Newton–Cotes quadrature formulas.

The quadrature formula (22) is interpolatory, and its coefficients are defined by formula (10):

$$
C_k^{(n)} = \int_a^b \frac{\omega_{n+1}(x)\,dx}{(x - x_k^{(n)}) \omega_{n+1}'(x_k^{(n)})},\tag{23}
$$  

where  

$$
\omega_{n+1}(x) = (x - x_0^{(n)})(x - x_1^{(n)}) \cdots (x - x_n^{(n)}).\tag{24}
$$

The quotient in (23) is a polynomial after cancelling the factor $x-x_k^{(n)}$; its apparent singularity is removable. In that integral, let us make the change of variable $x = a + h_n t$. We then have

$$
x - x_k^{(n)} = h_n (t - k),\tag{25}
$$

and

$$
\omega_{n+1}(x) = h_n^{n+1} t(t - 1)(t - 2) \cdots (t - n),\tag{26}
$$

while

$$
\omega_{n+1}'(x_k^{(n)}) = h_n^n (-1)^{n-k} k!(n - k)!.\tag{27}
$$

Therefore,

$$
C_k^{(n)} = \frac{h_n (-1)^{n-k}}{k!(n - k)!} \int_0^n \frac{t(t - 1)(t - 2) \cdots (t - n)}{t - k}\,dt.\tag{28}
$$

The obtained value of the coefficient $C_k^{(n)}$ depends on the length of the interval through $h_n = (b-a)/n$.  
Let us write

$$
C_k^{(n)} = (b - a) B_k^{(n)},\tag{29}
$$

where

$$
B_k^{(n)} = \frac{(-1)^{n - k}}{k!(n - k)!n} \int_0^n \frac{t(t - 1) \cdots (t - n)}{t - k}\,dt.\tag{30}
$$

The values $B_k^{(n)}$ do not depend on the interval $[a, b]$, and thus can be tabulated.  
The Newton–Cotes quadrature formula takes the form

$$
\int_a^b f(x)\,dx \approx (b - a) \sum_{k=0}^n B_k^{(n)} f(x_k^{(n)}).\tag{31}
$$

The coefficients $B_k^{(n)}$ for $n = 1, 2, \ldots, 10$ were first computed by Cotes.

For the coefficients listed in the table, the following symmetry relation holds:

$$
B_k^{(n)} = B_{n - k}^{(n)}, \quad (k = 0, 1, \ldots, n).\tag{32}
$$

This equality holds for any $n$.  
We now prove a lemma, from which this result follows as a special case.

**Lemma 1 (symmetry of weights).**
Let the quadrature formula

$$
\int_a^b w(x) f(x)\,dx \approx \sum_{k=1}^N C_k f(x_k)\tag{33}
$$
satisfy the following conditions:

1) the integration interval $[a, b]$ is finite;

2) the weight function $w(x)$ takes equal values at points symmetric with respect to  
the midpoint $c = (a + b)/2$ of the interval $[a, b]$, i.e.

$$
w(a + b - x) = w(x), \quad x \in [a, b];\tag{34}
$$

3) the nodes are symmetric with respect to the midpoint, i.e., if $x_k$ is a node, then $a + b - x_k$ is also a node for $k = 1, 2, \ldots, N$.

If the quadrature formula (33) is interpolatory, then the coefficients corresponding to the nodes symmetric with respect to $c$ are equal.

**Proof.**
Let us set $x_{-j} = a + b - x_j$, $j = 1, 2, \ldots, N$.  
According to condition 3 of the lemma, the points $x_{-j}$ are also nodes, and the nodes $x_{-j}$ and $x_j$ are symmetric with respect to $c$.  
Let us denote the corresponding coefficients by $C_{-j}$ and $C_j$.  
We need to show that $C_{-j} = C_j$.

Since the formula (33) is interpolatory, we have

$$
C_{-j} = \int_a^b \frac{w(x)\,\omega_N(x)}{(x - x_{-j})\,\omega_N'(x_{-j})}\,dx.\tag{35}
$$

Let us make the change of variable in this integral: $x = a + b - t$.  
According to condition 3 of the lemma, the set $\{x_{-j}\,|\, j = 1, 2, \ldots, N\}$ coincides with the set of all nodes of the quadrature formula.  
Therefore,

$$
\begin{aligned}
\omega_N(a+b-t)&=\prod_{k=1}^N(a+b-t-x_k)\\
&=(-1)^N\prod_{k=1}^N\bigl(t-(a+b-x_k)\bigr)=(-1)^N\omega_N(t).
\end{aligned}\tag{36}
$$

We have thus obtained the relation

$$
\omega_N(x) = (-1)^N \omega_N(t),\tag{37}
$$

where $x$ and $t$ are related by $t = a + b - x$.  
Differentiating both sides of (37) with respect to $x$, we get

$$
\omega_N'(x) = (-1)^{N+1} \omega_N'(t).\tag{38}
$$

This equality connects the values of the derivative $\omega_N'(x)$ at points symmetric with respect to $c$.  
In particular,

$$
\omega_N'(x_{-j}) = (-1)^{N+1} \omega_N'(x_j).\tag{39}
$$

Using equalities (37), (39), and condition 2) of the lemma, we rewrite the integral (35) in the new variable of integration:

$$
\begin{aligned}
C_{-j}&=\int_b^a\frac{w(a+b-t)(-1)^N\omega_N(t)}{(a+b-t-x_{-j})(-1)^{N+1}\omega_N'(x_j)}(-dt)\\
&=\int_a^b\frac{w(t)\omega_N(t)}{(t-x_j)\omega_N'(x_j)}\,dt=C_j.
\end{aligned}\tag{40}
$$

Thus, the lemma is proven.

The Newton–Cotes quadrature formula (31) satisfies the conditions of the lemma; therefore, we obtain  
$B_k^{(n)} = B_{n - k}^{(n)}$, for $k = 0, 1, \ldots, n$.

We now need a theorem concerning the mean value. Let us state it as follows:

**Theorem 3 (weighted mean value theorem).**
Let the functions $f(x)$ and $g(x)$ be Riemann integrable on the finite interval $[a, b]$, and assume that:

1) $f(x)$ is bounded, i.e. $m \le f(x) \le M$ for all $x \in [a, b]$;  
2) $g(x)$ does not change sign on $[a, b]$.

Then the function $f(x) g(x)$ is integrable, and

$$
\int_a^b f(x) g(x)\,dx = \mu \int_a^b g(x)\,dx,\tag{41}
$$

where $m \le \mu \le M$.  
If $\int_a^b g=0$, both sides vanish. If $f(x)$ is continuous on $[a, b]$, then there exists a point $\eta \in [a, b]$ such that $\mu = f(\eta)$.

**Theorem 4 (degree of Newton–Cotes rules).**
The **algebraic degree of accuracy** $d_n$ of the Newton–Cotes quadrature formula (31) is given by

$$
d_n =
\begin{cases}
n, & \text{if } n \text{ is odd}, \\
n + 1, & \text{if } n \text{ is even}.
\end{cases}\tag{42}
$$

**Proof.**
We will prove that the quadrature formula (31) is exact for polynomials of degree not higher than $d_n$.  
Since the quadrature formula (31) is interpolatory, by Theorem 1 it is exact for all polynomials of degree not exceeding $n$.  
This gives only the lower bound $d_n\ge n$; non-exactness at the next relevant degree remains to be proved.

Let $n$ be even.  
We will show that for the polynomial $(x-c)^{n+1}$, the quadrature formula (31) is exact.  
Indeed, the integral over $[a, b]$ of $(x - c)^{n+1}$ equals zero.  
The quadrature sum for it is also zero because for symmetric nodes with respect to the midpoint $c$, the function takes equal absolute values but opposite signs, and since the coefficients corresponding to symmetric nodes are equal, the positive and negative terms cancel each other.

Since the coefficients corresponding to symmetric nodes are equal,  
and since the formula (31) is exact for polynomials of degree not higher than $n$,  
it follows that it is also exact for $(x - c)^{n+1}$.  
Thus, the formula is exact for all polynomials of degree not higher than $n+1$, so $d_n\ge n+1$.

Let $n$ be even.  
The polynomial $\omega_{n+1}(x)$ from (24) has the following property:  
at points symmetric with respect to $c = (a + b)/2$,  
it takes values that are equal in magnitude but opposite in sign ($N = n + 1$).  
Therefore,

$$
\int_a^b \omega_{n+1}(x)\,dx = 0.\tag{43}
$$

Define the function

$$
\Omega(x) = \int_a^x \omega_{n+1}(t)\,dt,\tag{44}
$$

which is a polynomial of degree $n + 2$ satisfying the conditions

$$
\Omega(a) = \Omega(b) = 0.\tag{45}
$$

Consider the integrals

$$
I_j = \int_{a + jh_n}^{a + (j+1)h_n} \omega_{n+1}(x)\,dx, \quad j = 0, 1, \ldots, n - 1.\tag{46}
$$

We will show that for $n \ge 4$, the following inequalities hold:

$$
|I_0| > |I_1| > \cdots > |I_{n/2 - 1}|.\tag{47}
$$

In the integral $I_j$ (where $1 \le j \le n/2 - 1$),  
we make the substitution $x = y + h_n$. Then we obtain:

$$
\begin{aligned}
I_j&=\int_{a+(j-1)h_n}^{a+jh_n}\prod_{s=0}^n(y+h_n-x_s^{(n)})\,dy\\
&=\int_{a+(j-1)h_n}^{a+jh_n}\frac{y+h_n-x_0^{(n)}}{y-x_n^{(n)}}\omega_{n+1}(y)\,dy.
\end{aligned}\tag{48}
$$

Since $\omega_{n+1}(y)$ preserves its sign over the interval of integration,  
we can apply the Mean Value Theorem (Theorem 3) to the last integral:

$$
I_j = \frac{\eta_j + h_n - x_0^{(n)}}{\eta_j - x_n^{(n)}} I_{j-1},\tag{49}
$$

where

$$
a + (j - 1)h_n \le \eta_j \le a + jh_n.\tag{50}
$$

The factor in equality (49), for all admissible values of $j$ ($1 \le j \le n/2 - 1$),  
is negative, and its absolute value satisfies the inequality

$$
\frac{h_n + \eta_j - a}{h_n n - (\eta_j - a)} \le \frac{h_n + (n/2 - 1)h_n}{h_n n - (n/2 - 1)h_n} = \frac{n}{n + 2} < 1.\tag{51}
$$

Here, the second of inequalities (50) for $j = n/2 - 1$ was used.  
From (49) we obtain $|I_j| < |I_{j-1}|$. Thus, inequality (47) is proven.

Since  
$\operatorname{sgn} I_j = (-1)^j$ and $I_j = -I_{n - 1 - j}$ for $j = 0, 1, \ldots, n - 1$,  
the alternating partial integrals on the left half remain positive by (47), and symmetry gives $\Omega(a+b-x)=\Omega(x)$. For $n=2$ this follows directly from $I_0>0$. Thus

$$
\Omega(x) = \int_a^x \omega_{n+1}(t)\,dt > 0, \quad a < x < b.\tag{52}
$$

We now prove that the quadrature formula (31) is **not exact** for the polynomial  
$x \omega_{n+1}(x)$ of degree $n + 2$.  
Since this polynomial vanishes at the nodes of formula (31),  
it suffices to show that its integral over $[a, b]$ is nonzero.  
Integrating by parts and taking into account (45) and (52), we have:

$$
\int_a^b x \omega_{n+1}(x)\,dx = x \Omega(x)\big|_a^b - \int_a^b \Omega(x)\,dx = -\int_a^b \Omega(x)\,dx < 0.\tag{53}
$$

Thus, we have proved that $d_n = n + 1$ for even $n$.

For odd $n$, $\omega_{n+1}$ is symmetric about $c$. Its subinterval integrals satisfy $I_j=I_{n-1-j}$, start negative, and alternate in sign. The same shift comparison used above gives strictly decreasing magnitudes up to the middle, including the central interval: for $n=2m+1$, the multiplier has magnitude less than one in the interior for $1\le j\le m$. Pairing the symmetric intervals therefore gives $\int_a^b\omega_{n+1}(x)\,dx<0$. For $n=1$ the integrand is negative on $(a,b)$ directly. Since the nodal sum for $\omega_{n+1}$ is zero, the rule is not exact at degree $n+1$. Hence $d_n=n$ for odd $n$.

Let us now consider specific **Newton–Cotes quadrature formulas** for $n = 1, 2, 3$.  

For $n = 1$, the quadrature formula (31) has the form

$$
\int_a^b f(x)\,dx \approx \frac{b - a}{2} [f(a) + f(b)].\tag{54}
$$

It is called the **trapezoidal rule** (see Figure 1 below).  
The algebraic degree of accuracy of this formula is 1.

For $n = 2$, the quadrature formula (31) has the form

$$
\int_a^b f(x)\,dx \approx \frac{b - a}{6} \left[ f(a) + 4f\!\left( \frac{a + b}{2} \right) + f(b) \right].\tag{55}
$$

It is called **Simpson’s formula**.  
The algebraic degree of accuracy is 3.

*Figure 1 below illustrates the trapezoidal rule: the straight segment joins the endpoint values of the function.*

For $n = 3$, the quadrature formula (31) takes the form

$$
\int_a^b f(x)\,dx \approx \frac{b - a}{8} \left[ f(a) + 3f(a + h_3) + 3f(a + 2h_3) + f(b) \right].\tag{56}
$$

It is called the **three-eighths rule**, the name originating from the coefficient $3/8$.  
The algebraic degree of accuracy is also 3.

The **error** or **remainder term** of a quadrature formula is the difference between the true integral and the quadrature sum.  
We denote the error of the Newton–Cotes quadrature formula (31) by

$$
R^{(n)}(f) = \int_a^b f(x)\,dx - (b - a)\sum_{k=0}^n B_k^{(n)} f(x_k^{(n)}).\tag{57}
$$

Representations for $R^{(n)}(f)$ exist under the assumption that the function $f(x)$  
is sufficiently smooth (i.e., has a sufficient number of derivatives).  
These representations differ for even and odd $n$, and are derived using different techniques.  
We will not consider the general case for arbitrary $n$,  
but will limit ourselves to the two simplest and most typical cases: $n = 1$ and $n = 2$.

For $n = 1$, we are dealing with the error representation for the **trapezoidal rule** (formula (54)).  
Let us assume that the function $f(x)$ has a continuous second derivative on $[a, b]$.  
Let $P_1(x)$ be the **interpolation polynomial** of the function $f(x)$ constructed through the endpoints $a$ and $b$.

```python
import numpy as np
import matplotlib.pyplot as plt

a, b = 0.0, 1.0
def f(x):
    t = (x - a) / (b - a)
    return 0.25 + 0.9*(1.5*t - 0.5*t**2)

xs = np.linspace(a, b, 400)
ys = f(xs)

plt.figure(figsize=(4, 3))

ymax = float(ys.max()) * 1.15
plt.annotate('', xy=(b + 0.08, 0), xytext=(a - 0.12, 0),
             arrowprops=dict(arrowstyle='->', linewidth=1.2))
plt.annotate('', xy=(a, ymax), xytext=(a, -0.02*ymax),
             arrowprops=dict(arrowstyle='->', linewidth=1.2))

plt.plot([a, b], [0, 0], linewidth=1.2)
plt.plot(xs, ys, linewidth=2.0)
plt.plot([a, b], [f(a), f(b)], linewidth=2.0)
tick_h = ymax * 0.04
plt.plot([a, a], [0, tick_h], linewidth=1.2)
plt.plot([b, b], [0, tick_h], linewidth=1.2)
plt.text((a + b) * 0.58, f(0.58*(b-a)+a) + 0.03*ymax, r'$y=f(x)$')
plt.text(a, -0.08*ymax, r'$a$', ha='center', va='top')
plt.text(b, -0.08*ymax, r'$b$', ha='center', va='top')
plt.xlim(a - 0.12, b + 0.1)
plt.ylim(-0.1*ymax, ymax)
plt.axis('off')
plt.tight_layout()
plt.show()
```

By the standard linear interpolation remainder, for each $x\in(a,b)$ we have
$$
f(x)-P_1(x)=\frac{f''(\xi)}{2}\,(x-a)(x-b),\qquad \xi=\xi(x)\in(a,b).\tag{58}
$$

To integrate rigorously, define $q_2(x)=2[f(x)-P_1(x)]/[(x-a)(x-b)]$ for $a<x<b$. The remainder above implies $m_2\le q_2(x)\le M_2$, with $m_2,M_2$ defined in (62). This quotient is continuous inside the interval and bounded; assigning any finite endpoint values makes it Riemann integrable. No continuity of a choice of $\xi(x)$ is assumed. Integrating gives:
$$
\int_a^b f(x)\,dx-\int_a^b P_1(x)\,dx
= \frac12\int_a^b q_2(x)\,(x-a)(x-b)\,dx.\tag{59}
$$

Since the trapezoidal rule is exact for $P_1(x)$,
$$
\int_a^b P_1(x)\,dx=\frac{b-a}{2}\,[P_1(a)+P_1(b)]
=\frac{b-a}{2}\,[f(a)+f(b)].\tag{60}
$$

Thus, the left–hand side of (59) is precisely the error $R^{(1)}(f)$ of the trapezoidal rule, and (59) can be written as
$$
R^{(1)}(f)=\frac12\int_a^b q_2(x)\,(x-a)(x-b)\,dx.\tag{61}
$$

The polynomial $(x-a)(x-b)$ is negative on $(a,b)$. We use bounds on the quotient rather than assume that the intermediate point $\xi$ depends continuously on $x$.

Set
$$
m_2=\min_{a\le x\le b} f''(x),\qquad
M_2=\max_{a\le x\le b} f''(x).\tag{62}
$$

Integrating the bounds on $q_2$ against the fixed-sign factor gives a number $\mu_2$ such that
$$
R^{(1)}(f)=\frac{\mu_2}{2}\int_a^b (x-a)(x-b)\,dx,\tag{63}
$$
where $m_2\le \mu_2\le M_2$.

It is easy to check that
$$
\int_a^b (x-a)(x-b)\,dx=-\frac{(b-a)^3}{6}.\tag{64}
$$

Since $f''(x)$ is continuous and $m_2\le \mu_2\le M_2$, by the intermediate value theorem there exists a point $\eta\in[a,b]$ such that $\mu_2=f''(\eta)$.  
Therefore (63) becomes
$$
R^{(1)}(f)=-\,\frac{(b-a)^3}{12}\,f''(\eta),\qquad \eta\in[a,b].\tag{65}
$$

This is the **representation of the error term of the trapezoidal rule**.

For odd $n$ (and, in fact, for any $n\ge1$), interpolation gives the safe general bound
$$
|R^{(n)}(f)|\le\frac{M_{n+1}}{(n+1)!}\int_a^b|\omega_{n+1}(x)|\,dx,\quad
M_{n+1}=\max_{[a,b]}|f^{(n+1)}|.\tag{66}
$$
Here $f\in C^{n+1}([a,b])$ and $\omega_{n+1}$ is the monic polynomial with the quadrature nodes as its roots. Since this polynomial changes sign for $n>1$, the simple fixed-sign mean value argument used for trapezoids does not directly give a general single-point remainder.

Let us now derive the expression for the **error of Simpson’s formula** (55).  
Assume that the function $f(x)$ has a continuous fourth derivative on $[a,b]$.  
We cannot proceed exactly as in the trapezoidal case, since  
$\omega_3(x) = (x - a)(x - c)(x - b)$ changes sign on $[a,b]$,  
and thus the Mean Value Theorem cannot be directly applied.

Let $P_3(x)$ denote the **interpolating Hermite polynomial** of $f(x)$ satisfying the conditions

$$
P_3(a) = f(a), \qquad P_3(c) = f(c), \qquad P_3'(c) = f'(c), \qquad P_3(b) = f(b),\tag{67}
$$

where $c = \frac{a+b}{2}$.

The standard **Hermite interpolation remainder** gives, at points other than the nodes,

$$
f(x) - P_3(x) = \frac{f^{(IV)}(\xi)}{4!}(x - a)(x - c)^2(x - b), \qquad \xi = \xi(x) \in (a,b).\tag{68}
$$

Define $q_4(x)=24[f(x)-P_3(x)]/[(x-a)(x-c)^2(x-b)]$ away from the nodes. As above, the remainder bounds $q_4$ between the minimum and maximum of $f^{(4)}$; the quotient is continuous off the nodes and bounded, so it is integrable after assigning finite values at the nodes. Integrating from $a$ to $b$,  
and taking into account that

$$
\int_a^b P_3(x)\,dx = \frac{b - a}{6}\,[f(a) + 4f(c) + f(b)],\tag{69}
$$

we obtain

$$
R^{(2)}(f) = \frac{1}{24}\int_a^b q_4(x)\,(x - a)(x - c)^2(x - b)\,dx,\tag{70}
$$

where $R^{(2)}(f)$ is the **error of Simpson’s rule**.

The factor $(x-a)(x-c)^2(x-b)$ is nonpositive and its integral is negative. The bounded quotient $q_4$ satisfies
$$
m_4 \le q_4(x) \le M_4,\tag{71}
$$
where
$$
m_4 = \min_{a \le x \le b} f^{(IV)}(x), \qquad M_4 = \max_{a \le x \le b} f^{(IV)}(x).\tag{72}
$$

Thus, from (70) we obtain
$$
R^{(2)}(f) = \frac{\mu_4}{24} \int_a^b (x - a)(x - c)^2(x - b)\,dx,\tag{73}
$$
where $m_4 \le \mu_4 \le M_4$.

To compute the integral on the right-hand side of (73), make the substitution $x - c = t$.  
Then we have
$$
\begin{aligned}
\int_a^b(x-a)(x-c)^2(x-b)\,dx
&=\int_{-\alpha}^{\alpha}t^2(t^2-\alpha^2)\,dt\\
&=-\frac{4\alpha^5}{15}=-\frac{(b-a)^5}{120}.
\end{aligned}\tag{74}
$$
where $\alpha = (b - a)/2$.

Since $f^{(IV)}(x)$ is continuous on $[a,b]$ and satisfies $m_4 \le \mu_4 \le M_4$,  
by the intermediate value theorem there exists a point $\eta \in [a,b]$ such that $\mu_4 = f^{(IV)}(\eta)$.  
Hence, equality (73) can be written as
$$
R^{(2)}(f) = -\,\frac{(b - a)^5}{2880}\, f^{(IV)}(\eta), \qquad \eta \in [a,b].\tag{75}
$$

We have thus obtained the desired **representation of the error term for Simpson’s rule**.

For even $n\ge2$, use a degree-at-most-$n+1$ Hermite interpolant that matches all $n+1$ nodal values and also $f'(c)$ at the midpoint node $c=(a+b)/2$. Since the Newton–Cotes rule is exact through degree $n+1$, its error is bounded by
$$
|R^{(n)}(f)|\le\frac{M_{n+2}}{(n+2)!}\int_a^b|(x-c)\omega_{n+1}(x)|\,dx,\quad
M_{n+2}=\max_{[a,b]}|f^{(n+2)}|.\tag{76}
$$
This requires $f\in C^{n+2}([a,b])$. The Hermite remainder contains $(x-c)\omega_{n+1}(x)$, a polynomial of degree $n+2$. For $n=2$ the bound agrees with the absolute-value version of (75).

Now let us consider the behavior of the coefficients $B_k^{(n)}$ as $n$ increases without bound.  
The first table of coefficients $B_k^{(n)}$ for $n = 1, 2, \ldots, 10$ was computed by **Cotes**.  
It turned out that for $n = 8$ and $n = 10$, negative coefficients appear.  
Later computations confirmed that for $n > 10$, the coefficients $B_k^{(n)}$ can be both positive and negative.

To quantify sensitivity, consider
$$
\gamma_n=\sum_{k=0}^n|B_k^{(n)}|.\tag{77}
$$
The classical Newton–Cotes weights have an unbounded sequence of absolute sums (see [1, 2]). Exactness for constants gives $\sum_{k=0}^nB_k^{(n)}=1$, so $\gamma_n>1$ signals negative weights. Negative weights first occur at $n=8$; all weights are positive again for $n=9$. The large-$n$ behavior should not be inferred from monotonicity at small $n$.

Let us now consider the **quadrature formula (33)** in its general form, assuming that it is exact for $f(x) = 1$:

$$
\sum_{k=1}^N C_k = \mu_0 = \int_a^b w(x)\,dx,\tag{78}
$$

where we assume real weights, $w\ge0$, and $\mu_0>0$.

Let us explain the meaning of the **sum of the absolute values of the coefficients**,  
defined as $\gamma = \sum_{k=1}^N |C_k|$, in the context of integration using formula (33).

When evaluating an integral using (33), the quadrature sum is computed approximately.  
The values $f(x_k)$ are replaced by their approximate counterparts $\tilde{f}(x_k)$,  
so that the computational error is $\varepsilon_k = f(x_k) - \tilde{f}(x_k)$.  
Suppose that $|\varepsilon_k| \le \varepsilon$ for all $k = 1, 2, \ldots, N$,  
and that the sum $\sum_{k=1}^N C_k \tilde{f}(x_k)$ is computed exactly  
(with exact weights and exact summation for this data-error estimate).  

Then the propagated data error satisfies
$$
\left|\sum_{k=1}^N C_k\varepsilon_k\right|\le\varepsilon\sum_{k=1}^N|C_k|=\varepsilon\gamma.\tag{79}
$$

This estimate is not necessarily exact, since the actual error might be smaller,  
but it gives an **upper bound** on the computational error,  
proportional to $\gamma$.  
Thus, the larger the value of $\gamma$,  
the greater the potential computational error of the quadrature sum.

The value of $\gamma$ reaches its **minimum** when all coefficients $C_k\ge0$,  
for $k = 1, 2, \ldots, N$.  
Indeed, from (78) we have

$$
\gamma=\sum_{k=1}^N|C_k|\ge\left|\sum_{k=1}^N C_k\right|=\mu_0,\tag{80}
$$

Equality holds when all weights are nonnegative, giving $\gamma=\mu_0$.

This equality is achieved **if and only if** all real coefficients $C_k$ are **nonnegative** ($\mu_0>0$ ensures that at least one weight is positive).  
This explains the importance, in practical applications, of using **quadrature formulas with positive coefficients**.

For high-degree **Newton–Cotes formulas**, negative weights occur,  
and the sum of their absolute values, $\gamma_n$, grows rapidly.  
Therefore, applying Newton–Cotes formulas for large $n$ is impractical.

It is more advantageous to improve accuracy by **dividing the integration interval into smaller subintervals**  
and applying the Newton–Cotes quadrature formula with small $n$ to each part separately.  
This approach leads to the so-called **composite quadrature formulas**,  
which will be discussed in a subsequent lecture.

This bound concerns perturbations of the nodal values only; truncation error and rounding of the weights or the sum must be assessed separately.

**Numerical check.** The following exact-arithmetic calculation verifies the weights, symmetry, algebraic degrees, and values in Table 2. It also checks the signs and constants of the trapezoidal and Simpson remainders on simple polynomials. This is a verification of the displayed cases, rather than a proof for every number of nodes.

```python
from fractions import Fraction
from math import factorial


def newton_cotes_weights(n):
    """Exact normalized weights on [0, 1] with n+1 equally spaced nodes."""
    if not isinstance(n, int) or n < 1:
        raise ValueError("n must be a positive integer")
    polynomial = [1]  # Coefficients in increasing powers of t.
    for j in range(n + 1):
        product = [0] * (len(polynomial) + 1)
        for k, coefficient in enumerate(polynomial):
            product[k] -= j * coefficient
            product[k + 1] += coefficient
        polynomial = product

    weights = []
    for j in range(n + 1):
        # Divide the nodal polynomial by t-j before integrating.
        quotient = [0] * (n + 1)
        quotient[n] = polynomial[n + 1]
        for k in range(n - 1, -1, -1):
            quotient[k] = polynomial[k + 1] + j * quotient[k + 1]
        assert polynomial[0] + j * quotient[0] == 0
        integral = sum((Fraction(c * n**(k + 1), k + 1)
                        for k, c in enumerate(quotient)), Fraction(0))
        weights.append(integral / (n * (-1)**(n-j) * factorial(j) * factorial(n-j)))
    return weights


checked_weights = {}
print(f"{'n':>3} {'degree':>7} {'sum |B|':>16} {'negative weights':>17}")
for n_check in (*range(1, 11), 20, 30, 40, 50):
    weights = newton_cotes_weights(n_check)
    checked_weights[n_check] = weights
    degree = n_check + (n_check % 2 == 0)
    assert sum(weights) == 1 and weights == weights[::-1]
    for power in range(degree + 1):
        assert sum(w * Fraction(j, n_check)**power
                   for j, w in enumerate(weights)) == Fraction(1, power + 1)
    assert sum(w * Fraction(j, n_check)**(degree + 1)
               for j, w in enumerate(weights)) != Fraction(1, degree + 2)
    gamma = sum(map(abs, weights))
    print(f"{n_check:3d} {degree:7d} {float(gamma):16.6g} {sum(w < 0 for w in weights):17d}")

# Signed errors: exact integral minus quadrature sum, on [0, 1].
trapezoid_error = Fraction(1, 3) - sum(w * Fraction(j, 1)**2
                                    for j, w in enumerate(checked_weights[1]))
simpson_error = Fraction(1, 5) - sum(w * Fraction(j, 2)**4
                                  for j, w in enumerate(checked_weights[2]))
assert trapezoid_error == -Fraction(1, 6)
assert simpson_error == -Fraction(1, 120)
print("Trapezoid error for x^2:", trapezoid_error)
print("Simpson error for x^4:", simpson_error)
print("All exact-arithmetic checks passed.")
```
