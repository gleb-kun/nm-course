# Finite differences

We introduce tables of finite differences, study their relation to derivatives and data errors, and then use finite differences to describe Bernstein polynomials and smooth joins between Bézier curves.

## Table of finite differences

**Definition 1 (equally spaced nodes).** Let $y_i=f(x_i)$, $i=0,1,\ldots,n$, be a table of function values at nodes $x_0<x_1<\cdots<x_n$. If the distance between neighboring nodes is constant,

$$
h=x_i-x_{i-1}>0,\qquad x_i=x_0+ih,\qquad i=1,\ldots,n,\tag{1}
$$

then $h$ is called the **step of the table**, and the nodes are **equally spaced**.

**Definition 2 (forward finite differences).** The quantity $\Delta y_i=y_{i+1}-y_i$ is the first-order forward finite difference at $x_i$, with step $h$. The second-order difference is

$$
\Delta^2y_i=\Delta y_{i+1}-\Delta y_i.\tag{2}
$$

In the same way, differences of higher orders are defined recursively:

$$
\Delta^ky_i=\Delta^{k-1}y_{i+1}-\Delta^{k-1}y_i,
\qquad \Delta^0y_i=y_i.\tag{3}
$$

Here $1\le k\le n$ and $0\le i\le n-k$: a difference of order $k$ uses $k+1$ available values.

The table of forward finite differences is arranged as follows.
Each difference is placed between the two entries from which it is computed.

**Table 1. Forward finite differences.**

$$
\begin{array}{c|c|c|c|c|c|c}
x & y & \Delta y & \Delta^2y & \Delta^3y & \cdots & \Delta^ny \\
\hline
x_0 & y_0 & & & & & \\
& & \Delta y_0 & & & & \\
x_1 & y_1 & & \Delta^2y_0 & & & \\
& & \Delta y_1 & & \Delta^3y_0 & & \\
x_2 & y_2 & & \Delta^2y_1 & & \cdots & \\
& & \Delta y_2 & & \Delta^3y_1 & & \\
\vdots & \vdots & \vdots & \vdots & \vdots & \ddots & \Delta^ny_0 \\
x_{n-2} & y_{n-2} & & \Delta^2y_{n-3} & & \cdots & \\
& & \Delta y_{n-2} & & \Delta^3y_{n-3} & & \\
x_{n-1} & y_{n-1} & & \Delta^2y_{n-2} & & & \\
& & \Delta y_{n-1} & & & & \\
x_n & y_n & & & & &
\end{array}
$$

### Properties of finite differences

**Proposition 1 (binomial formula).** For $0\le k\le n$ and $0\le i\le n-k$,

$$
\Delta^ky_i=\sum_{l=0}^k(-1)^{k-l}\binom{k}{l}y_{i+l},
\qquad \binom{k}{l}=\frac{k!}{l!(k-l)!}.\tag{4}
$$

In particular,

$$
\Delta^2y_i=y_{i+2}-2y_{i+1}+y_i,\tag{5}
$$

$$
\Delta^3y_i=y_{i+3}-3y_{i+2}+3y_{i+1}-y_i,\tag{6}
$$

$$
\Delta^4y_i=y_{i+4}-4y_{i+3}+6y_{i+2}-4y_{i+1}+y_i.\tag{7}
$$

**Proof.** Induction in the recurrence (3), using Pascal's identity $\binom{k-1}{l-1}+\binom{k-1}{l}=\binom{k}{l}$, gives (4). <span style="float: right;">$\square$</span>

**Theorem 1 (finite differences and derivatives).** Let $1\le k\le n$, $0\le i\le n-k$, and $f\in C^k([x_i,x_{i+k}])$, where $x_{i+k}=x_i+kh$. Then

$$
\Delta^ky_i=h^kf^{(k)}(\xi),\qquad \xi\in(x_i,x_{i+k}).\tag{8}
$$

**Proof.** The proof has three steps. Write the nodes as

$$
x_{i+j}=x_i+jh,\qquad j=0,1,\ldots,k.\tag{9}
$$

**1. Finite and divided differences.** Denote the $k$-th divided difference by $f[x_i,\ldots,x_{i+k}]$. For equally spaced nodes,

$$
\Delta^kf(x_i)=k!h^k f[x_i,\ldots,x_{i+k}].\tag{10}
$$

This follows by induction from the two recurrence formulas

$$
f[x_i,\ldots,x_{i+k}]=\frac{f[x_{i+1},\ldots,x_{i+k}]-f[x_i,\ldots,x_{i+k-1}]}{x_{i+k}-x_i},\tag{11}
$$

$$
\Delta^kf(x_i)=\Delta^{k-1}f(x_{i+1})-\Delta^{k-1}f(x_i),\tag{12}
$$

and the equality $x_{i+k}-x_i=kh$.

**2. Mean value theorem for divided differences.** We have

$$
f[x_i,\ldots,x_{i+k}]=\frac{f^{(k)}(\xi)}{k!},\qquad \xi\in(x_i,x_{i+k}).\tag{13}
$$

Indeed, let $P$ be the interpolation polynomial of degree at most $k$ through these nodes and set

$$
E(x)=f(x)-P(x).\tag{14}
$$

The function $E$ vanishes at $k+1$ distinct points. Applying Rolle's theorem $k$ times gives $E^{(k)}(\xi)=0$ at some interior point. Since

$$
P^{(k)}(x)=k!f[x_i,\ldots,x_{i+k}],\tag{15}
$$

we obtain (13).

**3. Conclusion.** Substituting (13) into (10) gives

$$
\Delta^ky_i=k!h^k\frac{f^{(k)}(\xi)}{k!}=h^kf^{(k)}(\xi).\tag{16}
$$

Thus,

$$
\boxed{\Delta^ky_i=h^kf^{(k)}(\xi)},\qquad \xi\in(x_i,x_i+kh).\tag{17}
$$

The theorem is proved. <span style="float: right;">$\square$</span>

**Remark 1.** For $k=1$, (8) is the usual mean value theorem: $f(x_i+h)-f(x_i)=hf'(\xi)$.

**Corollary 1 (polynomials).** For a polynomial $P_d(x)=\sum_{m=0}^d a_mx^m$ of degree $d$,

$$
\Delta^dP_d(x_i)=h^dd!a_d,\qquad \Delta^kP_d(x_i)=0\quad(k>d),\tag{18}
$$

whenever the required nodes are available. The first equality follows from $P_d^{(d)}=d!a_d$; each further difference of this constant is zero.

Finite differences have many practical applications. If $f^{(k)}$ varies only slightly on $[x_i,x_i+kh]$, then (8) gives the numerical differentiation formula

$$
f^{(k)}(x)\approx\frac{\Delta^ky_i}{h^k},\qquad x\in[x_i,x_i+kh].\tag{19}
$$

For example, if $f\in C^{k+1}$ on this interval, then

$$
\left|f^{(k)}(x)-\frac{\Delta^ky_i}{h^k}\right|\le khM_{k+1},
\qquad M_{k+1}=\max_{[x_i,x_i+kh]}|f^{(k+1)}|.\tag{20}
$$

Indeed, (8) reduces the difference to $f^{(k)}(x)-f^{(k)}(\xi)$. For fixed $k$ and bounded derivatives on a fixed neighborhood, this gives at least first-order accuracy as $h\to0$. We will return to endpoint formulas at the end of the lecture.

Finite differences also occur in Newton interpolation. We next consider error propagation, estimation of the noise level in a table, and the construction of Bézier curves.

**Proposition 2 (propagation of data errors).** Let $y_i^*$ be the available values and $\varepsilon_i=y_i-y_i^*$ their signed errors. If $|\varepsilon_i|\le\varepsilon$, then

$$
\varepsilon_i^{(k)}=\Delta^ky_i-\Delta^ky_i^*
=\sum_{l=0}^k(-1)^{k-l}\binom{k}{l}\varepsilon_{i+l},\tag{21}
$$

and

$$
|\varepsilon_i^{(k)}|\le\sum_{l=0}^k\binom{k}{l}\varepsilon=2^k\varepsilon.\tag{22}
$$

**Proof.** Apply (4) to $y-y^*$ and use the triangle inequality. For a given difference, the bound is attained by $\varepsilon_{i+l}=(-1)^{k-l}\varepsilon$. Thus it is sharp, but the actual errors need not grow with $k$; for example, a constant error cancels in every difference of positive order. <span style="float: right;">$\square$</span>

For the differentiation quotient, the propagated data error is bounded by

$$
\left|\frac{\Delta^ky_i-\Delta^ky_i^*}{h^k}\right|\le\frac{2^k\varepsilon}{h^k}.\tag{23}
$$

These bounds assume exact arithmetic when forming the differences; additional floating-point rounding is a separate source of error.

## Examples

### Estimation of the noise level in a table

In practice, a table of approximate values of a smooth function is often obtained by measurement or computation. We would like to estimate its noise level. The worst-case bound (22) does not fully answer this statistical question.

**Definition 3 (noise model).** Assume that the errors $\varepsilon_j$ are independent random variables with

$$
\mathbb E\varepsilon_j=0,\qquad \operatorname{Var}(\varepsilon_j)=\mathbb E\varepsilon_j^2=\sigma^2.\tag{24}
$$

We call $\sigma$ the **root-mean-square noise level** of the table. Normality is not required. Pairwise uncorrelated errors with the same variance are sufficient for the variance calculation below.

**Proposition 3 (variance of a finite-difference error).** Under (24), $\mathbb E\varepsilon_i^{(k)}=0$, and its variance satisfies

$$
(\sigma^{(k)})^2
=\mathbb E\left[\left(\sum_{l=0}^k(-1)^{k-l}\binom{k}{l}\varepsilon_{i+l}\right)^2\right].\tag{25}
$$

**Proof.** Expanding the square gives

$$
\begin{aligned}
(\sigma^{(k)})^2
&=\sum_{l=0}^k\binom{k}{l}^2\mathbb E\varepsilon_{i+l}^2\\
&\quad+\sum_{\substack{0\le l,r\le k\\l\ne r}}(-1)^{l+r}\binom{k}{l}\binom{k}{r}\mathbb E[\varepsilon_{i+l}\varepsilon_{i+r}].
\end{aligned}\tag{26}
$$

The mixed terms vanish, and Vandermonde's identity gives

$$
\sum_{l=0}^k\binom{k}{l}^2=\binom{2k}{k},\qquad
\mathbb E[\varepsilon_{i+l}\varepsilon_{i+r}]=0\quad(l\ne r).\tag{27}
$$

Consequently,

$$
(\sigma^{(k)})^2=\binom{2k}{k}\sigma^2,\tag{28}
$$

or, in terms of root-mean-square errors,

$$
\sigma^{(k)}=\sqrt{\binom{2k}{k}}\,\sigma.\tag{29}
$$

The amplification factor is smaller than the worst-case factor in (22), since

$$
\sqrt{\binom{2k}{k}}<2^k,\qquad k\ge1.\tag{30}
$$

The last inequality follows from $\binom{2k}{k}<\sum_{j=0}^{2k}\binom{2k}{j}=4^k$. <span style="float: right;">$\square$</span>

**Remark 2 (estimating the noise).** In some tables, finite differences decrease for the first few orders and then grow and oscillate. This can indicate that the noise dominates the differences of the underlying function, but smoothness alone does not guarantee this pattern.

Choose an order $1\le p\le n$ for which $\Delta^pf(x_i)$ is negligible relative to the noise. Then

$$
\begin{aligned}
\widehat{(\sigma^{(p)})^2}&=\frac{1}{n-p+1}\sum_{i=0}^{n-p}(\Delta^py_i^*)^2,\\
\widehat{\sigma}&=\left[\frac{\sum_{i=0}^{n-p}(\Delta^py_i^*)^2}{(n-p+1)\binom{2p}{p}}\right]^{1/2}.
\end{aligned}\tag{31}
$$

For a fixed choice of $p$, the corresponding variance estimator satisfies

$$
\mathbb E[\widehat{\sigma}^{\,2}]
=\sigma^2+\frac{\sum_{i=0}^{n-p}(\Delta^pf(x_i))^2}{(n-p+1)\binom{2p}{p}}.\tag{32}
$$

Thus it is unbiased for $\sigma^2$ if those function differences vanish, and otherwise includes a nonnegative signal contribution. The square-root estimator need not be unbiased for $\sigma$. Differences in neighboring rows share input errors and are correlated; they are not independent repetitions. A short table therefore gives only a rough estimate, and systematic or correlated errors require a different model.

**Example 1 (a table of logarithms).** Let us estimate the noise level for $f(x)=\ln x$ at the eleven nodes $x_i=1+0.1i$, $i=0,\ldots,10$. To make the example reproducible, we add independent normal errors with standard deviation $\sigma=2\cdot10^{-5}$, using random seed 42. Only the displayed table is rounded.

For this realization, the column root-mean-square values begin to increase at order $p=5$. The six fifth differences give

$$
\widehat{\sigma^{(5)}}=\left[\frac16\sum_{i=0}^5(\Delta^5y_i^*)^2\right]^{1/2}
\approx2.1035091\cdot10^{-4},\tag{33}
$$

and hence

$$
\widehat{\sigma}=\frac{\widehat{\sigma^{(5)}}}{\sqrt{\binom{10}{5}}}
=\frac{\widehat{\sigma^{(5)}}}{\sqrt{252}}
\approx1.3250862\cdot10^{-5}.\tag{34}
$$

This is an estimate from one short noisy table, not the known standard deviation itself. The code below constructs the actual data used in these calculations.

```python
import numpy as np
from math import comb

x = np.linspace(1.0, 2.0, 11)
sigma = 2e-5
rng = np.random.default_rng(42)
y_star = np.log(x) + rng.normal(0.0, sigma, len(x))
differences = [np.diff(y_star, n=k) for k in range(8)]

print(" x       y*       Δy*       Δ²y*      Δ³y*      Δ⁴y*      Δ⁵y*      Δ⁶y*      Δ⁷y*")
for i in range(len(x)):
    row = [f"{column[i]: .6f}" if i < len(column) else "         " for column in differences]
    print(f"{x[i]:.1f} " + " ".join(row))

print("\nOrder   column RMS")
for k in range(1, 8):
    print(f"{k:3d}     {np.sqrt(np.mean(differences[k]**2)):.8e}")

p = 5
sigma_p = np.sqrt(np.mean(differences[p]**2))
sigma_estimate = sigma_p / np.sqrt(comb(2*p, p))
print(f"\nEstimated noise: {sigma_estimate:.8e}")
print(f"Model standard deviation: {sigma:.8e}")
```

**Remark 3 (interpretation of the example).** The estimate is of the same order as the prescribed noise, but it varies between realizations. It is an RMS estimate, not a guaranteed bound on each table entry, so it does not by itself justify declaring particular decimal places useless. Even a noiseless nonpolynomial function has nonzero high-order differences; these must not automatically be identified with noise.

### Bézier curves

The next application is the construction of Bézier curves [2]. Finite differences will appear in the coefficients and in the conditions for joining two curves.

#### Constructing a single Bézier curve

**Definition 4 (Bernstein form).** For $n\ge1$ and real coefficients $y_0,\ldots,y_n$, the polynomial

$$
B(x)=\sum_{k=0}^ny_k\binom{n}{k}x^k(1-x)^{n-k}\tag{35}
$$

is a polynomial of degree at most $n$ in **Bernstein form**. Its basis polynomials are

$$
p_{nk}(x)=\binom{n}{k}x^k(1-x)^{n-k},\qquad k=0,\ldots,n.\tag{36}
$$

We also set $p_{00}=1$, as needed in the recursion below. At the endpoints,

$$
B(0)=y_0,\qquad B(1)=y_n.\tag{37}
$$

The coefficients are arbitrary control values; in general, $B(k/n)\ne y_k$ at interior indices. This is not interpolation at all control points.

**Proposition 4 (Bernstein basis properties).** The basis polynomials satisfy

$$
p_{nk}(x)>0\quad(0<x<1),\qquad p_{nk}(x)\ge0\quad(0\le x\le1),\tag{38}
$$

and

$$
\sum_{k=0}^np_{nk}(x)=1\qquad\text{for all real }x.\tag{39}
$$

**Proof.** Positivity follows directly from (36); the sum is $(x+(1-x))^n=1$ by the binomial theorem. Thus, on $[0,1]$, $B(x)$ is a convex combination of its control values and lies between their minimum and maximum. <span style="float: right;">$\square$</span>

**Lemma 1 (linear reproduction).** For $n\ge1$ and all real $x$,

$$
\frac1n\sum_{k=0}^nk\,p_{nk}(x)=x.\tag{40}
$$

**Proof.** First take $0<x<1$, where all basis polynomials are positive. Logarithmic differentiation gives

$$
\bigl(\ln p_{nk}(x)\bigr)'=\frac{k}{x}-\frac{n-k}{1-x}
=\frac{k-nx}{x(1-x)}.\tag{41}
$$

Since

$$
\bigl(\ln p_{nk}(x)\bigr)'=\frac{p'_{nk}(x)}{p_{nk}(x)},\tag{42}
$$

we obtain

$$
p'_{nk}(x)=\frac{n}{x(1-x)}\left(\frac{k}{n}-x\right)p_{nk}(x).\tag{43}
$$

Differentiating (39) yields

$$
0=\frac{n}{x(1-x)}\sum_{k=0}^n\left(\frac{k}{n}-x\right)p_{nk}(x).\tag{44}
$$

Using (39) again gives (40) on $(0,1)$. Both sides are polynomials, so equality extends to all real $x$. <span style="float: right;">$\square$</span>

**Theorem 2 (one step of de Casteljau's algorithm).** Set $y_k^{(0)}=y_k$. For a fixed real $x$,

$$
B(x)=\sum_{k=0}^{n-1}y_k^{(1)}p_{n-1,k}(x),\tag{45}
$$

where

$$
y_k^{(1)}=(1-x)y_k^{(0)}+xy_{k+1}^{(0)},\qquad k=0,\ldots,n-1.\tag{46}
$$

**Proof.** Substitution gives

$$
\begin{aligned}
\sum_{k=0}^{n-1}y_k^{(1)}p_{n-1,k}(x)
&=\sum_{k=0}^{n-1}y_k\binom{n-1}{k}x^k(1-x)^{n-k}
\\&\quad+\sum_{k=1}^ny_k\binom{n-1}{k-1}x^k(1-x)^{n-k}\\
&=y_0(1-x)^n+y_nx^n\\
&\quad+\sum_{k=1}^{n-1}y_k\left[\binom{n-1}{k}+\binom{n-1}{k-1}\right]x^k(1-x)^{n-k}.
\end{aligned}\tag{47}
$$

By Pascal's identity,

$$
\binom{n-1}{k}+\binom{n-1}{k-1}=\binom{n}{k},\qquad 1\le k\le n-1.\tag{48}
$$

Thus,

$$
\sum_{k=0}^{n-1}y_k^{(1)}p_{n-1,k}(x)
=\sum_{k=0}^ny_k\binom{n}{k}x^k(1-x)^{n-k}=B(x).\tag{49}
$$

The identity also holds for $n=1$, with the interior sum empty. The coefficients $y_k^{(1)}$ depend on $x$; this is an evaluation identity, not a degree reduction with constant coefficients. <span style="float: right;">$\square$</span>

**Corollary 2 (de Casteljau recursion).** The transformation (45) can be repeated. For example, for $n\ge2$,

$$
B(x)=\sum_{k=0}^{n-2}y_k^{(2)}p_{n-2,k}(x),\qquad
y_k^{(2)}=(1-x)y_k^{(1)}+xy_{k+1}^{(1)}.\tag{50}
$$

In general, define

$$
y_k^{(r)}=(1-x)y_k^{(r-1)}+xy_{k+1}^{(r-1)},
\qquad r=1,\ldots,n,\quad k=0,\ldots,n-r,\tag{51}
$$

with

$$
y_k^{(0)}=y_k,\qquad k=0,\ldots,n.\tag{52}
$$

The last step gives

$$
B(x)=\sum_{k=0}^1y_k^{(n-1)}p_{1k}(x)
=(1-x)y_0^{(n-1)}+xy_1^{(n-1)}=y_0^{(n)}.\tag{53}
$$

This evaluates $B(x)$ at a fixed $x$. For $x\in[0,1]$, every step is a convex combination, including at the endpoints.

#### Geometric interpretation and vector Bézier curves

**Definition 5 (Bézier curve).** Let $Y_0,\ldots,Y_n\in\mathbb R^s$ be control points. The vector polynomial

$$
\mathbf B(x)=\sum_{k=0}^np_{nk}(x)Y_k,\qquad 0\le x\le1,\tag{54}
$$

parametrizes a **Bézier curve** connecting the endpoints

$$
\mathbf B(0)=Y_0,\qquad \mathbf B(1)=Y_n.\tag{55}
$$

The scalar recursion applies componentwise:

$$
Y_k^{(r)}=(1-x)Y_k^{(r-1)}+xY_{k+1}^{(r-1)},
\qquad r=1,\ldots,n,\quad k=0,\ldots,n-r,\tag{56}
$$

with

$$
Y_k^{(0)}=Y_k,\qquad k=0,\ldots,n,\qquad \mathbf B(x)=Y_0^{(n)}.\tag{57}
$$

As $x$ traverses $[0,1]$, this final point traces the curve.

**Proposition 5 (convex hull property).** For $x\in[0,1]$, the point $\mathbf B(x)$ lies in the convex hull of $Y_0,\ldots,Y_n$.

**Proof.** By (38) and (39), the coefficients in (54) are nonnegative and sum to one. In particular, the curve interpolates its first and last control points but generally does not pass through the interior control points. Convexity is asserted only for $x\in[0,1]$. <span style="float: right;">$\square$</span>

**Example 2 (the graph of a Bernstein polynomial).** Take $s=2$ and $Y_k=(k/n,y_k)$. Lemma 1 gives

$$
\mathbf B(x)=(x,B(x)),\qquad x\in[0,1].\tag{58}
$$

At level $r$ of (56), the abscissa of $Y_k^{(r)}$ is

$$
\frac{k+rx}{n},\qquad k=0,\ldots,n-r.\tag{59}
$$

This follows by induction in the affine recursion. Thus the same construction evaluates the graph of $B$.

Figure 1 illustrates $n=3$ at a fixed parameter $x=x_*$. The control points are $Y_k=(k/3,y_k)$, the intermediate points are $Y_k^{(r)}$ from (56), and the final point is $Y_0^{(3)}=\mathbf B(x_*)=(x_*,B(x_*))$.

**Example 3 (plane and spatial curves).** Arbitrary control points $Y_k=(y_{k1},y_{k2})$ generate a plane Bézier curve. Three-component points $Y_k=(y_{k1},y_{k2},y_{k3})$ generate a spatial curve by exactly the same formula (54).

#### Smooth joins between Bézier curves: composite curves

Let us consider the simplest problem of constructing smooth composite Bézier curves. We first establish two power representations in terms of finite differences.

**Lemma 2 (forward-difference representation).** The polynomial (35) can be written as

$$
B(x)=\sum_{i=0}^n\binom{n}{i}\Delta^iy_0\,x^i,\tag{60}
$$

where

$$
\Delta^iy_0=\sum_{k=0}^i(-1)^{i-k}\binom{i}{k}y_k.\tag{61}
$$

**Proof.** Expanding $(1-x)^{n-k}$ gives

$$
B(x)=\sum_{k=0}^n\sum_{j=0}^{n-k}y_k\binom{n}{k}\binom{n-k}{j}(-1)^jx^{k+j}.\tag{62}
$$

Use $i=k+j$ and the identity

$$
\binom{n}{k}\binom{n-k}{i-k}=\binom{n}{i}\binom{i}{k},\qquad 0\le k\le i\le n.\tag{63}
$$

Collecting powers of $x$ gives

$$
B(x)=\sum_{i=0}^n\binom{n}{i}
\left[\sum_{k=0}^i(-1)^{i-k}\binom{i}{k}y_k\right]x^i
=\sum_{i=0}^n\binom{n}{i}\Delta^iy_0\,x^i.\tag{64}
$$

The lemma is proved. <span style="float: right;">$\square$</span>

**Definition 6 (backward differences).** For coefficients indexed to the left of zero, set $\nabla y_j=y_j-y_{j-1}$ and repeat this operation. In particular,

$$
\nabla^iy_0=\sum_{k=0}^i(-1)^k\binom{i}{k}y_{-k},\qquad \nabla^0y_0=y_0.\tag{65}
$$

Here we are taking differences of the coefficient sequence, not assuming that an underlying function has been sampled at physical nodes with a specified step.

The polynomial $B$ corresponds to $[0,1]$. For $[-1,0]$, introduce a Bernstein polynomial of degree at most $m\ge1$:

$$
\begin{aligned}
Q(x)&=\sum_{k=0}^my_{-m+k}\binom{m}{k}(1+x)^k(-x)^{m-k}\\
&=\sum_{k=0}^m(-1)^ky_{-k}\binom{m}{k}x^k(1+x)^{m-k}.
\end{aligned}\tag{66}
$$

The first line uses the local parameter $1+x\in[0,1]$; the second follows by reversing the index. In particular, $Q(-1)=y_{-m}$ and $Q(0)=y_0=B(0)$.

**Lemma 3 (backward-difference representation).** The polynomial $Q$ admits the representation

$$
Q(x)=\sum_{i=0}^m\binom{m}{i}\nabla^iy_0\,x^i.\tag{67}
$$

**Proof.** Expand the second line of (66):

$$
Q(x)=\sum_{k=0}^m\sum_{j=0}^{m-k}(-1)^ky_{-k}\binom{m}{k}\binom{m-k}{j}x^{k+j}.\tag{68}
$$

Set $i=k+j$ and use (63), with $m$ in place of $n$. Then

$$
Q(x)=\sum_{i=0}^m\binom{m}{i}
\left[\sum_{k=0}^i(-1)^k\binom{i}{k}y_{-k}\right]x^i
=\sum_{i=0}^m\binom{m}{i}\nabla^iy_0\,x^i.\tag{69}
$$

The bracket is precisely (65), proving the lemma. <span style="float: right;">$\square$</span>

**Definition 7 (composite function).** Join the two polynomials at their common endpoint:

$$
S(x)=\begin{cases}
Q(x),&-1\le x\le0,\\
B(x),&0\le x\le1.
\end{cases}\tag{70}
$$

Because $Q(0)=B(0)=y_0$, the function $S$ is well defined and continuous. We now ask when its derivatives also match.

**Theorem 3 (smoothness of a join).** Let $0\le r\le\min(m,n)$. The function (70) belongs to $C^r([-1,1])$ if and only if

$$
\binom{m}{i}\nabla^iy_0=\binom{n}{i}\Delta^iy_0,
\qquad i=1,\ldots,r.\tag{71}
$$

For $m=n$, these conditions simplify to

$$
\nabla^iy_0=\Delta^iy_0,\qquad i=1,\ldots,r.\tag{72}
$$

**Proof.** By (60) and (67),

$$
Q^{(i)}(0)=i!\binom{m}{i}\nabla^iy_0,\qquad
B^{(i)}(0)=i!\binom{n}{i}\Delta^iy_0.\tag{73}
$$

Each piece is a polynomial and hence smooth on its own interval. Since the values already match, $S\in C^r$ precisely when these derivatives match for $i=1,\ldots,r$, which is (71). For $r=0$, the condition is empty and continuity is automatic. For higher orders than a piece's degree, its derivative is zero; smoothness must still be checked by matching the derivatives directly. <span style="float: right;">$\square$</span>

**Remark 4 (other parameter intervals and vector curves).** The intervals in Theorem 3 both have length one. For Bernstein pieces on $[-a,0]$ and $[0,b]$, with $a,b>0$ and local parameters $(x+a)/a$ and $x/b$, the conditions become

$$
\frac{\binom{m}{i}}{a^i}\nabla^iy_0
=\frac{\binom{n}{i}}{b^i}\Delta^iy_0,\qquad i=1,\ldots,r.\tag{74}
$$

The same conditions hold componentwise for vector control points. They describe $C^r$ smoothness with respect to the common parameter. Merely making tangent directions agree is a weaker geometric condition and does not ensure equal first derivatives.

**Example 4 (a $C^2$ join of different degrees).** Let $m=5$ on $[-1,0]$ and $n=3$ on $[0,1]$. Choose the common value $y_0$ and right-hand control values $y_1,y_2,y_3$. The first two matching conditions in (71) determine the two left-hand values nearest the join:

$$
\begin{aligned}
y_{-1}&=y_0-\frac{3}{5}(y_1-y_0),\\
y_{-2}&=2y_{-1}-y_0+\frac{3}{10}(y_2-2y_1+y_0).
\end{aligned}\tag{75}
$$

The remaining left-hand values $y_{-5},y_{-4},y_{-3}$ determine the shape away from the join. By construction,

$$
\begin{aligned}
Q'(0)&=5(y_0-y_{-1})=3(y_1-y_0)=B'(0),\\
Q''(0)&=20(y_0-2y_{-1}+y_{-2})=6(y_2-2y_1+y_0)=B''(0).
\end{aligned}\tag{76}
$$

Thus the two pieces form a $C^2$ join. Higher smoothness would require additional derivative-matching conditions.

### Using tables with a constant step: numerical differentiation

We return to the formulas discussed in the [first lecture](https://github.com/gleb-kun/nm-course/blob/master/lectures/01.1/lecture01_eng.ipynb), in the section “Using tables with a constant step.” Formula numbers in the present lecture continue consecutively.

For $y_i=f(x_i)$, $x_i=x_0+ih$, the highest divided-difference approximation takes the form

$$
f^{(n)}(x)\approx\frac{\Delta^ny_0}{h^n},\qquad x\in[x_0,x_n].\tag{77}
$$

Its justification and first-order error bound follow from (8) and (20), with $k=n$.

At a boundary point, symmetric stencils may require unavailable values. Instead, differentiate the degree-at-most-$n$ interpolation polynomial $P_n$:

$$
f^{(k)}(x_0)\approx P_n^{(k)}(x_0),\qquad
f^{(k)}(x_n)\approx P_n^{(k)}(x_n),\qquad 1\le k\le n.\tag{78}
$$

Differentiating Newton's forward and backward interpolation polynomials gives

$$
f'(x_0)\approx\frac1h\sum_{j=1}^n\frac{(-1)^{j-1}}j\Delta^jy_0,\tag{79}
$$

$$
f'(x_n)\approx\frac1h\sum_{j=1}^n\frac1j\nabla^jy_n.\tag{80}
$$

For fixed $n\ge1$ and $f\in C^{n+1}$ on a fixed neighborhood containing the shrinking stencil, both first-derivative formulas have error $O(h^n)$ as $h\to0$. This order statement concerns exact input values; errors in the differences must also be propagated through the weighted sums.

**Example 5 (three-point endpoint formulas).** For $n=2$, (79) becomes

$$
f'(x_0)\approx\frac1h\left(\Delta y_0-\frac12\Delta^2y_0\right).\tag{81}
$$

Since $\Delta y_0=y_1-y_0$ and $\Delta^2y_0=y_2-2y_1+y_0$,

$$
\begin{aligned}
f'(x_0)&\approx\frac1h\left[(y_1-y_0)-\frac12(y_2-2y_1+y_0)\right]\\
&=\frac{-3y_0+4y_1-y_2}{2h}.
\end{aligned}\tag{82}
$$

Thus,

$$
f'(x_0)\approx\frac{-3f(x_0)+4f(x_1)-f(x_2)}{2h}.\tag{83}
$$

At the right endpoint, (80) gives

$$
f'(x_n)\approx\frac1h\left(\nabla y_n+\frac12\nabla^2y_n\right).\tag{84}
$$

Using $\nabla y_n=y_n-y_{n-1}$ and $\nabla^2y_n=y_n-2y_{n-1}+y_{n-2}$,

$$
\begin{aligned}
f'(x_n)&\approx\frac1h\left[(y_n-y_{n-1})+\frac12(y_n-2y_{n-1}+y_{n-2})\right]\\
&=\frac{y_{n-2}-4y_{n-1}+3y_n}{2h}.
\end{aligned}\tag{85}
$$

Therefore,

$$
f'(x_n)\approx\frac{f(x_{n-2})-4f(x_{n-1})+3f(x_n)}{2h}.\tag{86}
$$

Both formulas are exact for polynomials of degree at most two and have error $O(h^2)$ for $f\in C^3$ on the corresponding shrinking intervals, with the third derivative uniformly bounded there.
