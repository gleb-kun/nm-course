# Практическое занятие №11.1. Численное интегрирование

## Приближенное вычисление определенных интегралов

Самостоятельно реализуйте квадратурные формулы на одном отрезке и составные формулы, оцените их точность и исследуйте зависимость погрешностей от шага сетки и выбранного метода.

### Исходные данные и требования к реализации

Рассмотрим определенный интеграл
$$
I=\int_a^b f(x)\,dx,\quad H=b-a>0.
$$
Для составных формул используйте равномерную сетку
$$
x_i=a+ih,\quad i=0,1,\ldots,n,\quad h=\frac{H}{n}.
$$
Здесь $n$ — число промежутков, а сетка содержит $n+1$ узел. Символ $H$ обозначает длину всего отрезка, $h$ — шаг сетки составной формулы.

В качестве тестовых функций используйте $f(x)=\sin(x)$, $f(x)=\cos(x)$ и $f(x)=\exp(x)$ на $[a,b]=[0,2]$. Аналитически найдите их точные интегралы и необходимые ниже оценки производных. При сгущении сетки сохраняйте отрезок неизменным. В обязательной части вес единичный: $w(x)\equiv1$.

- Реализуйте формулы левых, правых и средних прямоугольников, трапеций, Симпсона и «трех восьмых» — как на одном отрезке, так и в составном виде.
- Передавайте подынтегральную функцию, концы отрезка и, для составных формул, $n$ как аргументы функций. Аналитические интегралы и производные являются справочными данными для проверки, а не входными данными квадратурного вычисления.
- Требуйте конечные концы отрезка $a<b$ и целое положительное $n$. Для составной формулы Симпсона нужно четное $n\ge2$; для составной формулы «трех восьмых» — $n\ge3$, кратное трем. Сообщайте о недопустимых данных, не изменяя $n$ автоматически.
- Вычисляйте узлы через целочисленные индексы, а не последовательным прибавлением $h$. При необходимости вычисляйте значения в серединах промежутков $x_i+h/2$.
- Можно использовать библиотеки для работы с массивами, таблицами и графиками. Квадратурные суммы реализуйте самостоятельно; готовые функции численного интегрирования допускаются только для дополнительной проверки. В обязательных экспериментах используйте аналитически найденные интегралы как эталонные значения.

### Краткая справка

Построение квадратурных формул и их остаточные члены разобраны в [лекции по численному интегрированию](../../lectures/02.1/lecture02_rus.ipynb).

Для любой формулы различайте погрешность со знаком $R$, абсолютную погрешность $E$ и относительную погрешность $\delta$:
$$
R=I-Q,\quad E=|I-Q|,\quad \delta=\frac{E}{|I|}\quad(I\ne0).
$$
При $I=0$ относительная погрешность не определена. В этом случае указывайте `--` и используйте абсолютную погрешность. Малое ненулевое значение $|I|$ также может делать относительную погрешность малоинформативной.

Обозначим приближение на одном отрезке через $Q_m^{\mathrm{single}}$, где $m$ указывает выбранную формулу. В следующей таблице $c=(a+b)/2$, а у каждого остаточного члена своя точка $\xi\in[a,b]$. Достаточно существования непрерывной производной функции $f$, входящей в соответствующий остаточный член.

#### Формулы на одном отрезке

| Формула | $Q_m^{\mathrm{single}}$ | $R_m^{\mathrm{single}}=I-Q_m^{\mathrm{single}}$ |
| --- | --- | --- |
| Левых прямоугольников | $Hf(a)$ | $\dfrac{H^2}{2}f'(\xi)$ |
| Правых прямоугольников | $Hf(b)$ | $-\dfrac{H^2}{2}f'(\xi)$ |
| Средних прямоугольников | $Hf(c)$ | $\dfrac{H^3}{24}f''(\xi)$ |
| Трапеций | $\dfrac{H}{2}[f(a)+f(b)]$ | $-\dfrac{H^3}{12}f''(\xi)$ |
| Симпсона | $\dfrac{H}{6}[f(a)+4f(c)+f(b)]$ | $-\dfrac{H^5}{2880}f^{(4)}(\xi)$ |
| «Трех восьмых» | $\dfrac{H}{8}[f(a)+3f(a+H/3)+3f(a+2H/3)+f(b)]$ | $-\dfrac{H^5}{6480}f^{(4)}(\xi)$ |

Для априорной оценки абсолютной погрешности $B_m^{\mathrm{single}}$ замените производную в остаточном члене на
$$
M_r=\max_{x\in[a,b]}|f^{(r)}(x)|,\quad r=1,2,4,
$$
и возьмите модуль коэффициента. Здесь $r=1$ используется для левых и правых прямоугольников, $r=2$ — для средних прямоугольников и трапеций, $r=4$ — для формул Симпсона и «трех восьмых». Если точный максимум неизвестен, используйте обоснованную верхнюю границу для $M_r$.

#### Составные формулы

Постройте составное приближение $Q_m(h)$ суммированием формул на отдельных частях всего отрезка. Для прямоугольников и трапеций используйте части ширины $h$, для Симпсона — $2h$, для «трех восьмых» — $3h$. Во всех случаях $n$ — число промежутков базовой сетки, а не число частей, на которых применяется отдельная формула. Веса общих узлов получаются сложением их вкладов из соседних частей.

В таблице приведены априорная оценка $|I-Q_m(h)|\le B_m(h)$, ожидаемый глобальный порядок $p$ и алгебраическая степень точности $d$.

| Формула | Ограничение на сетку | $B_m(h)$ | $p$ | $d$ |
| --- | --- | --- | --- | --- |
| Левых прямоугольников | $n\ge1$ | $\dfrac{Hh}{2}M_1$ | 1 | 0 |
| Правых прямоугольников | $n\ge1$ | $\dfrac{Hh}{2}M_1$ | 1 | 0 |
| Средних прямоугольников | $n\ge1$ | $\dfrac{Hh^2}{24}M_2$ | 2 | 1 |
| Трапеций | $n\ge1$ | $\dfrac{Hh^2}{12}M_2$ | 2 | 1 |
| Симпсона | $n\ge2$, четное | $\dfrac{Hh^4}{180}M_4$ | 4 | 3 |
| «Трех восьмых» | $n\ge3$, кратное трем | $\dfrac{Hh^4}{80}M_4$ | 4 | 3 |

Для этих оценок требуется $f\in C^r([a,b])$ при соответствующем $r=1,2,4$. Порядки относятся к $h\to0$ при фиксированных функции и отрезке. Они характеризуют погрешность аппроксимации; округление и погрешности значений функции нужно учитывать отдельно. Для формул на одном отрезке алгебраические степени точности те же.
