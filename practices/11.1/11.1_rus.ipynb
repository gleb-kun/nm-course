### Краткая справка

Квадратурные формулы разобраны в [лекции по численному интегрированию](../../lectures/02.1/lecture02_rus.ipynb).

**Алгебраическая степень точности.** Квадратурная формула имеет степень точности $d$, если она точна для любого полинома степени не выше $d$, но не точна для $x^{d+1}$. В силу линейности достаточно проверить мономы $1,x,\ldots,x^d$ и первый моном $x^{d+1}$, на котором точность нарушается.

Для метода $m$ используйте погрешность со знаком, абсолютную и относительную погрешности
$$
R_m=I-Q_m,\quad E_m=|R_m|,\quad
\delta_m=\frac{E_m}{|I|}\quad(I\ne0).
$$
При $I=0$ относительная погрешность не определена, даже если $Q_m=0$. Указывайте `--` и используйте абсолютную погрешность.

В следующей таблице $c=(a+b)/2$, а у каждого остаточного члена своя точка $\xi\in[a,b]$. Предполагается непрерывность производной, входящей в остаточный член, на $[a,b]$.

#### Квадратурные формулы на $[a, b]$

| Формула | $Q_m$ | $R_m=I-Q_m$ |
| --- | --- | --- |
| Левых прямоугольников | $Hf(a)$ | $\dfrac{H^2}{2}f'(\xi)$ |
| Правых прямоугольников | $Hf(b)$ | $-\dfrac{H^2}{2}f'(\xi)$ |
| Средних прямоугольников | $Hf(c)$ | $\dfrac{H^3}{24}f''(\xi)$ |
| Трапеций | $\dfrac{H}{2}[f(a)+f(b)]$ | $-\dfrac{H^3}{12}f''(\xi)$ |
| Симпсона | $\dfrac{H}{6}[f(a)+4f(c)+f(b)]$ | $-\dfrac{H^5}{2880}f^{(4)}(\xi)$ |
| «Трех восьмых» | $\dfrac{H}{8}[f(a)+3f(a+H/3)+3f(a+2H/3)+f(b)]$ | $-\dfrac{H^5}{6480}f^{(4)}(\xi)$ |

Для априорной оценки абсолютной погрешности $B_m$ замените производную в остаточном члене на
$$
M_r=\max_{x\in[a,b]}|f^{(r)}(x)|,\quad r=1,2,4,
$$
и возьмите модуль коэффициента. Здесь $r=1$ используется для левых и правых прямоугольников, $r=2$ — для средних прямоугольников и трапеций, $r=4$ — для формул Симпсона и «трех восьмых». Если точный максимум неизвестен, используйте обоснованную верхнюю границу для $M_r$.

### Моментные условия

Перейдите к отрезку $[0,1]$ заменой $x=a+Ht$. Запишите каждую формулу в виде
$$
Q_m(f)=H\sum_{j=1}^{N_m}\beta_{mj}f(a+Ht_{mj}),\quad 0\leqslant t_{mj}\leqslant1,
$$
где $N_m$ — число различных узлов, $t_{mj}$ — нормированные узлы, $\beta_{mj}$ — нормированные веса. Точность на одночленах степеней $0,\ldots,d$ равносильна условиям
$$
\sum_{j=1}^{N_m}\beta_{mj}t_{mj}^k=\frac{1}{k+1},\quad k=0,1,\ldots,d.
$$
В частности, условие при $k=0$ означает, что сумма нормированных весов равна единице. Самостоятельно определите узлы и веса из формул выше.
