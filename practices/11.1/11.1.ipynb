### Brief reference

Quadrature formulas and their algebraic degrees of accuracy are discussed in the [lecture on numerical integration](../../lectures/02.1/lecture02_eng.ipynb).

**Algebraic degree of accuracy.** A quadrature formula has degree $d$ if it is exact for every polynomial of degree at most $d$, but is not exact for $x^{d+1}$. By linearity, it is sufficient to check the monomials $1,x,\ldots,x^d$ and the first failing monomial $x^{d+1}$.

For a rule $m$, use the signed, absolute, and relative errors
$$
R_m=I-Q_m,\qquad E_m=|R_m|,\qquad
\delta_m=\frac{E_m}{|I|}\quad(I\ne0).
$$
If $I=0$, relative error is undefined, even when $Q_m=0$. Report `--` and use the absolute error.

In the table below, $c=(a+b)/2$ and each remainder has its own $\xi\in[a,b]$. Assume the derivative appearing in the remainder is continuous on $[a,b]$.

#### Single-panel formulas

| Rule | $Q_m$ | $R_m=I-Q_m$ |
| --- | --- | --- |
| Left rectangle | $Hf(a)$ | $\dfrac{H^2}{2}f'(\xi)$ |
| Right rectangle | $Hf(b)$ | $-\dfrac{H^2}{2}f'(\xi)$ |
| Midpoint | $Hf(c)$ | $\dfrac{H^3}{24}f''(\xi)$ |
| Trapezoidal | $\dfrac{H}{2}[f(a)+f(b)]$ | $-\dfrac{H^3}{12}f''(\xi)$ |
| Simpson | $\dfrac{H}{6}[f(a)+4f(c)+f(b)]$ | $-\dfrac{H^5}{2880}f^{(4)}(\xi)$ |
| Three-eighths | $\dfrac{H}{8}[f(a)+3f(a+H/3)+3f(a+2H/3)+f(b)]$ | $-\dfrac{H^5}{6480}f^{(4)}(\xi)$ |

For an a priori absolute-error bound $B_m$, replace the derivative in the remainder by
$$
M_r=\max_{x\in[a,b]}|f^{(r)}(x)|,\qquad r=1,2,4,
$$
and take the absolute value of the coefficient. Here $r=1$ is used for the left and right rectangles, $r=2$ for midpoint and trapezoidal rules, and $r=4$ for Simpson and three-eighths rules. Use a proven upper bound for $M_r$ if the exact maximum is not available.

### Moment conditions

Map the interval to $[0,1]$ by $x=a+Ht$. Write each rule as
$$
Q_m(f)=H\sum_{j=1}^{N_m}\beta_{mj}f(a+Ht_{mj}),\qquad 0\le t_{mj}\le1,
$$
where $N_m$ is the number of distinct nodes, $t_{mj}$ are normalized nodes, and $\beta_{mj}$ are normalized weights. Exactness on monomials of degrees $0,\ldots,d$ is equivalent to
$$
\sum_{j=1}^{N_m}\beta_{mj}t_{mj}^k=\frac{1}{k+1},\qquad k=0,1,\ldots,d.
$$
In particular, the $k=0$ condition states that the normalized weights sum to one. Use the formulas above to determine the nodes and weights yourself.
