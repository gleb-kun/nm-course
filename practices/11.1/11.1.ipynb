## Approximate calculation of definite integrals

Implement single-panel and composite quadrature rules independently, estimate their accuracy, and investigate how the errors depend on the grid spacing and the chosen rule.

### Input data and implementation requirements

Consider the definite integral
$$
I=\int_a^b f(x)\,dx,\qquad H=b-a>0.
$$
For composite rules, use the uniform grid
$$
x_i=a+ih,\quad i=0,1,\ldots,n,\qquad h=\frac{H}{n}.
$$
Here $n$ is the number of subintervals, and the grid contains $n+1$ nodes. The symbol $H$ denotes the length of the whole interval; $h$ denotes the composite grid spacing.

Use $f(x)=\sin(x)$, $f(x)=\cos(x)$, and $f(x)=\exp(x)$ on $[a,b]=[0,2]$ as test functions. Find their exact integrals and the derivative bounds needed below analytically. Keep the interval fixed when refining the grid. The required part uses unit weight, $w(x)\equiv1$.

- Implement the left rectangle, right rectangle, midpoint, trapezoidal, Simpson, and three-eighths rules, both on one panel and in composite form.
- Pass the integrand, endpoints, and, for composite rules, $n$ as function arguments. Analytical integrals and derivatives are reference data for checking the results, not inputs to the quadrature calculation.
- Require finite endpoints $a<b$ and a positive integer $n$. Composite Simpson requires even $n\ge2$; composite three-eighths requires $n\ge3$ divisible by three. Report inadmissible inputs rather than silently changing $n$.
- Build nodes using integer indices rather than repeated addition of $h$. Evaluate midpoint values at $x_i+h/2$ where required.
- You may use libraries for arrays, tables, and plots. Implement the quadrature sums yourself; ready-made numerical integration functions may be used only as an additional check. Use the analytically obtained integrals as the reference values in the required experiments.

### Brief reference

The construction of quadrature rules and their remainders is covered in the [lecture on numerical integration](../../lectures/02.1/lecture02_eng.ipynb).

For any rule, distinguish the signed error $R$, the absolute error $E$, and the relative error $\delta$:
$$
R=I-Q,\qquad E=|I-Q|,\qquad \delta=\frac{E}{|I|}\quad(I\ne0).
$$
If $I=0$, the relative error is undefined. Report `--` in this case and use the absolute error. A small nonzero $|I|$ can also make the relative error misleading.

Let $Q_m^{\mathrm{single}}$ denote a single-panel approximation, where $m$ identifies the rule. In the following table, $c=(a+b)/2$, and each remainder has its own point $\xi\in[a,b]$. It is sufficient that $f$ have the continuous derivative appearing in that remainder.

#### Single-panel formulas

| Rule | $Q_m^{\mathrm{single}}$ | $R_m^{\mathrm{single}}=I-Q_m^{\mathrm{single}}$ |
| --- | --- | --- |
| Left rectangle | $Hf(a)$ | $\dfrac{H^2}{2}f'(\xi)$ |
| Right rectangle | $Hf(b)$ | $-\dfrac{H^2}{2}f'(\xi)$ |
| Midpoint | $Hf(c)$ | $\dfrac{H^3}{24}f''(\xi)$ |
| Trapezoidal | $\dfrac{H}{2}[f(a)+f(b)]$ | $-\dfrac{H^3}{12}f''(\xi)$ |
| Simpson | $\dfrac{H}{6}[f(a)+4f(c)+f(b)]$ | $-\dfrac{H^5}{2880}f^{(4)}(\xi)$ |
| Three-eighths | $\dfrac{H}{8}[f(a)+3f(a+H/3)+3f(a+2H/3)+f(b)]$ | $-\dfrac{H^5}{6480}f^{(4)}(\xi)$ |

For an a priori absolute-error bound $B_m^{\mathrm{single}}$, replace the derivative in the remainder by
$$
M_r=\max_{x\in[a,b]}|f^{(r)}(x)|,\qquad r=1,2,4,
$$
and take the absolute value of the coefficient. Here $r=1$ is used for the left and right rectangles, $r=2$ for midpoint and trapezoidal rules, and $r=4$ for Simpson and three-eighths rules. Use a proven upper bound for $M_r$ if the exact maximum is not available.

#### Composite formulas

Construct a composite approximation $Q_m(h)$ by summing single-panel rules over the whole interval. Use panels of width $h$ for rectangle and trapezoidal rules, width $2h$ for Simpson, and width $3h$ for three-eighths. In every case, $n$ counts the subintervals of the base grid, not the number of panels. Shared nodes must receive the weights obtained by adding their panel contributions.

The table gives the a priori bound $|I-Q_m(h)|\le B_m(h)$, the expected global order $p$, and the algebraic degree of accuracy $d$.

| Rule | Grid restriction | $B_m(h)$ | $p$ | $d$ |
| --- | --- | --- | --- | --- |
| Left rectangle | $n\ge1$ | $\dfrac{Hh}{2}M_1$ | 1 | 0 |
| Right rectangle | $n\ge1$ | $\dfrac{Hh}{2}M_1$ | 1 | 0 |
| Midpoint | $n\ge1$ | $\dfrac{Hh^2}{24}M_2$ | 2 | 1 |
| Trapezoidal | $n\ge1$ | $\dfrac{Hh^2}{12}M_2$ | 2 | 1 |
| Simpson | $n\ge2$, even | $\dfrac{Hh^4}{180}M_4$ | 4 | 3 |
| Three-eighths | $n\ge3$, divisible by 3 | $\dfrac{Hh^4}{80}M_4$ | 4 | 3 |

The stated bounds require $f\in C^r([a,b])$ for the corresponding $r=1,2,4$. The orders refer to $h\to0$ for a fixed function and interval. They describe truncation error; roundoff and errors in function values must be considered separately. The same algebraic degrees apply to the single-panel formulas.
