# Практическое занятие №11.2. Составные квадратурные формулы

## Приближенное интегрирование составными формулами

Самостоятельно реализуйте составные квадратурные формулы, оцените их точность и исследуйте зависимость погрешностей от шага сетки и выбранного метода.

### Исходные данные и требования к реализации

Рассмотрим определенный интеграл
$$
I=\int_a^b f(x)\,dx,\quad H=b-a>0.
$$
Используйте равномерную сетку
$$
x_i=a+ih,\quad i=0,1,\ldots,n,\quad h=\frac{H}{n}.
$$
Здесь $n$ — число промежутков, а сетка содержит $n+1$ узел. Символ $H$ обозначает длину всего отрезка, $h$ — шаг сетки.

В качестве тестовых функций используйте $f(x)=\sin(x)$, $f(x)=\cos(x)$ и $f(x)=\exp(x)$ на $[a,b]=[0,2]$. Аналитически найдите их точные интегралы и необходимые ниже оценки производных. При сгущении сетки сохраняйте отрезок неизменным. Вес единичный: $w(x)\equiv1$.

- Реализуйте составные формулы левых, правых и средних прямоугольников, трапеций, Симпсона и «трех восьмых».
- Передавайте подынтегральную функцию, концы отрезка и $n$ как аргументы функций. Аналитические интегралы и производные являются справочными данными для проверки, а не входными данными квадратурного вычисления.
- Требуйте конечные концы отрезка $a<b$ и целое положительное $n$. Для составной формулы Симпсона нужно четное $n\ge2$; для составной формулы «трех восьмых» — $n\ge3$, кратное трем. Сообщайте о недопустимых данных, не изменяя $n$ автоматически.
- Вычисляйте узлы через целочисленные индексы, а не последовательным прибавлением $h$. При необходимости вычисляйте значения в серединах промежутков $x_i+h/2$.
- Можно повторно использовать собственные реализации локальных формул из работы 11.1. Включите используемые функции или модули в сдаваемые материалы, чтобы они работали в новом сеансе.
- Можно использовать библиотеки для массивов, таблиц и графиков. Квадратурные суммы реализуйте самостоятельно; готовые функции интегрирования допускаются только для дополнительной проверки. В обязательных экспериментах используйте аналитически найденные интегралы как эталонные значения.

### Краткая справка

Для составного приближения $Q_m(h)$, где $m$ указывает выбранную формулу, используйте
$$
R_m(h)=I-Q_m(h),\quad E_m(h)=|R_m(h)|,\quad
\delta_m(h)=\frac{E_m(h)}{|I|}\quad(I\ne0).
$$
При $I=0$ относительная погрешность не определена. В этом случае указывайте `--` и используйте абсолютную погрешность. Малое ненулевое значение $|I|$ также может делать относительную погрешность малоинформативной.

Для априорных оценок ниже определим
$$
M_r=\max_{x\in[a,b]}|f^{(r)}(x)|,\quad r=1,2,4.
$$
Если точный максимум неизвестен, используйте аналитически обоснованную верхнюю границу.

#### Составные формулы

Постройте составное приближение $Q_m(h)$ суммированием формул на отдельных частях всего отрезка. Для прямоугольников и трапеций используйте части ширины $h$, для Симпсона — $2h$, для «трех восьмых» — $3h$. Во всех случаях $n$ — число промежутков базовой сетки, а не число частей, на которых применяется отдельная формула. Веса общих узлов получаются сложением их вкладов из соседних частей.

В таблице приведены априорная оценка $|I-Q_m(h)|\le B_m(h)$, ожидаемый глобальный порядок $p$ и алгебраическая степень точности $d$.

| Формула | Ограничение на сетку | $B_m(h)$ | $p$ | $d$ |
| --- | --- | --- | --- | --- |
| Левых прямоугольников | $n\ge1$ | $\dfrac{Hh}{2}M_1$ | 1 | 0 |
| Правых прямоугольников | $n\ge1$ | $\dfrac{Hh}{2}M_1$ | 1 | 0 |
| Средних прямоугольников | $n\ge1$ | $\dfrac{Hh^2}{24}M_2$ | 2 | 1 |
| Трапеций | $n\ge1$ | $\dfrac{Hh^2}{12}M_2$ | 2 | 1 |
| Симпсона | $n\ge2$, четное | $\dfrac{Hh^4}{180}M_4$ | 4 | 3 |
| «Трех восьмых» | $n\ge3$, кратное трем | $\dfrac{Hh^4}{80}M_4$ | 4 | 3 |

Для этих оценок требуется $f\in C^r([a,b])$ при соответствующем $r=1,2,4$. Порядки относятся к $h\to0$ при фиксированных функции и отрезке. Они характеризуют погрешность аппроксимации; округление и погрешности значений функции нужно учитывать отдельно. Для формул на одном отрезке алгебраические степени точности те же.
