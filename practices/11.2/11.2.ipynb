### Input data and implementation requirements

$$

### Brief reference

For a composite approximation $Q_m(h)$, where $m$ identifies the rule, use
$$
R_m(h)=I-Q_m(h),\qquad E_m(h)=|R_m(h)|,\qquad
\delta_m(h)=\frac{E_m(h)}{|I|}\quad(I\ne0).
$$
If $I=0$, the relative error is undefined. Report `--` in this case and use the absolute error. A small nonzero $|I|$ can also make the relative error misleading.

For the a priori bounds below, define
$$
M_r=\max_{x\in[a,b]}|f^{(r)}(x)|,\qquad r=1,2,4.
$$
Use an analytically justified upper bound if the exact maximum is not available.

#### Composite formulas

Construct a composite approximation $Q_m(h)$ by summing single-panel rules over the whole interval. Use panels of width $h$ for rectangle and trapezoidal rules, width $2h$ for Simpson, and width $3h$ for three-eighths. In every case, $n$ counts the subintervals of the base grid, not the number of panels. Shared nodes must receive the weights obtained by adding their panel contributions.

The table gives the a priori bound $|I-Q_m(h)|\le B_m(h)$, the expected global order $p$, and the algebraic degree of accuracy $d$.

| Rule | Grid restriction | $B_m(h)$ | $p$ | $d$ |
| --- | --- | --- | --- | --- |
| Left rectangle | $n\ge1$ | $\dfrac{Hh}{2}M_1$ | 1 | 0 |
| Right rectangle | $n\ge1$ | $\dfrac{Hh}{2}M_1$ | 1 | 0 |
| Midpoint | $n\ge1$ | $\dfrac{Hh^2}{24}M_2$ | 2 | 1 |
| Trapezoidal | $n\ge1$ | $\dfrac{Hh^2}{12}M_2$ | 2 | 1 |
| Simpson | $n\ge2$, even | $\dfrac{Hh^4}{180}M_4$ | 4 | 3 |
| Three-eighths | $n\ge3$, divisible by 3 | $\dfrac{Hh^4}{80}M_4$ | 4 | 3 |

The stated bounds require $f\in C^r([a,b])$ for the corresponding $r=1,2,4$. The orders refer to $h\to0$ for a fixed function and interval. They describe truncation error; roundoff and errors in function values must be considered separately. The same algebraic degrees apply to the single-panel formulas.
