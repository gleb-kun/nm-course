## Finding the first and second derivatives of a tabulated function

Implement numerical differentiation independently, estimate its accuracy, and investigate how the errors depend on the grid spacing and the position of a node.

### Input data and implementation requirements

Consider a uniform grid on the closed interval $[a,b]$:
$$
x_i=a+ih,\quad i=0,1,\ldots,n,\quad h=\frac{b-a}{n}>0.
$$
Here $n\geqslant 3$ is the number of subintervals, so the table contains $n+1\geqslant 4$ nodes.

Use $f(x)=\sin(x)$, $f(x)=\cos(x)$, and $f(x)=\exp(x)$ as test functions. Find their exact first and second derivatives analytically to use as reference values.

For the experiments, use $[a,b]=[0,2]$. Keep this interval fixed when refining the grid.

- Write your own implementation of the second-order finite difference formulas. It must accept tabulated function values and the grid spacing, without using the analytical form of the function.
- Return one approximation to each derivative at every node, including both endpoints, using only values from the given table.
- Validate the input: the grid spacing must be finite and positive, and at least four tabulated values are required.
- Build grid nodes using integer indices rather than repeated addition of $h$.
- You may use libraries for arrays, tables, and plots. Do not replace your implementation with built-in numerical differentiation, automatic differentiation, or symbolic differentiation. Analytical derivatives are used only to check the results (for checking you can use, for example, `scipy.differentiate` or `sympy.diff`).

### Brief reference

The formulas, accuracy estimates, and effects of roundoff are covered in the [lecture on numerical differentiation](../../lectures/01.1/lecture01_eng.ipynb).

Use three-point centered formulas at interior nodes, three-point one-sided formulas for the first derivative at endpoints, and four-point one-sided formulas for the second derivative at endpoints. All these formulas have $O(h^2)$ truncation error for $f\in C^4([a,b])$. The lecture's introductory example uses first-order endpoint differences. This lab requires second-order endpoint formulas.

For derivatives of order $k=1.2$, use the following notations for absolute and relative errors:
$$
\Delta_k(x_i)=\left|f^{(k)}(x_i)-f^{(k)}_{\mathrm{ND}}(x_i)\right|,\quad
\delta_k(x_i)=\frac{\Delta_k(x_i)}{\left|f^{(k)}(x_i)\right|},
$$
where $f^{(k)}_{\mathrm{ND}}$ is the numerical approximation.
