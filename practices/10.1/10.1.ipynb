## Finding the derivatives of a tabulated function using numerical differentiation formulas

### Preparatory stage

You need to write a program that performs the following steps.

- Prints a table of $(n+1)$ values ​​of some function $f(x)$ at equally spaced nodes $x_i=a+ih$, $i=0,1,\ldots, n$ with a step of $h$.
- Use $f(x)=\sin(x)$, $f(x)=\cos(x)$ and $f(x)=\exp(x)$ as input functions.
- Input values:
    - $n + 1$ is the number of values ​​in the table,
    - $a$ is the beginning of the interval $[a, b)$,
    - $h>0$ are the input parameters of the problem.

```python
import math
def m_sin(x):
    return math.sin(x)

def m_cos(x):
    return math.cos(x)

def m_exp(x):
    return math.exp(x)

x_values = []
h = 0.2
a = 0

f_value_sin = []
f_value_cos = []
f_value_exp = []
for i in range(10):
    node = a + i * h
    x_values.append(node)
    f_value_sin.append(m_sin(node))
    f_value_cos.append(m_cos(node))
    f_value_exp.append(m_exp(node))

print(f_value_cos)
print(f_value_sin)
print(f_value_exp)
print(x_values)
```

### Solution of the problem of numerical differentiation

#### Result of section 1.1.1.

Let the function $f(x)$ be defined in a table—that is, its values ​​at the nodes are known:
$$
x_0, \, x_1, \, x_2, \, \ldots, \, x_n,\tag{1}
$$
and the corresponding function values:
$$
f(x_0), \, f(x_1), \, f(x_2), \, \ldots, \, f(x_n).\tag{2}
$$

It is required to find the values ​​of the first and second derivatives of $f(x)$ at these nodes.

#### Numerical Differentiation Formulas

To approximate the derivatives, simple second-order difference formulas are used,
i.e., with an error of order $O(h^2)$, where $h$ is the grid spacing:
$$
h = x_{i+1} - x_i = \text{const}.\tag{3}
$$

#### First Derivative

- For internal nodes, when $1 \leqslant i \leqslant n-1$, we use the central difference scheme:
$$
f'(x_i) \approx \frac{f(x_{i+1}) - f(x_{i-1})}{2h}.\tag{4}
$$

- For the extreme grid points, one-sided second-order formulas can be used:
$$
f'(x_0) \approx \frac{-3f(x_0) + 4f(x_1) - f(x_2)}{2h},\tag{5}
$$

$$
f'(x_n) \approx \frac{3f(x_n) - 4f(x_{n-1}) + f(x_{n-2})}{2h}.\tag{6}
$$

#### Second derivative

For internal nodes:
$$
f''(x_i) \approx \frac{f(x_{i+1}) - 2f(x_i) + f(x_{i-1})}{h^2}.\tag{7}
$$

For the outermost nodes, we can apply second-order approximation formulas:
$$
f''(x_0) \approx \frac{1}{h^2} \Big( f(x_0) - 5 f(x_1) + 4 f(x_2) - f(x_3) \Big),\tag{8}
$$
$$
f''(x_n) \approx \frac{1}{h^2} \Big( f(x_n) - 5 f(x_{n-1}) + 4 f(x_{n-2}) - f(x_{n-3}) \Big).\tag{9}
$$

### Calculation Errors

Each of the formulas given has an error of order $O(h^2)$,
meaning that the error decreases by approximately a factor of 4 when the step size $h$ is halved:
$$
\varepsilon \sim C h^2.\tag{10}
$$

#### Absolute Error

The absolute error indicates how much the approximate value differs from the true (analytical) value.

For the first derivative:
$$
\Delta_1(x_i) = \big| f'(x_i)_E - f'(x_i)_{\text{ND}} \big|,\tag{11}
$$

where
- $f'(x_i)_E=f'(x_i)$ is the true value of the derivative (e.g., found analytically),
- $f'(x_i)_{\text{ND}}\approx \widetilde f'(x_i)$ is the numerically calculated value of the derivative.

Similarly, for the second derivative:
$$
\Delta_2(x_i) = \big| f''(x_i)_E - f''(x_i)_{\text{ND}} \big|.\tag{12}
$$

#### Relative Error

For the first derivative:
$$
\delta_1(x_i) =
\frac{\big| f'(x_i)_E - f'(x_i)_{\text{ND}} \big|}{\big| f'(x_i)_E \big|}.\tag{13}
$$

For the second derivative:
$$
\delta_2(x_i) =
\frac{\big| f''(x_i)_E - f''(x_i)_{\text{ND}} \big|}{\big| f''(x_i)_E \big|}.\tag{14}
$$

```python
import numpy as np
import pandas as pd

# Number of data rows
n = 5

# Generate random numbers
x_i = np.linspace(0, 1, n)
f_x = np.random.uniform(-1, 1, n)
f1_num = np.random.uniform(-1, 1, n)
abs_err1 = np.random.uniform(0, 0.1, n)
rel_err1 = np.random.uniform(0, 0.05, n)
f2_num = np.random.uniform(-1, 1, n)
abs_err2 = np.random.uniform(0, 0.1, n)
rel_err2 = np.random.uniform(0, 0.05, n)

# Create a table
table = pd.DataFrame({ 
r"$x_i$": x_i, 
r"$f(x_i)$": f_x, 
r"$f'(x_i)_{\text{ND}}$": f1_num, 
r"$|f'(x_i)_E - f'(x_i)_{\text{ND}}|$": abs_err1, 
r"Relative error for $f'$": rel_err1, 
r"$f''(x_i)_{\text{ND}}$": f2_num, 
r"$|f''(x_i)_E - f''(x_i)_{\text{ND}}|$": abs_err2, 
r"Relative error for $f''$": rel_err2
})

# Configure the display of numbers
pd.options.display.float_format = "{: .5f}".format

print(table)
```
