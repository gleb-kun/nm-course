# Практическое занятие №10.1. Численное дифференцирование

## Нахождение первой и второй производных таблично заданной функции

Самостоятельно реализуйте численное дифференцирование, оцените его точность и исследуйте зависимость погрешностей от шага сетки и положения узла.

### Исходные данные и требования к реализации

Рассмотрим равномерную сетку на замкнутом отрезке $[a,b]$:
$$
x_i=a+ih,\quad i=0,1,\ldots,n,\quad h=\frac{b-a}{n}>0.
$$
Здесь $n\geqslant 3$ — число промежутков, поэтому таблица содержит $n+1\geqslant 4$ узла.

В качестве тестовых функций используйте $f(x)=\sin(x)$, $f(x)=\cos(x)$ и $f(x)=\exp(x)$. Найдите их точные первую и вторую производные аналитически для проверки результатов.

- Самостоятельно реализуйте разностные формулы второго порядка. Реализация должна принимать табличные значения функции и шаг сетки, не используя аналитический вид функции.
- Вычисляйте по одному приближению каждой производной во всех узлах, включая оба конца отрезка, используя только значения из заданной таблицы.
- Проверяйте входные данные: шаг сетки должен быть конечным и положительным, а таблица должна содержать не менее четырех значений.
- Вычисляйте узлы сетки через целочисленные индексы, а не последовательным прибавлением $h$.
- Можно использовать библиотеки для работы с массивами, таблицами и графиками. Не заменяйте собственную реализацию готовыми средствами численного, автоматического или символьного дифференцирования. Аналитические производные используются только для проверки результатов.

В обязательных экспериментах используйте $[a,b]=[0,2]$. При сгущении сетки сохраняйте этот отрезок неизменным.

### Краткая справка

Формулы, оценки точности и влияние округления разобраны в [лекции по численному дифференцированию](../../lectures/01.1/lecture01_rus.ipynb).

Во внутренних узлах используйте трехточечные центральные формулы, на границах — трехточечные односторонние формулы для первой производной и четырехточечные односторонние формулы для второй. Все эти формулы имеют погрешность аппроксимации $O(h^2)$ при $f\in C^4([a,b])$. Во вводном примере лекции на границах применяются разности первого порядка; в этой работе требуются граничные формулы второго порядка.

Для производных порядка $k=1,2$ используйте обозначения
$$
\Delta_k(x_i)=\left|f^{(k)}(x_i)-f^{(k)}_{\mathrm{ND}}(x_i)\right|,\quad
\delta_k(x_i)=\frac{\Delta_k(x_i)}{\left|f^{(k)}(x_i)\right|}.
$$
Здесь $f^{(k)}_{\mathrm{ND}}$ — численное приближение. Если точная производная равна нулю, отмечайте относительную погрешность как `--` и указывайте абсолютную погрешность.

### Обязательные задания

#### 1. Реализация и проверка формул

Реализуйте формулы первой и второй производных для внутренних и граничных узлов. Перед исследованием тестовых функций выполните следующие проверки:

- Проверьте постоянный, линейный и квадратичный полиномы на нескольких равномерных сетках, включая сетку ровно из четырех узлов. Обе производные должны совпадать с точными значениями с точностью до округления при вычислениях с плавающей точкой.
- Проверьте кубический полином: формулы второй производной должны воспроизводить его точную вторую производную с точностью до округления, в том числе на границах. Формулы первой производной не обязаны быть точными для кубического полинома.
- Проверьте, что каждый выходной массив содержит ровно $n+1$ элементов, а первый и последний элементы соответствуют концам отрезка.
- Проверьте, что недопустимый шаг сетки и недостаточное число табличных значений приводят к понятному сообщению об ошибке.

Опишите проверки и приведите полученные погрешности. Для оценки расхождений из-за округления используйте численный допуск, а не строгое равенство. Затем составьте требуемую таблицу для каждой из трех тестовых функций.

#### 2. Исследование сходимости

Для каждой тестовой функции на $[0,2]$ используйте $n=10,20,40,80,160,320$, каждый раз пересчитывая шаг $h=2/n$. Для производных порядка $k=1,2$ вычислите максимальную абсолютную погрешность:
$$
E_k(h)=\max_{0\leqslant i\leqslant n}\Delta_k(x_i).
$$
Для последовательных сеток оцените наблюдаемый порядок:
$$
p_k(h)=\log_2\frac{E_k(h)}{E_k(h/2)}.
$$

- Для каждой функции приведите таблицу со столбцами $h$, $E_1(h)$, $p_1(h)$, $E_2(h)$ и $p_2(h)$. Относите $p_k(h)$ к паре шагов $h$ и $h/2$; для самой мелкой сетки оставьте его неопределенным.
- Постройте графики $E_1(h)$ и $E_2(h)$ в зависимости от $h$ в логарифмическом масштабе по обеим осям и добавьте линию сравнения, пропорциональную $h^2$.
- Объясните, приближаются ли измеренные порядки к теоретическому значению, и отметьте отклонения от ожидаемой закономерности.

Если одна из погрешностей в оценке порядка равна нулю, отметьте порядок как неопределенный, не выполняя деление на ноль или вычисление логарифма нуля. Не интерпретируйте результаты, в которых преобладает округление, как сходимость погрешности аппроксимации.

#### 3. Сравнение граничных и внутренних узлов

Для обеих производных и всех сеток из задания 2 отдельно вычислите:
$$
E_{k,\mathrm{end}}(h)=\max\{\Delta_k(x_0),\Delta_k(x_n)\},\quad
E_{k,\mathrm{int}}(h)=\max_{1\leqslant i\leqslant n-1}\Delta_k(x_i),\quad k=1,2.
$$

- Сравните эти погрешности с помощью таблиц или графиков.
- Укажите, какие узлы определяют максимальную погрешность для каждой функции и производной.
- Объясните, почему формулы с одинаковым порядком точности могут давать разные по величине погрешности; обоснуйте ответ коэффициентами, расположением узлов формулы или разложением Тейлора.
- Обсудите, почему относительная погрешность может быть неопределенной или вводить в заблуждение вблизи нуля точной производной, даже если абсолютная погрешность мала. Приведите примеры для тестовых функций.
