## Сравнение одношаговых и многошаговых методов

### Исходные данные и требования к реализации

Пусть дана задача Коши
$$
y'=\cos(1.75x+y)+1.25(x-y),\quad y(0)=0,\quad 0\leqslant x\leqslant1. \tag{1}
$$
Используем равномерную сетку $x_i=ih$, $i=0,\ldots,N$, где $h=1/N$. Получите таблицу решений с шагом $h=0.1$, затем повторите вычисления с шагами $h=0.05$ и $h=0.025$.

Самостоятельно реализуйте классический метод Рунге–Кутты четвертого порядка и оба метода Адамса пятого порядка, Адамса–Башфорта и Адамса–Мултона. На каждой сетке одинаковые стартовые значения $y_0,\ldots,y_4$ для обоих методов Адамса получите методом Рунге–Кутты с шагом этой сетки. Для сравнения вычислите эталонное численное решение $y_{\mathrm{ref}}$ библиотечным методом. На каждой сетке сравнивайте значения всех методов и эталона в одних и тех же узлах.

### Краткая справка

Метод Рунге–Кутты является одношаговым. Его классическая формула четвертого порядка использует четыре стадии. В обозначениях, где $k_j$ уже содержит множитель $h$,
$$
\begin{aligned}
k_1&=h f(x_m,y_m),\quad k_2=h f(x_m+h/2,y_m+k_1/2),\\
k_3&=h f(x_m+h/2,y_m+k_2/2),\quad k_4=h f(x_m+h,y_m+k_3),\\
y_{m+1}&=y_m+\frac{k_1+2k_2+2k_3+k_4}{6}.
\end{aligned} \tag{2}
$$
При достаточной гладкости данных и липшицевости $f$ по $y$ локальная погрешность шага из точного значения равна $O(h^5)$, а глобальная погрешность на фиксированном отрезке равна $O(h^4)$. Для $y'=f(x)$ метод дает формулу Симпсона на одном шаге.

#### Явный метод Адамса (экстраполяционный)

Для приближения интеграла на отрезке $[x_m,x_{m+1}]$ метод Адамса–Башфорта использует интерполяционный многочлен правой части в пяти известных узлах $x_{m-4},\ldots,x_m$. На отрезке шага этот многочлен используют для экстраполяции. Обозначим $q_j=h f(x_j,y_j)$. Формула пятого порядка имеет вид
$$
y_{m+1}=y_m+\frac{1901q_m-2774q_{m-1}+2616q_{m-2}-1274q_{m-3}+251q_{m-4}}{720}. \tag{3}
$$
Правая часть (3) зависит только от уже вычисленных значений, поэтому метод явный. Применяйте (3) для $m=4,\ldots,N-1$. Первое новое значение вычисляется в узле $x_5$.

#### Неявный метод Адамса (интерполяционный)

В методе Адамса–Мултона новый узел $x_{m+1}$ входит в число пяти узлов интерполяции $x_{m-3},\ldots,x_{m+1}$. Отрезок интегрирования $[x_m,x_{m+1}]$ лежит в пределах этих узлов. Формула пятого порядка имеет вид
$$
y_{m+1}=y_m+\frac{251q_{m+1}+646q_m-264q_{m-1}+106q_{m-2}-19q_{m-3}}{720}. \tag{4}
$$
Здесь $q_{m+1}=h f(x_{m+1},y_{m+1})$ зависит от неизвестного $y_{m+1}$. Поэтому метод (4) неявный. В этой работе применяйте (4) с $m=4$, чтобы сравнить оба метода Адамса с одинаковыми стартовыми значениями.

Для гладкой задачи с правой частью, липшицевой по $y$, и погрешностью стартовых значений $O(h^5)$ глобальная погрешность обоих методов Адамса на фиксированном отрезке равна $O(h^5)$ при точном решении уравнений на каждом шаге. Четыре шага классического метода Рунге–Кутты четвертого порядка из точного начального значения удовлетворяют этому требованию при $h\to0$.
