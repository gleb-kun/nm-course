## Сравнение одношаговых и многошаговых методов

Найдите приближенные решения задачи Коши первого порядка методом Рунге–Кутты четвертого порядка и экстраполяционным методом Адамса пятого порядка. Исследуйте влияние шага и стартовых значений.

### Исходные данные и требования к реализации

Пусть дана задача Коши
$$
y'=\cos(1.75x+y)+1.25(x-y),\quad y(0)=0,\quad 0\leqslant x\leqslant1. \tag{1}
$$
Используем равномерную сетку $x_i=ih$, $i=0,\ldots,N$, где $h=1/N$. Получите таблицу решений с шагом $h=0.1$, затем повторите вычисления с шагами $h=0.05$ и $h=0.025$.

Самостоятельно реализуйте классический метод Рунге–Кутты четвертого порядка (РК4) и метод Адамса–Башфорта пятого порядка (АБ5). На каждой сетке начальные приближения для АБ5 получите методом РК4 с шагом этой сетки. Для сравнения вычислите эталонное численное решение $y_{\mathrm{ref}}$ библиотечным методом. На каждой сетке сравнивайте значения всех методов и эталона в одних и тех же узлах.

### Краткая справка

Метод Рунге–Кутты является одношаговым. Его классическая формула четвертого порядка использует четыре стадии. В обозначениях, где $k_j$ уже содержит множитель $h$,
$$
y_{m+1}=y_m+\frac{k_1+2k_2+2k_3+k_4}{6}. \tag{2}
$$
Формулы стадий приведены в [лекции 07.2](../../lectures/07.2/lecture07_2_rus.ipynb) в обозначениях для наклонов $k_j/h$. При достаточной гладкости данных и липшицевости $f$ по $y$ локальная погрешность шага из точного значения равна $O(h^5)$, а глобальная погрешность на фиксированном отрезке равна $O(h^4)$. Для $y'=f(x)$ метод дает формулу Симпсона на одном шаге.

#### Экстраполяционный метод Адамса

Для приближения интеграла на одном шаге экстраполяционный метод Адамса использует интерполяционный многочлен правой части в предыдущих узлах. Для АБ5 обозначим $q_j=h f(x_j,y_j)$. Рабочая формула АБ5 имеет вид
$$
y_{m+1}=y_m+\frac{1901q_m-2774q_{m-1}+2616q_{m-2}-1274q_{m-3}+251q_{m-4}}{720},
\quad m=4,\ldots,N-1. \tag{3}
$$
Таким образом, начало таблицы составляют $y_0,\ldots,y_4$. Первое значение по (3) вычисляется в узле $x_5$. Для гладкой задачи с липшицевой правой частью и погрешностью стартовых значений $O(h^5)$ глобальная погрешность на фиксированном отрезке равна $O(h^5)$. Четыре шага РК4 из точного начального значения удовлетворяют этому требованию при $h\to0$.

### Дополнительное задание. Интерполяционный метод Адамса

В правой части интерполяционного метода Адамса есть $q_{m+1}=h f(x_{m+1},y_{m+1})$. Формула пятого порядка имеет вид
$$
y_{m+1}=y_m+\frac{251q_{m+1}+646q_m-264q_{m-1}+106q_{m-2}-19q_{m-3}}{720}. \tag{4}
$$
Для сравнения запустите явный и неявный методы с одинаковых значений РК4 и применяйте (4) с $m=4$. Для неявного уравнения можно использовать простые итерации, взяв в качестве начального приближения прогноз Адамса–Башфорта по текущим значениям. Проверьте, что в задаче (1) при выбранных шагах итерация является сжатием. Укажите допуск и проверьте невязку неявного уравнения на каждом шаге. Сравните два метода Адамса на исходной сетке и на одной более мелкой.
