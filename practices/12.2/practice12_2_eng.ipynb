For comparison, compute a numerical reference $y_{\mathrm{ref}}$ using a library solver. Use the same nodes for every method and the reference on a given grid.

### Brief reference

The Runge–Kutta method is one-step. At each step its classical fourth-order form uses four stages. With $k_j$ including the factor $h$,
$$
\begin{aligned}
k_1&=h f(x_m,y_m),\quad k_2=h f(x_m+h/2,y_m+k_1/2),\\
k_3&=h f(x_m+h/2,y_m+k_2/2),\quad k_4=h f(x_m+h,y_m+k_3),\\
y_{m+1}&=y_m+\frac{k_1+2k_2+2k_3+k_4}{6}.
\end{aligned} \tag{2}
$$
For sufficiently smooth data with $f$ Lipschitz in $y$, the local error per step started from the exact value is $O(h^5)$ and the global error on a fixed interval is $O(h^4)$. When $y'=f(x)$, the method gives Simpson's rule on one step.

#### Explicit Adams method (extrapolation)

To approximate the integral over $[x_m,x_{m+1}]$, the Adams–Bashforth method uses an interpolation polynomial of the right-hand side at the five known nodes $x_{m-4},\ldots,x_m$. On the step interval it extrapolates from those nodes. Write $q_j=h f(x_j,y_j)$. The fifth-order formula is
$$
y_{m+1}=y_m+\frac{1901q_m-2774q_{m-1}+2616q_{m-2}-1274q_{m-3}+251q_{m-4}}{720}. \tag{3}
$$
The right-hand side of (3) depends only on values already computed, so the update is explicit. Apply (3) for $m=4,\ldots,N-1$. The first new value is at $x_5$.

#### Implicit Adams method (interpolation)

The Adams–Moulton method includes the new node $x_{m+1}$ among the five interpolation nodes $x_{m-3},\ldots,x_{m+1}$. The integration interval $[x_m,x_{m+1}]$ lies within their range. The fifth-order formula is
$$
y_{m+1}=y_m+\frac{251q_{m+1}+646q_m-264q_{m-1}+106q_{m-2}-19q_{m-3}}{720}. \tag{4}
$$
Here $q_{m+1}=h f(x_{m+1},y_{m+1})$ depends on the unknown $y_{m+1}$. Thus (4) is implicit. In this work, apply it from $m=4$ to compare both Adams methods with the same starting values.

For a smooth problem with $f$ Lipschitz in $y$ and starting errors $O(h^5)$, both Adams methods have global errors $O(h^5)$ on a fixed interval when their step equations are solved exactly. Four classical fourth-order Runge–Kutta steps from the exact initial value meet this starting-error requirement as $h\to0$.
