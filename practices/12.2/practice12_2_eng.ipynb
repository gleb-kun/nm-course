### Brief reference

The Runge–Kutta method is one-step. At each step its classical fourth-order form uses four stages. The update, in the convention where $k_j$ includes $h$, is
$$
y_{m+1}=y_m+\frac{k_1+2k_2+2k_3+k_4}{6}. \tag{2}
$$
The stage formulas are given in [Lecture 07.2](../../lectures/07.2/lecture07_2_eng.ipynb) in terms of slopes $k_j/h$. For sufficiently smooth data with $f$ Lipschitz in $y$, the local error per step started from the exact value is $O(h^5)$ and the global error on a fixed interval is $O(h^4)$. When $y'=f(x)$, the method gives Simpson's rule on one step.

#### Adams extrapolation

To approximate the integral over one step, the Adams extrapolation method uses an interpolation polynomial of the right-hand side at previous nodes. For AB5 write $q_j=h f(x_j,y_j)$. The AB5 update is
$$
y_{m+1}=y_m+\frac{1901q_m-2774q_{m-1}+2616q_{m-2}-1274q_{m-3}+251q_{m-4}}{720},
\quad m=4,\ldots,N-1. \tag{3}
$$
Thus the beginning of the table consists of $y_0,\ldots,y_4$. The first value calculated by (3) is at $x_5$. For a smooth problem with a Lipschitz right-hand side and starting errors $O(h^5)$, the global error is $O(h^5)$ on a fixed interval. Four RK4 steps starting from the exact initial value meet the starting-error requirement as $h\to0$.
