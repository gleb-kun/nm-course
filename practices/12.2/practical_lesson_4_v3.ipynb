# Numerical Solution of the Cauchy Problem for a First-Order Ordinary Differential Equation

## Problem Statement

Consider the Cauchy problem for the ordinary differential equation

$$
y' = f(x, y), \quad y(x_0) = y_0.
\tag{1}
$$

We will assume that this problem has a unique solution on the interval $[x_0, b]$.
Rewrite (1) in integral form:

$$
y(x) = y_0 + \int_{x_0}^{x} f(t, y(t))\, dt.
\tag{2}
$$

Numerical methods for solving this equation aim to approximate the values of the hypothetical solution $y(x)$ at points  
$x_1, x_2, \ldots, x_N \in [x_0, b]$.

For simplicity, we assume that the points (nodes) $x_1, \ldots, x_N$ are equally spaced, i.e.  
$x_k = x_0 + kh$, where $h = (b - x_0)/N$.

In all considered methods for solving problem (1), the values at the nodes are computed sequentially.  
That is, assuming that  
$y_1 \approx y(x_1), \ldots, y_m \approx y(x_m)$  
are already known, we then compute  
$y_{m+1} \approx y(x_{m+1})$.

## Euler’s Method and the Improved Euler Method

The simplest way to obtain a numerical solution of equation (2) is to compute the integral on the right-hand side using some quadrature formula.  
In this case, to compute $y_{m+1}$ one can use only the value $y_m$.  
Note that such methods for solving a differential equation produce an error of the same order as the corresponding quadrature formulas.

### Euler’s Method

Apply the left rectangle rule:

$$
y_{m+1} = y_m + \int_{x_m}^{x_m + h} f(t, y(t))\, dt = y_m + h f(x_m, y_m) + O(h^2),\tag{f1}
$$

i.e., the computational formula of the Euler method is

$$
y_{m+1} = y_m + h f(x_m, y_m), \quad m = 0,\, 1, \ldots, N-1.
\tag{3}
$$

The local truncation error per step is $O(h^2)$ (assuming all input data are exact).  
On the whole interval $[x_0, b]$, the global error is no better than $O(h)$, but in practice it may be significantly worse due to the nonlinear growth of error during integration.

### Improved Euler Method

To compute the integral on the right-hand side of (2), apply the midpoint rectangle rule.  
To do this, introduce an additional point halfway between $x_m$ and $x_{m+1}$.  
Denote it by $x_{m+\tfrac12} = x_m + \tfrac{h}{2}$.

At this point, compute an approximate value of the solution using the standard Euler method, i.e., using the left rectangle rule:

$$
y_{m+\tfrac12} = y_m + \frac{h}{2} f(x_m, y_m).
\tag{f2}
$$

Now compute the value at $x_{m+1}$ using the midpoint rectangle rule:

$$
\begin{gathered}
y_{m+1} = y_m + \int_{x_m}^{x_m + h} f(t, y(t))\, dt 
\approx y_m + h f\!\left(x_{m+\tfrac12},\, y_{m+\tfrac12}\right)
=\\[4pt]
= y_m + h f\!\left(x_m + \frac{h}{2},\, y_m + \frac{h}{2} f(x_m, y_m)\right).
\end{gathered}
\tag{4}
$$

The ordinary Euler method can also be refined in another way.  
Assume that the value $y_{m+1}$ is first computed using the left rectangle formula, i.e., as in formula (3), and then the corresponding integral is computed using the trapezoidal rule.

Thus, let

$$
\widetilde{y}_{m+1} = y_m + h f(x_m, y_m).
\tag{f3}
$$

Then

$$
y_{m+1} = y_m + \int_{x_m}^{x_m + h} f(t, y(t))\, dt 
\approx y_m + \frac{h}{2} \left( f(x_m, y_m) + f\!\left(x_{m+1}, \widetilde{y}_{m+1}\right) \right).
\tag{5}
$$

In conclusion, note that both proposed modifications of the Euler method yield an error no better than $O(h^2)$, but the actual error may be significantly worse due to the nonlinear growth of error during integration, just as in the previous section.

## Runge–Kutta Method of Fourth Order

The computational formulas of the method are:

$$
y_{m+1} = y_m + \frac{1}{6}\left(k_1 + 2k_2 + 2k_3 + k_4\right),
\tag{6}
$$

where

$$
k_1 = h f(x_m, y_m),
$$

$$
k_2 = h f\!\left(x_m + \frac{h}{2},\, y_m + \frac{k_1}{2}\right),
$$

$$
k_3 = h f\!\left(x_m + \frac{h}{2},\, y_m + \frac{k_2}{2}\right),
$$

$$
k_4 = h f\!\left(x_m + h,\, y_m + k_3\right),
\qquad m = 0, 1, \ldots, N-1.
\tag{7}
$$

It turns out that if $y_m = y(x_m)$ (i.e., the equality is exact), then  
$y_{m+1} - y(x_{m+1}) = O(h^5)$.

Over the entire interval, the global error of the method is no better than $O(h^4)$ for the reasons mentioned in §1.2.1 (Euler’s Method).

If the equation has the form $y' = f(x)$, then $k_2 = k_3$, and one can see that the Runge–Kutta formula results from applying Simpson’s rule, or, more generally, a generalized Simpson rule.

**Remark 1.** All methods considered above are one-step methods, i.e., to obtain the solution at the next point, only the solution at the previous point is used.  
In multistep methods, the step size may vary.


### Runge’s Rule (Double Computation) for Practical Error Estimation. Richardson Extrapolation

Assume that the method used to compute the values of the solution of problem (1) is fixed and has accuracy order $s$.  
This means that $y(x_i) - y_i = O(h^s)$ for $i = 1, 2, \ldots, N$.  
Here we use the notation introduced earlier.

Compute the value at the point $x_m$ using step sizes $h$ and $h/2$.  
Denote the obtained values by $y_m^{(h)}$ and $y_m^{(h/2)}$, respectively.  
Then the leading term of the error is given by

$$
R_m^{(h)} = \frac{y_m^{(h/2)} - y_m^{(h)}}{2^s - 1}.
\tag{f4}
$$

For sufficiently small $h$ one can state that if $\lvert R_m^{(h)} \rvert < \varepsilon$,  
then $\lvert y(x_m) - y_m^{(h/2)} \rvert < \varepsilon$.

Richardson extrapolation consists in refining the value at the point $x_m$ using the formula

$$
\hat{y}_m = y_m^{(h/2)} + R_m^{(h)}.
\tag{8}
$$

Note that as a result of refinement using formula (8), one obtains a method with a higher order of accuracy than the original one.

## Adams Methods

Earlier we noted that the nodes $x_1, \ldots, x_N$ are assumed to be equally spaced for convenience.  
In all previously discussed methods this was indeed only a convenience, since all of those methods were one-step methods, i.e., the value $y_{m+1}$ was computed solely from $y_m$.

In particular, all previously considered methods allow a variable step size of the argument $(x_{m+1} - x_m = h_m)$.

Now we fix the step $h = (b - x_0)/N$ and assume that all nodes $x_i = x_0 + ih$, $i = 0, 1, \ldots, N$, are equally spaced.

When computing $y_{m+1}$, the values of the solution at the previous $k + 1$ nodes  
$y_{m-k}, \ldots, y_m$  
will be used.

Assume that approximate values of $y(x)$ are known at points  
$x_0, x_1, \ldots, x_m$,  
i.e.  
$y_i \approx y(x_i)$, $i = 0, 1, \ldots, m$,  
with $k \le m < N$  
(these values may have been obtained by one of the methods discussed earlier).  
From this point on, we assume $m \ge k$.

### Adams Extrapolation Method

We begin, as usual, with formula (2):

$$
y_{m+1} = y_m + \int_{x_m}^{x_m + h} f(x, y(x))\, dx.
\tag{f5}
$$

Replace the integrand $f(x, y(x))$ with an interpolation polynomial constructed using the nodes  
$x_{m-k}, \ldots, x_m$.

Since the approximate value $y_{m+1}$ is obtained at a point lying outside the interval containing all interpolation nodes, the method is called an **extrapolation** method.

Depending on the form of the polynomial, various formulas of the method are obtained.  
First, assume that the function $f(x, y(x))$ is replaced with a Newton interpolation polynomial for the end of the table.

Recall that the values $y_{m-k}, \ldots, y_m$ are considered known.  
Let us define  
$q_j = h\, f(x_j, y_j)$.

The polynomial has the form

$$
P_k(x_m + th) = q_m + t \Delta q_{m-1} + \ldots +
\frac{t(t+1)\cdots(t+k-1)}{k!}\, \Delta^k q_{m-k},
\tag{9}
$$

where finite differences are computed by the rule

$$
\Delta^j q_s = \Delta^{\,j-1} q_{s+1} - \Delta^{\,j-1} q_s.
\tag{10}
$$

Then

$$
\int_{x_m}^{x_m + h} f(x, y(x))\, dx
\approx
\int_0^1 \sum_{j=0}^{k} \frac{t(t+1)\cdots(t+j-1)}{j!}\, \Delta^j q_{m-j}\, dt.
\tag{11}
$$

To simplify the formula, it is convenient to introduce the notation

$$
a_j = \frac{1}{j!} \int_0^1 t(t+1)\cdots(t+j-1)\, dt.
\tag{12}
$$

Thus we obtain the computational formula

$$
y_{m+1} = y_m + \sum_{j=0}^{k} a_j\, \Delta^j q_{m-j}.
\tag{13}
$$

Formula (13) can be applied starting from $m = k$ for  
$m = k, k+1, \ldots, N-1$.

If the solution $y(x)$ is a polynomial of degree no greater than $k+1$, then the Adams extrapolation method gives the exact value of the solution  
(assuming that all computations are performed exactly).

The local truncation error per step is $O(h^{k+2})$, while the global error over the entire interval is no better than $O(h^{k+1})$  
(see §1.2.1, Euler’s Method).

For $k = 4$ we obtain the formula

$$
y_{m+1}
= y_m
+ q_m
+ \frac{1}{2}\,\Delta q_{m-1}
+ \frac{5}{12}\,\Delta^2 q_{m-2}
+ \frac{3}{8}\,\Delta^3 q_{m-3}
+ \frac{251}{720}\,\Delta^4 q_{m-4}.
\tag{14}
$$

The beginning of the table is the part whose cell values must be known  
in order to apply the Adams extrapolation method.

The solution values at the initial points of the table should be computed  
using an appropriate one-step method.

Advantages of the Adams method compared with the Runge–Kutta method:

- efficiency;
- transparent control — the last finite differences allow one to judge the accuracy of the result.

The disadvantage of the Adams method compared with Runge–Kutta  
is its multistep nature, i.e., the fact that the solution at the next point depends on the solution at several previous points,  
and the nodes must be equally spaced.

By using the Lagrange form of the interpolation polynomial or by expressing the finite differences  
in (14) in terms of function values, one can obtain a difference-free formula  
for the Adams extrapolation method:

$$
y_{m+1} \approx y_m + \sum_{j=0}^{k} b_{k j}\, q_{m-j},
\tag{15}
$$

where

$$
b_{k j}
=
\frac{(-1)^j}{j!\,(k-j)!}
\int_{0}^{1}
\frac{t(t+1)\cdots(t+k)}{t + j}\, dt,
\qquad j = 0, 1, 2, \ldots, k.
\tag{16}
$$

The numbers $b_{k j}$ do not depend on $m$ or on $h$, but they do depend on the order of the method.

The computational formulas of the difference-free Adams extrapolation method of various orders  
are given below in Table 2.

Note that the algorithm for computing the solution using the difference-free formula  
is simpler to implement than the finite-difference version,  
but the transparent error control is absent.

Recall that these methods have order $k+1$.

### Adams Interpolation Method

Let  
$h = (b - x_0)/N$  
and  
$x_i = x_0 + ih,\quad i = 0, 1, \ldots, N.$

Assume that approximate values of $y(x)$ are known at the points  
$x_0, x_1, \ldots, x_m$,  
i.e.  
$y(x_i) \approx y_i,\; i = 0, 1, \ldots, m,\; k \le m < N.$

Denote  
$q_i = h f(x_i, y_i)$.

Approximate the function $f(x, y(x))$ in

$$
y_{m+1} = y_m + \int_{x_m}^{x_{m+1}} f(x, y(x))\, dx
\tag{f5}
$$

by a Newton interpolation polynomial of degree $k$ for the end of the table, using the nodes  
$x_{m+1}, x_m, \ldots, x_{m+1-k}$.  
Integrating it, we obtain the computational formula of the method:

$$
y_{m+1} = y_m + \sum_{j=0}^{k} a_j^*\, \Delta^j q_{m+1-j},
\tag{17}
$$

where

$$
a_j^* = \frac{1}{j!} \int_{-1}^{0} t(t+1)\cdots(t+j-1)\, dt.
\tag{18}
$$

As we can see, the right-hand side of (17) contains  
$q_{m+1} = h f(x_{m+1}, y_{m+1})$,  
i.e., formula (17) is **an equation with respect to $y_{m+1}$**.

The Adams interpolation method is an *implicit method*.  
It is recommended to apply an iterative procedure to equation (17).

As the initial approximation, one may take $y_{m+1}$ computed by the extrapolation method;  
let us denote it by $y_{m+1}^{(0)}$.

Compute

$$
q_{m+1}^{(0)} = h\, f(x_{m+1}, y_{m+1}^{(0)}), \qquad
\Delta q_m^{(0)} = q_{m+1}^{(0)} - q_m,
\tag{f6}
$$

$$
\Delta^2 q_{m-1}^{(0)},\; \ldots,\; \Delta^k q_{m+1-k}^{(0)}.
\tag{f7}
$$

Using these values, compute $y_{m+1}^{(1)}$ from formula (17).

Check the condition

$$
|y_{m+1}^{(1)} - y_{m+1}^{(0)}| < \varepsilon,
\tag{f8}
$$

where $\varepsilon$ is the prescribed accuracy; all solutions at previous points must be computed with this accuracy.

If the condition is not satisfied, recompute the iteration until it is fulfilled.

Formula (17) can be applied for  
$m = k,\, k+1, \ldots, N - 1$.

If the solution $y(x)$ is a polynomial of degree no greater than $k+1$, then the Adams interpolation method  
yields the exact value of the solution.

The local truncation error per step is $O(h^{k+2})$, while the global error over the entire interval is no better than $O(h^{k+1})$  
(see 1.2.1 Euler’s Method).

For $k = 4$ we obtain the formula:

$$
\begin{aligned}
y_{m+1}
=\, y_m + q_{m+1}
&- \frac12\, \Delta q_m
- \frac{1}{12}\, \Delta^2 q_{m-1}
- \frac{1}{24}\, \Delta^3 q_{m-2} \\
&- \frac{19}{720}\, \Delta^4 q_{m-3}.
\end{aligned}
\tag{19}
$$

By using the Lagrange form of the interpolation polynomial,  
or by expressing the finite differences in (19) in terms of function values,  
one can obtain a difference-free formula of the Adams interpolation method:

$$
y_{m+1}
= y_m
+ \sum_{j=-1}^{k-1} b_{k j}^*\, q_{m-j},
\tag{20}
$$

where

$$
\begin{gathered}
b_{k j}^*
=
\frac{(-1)^{\,j+1}}{(j+1)!\,(k-1-j)!}
\int_{0}^{1}
\frac{(t-1)t(t+1)\cdots(t+k-1)}{t + j}\, dt,\\
\qquad j = -1,\, 0,\, 1,\, 2, \ldots, k-1.
\end{gathered}
\tag{21}
$$

The numbers $b_{k j}^*$ do not depend on $m$ or on $h$.

Below are the computational formulas of the difference-free Adams interpolation method  
for $k = 0, 1, 2, 3, 4$ in **Table 3**.

|:---:|---------------------------|
| 0 | $y_{m+1} = y_m + q_{m+1}$ |

| 2 | $y_{m+1} = y_m + \frac{1}{12} (5 q_{m+1} + 8 q_m - q_{m-1})$ |
