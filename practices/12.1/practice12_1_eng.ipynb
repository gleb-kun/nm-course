### Brief reference

Let us write (1) in its integral form
$$
y(x)=y_0+\int_{x_0}^{x}f(t,y(t))\,dt. \tag{2}
$$

One of the simplest approaches to solving the Cauchy problem is based on Taylor's formula. For $y\in C^{p+1}$ on the step interval it gives
$$
\begin{aligned}
y(x+h)&=y(x)+\sum_{j=1}^p\frac{y^{(j)}(x)}{j!}h^j+R_{p+1}(x,h),\\
R_{p+1}(x,h)&=\frac{y^{(p+1)}(\xi)}{(p+1)!}h^{p+1},\quad \xi\in(x,x+h).
\end{aligned}\tag{3}
$$
Higher derivatives along a solution can be written as $y^{(k+1)}(x)=(D^k f)(x,y(x))$, where $D=\partial_x+f(x,y)\partial_y$. In particular,
$$
y''(x)=f_x(x,y(x))+f_y(x,y(x))f(x,y(x)). \tag{4}
$$
Dropping the remainder in (3) and evaluating the terms through $h^2$ at the current approximation gives the second-order Taylor method.

Euler's method is obtained by keeping only the terms through $h$. Geometrically, one Euler step follows the tangent to the integral curve through $(x_i,y_i)$. The improved Euler method (explicit midpoint) uses a half-step Euler approximation to the midpoint state and the slope at that midpoint for the full step. See the formulas and Euler diagrams in [Lecture 07.2](../../lectures/07.2/lecture07_2_eng.ipynb).

#### Error and step halving

Let $y_i^h$ be the numerical value at $x_i$ on the grid with step $h$. Denote the maximum absolute error on that grid by $E(h)=\max_i |y(x_i)-y_i^h|$.

At a node common to grids with steps $h$ and $h/2$, use Runge's estimate and Richardson extrapolation
$$
R_i=\frac{y_{2i}^{h/2}-y_i^h}{2^p-1},\quad
y_i^{\mathrm{ext}}=y_{2i}^{h/2}+R_i, \tag{5}
$$
where $p$ is the global order of the method. If the signed error has a leading term proportional to $h^p$ at the common node, $R_i$ estimates the signed error $y(x_i)-y_{2i}^{h/2}$ of the fine-grid value. It is an asymptotic estimate. Its accuracy should be checked against the exact solution.
