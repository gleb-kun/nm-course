## Problem Statement

The Cauchy problem for a first-order ordinary differential equation (ODE) is given by
\begin{equation}
    y'\left(x\right)=f\left(x,y\left(x\right)\right),\quad y\left(x_0\right)=y_0.
    \tag{1}
\end{equation}
We assume that this problem has a unique solution on the interval $[x_0, b]$. 
Let us write (1) in its integral form:
\begin{equation}
    y(x) = y_0 + \int_{x_0}^{x} f(t, y(t))\,dt.\tag{2}
\end{equation}

Numerical methods for solving this equation involve the approximate computation 
of the values of the hypothetical solution $y(x)$ at points 
$x_1, x_2, \ldots, x_n \in [x_0, b]$.
For simplicity, we will assume that the points (nodes) 
$x_1, \ldots, x_n$ are uniformly distributed, that is,
\begin{equation}
    x_i = x_0 + ih, \quad \text{where } h = \frac{b - x_0}{n}.\tag{3}
\end{equation}

In all the considered methods for solving problem (1), 
the values at the nodes are constructed sequentially, meaning that we assume 
the values $y_1 \approx y(x_1), \ldots, y_m \approx y(x_m)$ are already known, 
and we aim to compute $y_{m+1} \approx y(x_{m+1})$.

## Using the Taylor formula

One of the simplest approaches to solving the Cauchy problem is based on the use of the Taylor formula:
\begin{equation}
	y\left(x+h\right)=y\left(x\right)+y'\left(x\right)h+\frac{y''}{2}h^2+\ldots+\frac{y^{\left(p\right)}}{p!}h^p+R_{p+1}\left(x,h\right).\tag{4}
\end{equation}
Here $R_{p+1}\left(x,h\right)=\frac{y^{\left(p+1\right)}\left(\xi\right)}{\left(p+1\right)!}h^{p+1}$ is the remainder term of the Taylor formula, with $\xi\in\left[x,x+h\right]$.
Neglecting the remainder term, we obtain the approximate equality
\begin{equation}
	y\left(x+h\right)\approx y\left(x\right)+y'\left(x\right)h+\frac{y''}{2}h^2+\ldots+\frac{y^{\left(p\right)}}{p!}h^p.\tag{5}
\end{equation}

If the value of the solution $y\left(x\right)$ is known, then according to equality (1), the value of $y'(x)$ can also be considered known.
To compute a higher-order approximation, we modify the Taylor formula and also write expressions for the derivatives $y'(x+h)$ and $y''(x+h)$:
\begin{equation}
	y'(x+h) = y'(x) + y''(x)\,h + \frac{y^{(3)}(x)}{2!}h^2 + \ldots + \frac{y^{(p+1)}(x)}{p!}h^p + \tilde R_{p+1}(x,h),\tag{6}
\end{equation}
\begin{equation}
	y''(x+h) = y''(x) + y^{(3)}(x)\,h + \frac{y^{(4)}(x)}{2!}h^2 + \ldots + \frac{y^{(p+2)}(x)}{p!}h^p + \hat R_{p+1}(x,h).\tag{7}
\end{equation}

Let $y'(x) = f(x, y(x))$. Then all higher-order derivatives can be expressed through the partial derivatives of the function $f$.  
We introduce the notation for the total differentiation operator:
\begin{equation}
	D = \frac{d}{dx} = \frac{\partial}{\partial x} + f(x,y)\frac{\partial}{\partial y}.\tag{8}
\end{equation}
Then we obtain the recurrence relation
\begin{equation}
	y^{(k+1)}(x) = D^k f(x, y(x)).
\end{equation}

The first three derivatives can be written explicitly:
\begin{equation}
	\begin{gathered}
	y'(x) = f(x, y),\\
	y''(x) = f_x(x, y) + f_y(x, y)\,f(x, y),\\
	y^{(3)}(x) = f_{xx}(x, y) + 2 f_{xy}(x, y)\,f(x, y) + f_{yy}(x, y)\,[f(x, y)]^2 \nonumber \\
	\quad + f_y(x, y)\,\bigl(f_x(x, y) + f_y(x, y)\,f(x, y)\bigr).
	\end{gathered}\tag{9}
\end{equation}
Thus, to construct a higher-order Taylor method, it is necessary to compute the values of the partial derivatives of the function $f$ at the point $(x, y)$, find the corresponding derivatives $y^{(k)}(x)$, and substitute them into the Taylor expansion for $y(x+h)$.


## Euler method

The simplest and historically first numerical method for solving the Cauchy problem  
(1) is the Euler method.  
It can be obtained if, in the approximate equality (5), we retain only the first two terms  
(i.e., take $p = 1$). Then the formula takes the form:
$$
	y_{n+1} = y_n + h f(t_n, y_n).
	\tag{10}
$$

The geometric interpretation of one step of the Euler method consists in approximating  
the solution on the segment $[t_n, t_{n+1}]$ by the tangent line  
$y = y_n + y'(t_n)(t - t_n)$,  
drawn at the point $(t_n, y_n)$ to the integral curve passing through this point (see next Figure).

![](img/img1.png)

Thus, after performing $N$ steps, the unknown integral curve is replaced by a broken line (the Euler polygon), whose slope $k_n$ at the $n$-th segment equals the value $f(t_n, y_n)$ (see next Figure).

The Euler method is an explicit one-step method. Its local approximation error has the form
$$
\psi_n = \frac{h}{2}\, y''(\xi_n).\tag{11}
$$
