### Исходные данные и требования к реализации

Задача Коши для обыкновенного дифференциального уравнения (ОДУ) первого порядка имеет вид
$$
y'(x)=f(x,y(x)),\quad y(x_0)=y_0. \tag{1}
$$
Будем считать, что задача имеет единственное решение на $[x_0,b]$. Для простоты используйте равноотстоящие узлы $x_i=x_0+ih$, $i=0,\ldots,N$, где $h=(b-x_0)/N$. Начиная с $y_0$, последовательно вычисляйте значения $y_i\approx y(x_i)$.

Выберите один из следующих вариантов. Во всех обязательных экспериментах сохраняйте отрезок $[x_0,b]=[0,1]$.

1. $y'=1-y^2$, $y(0)=0$.
2. $y'=-y+x$, $y(0)=1$.
3. $y'=-2y+y^2$, $y(0)=1$.

Начните с $N=10$ ($h=0.1$). Для исследования сходимости возьмите $N=10,20,40,80$. Вычисляйте узлы по их целочисленным индексам и сравнивайте методы в одних и тех же узлах. Реализуйте методы самостоятельно. Передавайте правую часть, начальные данные и параметры сетки как аргументы. Метод Тейлора может также принимать частные производные $f$. Точное решение используйте для проверки результатов.

### Краткая справка

Запишем (1) в интегральной форме
$$
y(x)=y_0+\int_{x_0}^{x}f(t,y(t))\,dt. \tag{2}
$$

Один из простейших подходов к решению задачи Коши основан на формуле Тейлора. Если $y\in C^{p+1}$ на отрезке шага, то
$$
\begin{aligned}
y(x+h)&=y(x)+\sum_{j=1}^p\frac{y^{(j)}(x)}{j!}h^j+R_{p+1}(x,h),\\
R_{p+1}(x,h)&=\frac{y^{(p+1)}(\xi)}{(p+1)!}h^{p+1},\quad \xi\in(x,x+h).
\end{aligned}\tag{3}
$$
Старшие производные вдоль решения можно записать в виде $y^{(k+1)}(x)=(D^k f)(x,y(x))$, где $D=\partial_x+f(x,y)\partial_y$. В частности,
$$
y''(x)=f_x(x,y(x))+f_y(x,y(x))f(x,y(x)). \tag{4}
$$
Отбрасывая остаточный член в (3) и вычисляя слагаемые до $h^2$ включительно в текущем приближении, получаем метод Тейлора второго порядка.

Метод Эйлера получается, если оставить только слагаемые до $h$ включительно. Геометрически один шаг Эйлера идет по касательной к интегральной кривой, проходящей через $(x_i,y_i)$. В усовершенствованном методе Эйлера (явном методе средней точки) значение в середине шага приближают половинным шагом Эйлера, а наклон в этой середине используют для полного шага. Формулы и рисунки к методу Эйлера приведены в [лекции 07.2](../../lectures/07.2/lecture07_2_rus.ipynb).

#### Погрешность и сгущение сетки

Пусть $y_i^h$ — численное значение в точке $x_i$ на сетке с шагом $h$. Обозначим максимальную абсолютную погрешность на этой сетке через $E(h)=\max_i |y(x_i)-y_i^h|$.

В общем узле сеток с шагами $h$ и $h/2$ используйте оценку Рунге и экстраполяцию Ричардсона
$$
R_i=\frac{y_{2i}^{h/2}-y_i^h}{2^p-1},\quad
y_i^{\mathrm{ext}}=y_{2i}^{h/2}+R_i, \tag{5}
$$
где $p$ — порядок глобальной точности метода. Если в общем узле главный член погрешности со знаком пропорционален $h^p$, то $R_i$ оценивает погрешность со знаком $y(x_i)-y_{2i}^{h/2}$ на мелкой сетке. Это асимптотическая оценка. Ее точность нужно проверить по точному решению.
